# 3D Gaussian Wave Packet Visualization

This project visualizes a 3D Gaussian Wave Packet using Matplotlib and Isometric projection.

## Features
- **Isometric 3D View**: Time (t), Real (Re), and Imaginary (Im) axes.
- **Mirrored Projections**: Dynamic traces on three planes (XY, XZ, YZ) positioned on the 'far side' of the view.
- **Projected Phasors**: Arrows showing the component-wise projection of the wave's amplitude.
- **Video Animation**: Smooth MP4 video animation of the wave packet's evolution.

## How to Run
1. Open this notebook in Google Colab.
2. Run all cells.
3. The animation will be generated and displayed in the notebook.
4. You can download the generated `wave_packet_animation.mp4` file.

## Libraries Used
- `numpy`
- `matplotlib`
- `IPython.display`

## Physics and Simulation
The Gaussian wave packet is defined as:

$$f(t) = e^{-\gamma(t-t_0)^2} \cdot e^{i\omega t}$$

This simulation uses an isometric projection to map the complex components into a 3D visual space.

```python
"""
Gaussian Wave Packet — 3D Isometric Video Animation
f(t) = exp(-γ(t-t₀)²) · exp(iωt)
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.lines import Line2D
from matplotlib.collections import LineCollection
import matplotlib.colors as mc
from IPython.display import HTML

# ── Isometric projection ──────────────────────────────────────────────────────
COS30 = np.cos(np.pi / 6)
SIN30 = np.sin(np.pi / 6)

def iso(wx, wy, wz):
    sx =  (wx - wz) * COS30
    sy = -(wy)      + (wx + wz) * SIN30
    return sx, sy

def iso_arr(wx, wy, wz):
    sx =  (wx - wz) * COS30
    sy = -(wy)      + (wx + wz) * SIN30
    return sx, sy

# ── Physics ───────────────────────────────────────────────────────────────────
def wave(t, t0, gamma, omega):
    env = np.exp(-gamma * (t - t0) ** 2)
    return env * np.cos(omega * t), env * np.sin(omega * t), env

# ── Configuration ─────────────────────────────────────────────────────────────
TX = 3.2
AMP = 1.25
Z_WALL = AMP * 1.75
Y_WALL = AMP * 1.75
X_WALL = TX * 1.35
N_CURVE = 160
HIST = 80
T_VALS = np.linspace(-TX, TX, N_CURVE)

C_GREEN, C_BLUE, C_AMBER, C_PURPLE, C_YELLOW, C_AX = \
    '#14A05A', '#1482C8', '#C8640A', '#6432C8', '#C89600', '#666666'

def face_quad(ax, corners_world, color):
    xs, ys = zip(*[iso(wx,wy,wz) for wx,wy,wz in corners_world])
    ax.fill(list(xs)+[xs[0]], list(ys)+[ys[0]], color=color, alpha=0.06)
    ax.plot(list(xs)+[xs[0]], list(ys)+[ys[0]], color=color, alpha=0.25, lw=0.7)

def grid_line(ax, p0, p1, color):
    sx0,sy0 = iso(*p0); sx1,sy1 = iso(*p1)
    ax.plot([sx0,sx1],[sy0,sy1], color=color, alpha=0.18, lw=0.5, linestyle='--', dashes=(4,4))

# ── Animation Class ───────────────────────────────────────────────────────────
class VideoApp:
    def __init__(self):
        self.gamma, self.omega = 0.6, 5.0
        self.hist_blue, self.hist_amber, self.hist_purple = [], [], []
        self.fig, self.ax = plt.subplots(figsize=(10, 8), dpi=100)
        self.fig.patch.set_facecolor('#0E0E12')
        self.ax.set_facecolor('#0E0E12')
        self.ax.set_aspect('equal')
        self.ax.axis('off')
        self._init_artists()

    def _init_artists(self):
        self.lc_helix = LineCollection([], zorder=6)
        self.ax.add_collection(self.lc_helix)
        self.lc_blue = LineCollection([], zorder=4)
        self.lc_amber = LineCollection([], zorder=4)
        self.lc_purple = LineCollection([], zorder=4)
        for lc in (self.lc_blue, self.lc_amber, self.lc_purple): self.ax.add_collection(lc)

        self.line_blue_s,  = self.ax.plot([],[],color=C_BLUE,  alpha=0.2, lw=1, zorder=3)
        self.line_amber_s, = self.ax.plot([],[],color=C_AMBER, alpha=0.2, lw=1, zorder=3)
        self.line_purple_s, = self.ax.plot([],[],color=C_PURPLE, alpha=0.15, lw=0.8, ls='--', zorder=3)
        self.conn_b, = self.ax.plot([],[],color=C_BLUE, alpha=0.4, lw=0.8, ls='--', zorder=5)
        self.conn_a, = self.ax.plot([],[],color=C_AMBER, alpha=0.4, lw=0.8, ls='--', zorder=5)
        self.conn_p, = self.ax.plot([],[],color=C_PURPLE, alpha=0.4, lw=0.8, ls='--', zorder=5)

        self.arr_main = self.ax.annotate('', xy=(0,0), xytext=(0,0), arrowprops=dict(arrowstyle='->', color=C_YELLOW, lw=2))
        self.readout = self.fig.text(0.02, 0.05, '', color='white', family='monospace', fontsize=9)

    def update(self, frame):
        if frame == 0:
            face_quad(self.ax, [(-TX,-AMP,-Z_WALL),(TX,-AMP,-Z_WALL),(TX,AMP,-Z_WALL),(-TX,AMP,-Z_WALL)], C_BLUE)
            face_quad(self.ax, [(-TX,-Y_WALL,-AMP),(TX,-Y_WALL,-AMP),(TX,-Y_WALL,AMP),(-TX,-Y_WALL,AMP)], C_AMBER)
            face_quad(self.ax, [(-X_WALL,-AMP,-AMP),(-X_WALL,AMP,-AMP),(-X_WALL,AMP,AMP),(-X_WALL,-AMP,AMP)], C_PURPLE)

        t0 = TX * np.sin(frame * 0.05)
        re_t, im_t, env_t = wave(T_VALS, t0, self.gamma, self.omega)
        re0, im0, env0 = wave(t0, t0, self.gamma, self.omega)

        # Helix
        hx, hy = iso_arr(T_VALS, re_t, im_t)
        segs = [((hx[i-1],hy[i-1]),(hx[i],hy[i])) for i in range(1, N_CURVE)]
        self.lc_helix.set_segments(segs)
        self.lc_helix.set_color([(0.08, 0.78, 0.39, 0.3 + 0.7*env_t[i]) for i in range(len(segs))])

        # Projections
        bx, by = iso(t0, re0, -Z_WALL); amx, amy = iso(t0, -Y_WALL, im0); px, py = iso(-X_WALL, re0, im0)
        tipx, tipy = iso(t0, re0, im0); orx, ory = iso(t0, 0, 0)

        self.conn_b.set_data([tipx, bx], [tipy, by])
        self.conn_a.set_data([tipx, amx], [tipy, amy])
        self.conn_p.set_data([tipx, px], [tipy, py])
        self.arr_main.set_position((orx, ory)); self.arr_main.xy = (tipx, tipy)

        self.readout.set_text(f't0: {t0:+.2f} | Envelope: {env0:.2f}')
        return self.lc_helix, self.conn_b, self.conn_a, self.conn_p, self.arr_main

app = VideoApp()
anim = FuncAnimation(app.fig, app.update, frames=126, interval=40, blit=True)
plt.close()
HTML(anim.to_jshtml())
```
