# 3D Visualization of Electromagnetic Radiation: Near-Field Energy Flow

This notebook generates a high-fidelity 3D animation of the **Poynting vector field** surrounding a Hertzian dipole antenna. It specifically illustrates the "pinch-off" effect, where energy loops detach from the antenna to become self-propagating radio waves.

### Key Features:
- **Physics-Based Simulation:** Uses spherical coordinate derivations for the near-field Poynting vector.
- **Dynamic 3D Visualization:** Employs Matplotlib `quiver` plots with a vibrant 'magma' colormap.
- **Interactive Animation:** Features an oscillating camera perspective to showcase the toroidal energy shells.
- **Exportable:** Includes functionality to encode and download the animation as an MP4 file.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from mpl_toolkits.mplot3d import Axes3D

"""
Calculates and animates the 3D Poynting vector field for a Hertzian dipole.

The simulation focuses on the near-field region where the transition from
bound energy to radiating energy (detachment) is most visible.
"""

# Higher resolution for more fluid energy lines
num_points = 15
x = np.linspace(-2.2, 2.2, num_points)
y = np.linspace(-2.2, 2.2, num_points)
z = np.linspace(-2.2, 2.2, num_points)
X, Y, Z = np.meshgrid(x, y, z)

# Polar coordinates
R = np.sqrt(X**2 + Y**2 + Z**2)
Theta = np.arccos(Z / (R + 1e-9))
Phi = np.arctan2(Y, X)

def get_poynting_3d(t):
    """
    Computes the U, V, W components of the Poynting vector at time t.
    Based on the cross product of the Electric (E) and Magnetic (H) fields
    for an oscillating dipole source.
    """
    k = 2 * np.pi
    omega = 2 * np.pi
    phase = omega * t - k * R
    # Hertzian dipole physics
    Sr = (np.sin(Theta)**2 / (R**2 + 0.1)) * (np.cos(phase) + (1/(k*R + 0.1)) * np.sin(phase))
    Stheta = (np.sin(Theta) * np.cos(Theta) / (R**3 + 0.1)) * np.sin(phase)
    U = Sr * np.sin(Theta) * np.cos(Phi) + Stheta * np.cos(Theta) * np.cos(Phi)
    V = Sr * np.sin(Theta) * np.sin(Phi) + Stheta * np.cos(Theta) * np.sin(Phi)
    W = Sr * np.cos(Theta) - Stheta * np.sin(Theta)
    mask = R < 0.4
    U[mask] = 0; V[mask] = 0; W[mask] = 0
    return U, V, W

fig = plt.figure(figsize=(12, 10), facecolor='black')
ax = fig.add_subplot(111, projection='3d', facecolor='black')

# Energy scale legend
sm = plt.cm.ScalarMappable(cmap='magma', norm=plt.Normalize(vmin=0, vmax=1))
cbar = fig.colorbar(sm, ax=ax, shrink=0.5, aspect=10, pad=0.05)
cbar.set_label('Energy Intensity', color='white', size=12)
cbar.ax.yaxis.set_tick_params(color='white', labelcolor='white')

def update(frame):
    ax.clear()
    ax.set_facecolor('black')
    ax.axis('off')

    # Draw the physical source (Antenna)
    az = np.linspace(-0.6, 0.6, 12)
    atheta = np.linspace(0, 2*np.pi, 12)
    az, atheta = np.meshgrid(az, atheta)
    ax_cyl = 0.06 * np.cos(atheta)
    ay_cyl = 0.06 * np.sin(atheta)
    ax.plot_surface(ax_cyl, ay_cyl, az, color='gold', alpha=1.0, zorder=10)

    t = frame / 15.0
    U, V, W = get_poynting_3d(t)
    mag = np.sqrt(U**2 + V**2 + W**2)
    norm_mag = mag.flatten() / (mag.max() + 1e-9)

    # Vector field rendering
    ax.quiver(X, Y, Z, U, V, W, length=0.1, colors=plt.cm.magma(norm_mag),
              alpha=0.90, linewidth=1.0, pivot='middle')

    limit = 1.5
    ax.set_xlim(-limit, limit); ax.set_ylim(-limit, limit); ax.set_zlim(-limit, limit)
    ax.set_title("Oscillation and Detachment of Electromagnetic Energy Shells",
                 color='orange', fontsize=16, fontweight='bold', pad=25)

    # Camera rotation and oscillation
    tilt = 20 + 15 * np.sin(frame * 0.1)
    ax.view_init(elev=tilt, azim=frame * 3)

anim = FuncAnimation(fig, update, frames=100, interval=60)
plt.close()
HTML(anim.to_jshtml())
```

```python
from google.colab import files

# Define the filename
video_filename = 'Hertzian_Dipole_Energy_Detachment.mp4'

# Save the animation (requires ffmpeg to be installed in the environment)
print('Encoding video... please wait.')
anim.save(video_filename, writer='ffmpeg', fps=15, dpi=150)

# Trigger download
files.download(video_filename)
print(f'Done! {video_filename} is ready for download.')
```
