# From Random Walks to Option Prices

### A 9:16 short video that turns a swarm of random particles into the Black-Scholes call price — and proves it numerically along the way.

A production-grade, parametric Python animation that tells one of the most beautiful stories in quantitative finance in under 40 seconds: how *pure randomness*, watched carefully, produces a *fair price*. One thousand particles diffuse in log-price space, their density crystallises into the Gaussian heat kernel, the cloud is mapped back into asset prices to form the risk-neutral distribution, and the discounted expected payoff lands exactly on the closed-form Black-Scholes value.

---

## The idea in one line

> **Stochastic diffusion → Heat equation → Option price.**

The same physics that describes heat spreading through a metal bar describes uncertainty spreading through a stock price. That single bridge is what makes options pricing possible, and this video makes the bridge visible.

## The concepts, properly explained

### 1. Brownian motion in log-price space
Asset prices cannot go negative and tend to move in *percentage* terms, so the natural variable is the logarithm of price, `x = ln S`. Under the risk-neutral measure, `x` follows a simple stochastic differential equation:

```
dx = (r − ½σ²) dt + σ dW
```

- `r` is the risk-free rate, `σ` the volatility, and `dW` an increment of a Wiener process (the mathematical idealisation of Brownian motion).
- The term `−½σ²` is the **Itô correction**: it is the price of working in log-space and is what makes the discounted asset a martingale.
- Each of the 1000 particles is one realisation of this SDE. Watching them collectively is watching a probability distribution come to life.

### 2. From a stochastic process to the heat equation
A remarkable fact (the Fokker-Planck / Kolmogorov forward equation) is that the *probability density* `p(x, t)` of a diffusing particle does not evolve randomly at all — it evolves deterministically. After removing the drift, the density obeys the **heat equation**:

```
∂p/∂t = (σ²/2) ∂²p/∂x²
```

This is exactly the equation Fourier wrote in 1822 to describe heat flowing through a solid. Volatility `σ²` plays the role of thermal diffusivity. Randomness at the level of a single particle becomes smooth, predictable physics at the level of the crowd.

### 3. The Gaussian heat kernel
The fundamental solution of the heat equation, starting from a point source, is the **heat kernel** — a Gaussian that widens over time:

```
p(x, t) = 1 / √(2πσ²t) · exp( −(x − μt)² / (2σ²t) )
```

Its standard deviation grows as `σ√t`, which is precisely the diffusion envelope drawn in Scene 1. This is why the particle histogram and the analytic curve converge: they are the same object seen two ways — empirical and exact.

### 4. The risk-neutral distribution
Because `S = e^x` and `x` is normally distributed, `S` is **log-normally** distributed. Mapping the Gaussian through the exponential produces the risk-neutral density `q(S)` of the price at maturity `T`. "Risk-neutral" means we have chosen the drift (`r`) for which today's fair value of any payoff is simply its discounted expectation — no risk premia to estimate.

### 5. The Black-Scholes call price
A European call pays `(S_T − K)⁺` at maturity. Its fair value today is the discounted expected payoff under the risk-neutral density:

```
C = e^{−rT} · E[(S_T − K)⁺] = S₀ N(d₁) − K e^{−rT} N(d₂)
```

with

```
d₁ = [ ln(S₀/K) + (r + ½σ²)T ] / (σ√T),     d₂ = d₁ − σ√T
```

`N(·)` is the standard normal cumulative distribution function. The video computes this two ways and shows them side by side:

- **Closed form (Black-Scholes):** `C = 7.57`
- **Monte-Carlo (1000 particles):** `C = 7.56`

The agreement is not a coincidence — it is the entire theory made visible.

````python
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Concept: Stochastic diffusion to Black-Scholes (Brownian motion -> heat kernel -> option price)
GitHub: github.com/zombimann/Mathematical-video-animations-and-visualization
"""

import os
import math
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.backends.backend_agg import FigureCanvasAgg
import imageio.v2 as imageio

# ----------------------------------------------------------------------
# PARAMETERS  (everything is parametric for easy experimentation)
# ----------------------------------------------------------------------
CFG = dict(
    # ---- finance / physics ----
    S0=100.0,      # spot asset price
    K=105.0,       # option strike
    r=0.04,        # risk-free rate
    sigma=0.20,    # volatility
    T=1.0,         # maturity (years)
    N_PARTICLES=1000,
    STEPS=240,     # time steps of the random walk
    SEED=7,
    # ---- video ----
    WIDTH=1080, HEIGHT=1920, DPI=120, FPS=30,
    CRF=26, PRESET="medium",
    OUTPUT="diffusion_to_price.mp4",
    QUICK=False,   # set True for a fast low-fps preview
)

# ----------------------------------------------------------------------
# PALETTE  (one coherent, social-media friendly theme)
# ----------------------------------------------------------------------
BG        = "#0B0E16"   # near-black navy
CARD_BG   = "#121A2B"
WHITE     = "#F4F7FB"
MUTED     = "#8A98AD"
PRIMARY   = "#24D3C4"   # level primary colour (teal)
PRIMARY_D = "#0E5A55"
GOLD      = "#FFC857"   # analytic / heat kernel
ACCENT    = "#FF7A66"   # strike / payoff
PATH_CLR  = "#7AA2FF"   # secondary highlight

CMAP = LinearSegmentedColormap.from_list("diffusion", [BG, PRIMARY_D, PRIMARY, "#EAFFFB"])

WATERMARK = "(c) Mugambi Ndwiga / @craftsandengineering"

# ----------------------------------------------------------------------
# EASING
# ----------------------------------------------------------------------
def smoothstep(a, b, x):
    if b == a:
        return float(x >= b)
    t = float(np.clip((x - a) / (b - a), 0.0, 1.0))
    return t * t * (3 - 2 * t)

def ease(p):
    return p * p * (3 - 2 * p)

# numpy 2.x renamed trapz -> trapezoid; support both
trapz = getattr(np, "trapezoid", getattr(np, "trapz", None))

# ----------------------------------------------------------------------
# MODEL  (simulate the particles once, reuse everywhere)
# ----------------------------------------------------------------------
S0, K, r, sigma, T = CFG["S0"], CFG["K"], CFG["r"], CFG["sigma"], CFG["T"]
N = 200 if CFG["QUICK"] else CFG["N_PARTICLES"]
STEPS = 60 if CFG["QUICK"] else CFG["STEPS"]
rng = np.random.default_rng(CFG["SEED"])

x0 = math.log(S0)
mu = r - 0.5 * sigma ** 2                       # risk-neutral log drift
dt = T / STEPS
incr = mu * dt + sigma * math.sqrt(dt) * rng.standard_normal((STEPS, N))
X = np.empty((STEPS + 1, N))
X[0] = x0
X[1:] = x0 + np.cumsum(incr, axis=0)
t_grid = np.linspace(0.0, T, STEPS + 1)
S_T = np.exp(X[-1])

# log-price viewport
XMIN = x0 - 4.2 * sigma * math.sqrt(T)
XMAX = x0 + 4.2 * sigma * math.sqrt(T)

# heat-map of particle density over (x, t), row-normalised for visibility
NXB = 140
xbins = np.linspace(XMIN, XMAX, NXB + 1)
HEAT = np.zeros((STEPS + 1, NXB))
for k in range(STEPS + 1):
    h, _ = np.histogram(X[k], bins=xbins)
    m = h.max()
    HEAT[k] = h / m if m > 0 else h

PATH_IDX = rng.choice(N, size=4, replace=False)
PATH_COLORS = [GOLD, ACCENT, PATH_CLR, "#B388FF"]

# price-space (risk-neutral log-normal density)
m_lnS = x0 + mu * T
sd_lnS = sigma * math.sqrt(T)
SMIN, SMAX = 40.0, 250.0
Sgrid = np.linspace(SMIN, SMAX, 600)
q = (1.0 / (Sgrid * sd_lnS * math.sqrt(2 * math.pi))
     * np.exp(-(np.log(Sgrid) - m_lnS) ** 2 / (2 * sd_lnS ** 2)))
payoff_density = np.maximum(Sgrid - K, 0.0) * q          # (S-K)^+ * q(S)

# Black-Scholes call (closed form, no scipy dependency)
def Ncdf(z):
    return 0.5 * (1.0 + math.erf(z / math.sqrt(2.0)))

d1 = (math.log(S0 / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * math.sqrt(T))
d2 = d1 - sigma * math.sqrt(T)
C_BS = S0 * Ncdf(d1) - K * math.exp(-r * T) * Ncdf(d2)
C_MC = math.exp(-r * T) * np.mean(np.maximum(S_T - K, 0.0))

# decoration data for the title swarm
SW_N = 260
rng2 = np.random.default_rng(3)
SW_BASE = np.clip(rng2.normal(0, 0.45, (SW_N, 2)), -1, 1)
SW_PH = rng2.uniform(0, 2 * math.pi, SW_N)

# ----------------------------------------------------------------------
# FIGURE / CANVAS
# ----------------------------------------------------------------------
W, H, DPI = CFG["WIDTH"], CFG["HEIGHT"], CFG["DPI"]
FIG = plt.figure(figsize=(W / DPI, H / DPI), dpi=DPI)
CANVAS = FigureCanvasAgg(FIG)

MAIN_RECT = [0.135, 0.31, 0.78, 0.55]   # left, bottom, w, h -> top = 0.86

def grab():
    CANVAS.draw()
    return np.asarray(CANVAS.buffer_rgba())[..., :3].copy()

# ----------------------------------------------------------------------
# REUSABLE LAYOUT ELEMENTS
# ----------------------------------------------------------------------
def watermark():
    FIG.text(0.035, 0.013, WATERMARK, color=MUTED, alpha=0.6,
             fontsize=15, ha="left", va="bottom")

def header(text):
    FIG.text(0.5, 0.955, text, color=WHITE, fontsize=30,
             ha="center", va="center", weight="bold")

def chain(active):
    items = [("DIFFUSION", 0.205), ("HEAT EQUATION", 0.5), ("OPTION PRICE", 0.795)]
    for i, (label, x) in enumerate(items):
        on = (i == active)
        FIG.text(x, 0.905, label, color=PRIMARY if on else MUTED,
                 fontsize=16, ha="center", va="center",
                 weight="bold" if on else "normal", alpha=1.0 if on else 0.7)
    FIG.text(0.355, 0.905, ">", color=MUTED, fontsize=15, ha="center", va="center", alpha=0.6)
    FIG.text(0.645, 0.905, ">", color=MUTED, fontsize=15, ha="center", va="center", alpha=0.6)

def card(title, lines):
    ax = FIG.add_axes([0, 0, 1, 1]); ax.axis("off")
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    box = FancyBboxPatch((0.06, 0.052), 0.88, 0.205,
                         boxstyle="round,pad=0.004,rounding_size=0.018",
                         linewidth=1.3, edgecolor=PRIMARY, facecolor=CARD_BG, alpha=0.97)
    ax.add_patch(box)
    FIG.text(0.5, 0.233, title, color=PRIMARY, fontsize=18, ha="center",
             va="center", weight="bold")
    ys = [0.185, 0.138, 0.092]
    for (txt, col, fs), y in zip(lines, ys):
        FIG.text(0.5, y, txt, color=col, fontsize=fs, ha="center", va="center")

def main_axes():
    ax = FIG.add_axes(MAIN_RECT)
    ax.set_facecolor(BG)
    for s in ax.spines.values():
        s.set_color(MUTED); s.set_alpha(0.35)
    ax.tick_params(colors=MUTED, labelsize=14)
    return ax

def axlabel(ax, xl, yl):
    ax.set_xlabel(xl, color=WHITE, fontsize=18, labelpad=6)
    ax.set_ylabel(yl, color=WHITE, fontsize=18, labelpad=6)

# ----------------------------------------------------------------------
# SCENES
# ----------------------------------------------------------------------
def scene_title(p):
    watermark()
    ax = FIG.add_axes(MAIN_RECT); ax.axis("off")
    ax.set_xlim(-1, 1); ax.set_ylim(-1, 1)
    appear = smoothstep(0.0, 0.5, p)
    spread = 0.18 + 0.55 * smoothstep(0.15, 1.0, p)
    jit = 0.035 * np.sin(SW_PH + p * 11.0)
    px = SW_BASE[:, 0] * spread * 1.7 + jit
    py = 0.62 + SW_BASE[:, 1] * spread * 0.45 + jit * 0.6   # confined to the upper band
    ax.scatter(px, py, s=20, c=PRIMARY, alpha=0.30 * appear, edgecolors="none")

    tA = smoothstep(0.18, 0.5, p)
    FIG.text(0.5, 0.63, "FROM RANDOM WALKS", color=WHITE, fontsize=40,
             ha="center", va="center", weight="bold", alpha=tA)
    FIG.text(0.5, 0.565, "TO OPTION PRICES", color=PRIMARY, fontsize=40,
             ha="center", va="center", weight="bold", alpha=tA)
    sA = smoothstep(0.45, 0.8, p)
    FIG.text(0.5, 0.485, "Diffusion  >  Heat equation  >  Black-Scholes",
             color=MUTED, fontsize=20, ha="center", va="center", alpha=sA)
    hA = smoothstep(0.05, 0.35, p)
    FIG.text(0.5, 0.30, "What is the fair price of pure randomness?",
             color=GOLD, fontsize=19, ha="center", va="center", alpha=hA, style="italic")

def scene_walk(p):
    watermark(); header("1   RANDOM WALK"); chain(0)
    ax = main_axes()
    step = max(1, int(p * STEPS))
    ax.imshow(HEAT[:step + 1], extent=[XMIN, XMAX, 0, t_grid[step]],
              origin="lower", aspect="auto", cmap=CMAP, vmin=0, vmax=1,
              interpolation="bilinear")
    for c, idx in zip(PATH_COLORS, PATH_IDX):
        ax.plot(X[:step + 1, idx], t_grid[:step + 1], color=c, lw=2.0, alpha=0.9)
    tt = t_grid[1:step + 1]
    centre = x0 + mu * tt
    band = sigma * np.sqrt(tt)
    ax.plot(centre + band, tt, "--", color=WHITE, lw=1.5, alpha=0.55)
    ax.plot(centre - band, tt, "--", color=WHITE, lw=1.5, alpha=0.55)
    ax.set_xlim(XMIN, XMAX); ax.set_ylim(0, T)
    axlabel(ax, "log-price   x = ln S", "time   t   (years)")
    FIG.text(0.155, 0.835, "envelope  +/- sigma sqrt(t)", color=WHITE,
             fontsize=14, ha="left", va="center", alpha=0.7)
    card("Brownian motion in log-price",
         [(r"$dx=\left(r-\frac{1}{2}\sigma^{2}\right)dt+\sigma\,dW$", WHITE, 23),
          ("1000 particles take random steps;", MUTED, 16),
          ("the cloud widens with the square root of time.", MUTED, 16)])

def scene_kernel(p):
    watermark(); header("2   HEAT KERNEL"); chain(1)
    ax = main_axes()
    t0 = 0.06 * T
    tcur = t0 + (T - t0) * ease(p)
    step = int(round(tcur / T * STEPS))
    hist, edges = np.histogram(X[step], bins=64, range=(XMIN, XMAX), density=True)
    centres = 0.5 * (edges[:-1] + edges[1:])
    ax.fill_between(centres, hist, step="mid", color=PRIMARY, alpha=0.32)
    ax.plot(centres, hist, drawstyle="steps-mid", color=PRIMARY, lw=1.6, alpha=0.85)
    xs = np.linspace(XMIN, XMAX, 400)
    mean = x0 + mu * tcur
    sd = sigma * math.sqrt(tcur)
    g = 1.0 / (sd * math.sqrt(2 * math.pi)) * np.exp(-(xs - mean) ** 2 / (2 * sd ** 2))
    ax.plot(xs, g, color=GOLD, lw=3.0)
    ypeak = 1.0 / (sigma * math.sqrt(t0) * math.sqrt(2 * math.pi)) * 1.12
    ax.set_xlim(XMIN, XMAX); ax.set_ylim(0, ypeak)
    axlabel(ax, "log-price   x = ln S", "probability density   p(x,t)")
    FIG.text(0.165, 0.835, "particles", color=PRIMARY, fontsize=15, ha="left",
             va="center", weight="bold")
    FIG.text(0.165, 0.808, "heat kernel", color=GOLD, fontsize=15, ha="left",
             va="center", weight="bold")
    FIG.text(0.88, 0.835, "t = %.2f yr" % tcur, color=WHITE, fontsize=15,
             ha="right", va="center")
    card("Density solves the heat equation",
         [(r"$\frac{\partial p}{\partial t}=\frac{\sigma^{2}}{2}\,\frac{\partial^{2}p}{\partial x^{2}}$", WHITE, 22),
          (r"$p(x,t)=\frac{1}{\sqrt{2\pi\sigma^{2}t}}\;e^{-(x-\mu t)^{2}/(2\sigma^{2}t)}$", GOLD, 21),
          ("The spreading cloud is the Gaussian heat kernel.", MUTED, 16)])

def scene_price(p):
    watermark(); header("3   PRICE SPACE"); chain(2)
    ax = main_axes()
    draw_frac = smoothstep(0.0, 0.6, p)
    ncut = max(2, int(draw_frac * len(Sgrid)))
    ax.plot(Sgrid[:ncut], q[:ncut], color=PRIMARY, lw=3.0)
    ax.fill_between(Sgrid[:ncut], q[:ncut], color=PRIMARY, alpha=0.16)
    kA = smoothstep(0.5, 0.7, p)
    if kA > 0:
        ax.axvline(K, color=ACCENT, lw=2.2, alpha=kA)
        FIG.text(0.5, 0.83, "strike  K = %.0f" % K, color=ACCENT, fontsize=16,
                 ha="center", va="center", alpha=kA, weight="bold")
    shA = smoothstep(0.72, 1.0, p)
    if shA > 0:
        mask = Sgrid >= K
        ax.fill_between(Sgrid[mask], q[mask], color=ACCENT, alpha=0.30 * shA)
    ax.set_xlim(SMIN, SMAX); ax.set_ylim(0, q.max() * 1.18)
    axlabel(ax, "asset price   S", "risk-neutral density   q(S)")
    card("Map back to price:  S = e^x",
         [(r"$S=e^{x}\;\Rightarrow\;q(S)$  is log-normal", WHITE, 22),
          ("The Gaussian in x becomes the risk-neutral", MUTED, 16),
          ("distribution of the price at maturity.", MUTED, 16)])

def scene_bs(p):
    watermark(); header("4   OPTION PRICE"); chain(2)
    ax = main_axes()
    ax.plot(Sgrid, q * (payoff_density.max() / max(q.max(), 1e-9)) * 0.35,
            color=PRIMARY, lw=1.3, alpha=0.35)   # faint density for context
    ax.plot(Sgrid, payoff_density, color=ACCENT, lw=2.6, alpha=0.9)
    Sc = K + (SMAX - K) * ease(p)
    mask = (Sgrid >= K) & (Sgrid <= Sc)
    ax.fill_between(Sgrid[mask], payoff_density[mask], color=ACCENT, alpha=0.40)
    ax.axvline(K, color=ACCENT, lw=1.6, alpha=0.5)
    if mask.sum() > 1:
        cur = math.exp(-r * T) * trapz(payoff_density[mask], Sgrid[mask])
    else:
        cur = 0.0
    ax.set_xlim(SMIN, SMAX); ax.set_ylim(0, payoff_density.max() * 1.22)
    axlabel(ax, "asset price   S", "expected payoff density  (S-K)q(S)")
    FIG.text(0.5, 0.80, "C = $%.2f" % cur, color=GOLD, fontsize=34,
             ha="center", va="center", weight="bold")
    card("Discounted expected payoff = call price",
         [(r"$C=e^{-rT}\,\mathrm{E}\,[(S_T-K)^{+}]=S_0N(d_1)-Ke^{-rT}N(d_2)$", WHITE, 18),
          ("Black-Scholes   C = $%.2f" % C_BS, GOLD, 17),
          ("Monte-Carlo (1000 paths)   C = $%.2f" % C_MC, PRIMARY, 17)])

def scene_closing(p):
    FIG.patch.set_facecolor(PRIMARY)
    a = smoothstep(0.0, 0.4, p)
    FIG.text(0.5, 0.57, "Made by Mugambi Ndwiga", color="white", fontsize=34,
             ha="center", va="center", weight="bold", alpha=a)
    FIG.text(0.5, 0.5, "@craftsandengineering", color="white", fontsize=26,
             ha="center", va="center", alpha=a)

SCENES = [
    ("title",   scene_title,   3.5),
    ("walk",    scene_walk,    8.0),
    ("kernel",  scene_kernel,  8.0),
    ("price",   scene_price,   7.0),
    ("bs",      scene_bs,      9.0),
    ("closing", scene_closing, 1.8),
]
if CFG["QUICK"]:
    SCENES = [(n, f, 1.0) for n, f, _ in SCENES]

# ----------------------------------------------------------------------
# RENDER
# ----------------------------------------------------------------------
def render():
    fps = 10 if CFG["QUICK"] else CFG["FPS"]
    out = CFG["OUTPUT"]
    writer = imageio.get_writer(
        out, fps=fps, codec="libx264", macro_block_size=8,
        pixelformat="yuv420p",
        ffmpeg_params=["-crf", str(CFG["CRF"]), "-preset", CFG["PRESET"]],
    )
    total = 0
    for name, fn, dur in SCENES:
        nf = max(2, int(dur * fps))
        for i in range(nf):
            pf = i / (nf - 1)
            FIG.clf()
            FIG.patch.set_facecolor(PRIMARY if name == "closing" else BG)
            fn(pf)
            writer.append_data(grab())
            total += 1
    writer.close()
    size_mb = os.path.getsize(out) / 1e6
    print("frames=%d  size=%.2f MB  ->  %s" % (total, size_mb, out))
    print("Black-Scholes C = %.4f   Monte-Carlo C = %.4f" % (C_BS, C_MC))
    return out, size_mb

OUT, SIZE_MB = render()
````
