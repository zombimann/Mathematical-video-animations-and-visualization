# Butterfly Curve Animation

## Project Overview
This project generates a high-fidelity, mesmerizing video animation of the **Butterfly Curve**, a transcendental plane curve first discovered by Temple H. Fay. The visualization is designed for the general public, blending mathematical precision with cinematic aesthetics.

### The Mathematics
The curve is generated in the polar plane using the equation:

$$r = e^{\sin(t)} - 2\cos(4t) + \sin^5\left(\frac{t - \pi/2}{12}\right)$$

Where:
- **$e^{\sin(t)}$**: Provides the organic, exponential growth of the wings.
- **$2\cos(4t)$**: Creates the four-winged harmonic structure.
- **$\sin^5(\dots)$**: Modulates the curve over a long period ($24\pi$), adding intricate detail and asymmetry.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

"""
Butterfly Curve Animation (Cartesian Projection)
Author: Mugambi Ndwiga
Instagram: @craftsandengineering

Equation: r = exp(sin(t)) - 2*cos(4t) + sin((t - pi/2)/12)^5
"""

# Configuration
fps = 30
duration_sec = 30
num_frames = fps * duration_sec  # Exactly 900 frames
t_max = 24 * np.pi
t_vals = np.linspace(0, t_max, 10000)

def butterfly_r(theta):
    return np.exp(np.sin(theta)) - 2 * np.cos(4 * theta) + np.sin((theta - np.pi/2)/12)**5

r_vals = butterfly_r(t_vals)

# Convert Polar to Cartesian
x_vals = r_vals * np.cos(t_vals)
y_vals = r_vals * np.sin(t_vals)

# Figure Setup
plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(12, 14))
fig.patch.set_facecolor('#1a1a1a')
ax.set_facecolor('#1a1a1a')

# Layout styling
ax.grid(False)
for spine in ax.spines.values():
    spine.set_visible(False)

ax.axhline(0, color='#444444', linewidth=1.5, zorder=1)
ax.axvline(0, color='#444444', linewidth=1.5, zorder=1)
ax.set_xlabel("Abscissa (x)", color='gray', fontsize=12, loc='right')
ax.set_ylabel("Ordinate (y)", color='gray', fontsize=12, loc='top')
ax.tick_params(axis='both', colors='#555555', labelsize=10)

fig.suptitle("The Transcendental Butterfly Curve", fontsize=22, color='#00FFCC', y=0.95, fontweight='bold')
ax.set_title("Harmonic Oscillation & Exponential Growth (Cartesian)", color='gray', pad=20)

math_str = r"$r = e^{\sin(t)} - 2\cos(4t) + \sin^5\left(\frac{t - \pi/2}{12}\right)$"
fig.text(0.5, 0.12, math_str, color='white', fontsize=18, ha='center', style='italic')

watermark = "Author: Mugambi Ndwiga\n@craftsandengineering"
fig.text(0.85, 0.05, watermark, color='gray', fontsize=10, ha='right', alpha=0.7)

line, = ax.plot([], [], lw=2, alpha=0.9, zorder=3)
point, = ax.plot([], [], 'o', color='white', markersize=5, zorder=4)
telemetry_text = ax.text(0.05, 0.95, '', color='#00FFCC', fontsize=12, transform=ax.transAxes)

def init():
    line.set_data([], [])
    point.set_data([], [])
    telemetry_text.set_text('')
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.set_aspect('equal')
    return line, point, telemetry_text

def update(frame):
    idx = int((frame / num_frames) * len(t_vals))
    if idx == 0: idx = 1
    if idx >= len(t_vals): idx = len(t_vals) - 1

    line.set_data(x_vals[:idx], y_vals[:idx])
    point.set_data([x_vals[idx-1]], [y_vals[idx-1]])
    line.set_color(plt.cm.plasma(frame / num_frames))
    telemetry_text.set_text(f"θ: {t_vals[idx-1]:.2f} rad\nr: {r_vals[idx-1]:.3f}\nx: {x_vals[idx-1]:.2f}, y: {y_vals[idx-1]:.2f}")
    return line, point, telemetry_text

# Explicitly setting interval for exactly 30 seconds: 1000ms / 30fps = 33.33ms
ani = FuncAnimation(fig, update, frames=num_frames, init_func=init, blit=True, interval=1000/fps)

plt.close()
HTML(ani.to_html5_video())
```
