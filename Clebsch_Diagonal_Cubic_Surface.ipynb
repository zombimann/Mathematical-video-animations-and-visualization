# Clebsch Diagonal Cubic Surface Visualization

This notebook explores the visualization of the **Clebsch Diagonal Cubic Surface**, a fundamental object in algebraic geometry. It contains two distinct aesthetic implementations using the `Manim` mathematical animation engine:
1. **Version 1 (Parchment):** A classic, hand-drawn look using warm tones and calligraphic styling.
2. **Version 2 (Blueprint):** A modern, technical schematic look using deep blues and high-contrast lines.

## Background

The Clebsch surface is a smooth cubic surface discovered by Alfred Clebsch in 1871. It is uniquely significant due to its high degree of symmetry and its real-valued geometric properties.

### Key Concepts:
*   **The 27 Lines:** Every smooth cubic surface over complex numbers contains exactly 27 lines. The Clebsch surface is special because all 27 of these lines are **real**, making it visually and geometrically intuitive.
*   **Eckardt Points:** These are points where three of the 27 lines intersect. While a general cubic surface might have none, the Clebsch surface contains exactly **10 Eckardt points**.
*   **Symmetry ($S_5$):** The surface possesses the symmetry of the symmetric group $S_5$, the highest possible symmetry for a smooth cubic surface.
*   **The Equation:** In its diagonal form (homogeneous coordinates), it can be expressed as:
    $$\sum_{i=0}^4 x_i^3 = 0, \quad \sum_{i=0}^4 x_i = 0$$

## Project Structure

*   **Environment Setup:** Installation of system dependencies for LaTeX rendering and Manim (Cario, Pango, ffmpeg).
*   **Manim Logic:** Custom scenes leveraging `ThreeDScene` to render the algebraic shell and HUD elements.
*   **Animation:** Use of ambient camera rotations and creation animations to showcase the surface's geometry.

```python
%%manim -v WARNING -r 540,960 --fps 30 ClebschParchment

from manim import *
import numpy as np

config.frame_width  = 4.5
config.frame_height = 8.0

# ╔══════════════════════════════════════════════════════════════════╗
# ║                    ADJUST THESE PARAMETERS                       ║
# ╠══════════════════════════════════════════════════════════════════╣
# ║  SURFACE_Z    : vertical position  (higher = up on screen)       ║
# ║                 safe to change — moves along rotation axis,      ║
# ║                 so the surface stays centred during spin          ║
# ║  SURFACE_SCALE: overall size of the 3D geometry                  ║
# ║  IRREGULARITY : hand-drawn intensity  0=none  3=visible  6=max   ║
# ╚══════════════════════════════════════════════════════════════════╝

SURFACE_Z     =  1.5    # default: centred between header and panel
SURFACE_SCALE =  1.8    # default: fills ~85 % of portrait width
IRREGULARITY  =  2.0    # default: clearly visible without being extreme


class ClebschParchment(ThreeDScene):
    def construct(self):

        # ── PALETTE ───────────────────────────────────────────────────────────
        PAPER       = "#F4EFE6"
        PANEL_PAPER = "#EAE0D0"
        INK         = "#3A2A1D"
        MUTED_INK   = "#735C4A"
        ACCENT      = "#91694A"

        self.camera.background_color = PAPER
        self.set_camera_orientation(phi=75 * DEGREES, theta=-45 * DEGREES, zoom=0.62)

        CF = "TeX Gyre Chorus"   # Zapf Chancery calligraphic — handwritten feel
        fw = 4.5
        IR = IRREGULARITY

        # ── HEADER (top) ──────────────────────────────────────────────────────
        title    = Text("Clebsch Diagonal Cubic Surface",
                        font=CF, color=INK, font_size=28)
        subtitle = Text("One surface. 27 lines. 10 triple points.",
                        font=CF, color=MUTED_INK, font_size=16)
        subtitle.rotate(0.010 * IR)

        h_line = DashedLine(LEFT, RIGHT, color=ACCENT,
                            stroke_width=1.0 + 0.2 * IR,
                            dash_length=0.06 + 0.015 * IR,
                            dashed_ratio=0.55
                            ).scale_to_fit_width(fw * 0.88)

        header = VGroup(title, subtitle, h_line).arrange(DOWN, buff=0.18)
        header.to_edge(UP, buff=0.3).set_x(0)
        if header.width > fw * 0.9:
            header.scale_to_fit_width(fw * 0.9)

        # ── DASHBOARD (bottom) ────────────────────────────────────────────────
        panel_h  = 2.8
        panel_bg = RoundedRectangle(
            width=fw * 0.92, height=panel_h, corner_radius=0.18,
            fill_color=PANEL_PAPER, fill_opacity=0.92,
            stroke_color=INK, stroke_width=1.2)
        panel_bg.to_edge(DOWN, buff=0.3).set_x(0)
        panel_bg.rotate(0.008 * IR)

        math_label = Text("Clebsch Cubic Affine Equation",
                          font=CF, font_size=12, color=MUTED_INK)
        math_label.rotate(-0.010 * IR)

        math_eq = MathTex(r"x^3+y^3+z^3+1 \\ =(x+y+z+1)^3",
                          color=INK).scale(0.55)

        card_div = DashedLine(LEFT, RIGHT, color=MUTED_INK,
                              stroke_width=0.7 + 0.15 * IR,
                              dash_length=0.06 + 0.015 * IR,
                              dashed_ratio=0.5
                              ).scale_to_fit_width(fw * 0.80)

        s1 = VGroup(
            Text("27",  font=CF, font_size=36, color=INK),
            Text("Real Lines",     font=CF, font_size=11, color=MUTED_INK)
        ).arrange(DOWN, buff=0.05)
        s1.rotate( 0.010 * IR).shift(UP   * 0.013 * IR)

        s2 = VGroup(
            Text("10",  font=CF, font_size=36, color=INK),
            Text("Eckardt Points", font=CF, font_size=11, color=ACCENT)
        ).arrange(DOWN, buff=0.05)
        s2.rotate(-0.007 * IR).shift(DOWN * 0.017 * IR)

        s3 = VGroup(
            Text("S\u2085", font=CF, font_size=36, color=INK),
            Text("Symmetry",       font=CF, font_size=11, color=ACCENT)
        ).arrange(DOWN, buff=0.05)
        s3.rotate( 0.006 * IR).shift(UP   * 0.007 * IR)

        stats_row = VGroup(s1, s2, s3).arrange(RIGHT, buff=0.55)
        stats_row.shift(RIGHT * 0.02 * IR)

        panel_content = VGroup(math_label, math_eq, card_div,
                               stats_row).arrange(DOWN, buff=0.22)
        panel_content.move_to(panel_bg.get_center())

        self.add_fixed_in_frame_mobjects(
            header,
            panel_bg, math_label, math_eq, card_div, s1, s2, s3)

        # ── 3D SURFACE ────────────────────────────────────────────────────────
        def clebsch_shell(u, v):
            r = 1.35 + 0.35 * np.cos(3 * u) * np.sin(v)
            x = r * np.sin(v) * np.cos(u)
            y = r * np.sin(v) * np.sin(u)
            z = 1.1 * np.cos(v) + 0.2 * np.sin(3 * u)
            return np.array([x, y, z])

        mesh_surface = Surface(
            clebsch_shell,
            u_range=[0, 2 * PI], v_range=[0.1, PI - 0.1],
            resolution=(28, 28), should_make_jagged=False)
        mesh_surface.set_style(
            fill_color=PAPER, fill_opacity=0.65,
            stroke_color=MUTED_INK, stroke_width=0.8)

        mesh_surface.scale(SURFACE_SCALE)
        # ── CRITICAL: shift along OUT = world Z = rotation axis ──────────────
        # begin_ambient_camera_rotation orbits around world Z (OUT direction).
        # Any shift in X or Y would make the surface orbit off-centre.
        # Shifting along OUT (Z) is immune to the rotation — no drift ever.
        mesh_surface.shift(OUT * SURFACE_Z)

        # ── ANIMATION (~30 s at 30 fps) ───────────────────────────────────────
        self.play(FadeIn(header, shift=DOWN * 0.3), run_time=1.0)
        self.play(
            FadeIn(panel_bg, shift=UP * 0.3),
            FadeIn(math_label), FadeIn(math_eq),
            run_time=1.0)
        self.wait(0.5)

        self.play(Create(mesh_surface), run_time=4.0)

        self.begin_ambient_camera_rotation(rate=0.04)
        self.wait(1.5)

        self.play(
            AnimationGroup(
                Create(card_div),
                GrowFromCenter(s1), GrowFromCenter(s2), GrowFromCenter(s3),
                lag_ratio=0.2),
            run_time=2.5)

        self.wait(18.0)
        self.stop_ambient_camera_rotation()
```
