# Block Diagram Algebra: Parchment Calligraphy Edition

## Project Overview
This notebook contains a comprehensive implementation of Block Diagram Algebra simplification rules using the Manim animation engine. The project visualizes the mathematical reduction of complex control systems into their equivalent transfer functions.

### Visual Aesthetic
The animations utilize a specialized parchment aesthetic designed for high-resolution vertical video (9:16 portrait). Key features include:
* **Procedural Background**: Aged parchment texture with random age spots, paper grain, and vignette effects.
* **Calligraphic Typography**: Use of TeX Gyre Chorus for descriptive text and Mathpazo Palatino for LaTeX equations to simulate a hand-inked manuscript.
* **Ink Dynamics**: Implementation of translucent fills and 'ink-bleed' halos around blocks and junctions to mimic ink sinking into paper fibers.
* **Parametric Design**: A centralized configuration class (`ParchmentConfig`) managing all colors, geometry, and animation timings.

## Theoretical Background
Block diagram algebra is a shorthand pictorial representation of the cause-and-effect relationship between the input and output of a physical system. It is a fundamental tool in Control Systems Engineering for analyzing linear time-invariant (LTI) systems.

### Simplification Rules
The implementation covers the 10 essential rules as defined by standard texts (Ogata, Nise, Dorf & Bishop):
1.  **Cascade (Series) Connection**: Multiplication of transfer functions.
2.  **Parallel Connection**: Addition of transfer functions.
3.  **Feedback Loop**: The canonical closed-loop reduction formula.
4.  **Moving Summing Junctions**: Rules for shifting junctions ahead of or beyond blocks.
5.  **Moving Take-off Points**: Rules for shifting branch points while maintaining signal equivalence.
6.  **Summing Junction Properties**: Commutative and associative laws for multiple inputs.

### Complex Examples
Two real-world scenarios are modeled to demonstrate the application of these rules in sequence:
*  **Armature-Controlled DC Motor**: A multi-stage reduction of an electromechanical system.
*  **Multi-Loop Feedback System**: Simplification of nested loops through successive algebraic steps.

```python
"""
Block Diagram Algebra – Parchment Calligraphy Edition
======================================================
Author  : Mugambi Ndwiga / @craftsandengineering
Format  : 9×16 portrait  1080×1920  YouTube Shorts

AESTHETIC
  ┌─ Aged parchment background with procedural stains, grain, vignette
  ├─ TeX Gyre Chorus (Chancery calligraphy) for all display text
  ├─ mathpazo Palatino for all LaTeX / MathTex (elegant, hand-like)
  ├─ Every block/junction has an ink-bleed halo (ink sinking into paper)
  ├─ Block fills & diagram elements are slightly translucent (ink-on-paper)
  ├─ Text panels are mostly-translucent parchment overlays; text is opaque
  └─ Diagrams are dominant, large, and vertically centred in each frame

PARAMETRIC CUSTOMISATION
  All visual parameters live in ParchmentConfig below.
  Change any field to customise colours, sizes, timings, or layout.

RULE VERIFICATION SOURCES
  Ogata K. (2010) Modern Control Engineering, 5th ed. §3-5
  Nise N.S. (2020) Control Systems Engineering, 8th ed. §5-2
  Dorf R.C. & Bishop R.H. (2017) Modern Control Systems, 13th ed. §2-4
  Franklin G.F. et al. (2019) Feedback Control of Dynamic Systems, 8th ed.
"""

from __future__ import annotations
from dataclasses import dataclass, field
from typing import Optional
from manim import *
import numpy as np
import random as _rnd

# ══════════════════════════════════════════════════════════════════════════════
#  PARAMETRIC CONFIGURATION  ← edit anything here to customise the video
# ══════════════════════════════════════════════════════════════════════════════

@dataclass
class ParchmentConfig:
    """Single source of truth for every visual parameter in the video."""

    # ── Canvas ──────────────────────────────────────────────────────────────
    pixel_w: int   = 1080
    pixel_h: int   = 1920
    frame_w: float = 9.0
    frame_h: float = 16.0

    # ── Palette ─────────────────────────────────────────────────────────────
    # Background / paper
    parch_base:  str = "#F0E6CC"   # aged parchment (light)
    parch_dark:  str = "#DEC99A"   # darker wash for highlighted blocks
    parch_panel: str = "#EAD8B0"   # panel overlay tint

    # Ink colours
    ink:         str = "#1A0800"   # primary ink (near-black sepia)
    ink_faded:   str = "#3D1C00"   # faded ink for arrows / lines
    ink_border:  str = "#2C1000"   # block / junction borders (darker sepia)
    ink_red:     str = "#8B2000"   # red-sepia  → after-morph / highlighted blocks
    ink_purple:  str = "#4A1260"   # purple-ink → feedback path blocks
    ink_eq:      str = "#2C0A3A"   # indigo ink for equations
    ink_accent:  str = "#7A1800"   # deep red for titles / accents
    ink_step:    str = "#4A2000"   # warm sepia for step labels
    ink_final:   str = "#1A4A00"   # verdigris green for final results
    ink_wm:      str = "#A09060"   # faint amber for watermark

    # Stain colours (randomly sampled for age-spots)
    stain_colours: tuple = (
        "#8B6914", "#7A5C00", "#5C3D00",
        "#9A7020", "#6A4800", "#4A3010",
    )

    # ── Translucency ────────────────────────────────────────────────────────
    block_fill_alpha:  float = 0.14   # block / junction fill opacity — TRULY translucent
    bleed_alpha:       float = 0.10   # ink-bleed halo opacity (softer, less mass)
    panel_alpha:       float = 0.16   # text-panel background opacity
    diag_stroke_alpha: float = 0.90   # block border stroke opacity
    arrow_tip_alpha:   float = 0.85   # arrowhead fill opacity

    # ── Geometry ────────────────────────────────────────────────────────────
    block_w:     float = 2.6    # standard block width  (manim units)
    block_h:     float = 2.55   # standard block height — tall & dominant (stamp-like)
    block_stroke: float = 4.2   # block border stroke width (points)
    bleed_pad:   float = 0.04   # extra size of ink-bleed halo beyond block (kept tiny —
                                 # a larger value contaminates get_left()/get_right()
                                 # bounding-box queries used throughout for wire routing)
    sj_radius:   float = 0.44   # summing-junction circle radius — matches larger blocks
    sj_stroke:   float = 4.2    # summing-junction stroke width
    dot_radius:  float = 0.11   # branch / take-off dot radius
    stub_len:    float = 0.80   # arrow stub extending beyond block edge (tighter)
    label_pad:   float = 0.40   # gap between stub end and signal label (tighter)
    arrow_stroke: float = 2.9   # arrow / line stroke width
    arrow_tip:   float = 0.13   # max_tip_length_to_length_ratio
    arrow_min_stroke_ratio: float = 400  # max_stroke_width_to_length_ratio override —
                                 # Manim's default (5) auto-thins any arrow shorter than
                                 # ~stroke_width, making short inter-block connectors
                                 # render as near-invisible hairlines. High override
                                 # disables that auto-thinning entirely.

    # ── Typography ──────────────────────────────────────────────────────────
    font:        str  = "TeX Gyre Chorus"   # calligraphic font (Cairo)
    latex_pkg:   str  = r"\usepackage{mathpazo}\usepackage{microtype}"

    fs_title:    int  = 60    # intro/outro large title
    fs_section:  int  = 28    # rule scene heading
    fs_badge:    int  = 24    # "Rule N" badge text
    fs_block:    int  = 42    # block transfer-function label (normal) — bigger for taller blocks
    fs_block_sm: int  = 26    # block label when formula is large
    fs_signal:   int  = 26    # signal labels (X(s), Y(s), R, C, …)
    fs_eq:       int  = 34    # rule equation
    fs_step:     int  = 20    # complex-example step description
    fs_eq_step:  int  = 26    # complex-example step equation
    fs_wm:       int  = 13    # watermark

    # ── Layout (all in manim units; frame y ∈ [-8, 8], x ∈ [-4.5, 4.5]) ───
    title_y:     float =  6.8    # rule/section heading
    badge_y:     float =  5.75    # rule badge
    div_y:       float =  4.7    # decorative divider line
    diag_y:      float =  0.1    # diagram centroid — exact content-zone midpoint
    cdiag_y:     float =  2.3    # diagram centroid (complex examples)
    eq_y:        float = -5.0    # equation centre
    step_y:      float =  0.0    # step label (complex examples) — close beneath
                                  # the diagram so the connector cue stays short
    ceq_y:       float = -1.2    # step equation (complex examples)
    fbox_y:      float = -5.0    # final-result box centre
    diag_width:  float =  8.6    # target width for scale_to_fit_width (rules) — DOMINANT, fills frame
    cdiag_width: float =  8.7    # target width for complex diagrams — near-edge-to-edge
    panel_w:     float =  8.8    # text-panel width

    # ── Parchment texture ────────────────────────────────────────────────────
    stain_count:  int   = 24       # number of age-spot ellipses
    stain_seed:   int   = 271828   # RNG seed (fixed → same stains every render)
    grain_step:   float = 0.38     # spacing between paper-grain lines (manim units)

    # ── Timing (all in seconds) — tightened for a brisker, more graceful pace ──
    t_write:      float = 1.3    # Write() animation
    t_fade:       float = 0.6    # standard FadeIn
    t_morph:      float = 1.1    # ReplacementTransform
    t_hold_diag:  float = 1.0    # pause after diagram appears
    t_hold_eq:    float = 1.8    # pause after equation appears
    t_hold_step:  float = 1.4    # pause after each complex step
    t_hold_intro: float = 1.8    # pause on intro slide
    t_hold_outro: float = 2.2    # pause on outro slide
    t_hold_final: float = 2.6    # pause on complex final result


# ── Instantiate the default config (used everywhere below) ──────────────────
CFG = ParchmentConfig()

# ── Apply renderer settings from config ─────────────────────────────────────
config.pixel_width  = CFG.pixel_w
config.pixel_height = CFG.pixel_h
config.frame_width  = CFG.frame_w
config.frame_height = CFG.frame_h
config.background_color = CFG.parch_base

# ── LaTeX template ──────────────────────────────────────────────────────────
_TEX = TexTemplate()
_TEX.add_to_preamble(CFG.latex_pkg)


# ══════════════════════════════════════════════════════════════════════════════
#  BACKGROUND — procedural aged parchment
# ══════════════════════════════════════════════════════════════════════════════

def make_parchment_bg(cfg: ParchmentConfig = CFG) -> VGroup:
    """
    Layered aged-parchment background:
      1. Flat parchment base rectangle
      2. Random amber/umber age stains (ellipses, seeded RNG)
      3. Faint near-horizontal paper grain lines
      4. Dark corner vignette (old paper darkens at edges)
      5. Subtle central highlight (optical brightening)
    """
    rng = _rnd.Random(cfg.stain_seed)

    # 1. Base
    base = Rectangle(
        width=cfg.frame_w + 0.5, height=cfg.frame_h + 0.5,
        fill_color=cfg.parch_base, fill_opacity=1, stroke_width=0)

    # 2. Age stains
    stains = VGroup()
    for _ in range(cfg.stain_count):
        e = Ellipse(
            width=rng.uniform(0.3, 2.6),
            height=rng.uniform(0.12, 1.1),
            fill_color=rng.choice(cfg.stain_colours),
            fill_opacity=rng.uniform(0.03, 0.10),
            stroke_width=0)
        e.rotate(rng.uniform(0, PI))
        e.move_to([rng.uniform(-4.2, 4.2), rng.uniform(-7.6, 7.6), 0])
        stains.add(e)

    # 3. Paper grain
    grain = VGroup()
    for y in np.arange(-cfg.frame_h / 2 - 0.3,
                        cfg.frame_h / 2 + 0.3,
                        cfg.grain_step):
        grain.add(Line(
            [-cfg.frame_w / 2 - 0.1, y + rng.uniform(-0.04, 0.04), 0],
            [ cfg.frame_w / 2 + 0.1, y, 0],
            color="#7B5800",
            stroke_width=rng.uniform(0.3, 0.6),
            stroke_opacity=rng.uniform(0.03, 0.07)))

    # 4. Corner vignette — soft, SMALL, multi-layered for gradient falloff
    #    (three nested ellipses per corner, shrinking + fading, instead of
    #     one large flat blob — this reads as a gradient, not a stain)
    vign = VGroup()
    hw, hh = cfg.frame_w / 2, cfg.frame_h / 2
    for cx, cy in [(-hw, -hh), (hw, -hh), (-hw, hh), (hw, hh)]:
        for k, (wmul, hmul, op) in enumerate([
            (2.6, 3.0, 0.10), (1.8, 2.1, 0.08), (1.1, 1.3, 0.06),
        ]):
            vign.add(Ellipse(
                width=wmul, height=hmul,
                fill_color="#4A3010", fill_opacity=op,
                stroke_width=0).move_to([cx, cy, 0]))

    # 5. Central highlight
    centre = Ellipse(
        width=7.5, height=12,
        fill_color="#FFFFF8", fill_opacity=0.05, stroke_width=0)

    return VGroup(base, stains, grain, vign, centre)


def make_watermark(cfg: ParchmentConfig = CFG) -> Text:
    return Text(
        "© Mugambi Ndwiga / @craftsandengineering",
        font=cfg.font, color=cfg.ink_wm,
        font_size=cfg.fs_wm).to_corner(DR, buff=0.20)


# ══════════════════════════════════════════════════════════════════════════════
#  UI HELPERS
# ══════════════════════════════════════════════════════════════════════════════

def make_panel(y_centre: float, height: float,
               width: Optional[float] = None,
               alpha: Optional[float] = None,
               cfg: ParchmentConfig = CFG) -> Rectangle:
    """
    Mostly-translucent parchment panel with NO visible border —
    a soft tonal wash behind text, not a 'box'. Edges use rounded
    corners with zero stroke so the panel reads as a gentle shadow,
    not a UI element.
    """
    w = width if width is not None else cfg.panel_w
    a = alpha if alpha is not None else cfg.panel_alpha
    return RoundedRectangle(
        corner_radius=0.35, width=w, height=height,
        fill_color=cfg.parch_panel, fill_opacity=a,
        stroke_width=0).move_to([0, y_centre, 0])


def make_header_panel(cfg: ParchmentConfig = CFG) -> Rectangle:
    yc = (cfg.title_y + cfg.div_y) / 2
    h  = cfg.title_y - cfg.div_y + 0.8
    return make_panel(yc, h, cfg=cfg)


def make_eq_panel(cfg: ParchmentConfig = CFG) -> Rectangle:
    return make_panel(cfg.eq_y, 3.0, cfg=cfg)


def make_highlight(center, width: float, height: float,
                   color: str, cfg: ParchmentConfig = CFG) -> DashedVMobject:
    """
    Dashed rounded-rectangle drawn directly ON the diagram, around the
    specific block(s) a step's equation refers to. This is the primary
    visual cue tying text to graphics: the highlight and its equation
    always share the same colour, and a connector line (see below)
    physically links the two.
    """
    rect = RoundedRectangle(corner_radius=0.16, width=width, height=height,
                            stroke_color=color, stroke_width=3.2,
                            fill_opacity=0)
    rect.move_to(center)
    n_dash = max(14, int((width + height) * 5))
    return DashedVMobject(rect, num_dashes=n_dash, dashed_ratio=0.55)


def make_connector(top, bottom, color: str,
                   cfg: ParchmentConfig = CFG) -> DashedLine:
    """
    Short dashed tether from a highlighted diagram region down to its
    equation — the explicit visual link between graphics (dominant) and
    text (secondary) requested throughout: nothing floats disconnected.
    """
    return DashedLine(top, bottom, color=color,
                      stroke_width=2.2, dash_length=0.09)


def layout_between(left_x: float, right_x: float,
                   widths: list) -> tuple:
    """
    Evenly distribute N blocks (given their widths) between two fixed
    x-coordinates, with equal gaps before/between/after. Returns
    (list_of_centre_x, gap). Used to lay out a shrinking forward-path
    block chain across successive simplification stages while keeping
    the chain's start/end anchors (e.g. a summing junction and a branch
    point) completely stationary — so only the blocks that are actually
    changing move during a morph, never the fixed anchors.
    """
    n = len(widths)
    total_w = sum(widths)
    gap = (right_x - left_x - total_w) / (n + 1)
    centres = []
    x = left_x + gap
    for w in widths:
        centres.append(x + w / 2)
        x += w + gap
    return centres, gap


def safe_width(mob: Mobject, max_w: float = 7.6) -> Mobject:
    """Scale mob down if wider than max_w."""
    if mob.width > max_w:
        mob.scale_to_fit_width(max_w)
    return mob


def dominant_scale(mob: Mobject, target_w: float,
                   min_h: float = 3.2, max_w_hard: float = 8.7) -> Mobject:
    """
    Scale a diagram VGroup to be visually DOMINANT in a portrait frame.

    Wide-flat diagrams (e.g. two blocks side by side) would otherwise only
    fill a thin horizontal sliver after scale_to_fit_width(). This helper:
      1. Scales to fit target_w (normal horizontal fill)
      2. If the resulting height is still below min_h, scales further
         (uniformly — aspect ratio preserved) until height reaches min_h,
         even if that means exceeding target_w slightly
      3. Hard-caps total width at max_w_hard so nothing clips the frame

    This guarantees every diagram — regardless of its natural aspect
    ratio — claims a visually dominant footprint in the frame.
    """
    mob.scale_to_fit_width(target_w)
    if mob.height < min_h:
        scale_factor = min_h / mob.height
        new_width = mob.width * scale_factor
        if new_width <= max_w_hard:
            mob.scale_to_fit_height(min_h)
        else:
            mob.scale_to_fit_width(max_w_hard)
    return mob


# ══════════════════════════════════════════════════════════════════════════════
#  PRIMITIVE DIAGRAM BUILDERS
# ══════════════════════════════════════════════════════════════════════════════

def make_block(label: str,
               w: Optional[float] = None,
               h: Optional[float] = None,
               fill:   Optional[str] = None,
               border: Optional[str] = None,
               fs:     Optional[int] = None,
               cfg: ParchmentConfig = CFG) -> VGroup:
    """
    Transfer-function block with:
      • ink-bleed halo (larger rectangle, low opacity) — simulates ink
        seeping into paper fibres around the border
      • slightly translucent parchment fill (ink-on-paper feel)
      • MathTex label using Palatino via mathpazo
    """
    bw = w      if w      is not None else cfg.block_w
    bh = h      if h      is not None else cfg.block_h
    fc = fill   if fill   is not None else cfg.parch_base
    bc = border if border is not None else cfg.ink_border
    fs_ = fs    if fs     is not None else cfg.fs_block

    halo = Rectangle(
        width=bw + cfg.bleed_pad, height=bh + cfg.bleed_pad,
        fill_color=bc, fill_opacity=cfg.bleed_alpha, stroke_width=0)
    rect = Rectangle(
        width=bw, height=bh,
        fill_color=fc, fill_opacity=cfg.block_fill_alpha,
        stroke_color=bc, stroke_width=cfg.block_stroke,
        stroke_opacity=cfg.diag_stroke_alpha)
    tex = MathTex(label, tex_template=_TEX, color=cfg.ink, font_size=fs_)
    tex.move_to(rect.get_center())
    return VGroup(halo, rect, tex)


_SJ_DIRS = {"left": LEFT, "right": RIGHT, "top": UP, "bottom": DOWN}


def make_sj_sign(direction: str, symbol: str, sj_r: float,
                 cfg: ParchmentConfig = CFG) -> MathTex:
    """
    Build a single +/− sign positioned just inside a summing-junction's
    rim, in the given compass direction. Exposed separately so callers
    with non-compass (angled) input arrows — e.g. Rule 9's before-state —
    can still place a matching sign next to a custom-angled wire.
    """
    vec = _SJ_DIRS[direction] if isinstance(direction, str) else direction
    sym = "+" if symbol == "+" else "{-}"
    d = sj_r * 0.62
    return MathTex(sym, tex_template=_TEX, color=cfg.ink,
                   font_size=20).move_to(vec * d)


def make_sumjunction(ports: Optional[dict] = None,
                     r: Optional[float] = None,
                     cfg: ParchmentConfig = CFG) -> VGroup:
    """
    Summing junction: a plain circle with a +/− sign shown ONLY next to
    each port that actually has a wire attached. No crosshair — the
    earlier crosshair-plus-four-compass-signs design drew a sign at
    every compass point regardless of whether a wire was present there,
    which produced stray, meaningless +/− marks and visual clutter.

    ports: dict mapping direction -> '+' or '-', one entry per WIRED
           INPUT port. Valid directions: 'left', 'right', 'top', 'bottom'.
           The output port (almost always 'right') is never given a sign
           and should be omitted.
           Default: {'left': '+', 'bottom': '-'} — the common two-input
           negative-feedback junction (reference in from the left,
           feedback in from the bottom).

    Examples:
        make_sumjunction()                               # R(+), feedback(−), unlabeled output
        make_sumjunction({'top': '+', 'bottom': '+'})     # two additive inputs, output right (Rule 2)
        make_sumjunction({'left': '+'})                   # only the left port is compass-aligned;
                                                            # any other inputs use angled arrows and
                                                            # get their own sign via make_sj_sign()
    """
    ports = ports if ports is not None else {"left": "+", "bottom": "-"}
    sj_r = r if r is not None else cfg.sj_radius

    halo = Circle(sj_r + cfg.bleed_pad * 0.5,
                 fill_color=cfg.ink_border,
                 fill_opacity=cfg.bleed_alpha, stroke_width=0)
    circle = Circle(sj_r,
                    color=cfg.ink_border,
                    fill_color=cfg.parch_base,
                    fill_opacity=cfg.block_fill_alpha,
                    stroke_width=cfg.sj_stroke,
                    stroke_opacity=cfg.diag_stroke_alpha)

    signs = VGroup(*[
        make_sj_sign(direction, sign, sj_r, cfg=cfg)
        for direction, sign in ports.items()
    ])
    return VGroup(halo, circle, signs)


def make_branch_dot(cfg: ParchmentConfig = CFG) -> Dot:
    """Take-off / branch point dot."""
    return Dot(radius=cfg.dot_radius,
               color=cfg.ink, fill_opacity=0.88)


def layout_chain_x(widths: list, gap: float, start_x: float) -> list:
    """
    Compute non-overlapping x-centres for a horizontal chain of blocks.

    Given a list of block widths and a uniform gap between them, returns
    the x-coordinate of each block's centre such that consecutive blocks
    never overlap (accounting for the requested gap, which should already
    include any visual halo/bleed padding you want preserved).

    Example:
        widths = [0.95, 1.55, 0.95, 1.50]
        centres = layout_chain_x(widths, gap=0.35, start_x=-3.0)
        # → 4 x-coordinates, guaranteed correctly spaced, no overlap ever.

    This eliminates manual coordinate arithmetic (the source of the
    Complex-01 block-overlap bug) and makes chain layouts fully
    parametric — just change `widths` or `gap` to retune spacing.
    """
    centres = []
    x = start_x
    for i, w in enumerate(widths):
        if i == 0:
            x = start_x + w / 2
        else:
            x = centres[-1] + widths[i-1] / 2 + gap + w / 2
        centres.append(x)
    return centres


# ── Arrow / line builders (all respect CFG geometry) ─────────────────────────

def arr_h(x0: float, x1: float, y: float,
          color: Optional[str] = None,
          cfg: ParchmentConfig = CFG) -> Arrow:
    """
    Horizontal arrow from (x0,y) → (x1,y).

    max_stroke_width_to_length_ratio is explicitly overridden — Manim's
    default (5) auto-thins the stroke of any arrow shorter than a few
    stroke-widths, which made short inter-block connectors (a very
    common case in these diagrams) render as near-invisible hairlines.
    """
    c = color if color else cfg.ink_faded
    return Arrow([x0,y,0],[x1,y,0], buff=0, color=c,
                 stroke_width=cfg.arrow_stroke,
                 max_tip_length_to_length_ratio=cfg.arrow_tip,
                 max_stroke_width_to_length_ratio=cfg.arrow_min_stroke_ratio)

def arr_v(x: float, y0: float, y1: float,
          color: Optional[str] = None,
          cfg: ParchmentConfig = CFG) -> Arrow:
    """Vertical arrow from (x,y0) → (x,y1). See arr_h for the stroke-ratio note."""
    c = color if color else cfg.ink_faded
    return Arrow([x,y0,0],[x,y1,0], buff=0, color=c,
                 stroke_width=cfg.arrow_stroke,
                 max_tip_length_to_length_ratio=cfg.arrow_tip,
                 max_stroke_width_to_length_ratio=cfg.arrow_min_stroke_ratio)

def line_h(x0: float, x1: float, y: float,
           color: Optional[str] = None,
           cfg: ParchmentConfig = CFG) -> Line:
    c = color if color else cfg.ink_faded
    return Line([x0,y,0],[x1,y,0], color=c, stroke_width=cfg.arrow_stroke)

def line_v(x: float, y0: float, y1: float,
           color: Optional[str] = None,
           cfg: ParchmentConfig = CFG) -> Line:
    c = color if color else cfg.ink_faded
    return Line([x,y0,0],[x,y1,0], color=c, stroke_width=cfg.arrow_stroke)

def sig_label(latex_str: str, pos,
              fs: Optional[int] = None,
              cfg: ParchmentConfig = CFG) -> MathTex:
    """Signal label (X(s), Y(s), R, C, …)."""
    fs_ = fs if fs is not None else cfg.fs_signal
    return MathTex(latex_str, tex_template=_TEX,
                   color=cfg.ink, font_size=fs_).move_to(pos)

def display_text(s: str,
                 fs: Optional[int] = None,
                 color: Optional[str] = None,
                 cfg: ParchmentConfig = CFG, **kw) -> Text:
    """Calligraphic display text (non-math)."""
    return Text(s, font=cfg.font,
                color=color if color else cfg.ink,
                font_size=fs if fs else cfg.fs_section, **kw)


def make_badge(n: int, cfg: ParchmentConfig = CFG) -> VGroup:
    """Ink-stamp style 'Rule N' badge."""
    halo = RoundedRectangle(
        corner_radius=0.22, width=2.76, height=0.90,
        fill_color=cfg.ink_border, fill_opacity=cfg.bleed_alpha,
        stroke_width=0)
    pill = RoundedRectangle(
        corner_radius=0.20, width=2.60, height=0.74,
        fill_color=cfg.parch_base, fill_opacity=cfg.block_fill_alpha,
        stroke_color=cfg.ink_border, stroke_width=2.6)
    lab = Text(f"Rule {n}", font=cfg.font,
               color=cfg.ink, font_size=cfg.fs_badge, weight=BOLD)
    lab.move_to(pill)
    return VGroup(halo, pill, lab).move_to([0, cfg.badge_y, 0])


def make_divider(cfg: ParchmentConfig = CFG) -> Line:
    return Line([-4.2, cfg.div_y, 0], [4.2, cfg.div_y, 0],
                color=cfg.ink_border, stroke_width=1.4, stroke_opacity=0.45)


# ── Shorthand aliases for diagram functions (DRY) ────────────────────────────
# These keep the rule builder bodies concise while remaining parametric.

def _ah(x0,x1,y,cfg=CFG): return arr_h(x0,x1,y,cfg=cfg)
def _av(x,y0,y1,cfg=CFG): return arr_v(x,y0,y1,cfg=cfg)
def _lh(x0,x1,y,cfg=CFG): return line_h(x0,x1,y,cfg=cfg)
def _lv(x,y0,y1,cfg=CFG): return line_v(x,y0,y1,cfg=cfg)
def _s(lat,pos,fs=None,cfg=CFG): return sig_label(lat,pos,fs=fs,cfg=cfg)
def _blk(*a,cfg=CFG,**kw): return make_block(*a,cfg=cfg,**kw)
def _sj(ports=None,r=None,cfg=CFG): return make_sumjunction(ports,r,cfg)
def _dot(cfg=CFG): return make_branch_dot(cfg)


# ══════════════════════════════════════════════════════════════════════════════
#  CORE MORPH HELPER
# ══════════════════════════════════════════════════════════════════════════════

def morph_rule(scene: Scene,
               title_str: str,
               rule_n: int,
               build_before,
               build_after,
               equation: Mobject,
               cfg: ParchmentConfig = CFG) -> None:
    """
    Standard rule scene sequence:
      FadeIn header (panel + title + badge + divider)
      → FadeIn 'before' diagram
      → ReplacementTransform to 'after' diagram
      → FadeIn a dashed connector tethering the diagram to its equation
      → Write equation
      → FadeOut everything

    The connector is the explicit visual cue linking text to graphics:
    the equation never floats disconnected from the diagram it describes.

    build_before / build_after: callables returning a VGroup centred at ORIGIN.
    equation: a fully constructed MathTex (or VGroup) — positioned here.
    """
    hp    = make_header_panel(cfg)
    ep    = make_eq_panel(cfg)
    head  = display_text(title_str, fs=cfg.fs_section,
                         color=cfg.ink_accent, weight=BOLD
                         ).move_to([0, cfg.title_y, 0])
    badge = make_badge(rule_n, cfg)
    div   = make_divider(cfg)

    grp_b = build_before()
    grp_b.move_to([0, cfg.diag_y, 0])
    grp_a = build_after()
    grp_a.move_to([0, cfg.diag_y, 0])

    safe_width(equation, cfg.panel_w - 0.8)
    equation.move_to([0, cfg.eq_y, 0])

    scene.play(FadeIn(hp), FadeIn(ep),
               FadeIn(head), FadeIn(badge), FadeIn(div))
    scene.play(FadeIn(grp_b, run_time=cfg.t_fade))
    scene.wait(cfg.t_hold_diag)
    scene.play(ReplacementTransform(grp_b, grp_a), run_time=cfg.t_morph)
    scene.wait(0.4)

    conn = make_connector(
        [0, grp_a.get_bottom()[1] - 0.12, 0],
        [0, equation.get_top()[1] + 0.35, 0],
        cfg.ink_eq, cfg=cfg)
    scene.play(FadeIn(conn, run_time=cfg.t_fade))
    scene.play(Write(equation, run_time=1.2))
    scene.wait(cfg.t_hold_eq)
    scene.play(FadeOut(VGroup(hp, ep, head, badge, div, grp_a, equation, conn)))


# ══════════════════════════════════════════════════════════════════════════════
#  RULE ANIMATIONS
#
#  All 10 rules verified against:
#    Ogata (2010) §3-5 Table 3-1
#    Nise (2020) §5-2 Table 5.2
#    Dorf & Bishop (2017) §2-4
#
#  Diagram conventions:
#    • Arrows always start/end at object surfaces (buff=0, coords on edge)
#    • All VGroups use scale_to_fit_width(cfg.diag_width) → fills frame
#    • Feedback summing junctions: left='+', bottom='−' (IEEE convention)
#    • Highlighted (simplified) blocks use fill=parch_dark, border=ink_red
# ══════════════════════════════════════════════════════════════════════════════

def rule01_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 1 — Cascade (Series) connection
    Source: Ogata (2010) Table 3-1, Rule 1
    Y(s) = G₁(s)·G₂(s)·X(s)
    Two blocks in series reduce to a single block whose TF is the product.
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w / 2

    def before():
        gap = 0.5
        cx1, cx2 = -(HW + gap/2), (HW + gap/2)
        l1, r1 = cx1-HW, cx1+HW
        l2, r2 = cx2-HW, cx2+HW
        b1 = make_block("G_1", cfg=cfg).move_to([cx1, 0, 0])
        b2 = make_block("G_2", cfg=cfg).move_to([cx2, 0, 0])
        g = VGroup(b1, b2,
            arr_h(l1-S, l1, 0, cfg=cfg),
            arr_h(r1, l2, 0, cfg=cfg),
            arr_h(r2, r2+S, 0, cfg=cfg),
            sig_label("X(s)", [l1-S-P, 0, 0], cfg=cfg),
            sig_label("Y(s)", [r2+S+P, 0, 0], cfg=cfg))
        return dominant_scale(g, cfg.diag_width, min_h=3.4)

    def after():
        bw2 = 3.0; hw2 = bw2/2
        b = make_block("G_1 G_2", w=bw2, fill=cfg.parch_dark,
                       border=cfg.ink_red, cfg=cfg).move_to([0, 0, 0])
        l, r = -hw2, hw2
        g = VGroup(b,
            arr_h(l-S, l, 0, cfg=cfg),
            arr_h(r, r+S, 0, cfg=cfg),
            sig_label("X(s)", [l-S-P, 0, 0], cfg=cfg),
            sig_label("Y(s)", [r+S+P, 0, 0], cfg=cfg))
        return dominant_scale(g, cfg.diag_width, min_h=3.4)

    morph_rule(scene, "Rule 1: Cascade (Series)", 1, before, after,
        MathTex(r"Y(s) = G_1\,G_2\;X(s)",
                tex_template=_TEX, color=cfg.ink_eq, font_size=cfg.fs_eq),
        cfg=cfg)


def rule02_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 2 — Parallel connection
    Source: Ogata (2010) Table 3-1, Rule 2
    Y(s) = [G₁(s) + G₂(s)]·X(s)
    Two parallel branches each process X(s); outputs sum at a junction.
    The junction here has TOP and BOTTOM inputs (from G1, G2) and a
    RIGHT output — no LEFT port is wired, so none is drawn.
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w/2; SJR = cfg.sj_radius

    def before():
        sx, ex = -2.3, 2.3
        ty, by = 1.5, -1.5
        l1, r1 = -HW, HW
        g1 = make_block("G_1", cfg=cfg).move_to([0, ty, 0])
        g2 = make_block("G_2", cfg=cfg).move_to([0, by, 0])
        bp = make_branch_dot(cfg).move_to([sx, 0, 0])
        s  = make_sumjunction({"top": "+", "bottom": "+"}, cfg=cfg).move_to([ex, 0, 0])
        g = VGroup(g1, g2, bp, s,
            arr_h(sx-S, sx, 0, cfg=cfg),
            line_v(sx, 0, ty, cfg=cfg), arr_h(sx, l1, ty, cfg=cfg),
            line_v(sx, 0, by, cfg=cfg), arr_h(sx, l1, by, cfg=cfg),
            line_h(r1, ex, ty, cfg=cfg), line_v(ex, ty, SJR, cfg=cfg),
            line_h(r1, ex, by, cfg=cfg), line_v(ex, by, -SJR, cfg=cfg),
            arr_h(ex+SJR, ex+SJR+S, 0, cfg=cfg),
            sig_label("X(s)", [sx-S-P, 0, 0], cfg=cfg),
            sig_label("Y(s)", [ex+SJR+S+P, 0, 0], cfg=cfg))
        return dominant_scale(g, min(cfg.diag_width + 0.6, 8.7), min_h=3.4)

    def after():
        bw2 = 3.0; hw2 = bw2/2
        b = make_block("G_1 + G_2", w=bw2, fill=cfg.parch_dark,
                       border=cfg.ink_red, cfg=cfg).move_to([0, 0, 0])
        l, r = -hw2, hw2
        g = VGroup(b,
            arr_h(l-S, l, 0, cfg=cfg),
            arr_h(r, r+S, 0, cfg=cfg),
            sig_label("X(s)", [l-S-P, 0, 0], cfg=cfg),
            sig_label("Y(s)", [r+S+P, 0, 0], cfg=cfg))
        return dominant_scale(g, cfg.diag_width, min_h=3.4)

    morph_rule(scene, "Rule 2: Parallel Blocks", 2, before, after,
        MathTex(r"Y(s) = \bigl(G_1 + G_2\bigr)\,X(s)",
                tex_template=_TEX, color=cfg.ink_eq, font_size=cfg.fs_eq),
        cfg=cfg)


def rule03_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 3 — Closed-loop with feedback (negative feedback)
    Source: Ogata (2010) §3-5, Eq. 3-22; Nise (2020) §5-2
    C(s)/R(s) = G₁(s) / [1 + G₁(s)·H(s)]

    Sign convention (IEEE / standard control textbooks):
      left port of Σ: + (reference input R);  bottom port: − (feedback H).
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w/2; SJR = cfg.sj_radius

    def before():
        sx, gx, bpx = -2.6, 0.3, 2.3
        y0 = 0.6
        HWH = 1.7/2
        hy = y0 - 2.0

        s  = make_sumjunction(cfg=cfg).move_to([sx, y0, 0])
        g1 = make_block("G_1", cfg=cfg).move_to([gx, y0, 0])
        bp = make_branch_dot(cfg).move_to([bpx, y0, 0])
        hb = make_block("H", w=1.7, h=0.9, fill=cfg.parch_base,
                        border=cfg.ink_purple, cfg=cfg).move_to([gx, hy, 0])

        sl, sr = sx-SJR, sx+SJR
        s_bottom = y0-SJR
        gl, gr = gx-HW, gx+HW
        hl, hr = gx-HWH, gx+HWH

        grp = VGroup(s, g1, bp, hb,
            arr_h(sl-S, sl, y0, cfg=cfg),
            arr_h(sr, gl, y0, cfg=cfg),
            line_h(gr, bpx, y0, cfg=cfg),
            arr_h(bpx, bpx+S, y0, cfg=cfg),
            line_v(bpx, y0, hy, cfg=cfg),
            line_h(bpx, hr, hy, cfg=cfg),
            line_h(hl, sx, hy, cfg=cfg),
            arr_v(sx, hy, s_bottom, cfg=cfg),
            sig_label("R(s)", [sl-S-P, y0, 0], cfg=cfg),
            sig_label("C(s)", [bpx+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, min(cfg.diag_width + 0.6, 8.7), min_h=3.4)

    def after():
        bw2, bh2 = 3.2, 1.3
        hw2 = bw2/2
        b = make_block(r"\dfrac{G_1}{1+G_1 H}", w=bw2, h=bh2,
                       fill=cfg.parch_dark, border=cfg.ink_red,
                       fs=cfg.fs_block_sm, cfg=cfg).move_to([0, 0, 0])
        l, r = -hw2, hw2
        g = VGroup(b,
            arr_h(l-S, l, 0, cfg=cfg),
            arr_h(r, r+S, 0, cfg=cfg),
            sig_label("R(s)", [l-S-P, 0, 0], cfg=cfg),
            sig_label("C(s)", [r+S+P, 0, 0], cfg=cfg))
        return dominant_scale(g, cfg.diag_width, min_h=3.4)

    morph_rule(scene, "Rule 3: Feedback Loop", 3, before, after,
        MathTex(r"\frac{C(s)}{R(s)} = \frac{G_1}{1 + G_1 H}",
                tex_template=_TEX, color=cfg.ink_eq, font_size=cfg.fs_eq),
        cfg=cfg)


def rule04_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 4 — Move a summing junction AHEAD of (before) a block
    Source: Ogata (2010) Table 3-1, Rule 3; Nise (2020) Table 5.2, Rule 4
    Before : R→G→Σ(+X)→Y        After : R→Σ(+X/G)→G→Y
    Extra input X passes through 1/G to preserve equivalence:
      Before: Y = G·R + X.  After: Y = G·(R + X/G) = G·R + X.  ✓

    The auxiliary 1/G block in the 'after' diagram is deliberately
    centred at the SAME x-coordinate as the summing junction above it,
    so the connecting wire is a single straight vertical segment from
    the block's top edge to the junction's bottom edge — this avoids
    any horizontal wire needing to backtrack across the block's own
    interior (a routing bug present in earlier drafts of this diagram).
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w/2; SJR = cfg.sj_radius

    def before():
        gx, sx = -1.2, 1.7
        y0 = 0
        gl, gr = gx-HW, gx+HW
        sl, sr = sx-SJR, sx+SJR
        s_bottom = y0-SJR
        g = make_block("G", cfg=cfg).move_to([gx, y0, 0])
        s = make_sumjunction({"left": "+", "bottom": "+"}, cfg=cfg).move_to([sx, y0, 0])
        grp = VGroup(g, s,
            arr_h(gl-S, gl, y0, cfg=cfg),
            arr_h(gr, sl, y0, cfg=cfg),
            arr_h(sr, sr+S, y0, cfg=cfg),
            arr_v(sx, y0-1.6, s_bottom, cfg=cfg),
            sig_label("R", [gl-S-P, y0, 0], cfg=cfg),
            sig_label("X", [sx, y0-1.9, 0], cfg=cfg),
            sig_label("Y", [sr+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, cfg.diag_width, min_h=3.4)

    def after():
        sx, gx = -1.7, 1.2
        y0 = 0.6
        gi_y = -1.3
        HWG, HHG = 1.9/2, 0.9/2

        s  = make_sumjunction({"left": "+", "bottom": "+"}, cfg=cfg).move_to([sx, y0, 0])
        g  = make_block("G", fill=cfg.parch_dark, border=cfg.ink_red,
                        cfg=cfg).move_to([gx, y0, 0])
        gi = make_block(r"\tfrac{1}{G}", w=1.9, h=0.9,
                        cfg=cfg).move_to([sx, gi_y, 0])   # same x as s — no backtrack needed

        sl, sr = sx-SJR, sx+SJR
        s_bottom = y0-SJR
        gl, gr = gx-HW, gx+HW
        gil = sx-HWG
        gi_top = gi_y+HHG

        grp = VGroup(s, g, gi,
            arr_h(sl-S, sl, y0, cfg=cfg),
            arr_h(sr, gl, y0, cfg=cfg),
            arr_h(gr, gr+S, y0, cfg=cfg),
            arr_h(gil-S, gil, gi_y, cfg=cfg),
            arr_v(sx, gi_top, s_bottom, cfg=cfg),
            sig_label("R", [sl-S-P, y0, 0], cfg=cfg),
            sig_label("X", [gil-S-P, gi_y, 0], cfg=cfg),
            sig_label("Y", [gr+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, min(cfg.diag_width+0.2, 8.7), min_h=3.4)

    eq = MathTex(
        r"\Sigma \text{ moves before } G "
        r"\;\Rightarrow\; \text{branch gets } \tfrac{1}{G}",
        tex_template=_TEX, color=cfg.ink_eq, font_size=28)
    morph_rule(scene, "Rule 4: Move Sum Ahead of Block", 4, before, after, eq, cfg=cfg)


def rule05_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 5 — Move a summing junction BEYOND (after) a block
    Source: Ogata (2010) Table 3-1, Rule 4; Nise (2020) Table 5.2, Rule 5
    Before : R→Σ(+X)→G→Y        After : R→G→Σ(+GX)→Y
    Extra input X passes through G:
      Before: Y = G·(R + X) = G·R + G·X.  After: Y = G·R + G·X.  ✓

    Same edge-aligned wiring technique as Rule 4 — the auxiliary G block
    shares its x-coordinate with the junction above it.
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w/2; SJR = cfg.sj_radius

    def before():
        sx, gx = -1.7, 1.2
        y0 = 0
        sl, sr = sx-SJR, sx+SJR
        s_bottom = y0-SJR
        gl, gr = gx-HW, gx+HW
        s = make_sumjunction({"left": "+", "bottom": "+"}, cfg=cfg).move_to([sx, y0, 0])
        g = make_block("G", cfg=cfg).move_to([gx, y0, 0])
        grp = VGroup(s, g,
            arr_h(sl-S, sl, y0, cfg=cfg),
            arr_h(sr, gl, y0, cfg=cfg),
            arr_h(gr, gr+S, y0, cfg=cfg),
            arr_v(sx, y0-1.6, s_bottom, cfg=cfg),
            sig_label("R", [sl-S-P, y0, 0], cfg=cfg),
            sig_label("X", [sx, y0-1.9, 0], cfg=cfg),
            sig_label("Y", [gr+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, cfg.diag_width, min_h=3.4)

    def after():
        gx, sx = -1.2, 1.7
        y0 = 0.6
        gf_y = -1.3
        HWG, HHG = 1.9/2, 0.9/2

        g  = make_block("G", fill=cfg.parch_dark, border=cfg.ink_red,
                        cfg=cfg).move_to([gx, y0, 0])
        s  = make_sumjunction({"left": "+", "bottom": "+"}, cfg=cfg).move_to([sx, y0, 0])
        gf = make_block("G", w=1.9, h=0.9, cfg=cfg).move_to([sx, gf_y, 0])

        gl, gr = gx-HW, gx+HW
        sl, sr = sx-SJR, sx+SJR
        s_bottom = y0-SJR
        gfl = sx-HWG
        gf_top = gf_y+HHG

        grp = VGroup(g, s, gf,
            arr_h(gl-S, gl, y0, cfg=cfg),
            arr_h(gr, sl, y0, cfg=cfg),
            arr_h(sr, sr+S, y0, cfg=cfg),
            arr_h(gfl-S, gfl, gf_y, cfg=cfg),
            arr_v(sx, gf_top, s_bottom, cfg=cfg),
            sig_label("R", [gl-S-P, y0, 0], cfg=cfg),
            sig_label("X", [gfl-S-P, gf_y, 0], cfg=cfg),
            sig_label("Y", [sr+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, min(cfg.diag_width+0.2, 8.7), min_h=3.4)

    eq = MathTex(
        r"\Sigma \text{ moves beyond } G "
        r"\;\Rightarrow\; \text{branch gets } G",
        tex_template=_TEX, color=cfg.ink_eq, font_size=28)
    morph_rule(scene, "Rule 5: Move Sum Beyond Block", 5, before, after, eq, cfg=cfg)


def rule06_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 6 — Move a take-off point AHEAD of (before) a block
    Source: Ogata (2010) Table 3-1, Rule 6; Nise (2020) Table 5.2, Rule 6
    Before : X→G→bp(→Y, →Y)     After : X→bp→G→Y  and  bp→G(copy)→Y
    The branch must pass through a copy of G to keep Y = G·X on both outputs.
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w/2

    def before():
        gx, y0 = 0, 0
        gl, gr = gx-HW, gx+HW
        bpx = gr+0.5
        g = make_block("G", cfg=cfg).move_to([gx, y0, 0])
        bp = make_branch_dot(cfg).move_to([bpx, y0, 0])
        grp = VGroup(g, bp,
            arr_h(gl-S, gl, y0, cfg=cfg),
            line_h(gr, bpx, y0, cfg=cfg),
            arr_h(bpx, bpx+S, y0, cfg=cfg),
            arr_v(bpx, y0, y0-1.6, cfg=cfg),
            sig_label("X", [gl-S-P, y0, 0], cfg=cfg),
            sig_label("Y", [bpx+S+P, y0, 0], cfg=cfg),
            sig_label("Y", [bpx, y0-1.9, 0], cfg=cfg))
        return dominant_scale(grp, cfg.diag_width, min_h=3.4)

    def after():
        bpx, gx, y0 = -2.9, -0.05, 0.6
        gf_y = -1.3
        HWG = 1.9/2

        bp = make_branch_dot(cfg).move_to([bpx, y0, 0])
        g  = make_block("G", fill=cfg.parch_dark, border=cfg.ink_red,
                        cfg=cfg).move_to([gx, y0, 0])
        gf = make_block("G", w=1.9, h=0.9, cfg=cfg).move_to([gx, gf_y, 0])

        gl, gr = gx-HW, gx+HW
        gfl, gfr = gx-HWG, gx+HWG

        grp = VGroup(bp, g, gf,
            arr_h(bpx-S, bpx, y0, cfg=cfg),
            arr_h(bpx, gl, y0, cfg=cfg),
            arr_h(gr, gr+S, y0, cfg=cfg),
            line_v(bpx, y0, gf_y, cfg=cfg),
            arr_h(bpx, gfl, gf_y, cfg=cfg),
            arr_h(gfr, gfr+S, gf_y, cfg=cfg),
            sig_label("X", [bpx-S-P, y0, 0], cfg=cfg),
            sig_label("Y", [gr+S+P, y0, 0], cfg=cfg),
            sig_label("Y", [gfr+S+P, gf_y, 0], cfg=cfg))
        return dominant_scale(grp, min(cfg.diag_width+0.6, 8.7), min_h=3.4)

    eq = MathTex(
        r"\text{Take-off before } G "
        r"\;\Rightarrow\; \text{branch gets } G",
        tex_template=_TEX, color=cfg.ink_eq, font_size=28)
    morph_rule(scene, "Rule 6: Move Take-off Ahead", 6, before, after, eq, cfg=cfg)


def rule07_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 7 — Move a take-off point BEYOND (after) a block
    Source: Ogata (2010) Table 3-1, Rule 5; Nise (2020) Table 5.2, Rule 7
    Before : bp(→X)→G→Y         After : G→bp(→Y),  bp→1/G→X
    The branch passes through 1/G to recover X = Y/G = G·X/G = X.  ✓

    The branch's 1/G block is drawn as a straight vertical continuation
    (bp → block's top edge → block's bottom edge → further down to the
    X label) rather than exiting sideways, avoiding any ambiguous bend
    and keeping the wire fully clear of every other block.
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w/2

    def before():
        bpx, gx, y0 = -2.3, 0.4, 0
        gl, gr = gx-HW, gx+HW
        g = make_block("G", cfg=cfg).move_to([gx, y0, 0])
        bp = make_branch_dot(cfg).move_to([bpx, y0, 0])
        grp = VGroup(bp, g,
            arr_h(bpx-S, bpx, y0, cfg=cfg),
            arr_h(bpx, gl, y0, cfg=cfg),
            arr_h(gr, gr+S, y0, cfg=cfg),
            arr_v(bpx, y0, y0-1.6, cfg=cfg),
            sig_label("X", [bpx-S-P, y0, 0], cfg=cfg),
            sig_label("Y", [gr+S+P, y0, 0], cfg=cfg),
            sig_label("X", [bpx, y0-1.9, 0], cfg=cfg))
        return dominant_scale(grp, cfg.diag_width, min_h=3.4)

    def after():
        gx, y0 = -0.05, 0.6
        gi_y = -1.5
        HWG, HHG = 1.9/2, 0.9/2

        g  = make_block("G", fill=cfg.parch_dark, border=cfg.ink_red,
                        cfg=cfg).move_to([gx, y0, 0])
        gl, gr = gx-HW, gx+HW
        bpx = gr+0.5
        bp = make_branch_dot(cfg).move_to([bpx, y0, 0])
        gi = make_block(r"\tfrac{1}{G}", w=1.9, h=0.9,
                        cfg=cfg).move_to([bpx, gi_y, 0])   # same x as bp

        gi_top = gi_y+HHG
        gi_bot = gi_y-HHG

        grp = VGroup(g, bp, gi,
            arr_h(gl-S, gl, y0, cfg=cfg),
            line_h(gr, bpx, y0, cfg=cfg),
            arr_h(bpx, bpx+S, y0, cfg=cfg),
            line_v(bpx, y0, gi_top, cfg=cfg),
            arr_v(bpx, gi_bot, gi_bot-S, cfg=cfg),
            sig_label("X", [gl-S-P, y0, 0], cfg=cfg),
            sig_label("Y", [bpx+S+P, y0, 0], cfg=cfg),
            sig_label("X", [bpx, gi_bot-S-P, 0], cfg=cfg))
        return dominant_scale(grp, min(cfg.diag_width+0.2, 8.7), min_h=3.4)

    eq = MathTex(
        r"\text{Take-off beyond } G "
        r"\;\Rightarrow\; \text{branch gets } \tfrac{1}{G}",
        tex_template=_TEX, color=cfg.ink_eq, font_size=28)
    morph_rule(scene, "Rule 7: Move Take-off Beyond", 7, before, after, eq, cfg=cfg)


def _two_sj_row(label_a: str, label_b: str, cfg: ParchmentConfig = CFG) -> VGroup:
    """
    Shared diagram: two summing junctions in series, each with one
    additional bottom input (labelled label_a / label_b). Used by
    Rule 8 (commutative law) and Rule 9 ('after' state, associative law).
    """
    S = cfg.stub_len; P = cfg.label_pad; SJR = cfg.sj_radius
    s1x, s2x = -1.9, 1.9
    y0 = 0
    s1 = make_sumjunction({"left": "+", "bottom": "+"}, cfg=cfg).move_to([s1x, y0, 0])
    s2 = make_sumjunction({"left": "+", "bottom": "+"}, cfg=cfg).move_to([s2x, y0, 0])
    s1l, s1r = s1x-SJR, s1x+SJR
    s2l, s2r = s2x-SJR, s2x+SJR
    s_bottom = y0-SJR
    grp = VGroup(s1, s2,
        arr_h(s1l-S, s1l, y0, cfg=cfg),
        arr_h(s1r, s2l, y0, cfg=cfg),
        arr_h(s2r, s2r+S, y0, cfg=cfg),
        arr_v(s1x, y0-1.8, s_bottom, cfg=cfg),
        arr_v(s2x, y0-1.8, s_bottom, cfg=cfg),
        sig_label("R", [s1l-S-P, y0, 0], cfg=cfg),
        sig_label(label_a, [s1x, y0-2.1, 0], cfg=cfg),
        sig_label(label_b, [s2x, y0-2.1, 0], cfg=cfg),
        sig_label("Y", [s2r+S+P, y0, 0], cfg=cfg))
    return dominant_scale(grp, min(cfg.diag_width+0.8, 8.7), min_h=3.4)


def rule08_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 8 — Commutative law for summing junctions
    Source: Ogata (2010) Table 3-1, Rule 7
    R + A + B = R + B + A
    The order of inputs to a summing junction does not affect the output,
    demonstrated by swapping labels A and B between two cascaded junctions.
    """
    morph_rule(scene, "Rule 8: Summing Junction Order", 8,
        lambda: _two_sj_row("A", "B", cfg),
        lambda: _two_sj_row("B", "A", cfg),
        MathTex(r"R + A + B \;=\; R + B + A",
                tex_template=_TEX, color=cfg.ink_eq, font_size=cfg.fs_eq),
        cfg=cfg)


def rule09_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 9 — Associative law / splitting a summing junction
    Source: Ogata (2010) Table 3-1, Rule 8
    R + A + B = (R + A) + B
    One multi-input junction can always be replaced by two cascaded
    junctions. The single junction's A/B inputs enter at custom angles
    (not aligned to a compass point), so their +/− signs are placed
    manually rather than through the port-based compass API.
    """
    S = cfg.stub_len; P = cfg.label_pad; SJR = cfg.sj_radius

    def before():
        sx, y0 = 0, 0
        s = make_sumjunction({"left": "+"}, cfg=cfg).move_to([sx, y0, 0])
        sl, sr = sx-SJR, sx+SJR

        # Diagonal (45°) edge points — the exact circle boundary along the
        # down-left / down-right directions, i.e. SJR/√2 offset on each axis.
        diag = SJR / np.sqrt(2)
        sign_d = SJR * 0.62 / np.sqrt(2)   # matches compass-sign inset ratio

        kw = dict(buff=0, color=cfg.ink_faded, stroke_width=cfg.arrow_stroke,
                  max_tip_length_to_length_ratio=cfg.arrow_tip,
                  max_stroke_width_to_length_ratio=cfg.arrow_min_stroke_ratio)
        aA = Arrow([sx-1.0, y0-1.7, 0], [sx-diag, y0-diag, 0], **kw)
        aB = Arrow([sx+1.0, y0-1.7, 0], [sx+diag, y0-diag, 0], **kw)
        plusA = MathTex("+", tex_template=_TEX, color=cfg.ink,
                        font_size=20).move_to([sx-sign_d, y0-sign_d, 0])
        plusB = MathTex("+", tex_template=_TEX, color=cfg.ink,
                        font_size=20).move_to([sx+sign_d, y0-sign_d, 0])

        grp = VGroup(s, aA, aB, plusA, plusB,
            arr_h(sl-S, sl, y0, cfg=cfg),
            arr_h(sr, sr+S, y0, cfg=cfg),
            sig_label("R", [sl-S-P, y0, 0], cfg=cfg),
            sig_label("A", [sx-1.2, y0-2.0, 0], cfg=cfg),
            sig_label("B", [sx+1.2, y0-2.0, 0], cfg=cfg),
            sig_label("Y", [sr+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, cfg.diag_width, min_h=3.4)

    def after():
        return _two_sj_row("A", "B", cfg)

    morph_rule(scene, "Rule 9: Split Summing Junction", 9, before, after,
        MathTex(r"R + A + B \;=\; (R + A) + B",
                tex_template=_TEX, color=cfg.ink_eq, font_size=cfg.fs_eq),
        cfg=cfg)


def rule10_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    """
    Rule 10 — Moving a block from the feedback path to the forward path
    Source: Ogata (2010) Table 3-1, Rule 9; Nise (2020) §5-2
    Before : R→Σ(−)→G₁→C,  feedback: C→H→Σ
    After  : R→Σ(−)→G₁H→C/H,  feedback: unity gain (the block H has been
             absorbed into the forward path, converting the loop to
             unity feedback for direct comparison with a standard form).
    """
    S = cfg.stub_len; P = cfg.label_pad
    HW = cfg.block_w/2; SJR = cfg.sj_radius

    def before():
        sx, gx, bpx = -2.6, 0.3, 2.3
        y0 = 0.6
        HWH = 1.7/2
        hy = y0 - 2.0

        s  = make_sumjunction(cfg=cfg).move_to([sx, y0, 0])
        g1 = make_block("G_1", cfg=cfg).move_to([gx, y0, 0])
        bp = make_branch_dot(cfg).move_to([bpx, y0, 0])
        hb = make_block("H", w=1.7, h=0.9, fill=cfg.parch_base,
                        border=cfg.ink_purple, cfg=cfg).move_to([gx, hy, 0])

        sl, sr = sx-SJR, sx+SJR
        s_bottom = y0-SJR
        gl, gr = gx-HW, gx+HW
        hl, hr = gx-HWH, gx+HWH

        grp = VGroup(s, g1, bp, hb,
            arr_h(sl-S, sl, y0, cfg=cfg),
            arr_h(sr, gl, y0, cfg=cfg),
            line_h(gr, bpx, y0, cfg=cfg),
            arr_h(bpx, bpx+S, y0, cfg=cfg),
            line_v(bpx, y0, hy, cfg=cfg),
            line_h(bpx, hr, hy, cfg=cfg),
            line_h(hl, sx, hy, cfg=cfg),
            arr_v(sx, hy, s_bottom, cfg=cfg),
            sig_label("R", [sl-S-P, y0, 0], cfg=cfg),
            sig_label("C", [bpx+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, min(cfg.diag_width+0.6, 8.7), min_h=3.4)

    def after():
        sx, gx, bpx = -2.6, 0.3, 2.3
        y0 = 0.6
        bw2, hw2 = 3.0, 1.5
        hy = -1.4   # clear of gh's bottom edge (y0 - block_h/2)

        s  = make_sumjunction(cfg=cfg).move_to([sx, y0, 0])
        gh = make_block("G_1 H", w=bw2, fill=cfg.parch_dark,
                        border=cfg.ink_red, cfg=cfg).move_to([gx, y0, 0])
        bp = make_branch_dot(cfg).move_to([bpx, y0, 0])

        sl, sr = sx-SJR, sx+SJR
        s_bottom = y0-SJR
        gl, gr = gx-hw2, gx+hw2

        grp = VGroup(s, gh, bp,
            arr_h(sl-S, sl, y0, cfg=cfg),
            arr_h(sr, gl, y0, cfg=cfg),
            line_h(gr, bpx, y0, cfg=cfg),
            arr_h(bpx, bpx+S, y0, cfg=cfg),
            line_v(bpx, y0, hy, cfg=cfg),
            line_h(bpx, sx, hy, cfg=cfg),
            arr_v(sx, hy, s_bottom, cfg=cfg),
            sig_label("R", [sl-S-P, y0, 0], cfg=cfg),
            sig_label(r"\tfrac{C}{H}", [bpx+S+P, y0, 0], cfg=cfg))
        return dominant_scale(grp, min(cfg.diag_width+0.6, 8.7), min_h=3.4)

    morph_rule(scene, "Rule 10: Block in Feedback Path", 10, before, after,
        MathTex(r"\frac{C/H}{R} = \frac{G_1 H}{1 + G_1 H}",
                tex_template=_TEX, color=cfg.ink_eq, font_size=cfg.fs_eq),
        cfg=cfg)


# ══════════════════════════════════════════════════════════════════════════════
#  COMPLEX EXAMPLES — step helper
# ══════════════════════════════════════════════════════════════════════════════

def _step_with_highlight(scene: Scene, hl_center, hl_w: float, hl_h: float,
                         label: str, eq_str: str, cfg: ParchmentConfig = CFG,
                         fs: Optional[int] = None) -> None:
    """
    The core 'text is connected to graphics' cue for the complex examples:
      1. A dashed highlight box is drawn directly on the diagram, around
         the exact block(s) the upcoming step concerns.
      2. A dashed connector drops from that highlight to the step's label.
      3. The step description and its equation appear, in the SAME colour
         as the highlight/connector — so the association is unambiguous.
      4. Everything clears, then the caller performs the actual diagram
         morph (ReplacementTransform) for that step.
    """
    color = cfg.ink_eq
    hl = make_highlight(hl_center, hl_w, hl_h, color, cfg)
    conn_top = [hl_center[0], hl_center[1] - hl_h/2 - 0.04, 0]
    conn_bottom = [0, cfg.step_y + 0.32, 0]
    conn = make_connector(conn_top, conn_bottom, color, cfg)
    st = display_text(label, fs=cfg.fs_step, color=cfg.ink_step,
                      cfg=cfg).move_to([0, cfg.step_y, 0])
    eq = safe_width(MathTex(eq_str, tex_template=_TEX, color=color,
                            font_size=fs or cfg.fs_eq_step
                            ).move_to([0, cfg.ceq_y, 0]),
                    cfg.panel_w - 0.8)

    scene.play(FadeIn(hl, run_time=cfg.t_fade), FadeIn(conn, run_time=cfg.t_fade))
    scene.play(FadeIn(st, run_time=cfg.t_fade), Write(eq, run_time=1.0))
    scene.wait(cfg.t_hold_step)
    scene.play(FadeOut(VGroup(hl, conn, st, eq)))


# ══════════════════════════════════════════════════════════════════════════════
#  COMPLEX EXAMPLE 1 — Armature-Controlled DC Motor
#
#  Verified against:
#    Ogata (2010) §3-6, Example 3-7, pp. 82–85
#    Nise  (2020) §2-8, Example 2.16
#
#  Forward path:  G(s) = Ka·Kt / [(Ra + s·La)(Js + B)]
#  Feedback path: Kb  (back-EMF constant)
#  CLTF: C(s)/R(s) = Ka·Kt / [(Ra+s·La)(Js+B) + Ka·Kt·Kb]
#
#  The diagram is reduced ONE algebraic step at a time (four morphs:
#  cascade Ka with 1/(Ra+sLa) → cascade with Kt → cascade with 1/(Js+B)
#  → apply the feedback formula), matching the four equations shown.
#  The junction, branch point, and feedback block never move on screen
#  until the very last step — only the forward-path blocks that are
#  actually being combined change, so every morph reads as a direct,
#  legible simplification rather than a jump.
# ══════════════════════════════════════════════════════════════════════════════

def complex01_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    hp = make_header_panel(cfg)
    ti = display_text("Complex Example 1", fs=40, color=cfg.ink_accent,
                      weight=BOLD, cfg=cfg).move_to([0, cfg.title_y, 0])
    su = display_text("Armature-Controlled DC Motor", fs=24,
                      color=cfg.ink, cfg=cfg).move_to([0, cfg.badge_y, 0])
    scene.play(FadeIn(hp), FadeIn(ti), FadeIn(su))
    scene.wait(0.5)

    R2 = 0.27; F2 = 20; H2 = 0.76
    S = cfg.stub_len; P = cfg.label_pad
    Y0, hy = 0.8, -0.5
    sj1x, bpx = -3.3, 3.0
    sj1_right, sj1_left, sj1_bottom = sj1x+R2, sj1x-R2, Y0-R2

    # Fixed anchors — same instances reused (via .copy()) in every stage,
    # so the summing junction, branch point, and feedback block never
    # visibly move; only the forward-path blocks between them change.
    sj1_t = make_sumjunction(r=R2, cfg=cfg).move_to([sj1x, Y0, 0])
    bp_t  = make_branch_dot(cfg).move_to([bpx, Y0, 0])
    bKb_t = make_block("K_b", w=1.15, h=0.65, fill=cfg.parch_base,
                       border=cfg.ink_purple, fs=F2, cfg=cfg).move_to([0, hy, 0])
    wiring_t = VGroup(
        arr_h(sj1_left-S, sj1_left, Y0, cfg=cfg),
        line_v(bpx, Y0, hy, cfg=cfg),
        line_h(bpx, bKb_t.get_right()[0], hy, cfg=cfg),
        line_h(bKb_t.get_left()[0], sj1x, hy, cfg=cfg),
        arr_v(sj1x, hy, sj1_bottom, cfg=cfg),
        arr_h(bpx, bpx+S, Y0, cfg=cfg),
        sig_label("R(s)", [sj1_left-S-P, Y0, 0], fs=19, cfg=cfg),
        sig_label("C(s)", [bpx+S+P, Y0, 0], fs=19, cfg=cfg))

    def build_stage(widths_labels):
        """widths_labels: [(latex_label, width, fill, border), ...]"""
        widths = [w for _, w, _, _ in widths_labels]
        centres, _ = layout_between(sj1_right, bpx, widths)
        blocks, arrows = [], []
        prev_r = sj1_right
        for (label, w, fill, border), cx in zip(widths_labels, centres):
            fs_b = F2 if w < 1.3 else F2-4
            b = make_block(label, w=w, h=H2, fill=fill, border=border,
                          fs=fs_b, cfg=cfg).move_to([cx, Y0, 0])
            blocks.append(b)
            arrows.append(arr_h(prev_r, cx-w/2, Y0, cfg=cfg))
            prev_r = cx + w/2
        arrows.append(line_h(prev_r, bpx, Y0, cfg=cfg))
        g = VGroup(sj1_t.copy(), *blocks, VGroup(*arrows),
                  bp_t.copy(), bKb_t.copy(), wiring_t.copy())
        dominant_scale(g, cfg.cdiag_width, min_h=2.6)
        g.move_to([0, cfg.cdiag_y, 0])
        return g, blocks

    NORM = (cfg.parch_base, cfg.ink_border)
    HI   = (cfg.parch_dark, cfg.ink_red)

    sl = display_text("Step 0: Full System", fs=cfg.fs_step+2,
                      color=cfg.ink_step, weight=BOLD, cfg=cfg
                      ).move_to([0, cfg.div_y-0.6, 0])
    scene.play(FadeIn(sl))

    stage0, b0 = build_stage([
        ("K_a", 0.95, *NORM), (r"\frac{1}{R_a+sL_a}", 1.55, *NORM),
        ("K_t", 0.95, *NORM), (r"\frac{1}{Js+B}", 1.50, *NORM)])
    scene.play(FadeIn(stage0, run_time=cfg.t_fade))
    scene.wait(cfg.t_hold_diag)
    scene.play(FadeOut(sl))

    def hl_span(blocks, margin=0.10):
        l = blocks[0].get_left()[0]; r = blocks[-1].get_right()[0]
        cy = blocks[0].get_center()[1]
        h = max(b.height for b in blocks) + margin
        return [(l+r)/2, cy, 0], (r-l)+margin, h

    # ── Step 1: cascade Ka with 1/(Ra+sLa) ──────────────────────────────
    c, w, h = hl_span(b0[0:2])
    _step_with_highlight(scene, c, w, h,
        "Step 1: Cascade  Kₐ · 1/(Rₐ+sLₐ)",
        r"K_a \cdot \frac{1}{R_a+sL_a} = \frac{K_a}{R_a+sL_a}", cfg)

    stage1, b1 = build_stage([
        (r"\frac{K_a}{R_a+sL_a}", 1.9, *HI), ("K_t", 0.95, *NORM),
        (r"\frac{1}{Js+B}", 1.50, *NORM)])
    scene.play(ReplacementTransform(stage0, stage1), run_time=cfg.t_morph)
    scene.wait(0.4)

    # ── Step 2: cascade with Kt ──────────────────────────────────────────
    c, w, h = hl_span(b1[0:2])
    _step_with_highlight(scene, c, w, h,
        "Step 2: Cascade with Kₜ",
        r"\frac{K_a}{R_a+sL_a} \cdot K_t = \frac{K_a K_t}{R_a+sL_a}", cfg)

    stage2, b2 = build_stage([
        (r"\frac{K_a K_t}{R_a+sL_a}", 2.3, *HI), (r"\frac{1}{Js+B}", 1.50, *NORM)])
    scene.play(ReplacementTransform(stage1, stage2), run_time=cfg.t_morph)
    scene.wait(0.4)

    # ── Step 3: cascade with 1/(Js+B) — full forward path G(s) ──────────
    c, w, h = hl_span(b2[0:2])
    _step_with_highlight(scene, c, w, h,
        "Step 3: Full forward path  G(s)",
        r"G(s) = \frac{K_a K_t}{(R_a+sL_a)(Js+B)}", cfg)

    stage3, b3 = build_stage([("G(s)", 2.6, *HI)])
    scene.play(ReplacementTransform(stage2, stage3), run_time=cfg.t_morph)
    scene.wait(0.4)

    # ── Step 4: apply the feedback formula (Rule 3) ─────────────────────
    c = stage3.get_center(); w = stage3.width+0.15; h = stage3.height+0.15
    _step_with_highlight(scene, c, w, h,
        "Step 4: Apply feedback formula (Rule 3)",
        r"T(s) = \frac{G(s)}{1 + G(s)\,K_b}", cfg, fs=28)

    bf = make_block(r"\frac{K_a K_t}{(R_a{+}sL_a)(Js{+}B)+K_aK_tK_b}",
                    w=6.2, h=1.1, fill=cfg.parch_dark, border=cfg.ink_red,
                    fs=17, cfg=cfg).move_to([0, cfg.cdiag_y, 0])
    af = VGroup(bf,
        arr_h(bf.get_left()[0]-S,  bf.get_left()[0],    cfg.cdiag_y, cfg=cfg),
        arr_h(bf.get_right()[0],   bf.get_right()[0]+S, cfg.cdiag_y, cfg=cfg),
        sig_label("R(s)",[bf.get_left()[0]-S-P,  cfg.cdiag_y,0],fs=19,cfg=cfg),
        sig_label("C(s)",[bf.get_right()[0]+S+P, cfg.cdiag_y,0],fs=19,cfg=cfg))
    safe_width(af, 8.6)
    scene.play(ReplacementTransform(stage3, af), run_time=cfg.t_morph)
    scene.wait(0.5)

    fp  = make_panel(cfg.fbox_y, 3.2, alpha=0.20, cfg=cfg)
    bx  = RoundedRectangle(corner_radius=0.2, width=8.2, height=2.8,
                           fill_color=cfg.parch_panel,
                           fill_opacity=cfg.panel_alpha,
                           stroke_color=cfg.ink_border,
                           stroke_width=2.0).move_to([0, cfg.fbox_y, 0])
    lb  = display_text("Final Transfer Function", fs=22,
                       color=cfg.ink_accent, weight=BOLD,
                       cfg=cfg).next_to(bx, UP, buff=0.22)
    fe  = safe_width(MathTex(
        r"\frac{C(s)}{R(s)} = "
        r"\frac{K_a K_t}{(R_a+sL_a)(Js+B) + K_a K_t K_b}",
        tex_template=_TEX, color=cfg.ink_final, font_size=24
        ).move_to(bx), cfg.panel_w - 0.8)
    scene.play(FadeIn(fp), FadeIn(bx), FadeIn(lb), Write(fe, run_time=1.2))
    scene.wait(cfg.t_hold_final)
    scene.play(FadeOut(VGroup(hp, ti, su, af, fp, bx, lb, fe)))


# ══════════════════════════════════════════════════════════════════════════════
#  COMPLEX EXAMPLE 2 — Multi-Loop Feedback System
#
#  Verified against:
#    Ogata (2010) §3-6, Example 3-10, pp. 90–93
#    Nise  (2020) §5-5, Example 5.4
#
#  Inner loop (G₂,G₃ with H₂):  G₂₃ = G₂G₃ / (1 + G₂G₃H₂)
#  Outer loop (G₁,G₂₃ with H₁): C/R = G₁G₂₃ / (1 + G₁G₂₃H₁)
#  Expanded:  C(s)/R(s) = G₁G₂G₃ / [(1+G₂G₃H₂) + G₁G₂G₃H₁]
#
#  Three morphs, one per algebraic step: collapse the inner loop → cascade
#  G1 with the collapsed G23 → apply the outer feedback formula. The outer
#  summing junction, outer branch point, and H1 feedback block are fixed
#  anchors that never move until the final collapse.
# ══════════════════════════════════════════════════════════════════════════════

def complex02_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    hp = make_header_panel(cfg)
    ti = display_text("Complex Example 2", fs=40, color=cfg.ink_accent,
                      weight=BOLD, cfg=cfg).move_to([0, cfg.title_y, 0])
    su = display_text("Multi-Loop Feedback System", fs=24,
                      color=cfg.ink, cfg=cfg).move_to([0, cfg.badge_y, 0])
    scene.play(FadeIn(hp), FadeIn(ti), FadeIn(su))
    scene.wait(0.5)

    R2 = 0.27; F2 = 21; H2 = 0.76
    S = cfg.stub_len; P = cfg.label_pad
    Y0 = 0.9
    sj1x, obpx = -3.6, 3.0
    sj1_right, sj1_left, sj1_bottom = sj1x+R2, sj1x-R2, Y0-R2
    h1y = -1.6

    # Fixed outer-loop anchors, reused across every stage.
    sj1_t = make_sumjunction(r=R2, cfg=cfg).move_to([sj1x, Y0, 0])
    obp_t = make_branch_dot(cfg).move_to([obpx, Y0, 0])
    bH1_t = make_block("H_1", w=1.2, h=0.65, fill=cfg.parch_base,
                       border=cfg.ink_purple, fs=F2, cfg=cfg).move_to([0, h1y, 0])
    outer_wiring_t = VGroup(
        arr_h(sj1_left-S, sj1_left, Y0, cfg=cfg),
        line_v(obpx, Y0, h1y, cfg=cfg),
        line_h(obpx, bH1_t.get_right()[0], h1y, cfg=cfg),
        line_h(bH1_t.get_left()[0], sj1x, h1y, cfg=cfg),
        arr_v(sj1x, h1y, sj1_bottom, cfg=cfg),
        arr_h(obpx, obpx+S, Y0, cfg=cfg),
        sig_label("R(s)", [sj1_left-S-P, Y0, 0], fs=19, cfg=cfg),
        sig_label("C(s)", [obpx+S+P, Y0, 0], fs=19, cfg=cfg))

    def build_outer_stage(widths_labels):
        widths = [w for _, w, _, _ in widths_labels]
        centres, _ = layout_between(sj1_right, obpx, widths)
        blocks, arrows = [], []
        prev_r = sj1_right
        for (label, w, fill, border), cx in zip(widths_labels, centres):
            b = make_block(label, w=w, h=H2, fill=fill, border=border,
                          fs=F2, cfg=cfg).move_to([cx, Y0, 0])
            blocks.append(b)
            arrows.append(arr_h(prev_r, cx-w/2, Y0, cfg=cfg))
            prev_r = cx + w/2
        arrows.append(line_h(prev_r, obpx, Y0, cfg=cfg))
        g = VGroup(sj1_t.copy(), *blocks, VGroup(*arrows),
                  obp_t.copy(), bH1_t.copy(), outer_wiring_t.copy())
        dominant_scale(g, cfg.cdiag_width, min_h=2.6)
        g.move_to([0, cfg.cdiag_y, 0])
        return g, blocks

    NORM = (cfg.parch_base, cfg.ink_border)
    HI   = (cfg.parch_dark, cfg.ink_red)

    sl = display_text("Step 0: Two nested feedback loops", fs=cfg.fs_step+2,
                      color=cfg.ink_step, weight=BOLD, cfg=cfg
                      ).move_to([0, cfg.div_y-0.6, 0])
    scene.play(FadeIn(sl))

    # ── Stage 0: hand-placed full system (inner loop has its own internal
    #    structure — sj2/G2/G3/H2/bp1 — that doesn't fit the simple
    #    single-row forward-chain pattern used by every later stage) ────
    g1x = -2.4
    G1_0 = make_block("G_1", w=1.0, h=H2, fs=F2, cfg=cfg).move_to([g1x, Y0, 0])
    sj2x = -1.2
    sj2_0 = make_sumjunction(r=R2, cfg=cfg).move_to([sj2x, Y0, 0])
    G2_0 = make_block("G_2", w=1.0, h=H2, fs=F2, cfg=cfg).move_to([0.1, Y0, 0])
    G3_0 = make_block("G_3", w=1.0, h=H2, fs=F2, cfg=cfg).move_to([1.4, Y0, 0])
    bp1x = 2.2
    bp1_0 = make_branch_dot(cfg).move_to([bp1x, Y0, 0])
    h2y = -0.4
    H2_0 = make_block("H_2", w=1.0, h=0.65, fill=cfg.parch_base,
                      border=cfg.ink_purple, fs=F2, cfg=cfg).move_to([-0.5, h2y, 0])

    inner_wiring = VGroup(
        arr_h(sj1_right, g1x-0.5, Y0, cfg=cfg),
        arr_h(g1x+0.5, sj2x-R2, Y0, cfg=cfg),
        arr_h(sj2x+R2, G2_0.get_left()[0], Y0, cfg=cfg),
        arr_h(G2_0.get_right()[0], G3_0.get_left()[0], Y0, cfg=cfg),
        line_h(G3_0.get_right()[0], bp1x, Y0, cfg=cfg),
        line_h(bp1x, obpx, Y0, cfg=cfg),
        line_v(bp1x, Y0, h2y, cfg=cfg),
        line_h(bp1x, H2_0.get_right()[0], h2y, cfg=cfg),
        line_h(H2_0.get_left()[0], sj2x, h2y, cfg=cfg),
        arr_v(sj2x, h2y, Y0-R2, cfg=cfg))

    stage0 = VGroup(sj1_t.copy(), G1_0, sj2_0, G2_0, G3_0, H2_0, bp1_0,
                    inner_wiring, obp_t.copy(), bH1_t.copy(), outer_wiring_t.copy())
    dominant_scale(stage0, cfg.cdiag_width, min_h=2.6)
    stage0.move_to([0, cfg.cdiag_y, 0])
    scene.play(FadeIn(stage0, run_time=cfg.t_fade))
    scene.wait(cfg.t_hold_diag)
    scene.play(FadeOut(sl))

    # ── Step 1: reduce the inner loop (G2, G3 with H2) ──────────────────
    inner_group = [sj2_0, G2_0, G3_0, H2_0, bp1_0]
    l = min(m.get_left()[0] for m in inner_group)
    r = max(m.get_right()[0] for m in inner_group)
    t = max(m.get_top()[1] for m in inner_group)
    b = min(m.get_bottom()[1] for m in inner_group)
    hl_c = [(l+r)/2, (t+b)/2, 0]
    _step_with_highlight(scene, hl_c, (r-l)+0.15, (t-b)+0.15,
        "Step 1: Reduce inner loop  G₂·G₃  with  H₂",
        r"G_{23} = \frac{G_2 G_3}{1 + G_2 G_3 H_2}", cfg)

    stage1, s1blocks = build_outer_stage([
        ("G_1", 1.2, *NORM), (r"G_{23}", 1.9, *HI)])
    scene.play(ReplacementTransform(stage0, stage1), run_time=cfg.t_morph)
    scene.wait(0.4)

    # ── Step 2: cascade G1 with G23 ──────────────────────────────────────
    l = s1blocks[0].get_left()[0]; r = s1blocks[1].get_right()[0]
    cy = s1blocks[0].get_center()[1]
    hh = max(m.height for m in s1blocks) + 0.10
    _step_with_highlight(scene, [(l+r)/2, cy, 0], (r-l)+0.10, hh,
        "Step 2: Cascade  G₁ · G₂₃",
        r"G_1 \cdot G_{23} = G_1 G_{23}", cfg)

    stage2, s2blocks = build_outer_stage([(r"G_1 G_{23}", 2.6, *HI)])
    scene.play(ReplacementTransform(stage1, stage2), run_time=cfg.t_morph)
    scene.wait(0.4)

    # ── Step 3: apply the outer feedback formula (Rule 3) ───────────────
    c = stage2.get_center(); w = stage2.width+0.15; h = stage2.height+0.15
    _step_with_highlight(scene, c, w, h,
        "Step 3: Apply feedback formula (Rule 3)",
        r"T(s) = \frac{G_1 G_{23}}{1 + G_1 G_{23} H_1}", cfg, fs=28)

    bf2 = make_block(r"\frac{G_1 G_2 G_3}{(1+G_2G_3H_2)+G_1G_2G_3H_1}",
                     w=6.2, h=1.1, fill=cfg.parch_dark, border=cfg.ink_red,
                     fs=16, cfg=cfg).move_to([0, cfg.cdiag_y, 0])
    af2 = VGroup(bf2,
        arr_h(bf2.get_left()[0]-S,  bf2.get_left()[0],    cfg.cdiag_y, cfg=cfg),
        arr_h(bf2.get_right()[0],   bf2.get_right()[0]+S, cfg.cdiag_y, cfg=cfg),
        sig_label("R(s)",[bf2.get_left()[0]-S-P,  cfg.cdiag_y,0],fs=19,cfg=cfg),
        sig_label("C(s)",[bf2.get_right()[0]+S+P, cfg.cdiag_y,0],fs=19,cfg=cfg))
    safe_width(af2, 8.6)
    scene.play(ReplacementTransform(stage2, af2), run_time=cfg.t_morph)
    scene.wait(0.5)

    fp2 = make_panel(cfg.fbox_y, 3.4, alpha=0.20, cfg=cfg)
    bx2 = RoundedRectangle(corner_radius=0.2, width=8.2, height=3.0,
                           fill_color=cfg.parch_panel,
                           fill_opacity=cfg.panel_alpha,
                           stroke_color=cfg.ink_border,
                           stroke_width=2.0).move_to([0, cfg.fbox_y, 0])
    lb2 = display_text("Final Transfer Function", fs=22,
                       color=cfg.ink_accent, weight=BOLD,
                       cfg=cfg).next_to(bx2, UP, buff=0.22)
    fe2 = safe_width(MathTex(
        r"\frac{C(s)}{R(s)} = "
        r"\frac{G_1 G_2 G_3}{(1+G_2 G_3 H_2) + G_1 G_2 G_3 H_1}",
        tex_template=_TEX, color=cfg.ink_final, font_size=25
        ).move_to(bx2), cfg.panel_w - 0.8)
    scene.play(FadeIn(fp2), FadeIn(bx2), FadeIn(lb2), Write(fe2, run_time=1.2))
    scene.wait(cfg.t_hold_final)
    scene.play(FadeOut(VGroup(hp, ti, su, af2, fp2, bx2, lb2, fe2)))


def intro_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    ip    = make_panel(-0.3, 12.0, alpha=0.15, cfg=cfg)
    title = display_text("Block Diagram\nAlgebra",
                         fs=cfg.fs_title, color=cfg.ink, weight=BOLD,
                         t2c={"Algebra": cfg.ink_accent},
                         line_spacing=1.2, cfg=cfg)
    sub   = display_text("10 Simplification Rules\n+  2 Complex Examples",
                         fs=28, color=cfg.ink, line_spacing=1.4, cfg=cfg)
    auth  = display_text("Mugambi Ndwiga\n@craftsandengineering",
                         fs=20, color=cfg.ink_wm, line_spacing=1.3, cfg=cfg)
    grp = VGroup(title, sub, auth).arrange(DOWN, buff=1.1).move_to([0,-0.3,0])
    scene.play(FadeIn(ip))
    scene.play(Write(title, run_time=cfg.t_write))
    scene.play(FadeIn(sub,  shift=UP*.3, run_time=cfg.t_fade))
    scene.play(FadeIn(auth, run_time=cfg.t_fade))
    scene.wait(cfg.t_hold_intro)
    scene.play(FadeOut(VGroup(ip, grp)))


def outro_anim(scene: Scene, cfg: ParchmentConfig = CFG) -> None:
    op = make_panel(-0.3, 9.5, alpha=0.15, cfg=cfg)
    t1 = display_text("All 10 Rules",         fs=52, color=cfg.ink,
                      weight=BOLD, cfg=cfg)
    t2 = display_text("+ 2 Complex Examples", fs=36, color=cfg.ink_accent,
                      weight=BOLD, cfg=cfg)
    t3 = display_text("Contribute to this Open Source Project\nSee GitHub Sponsors or DM for details",
                      fs=22, color=cfg.ink, line_spacing=1.3, cfg=cfg)
    t4 = display_text("Mugambi Ndwiga\n@craftsandengineering",
                      fs=24, color=cfg.ink_wm, line_spacing=1.3, cfg=cfg)
    grp = VGroup(t1, t2, t3, t4).arrange(DOWN, buff=0.7).move_to([0,-0.3,0])
    scene.play(FadeIn(op))
    scene.play(Write(t1, run_time=1.2))
    scene.play(FadeIn(t2, shift=UP*.2, run_time=cfg.t_fade))
    scene.play(FadeIn(t3, run_time=cfg.t_fade))
    scene.play(FadeIn(t4, run_time=cfg.t_fade))
    scene.wait(cfg.t_hold_outro)
    scene.play(FadeOut(VGroup(op, grp)))


# ══════════════════════════════════════════════════════════════════════════════
#  MAIN VIDEO
# ══════════════════════════════════════════════════════════════════════════════

RULE_ANIMS = [
    rule01_anim, rule02_anim, rule03_anim, rule04_anim, rule05_anim,
    rule06_anim, rule07_anim, rule08_anim, rule09_anim, rule10_anim,
]

class BlockDiagramVideo(Scene):
    """
    Full video — all scenes in sequence.
    Render:   manim -pqh --fps 30 block_diagram_parchment.py BlockDiagramVideo
    Preview:  manim -pql --fps 15 block_diagram_parchment.py BlockDiagramVideo
    """
    def construct(self) -> None:
        self.add(make_parchment_bg(), make_watermark())
        intro_anim(self)
        for fn in RULE_ANIMS:
            fn(self)
        complex01_anim(self)
        complex02_anim(self)
        outro_anim(self)


# ══════════════════════════════════════════════════════════════════════════════
#  INDIVIDUAL SCENES — auto-generated (zero boilerplate)
# ══════════════════════════════════════════════════════════════════════════════

def _make_scene(fn, class_name: str) -> type:
    """Factory: build a named Scene subclass that runs one animation function."""
    def construct(self: Scene) -> None:
        self.add(make_parchment_bg(), make_watermark())
        fn(self)
    return type(class_name, (Scene,), {"construct": construct})

IntroScene            = _make_scene(intro_anim,      "IntroScene")
Rule01_Cascade        = _make_scene(rule01_anim,     "Rule01_Cascade")
Rule02_Parallel       = _make_scene(rule02_anim,     "Rule02_Parallel")
Rule03_FeedbackLoop   = _make_scene(rule03_anim,     "Rule03_FeedbackLoop")
Rule04_MoveSumForward = _make_scene(rule04_anim,     "Rule04_MoveSumForward")
Rule05_MoveSumBeyond  = _make_scene(rule05_anim,     "Rule05_MoveSumBeyond")
Rule06_MoveTakeoffAhead  = _make_scene(rule06_anim,  "Rule06_MoveTakeoffAhead")
Rule07_MoveTakeoffBeyond = _make_scene(rule07_anim,  "Rule07_MoveTakeoffBeyond")
Rule08_SwapSumJunctions  = _make_scene(rule08_anim,  "Rule08_SwapSumJunctions")
Rule09_SplitSumJunction  = _make_scene(rule09_anim,  "Rule09_SplitSumJunction")
Rule10_BlockInFeedback   = _make_scene(rule10_anim,  "Rule10_BlockInFeedback")
Complex01_DCMotor     = _make_scene(complex01_anim,  "Complex01_DCMotor")
Complex02_MultiLoop   = _make_scene(complex02_anim,  "Complex02_MultiLoop")
OutroScene            = _make_scene(outro_anim,      "OutroScene")
```
