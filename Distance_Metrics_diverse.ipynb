A parametric, production-grade explainer animation (vertical 9:16, YouTube Shorts

and answers it nine different ways. The point is the punchline: distance is not a
property of the objects. It is a property of the rule you choose to measure them.

## What the video shows

Two points, A and B, sit on a grid forming a clean 3-4-5 triangle. The same pair
is then measured under nine distance definitions, each presented in the setting
where it naturally lives. A closing summary card lines up all nine answers side by
side: one question, nine numbers.

The sequence is deliberately staged so the metrics build on one another, the three
classical "straight-line" metrics first, then their unifying generalization, then
a shift from position to direction, then metrics for discrete and geographic data.

---

## Background: the nine metrics

A distance (more formally a metric) is a function `d(A, B)` that returns a single
non-negative number describing how dissimilar two objects are. Most useful metrics
satisfy four properties: non-negativity, identity (`d(A, A) = 0`), symmetry, and
the triangle inequality. The nine below all measure the same two objects, yet they
disagree because each encodes a different notion of "closeness."

### 1. Euclidean distance

The familiar straight-line distance from the Pythagorean theorem.

```
d = sqrt(dx^2 + dy^2)
```

For our 3-4-5 triangle this is exactly 5. It is the special case p = 2 of the
Minkowski family below, and the default notion of distance in most of geometry and
physics.

### 2. Manhattan distance (taxicab / L1)

Distance if you can only travel along grid lines, like a taxi crossing a city of
square blocks.

```
d = |dx| + |dy|
```

Here that is 3 + 4 = 7. It is always at least as large as the Euclidean distance,
because diagonal shortcuts are forbidden. Common in routing, lattice problems, and
robust statistics.

### 3. Chebyshev distance (chessboard / L-infinity)

The number of moves a chess king needs, since a king can step diagonally as
cheaply as orthogonally.

```
d = max(|dx|, |dy|)
```

That is max(3, 4) = 4. It is the limiting case of the Minkowski family as p tends
to infinity, where only the largest coordinate difference matters.

### 4. Minkowski distance (the unifying family)

A single parameterized formula that contains the previous three as special cases.

```
d = ( |dx|^p + |dy|^p )^(1/p)
```

- p = 1 gives Manhattan
- p = 2 gives Euclidean
- p approaching infinity gives Chebyshev

The animation morphs the unit "ball" (the set of points exactly one unit away)
from a diamond (p = 1) to a circle (p = 2) and onward, making the family visible
as one continuous idea rather than three disconnected rules.

### 5. Cosine distance

A shift from position to direction. Treating A and B as vectors, cosine similarity
measures the angle between them and ignores their lengths entirely.

```
d = 1 - (A . B) / (||A|| ||B||)
```

Two vectors pointing the same way have distance 0 no matter how long they are.
This is the workhorse of text search, recommendation systems, and embeddings,
where the direction of a feature vector carries the meaning and its magnitude
often does not.

### 6. Hamming distance

For sequences of equal length, the number of positions where the symbols differ.
The animation re-encodes the two objects as 8-bit strings and highlights the
mismatched positions.

```
d = count of positions i where a_i != b_i
```

Foundational in coding theory and error detection and correction: it tells you how
many single-symbol errors it takes to turn one codeword into another.

### 7. Jaccard distance

For comparing sets, based on how much they overlap relative to their combined size.
The video frames it as two people's music tastes.

```
d = 1 - |A intersect B| / |A union B|
```

With two shared tags out of six total distinct tags, similarity is 2/6 and the
distance is about 0.67. Widely used for comparing collections, documents as
bag-of-words, and binary feature sets.

### 8. Sorensen-Dice distance

A close cousin of Jaccard on the same sets, but it weights the shared elements
more heavily by counting the intersection twice.

```
d = 1 - 2|A intersect B| / (|A| + |B|)
```

On the same example this gives 0.5 rather than 0.67. Comparing it directly with
Jaccard shows how two reasonable definitions of set overlap produce different
numbers from identical data, the whole thesis of the piece in miniature.

### 9. Haversine distance

Distance is not even fixed once you change the shape of space. The Haversine
formula gives the great-circle distance between two points on a sphere, the
shortest path along the surface rather than straight through the planet.

```
d = 2r * arcsin( sqrt( sin^2(dphi/2) + cos(phi1) cos(phi2) sin^2(dlambda/2) ) )
```

Using Nairobi and London with Earth's mean radius, the answer is roughly 6,821 km.
This is the standard formula behind real-world navigation and mapping.

### The finale

All nine answers are shown together: 5.00, 7.00, 4.00, 4.50, 0.008, 3, 0.67, 0.50,
and 6,821 km. Same two objects, nine legitimate answers. Distance is not universal;
it is a choice.

## The code

The entire animation lives in a single self-contained module, `nine_distances.py`.
It is built to be read, tweaked, and re-run.

- Renderer: Matplotlib `FuncAnimation` driving an H.264 (libx264) MP4.
- ffmpeg: provided automatically by the `imageio-ffmpeg` package, so no separate
  system install is required (works in Colab, Jupyter, and locally).
- Layout: a fixed 9 x 16 coordinate space with equal aspect, so circles stay
  perfectly round and every element has a deliberate, non-overlapping position.

### Design principles

- Parametric: all timing, colours, geometry, and subject data live in the `PARAMS`
  and `LAYOUT` dictionaries at the top. Change a value, re-render, see the result.
- Scene system: a timeline of named scenes, each a pure function of its local
  progress `p` in [0, 1], composited with shared easing helpers for smooth motion.
- Self-documenting: every formula shown on screen is tied to the visual that
  explains it, and the on-screen symbols match the definitions above.

## Customizing

A few of the most useful knobs in `PARAMS`:

- `dur`: per-scene durations in seconds, controlling pacing.
- `A`, `B`: the two protagonist points for the spatial metrics.
- `code_a`, `code_b`: the bit strings used for Hamming.
- `set_ana`, `set_ben`: the two tag sets used for Jaccard and Sorensen-Dice.
- `city_a`, `city_b`, `earth_r_km`: the two locations and radius used for Haversine.
- palette keys (`bg`, `accent`, `primary`, and so on): the full colour scheme.

Changing any of these automatically updates both the visuals and the computed
answers, since the nine results are derived from the same parameters.

````python
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Concept: The Same Two Things, Nine Different Distances
Repo: github.com/zombimann/Mathematical-video-animations-and-visualization

A parametric, production-grade explainer (YouTube Shorts 9:16) showing that the
"distance" between two objects depends entirely on how distance is defined.
Nine metrics, each shown in its natural habitat:
Euclidean, Manhattan, Chebyshev, Minkowski, Cosine, Hamming, Jaccard,
Sorensen-Dice, Haversine.

Renderer: Matplotlib + ffmpeg (via imageio-ffmpeg) -> compact H.264 mp4 (<10 MB).
Run render() to produce the file, then display_video() and offer_download().
"""

import math
import bisect
import numpy as np

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Circle, Rectangle, Arc, FancyArrowPatch

# Point ffmpeg at the bundled binary so this works in Colab, Jupyter, or locally.
try:
    import imageio_ffmpeg
    matplotlib.rcParams["animation.ffmpeg_path"] = imageio_ffmpeg.get_ffmpeg_exe()
except Exception:
    pass

from matplotlib.animation import FuncAnimation, FFMpegWriter


# ======================================================================
# PARAMETERS  (everything tweakable lives here)
# ======================================================================
PARAMS = dict(
    # --- canvas (9:16 vertical; coordinate space matches inches -> true circles)
    fig_w_in=9.0,
    fig_h_in=16.0,
    dpi=120,                 # 120 -> 1080x1920 final; lower (e.g. 45) for fast preview
    fps=30,

    # --- pacing (seconds)  (tighter middle; quicker onset; bigger finale beat)
    dur=dict(
        intro=4.0,
        euclidean=2.9, manhattan=2.9, chebyshev=2.9, minkowski=3.2,
        cosine=3.0, hamming=3.0, jaccard=3.0, dice=3.0, haversine=3.2,
        finale=5.6, closing=2.0,
    ),

    # --- palette
    bg="#0E1726", panel="#16223C", panel_edge="#2A3A5C",
    ink="#F4F7FF", muted="#9FB0CC",
    point_a="#FF6B6B", point_b="#4ECDC4",
    accent="#FFD166", primary="#6C7BFF",
    good="#8CE99A", warn="#FF8FA3",

    # --- the two protagonists (spatial scenes): a clean 3-4-5 triangle
    A=(0, 0), B=(3, 4),

    # --- subject data for non-spatial metrics
    code_a="10110100",
    code_b="11100101",
    set_ana=("jazz", "rock", "pop", "folk"),
    set_ben=("pop", "folk", "metal", "punk"),
    city_a=("Nairobi", -1.2921, 36.8219),
    city_b=("London", 51.5074, -0.1278),
    earth_r_km=6371.0,

    watermark="© Mugambi Ndwiga / @craftsandengineering",
)

# Layout bands in the 9x16 coordinate space
LAYOUT = dict(
    header_y=15.05,
    stage=(0.85, 5.55, 7.30, 8.35),     # x0, y0, w, h
    caption=(0.85, 1.75, 7.30, 3.05),   # x0, y0, w, h
    dots_y=1.18,
    wm_y=0.50,
)


# ======================================================================
# DERIVED VALUES  (the nine answers, computed once)
# ======================================================================
def _haversine(lat1, lon1, lat2, lon2, r):
    p1, p2 = math.radians(lat1), math.radians(lat2)
    dphi = math.radians(lat2 - lat1)
    dlmb = math.radians(lon2 - lon1)
    a = math.sin(dphi / 2) ** 2 + math.cos(p1) * math.cos(p2) * math.sin(dlmb / 2) ** 2
    return 2 * r * math.asin(math.sqrt(a))


def _minkowski(dx, dy, p):
    return (abs(dx) ** p + abs(dy) ** p) ** (1.0 / p)


_dx = PARAMS["B"][0] - PARAMS["A"][0]
_dy = PARAMS["B"][1] - PARAMS["A"][1]
_setA, _setB = set(PARAMS["set_ana"]), set(PARAMS["set_ben"])
_inter = len(_setA & _setB)
_union = len(_setA | _setB)
_ham = sum(c1 != c2 for c1, c2 in zip(PARAMS["code_a"], PARAMS["code_b"]))

# For the cosine lesson we treat the two protagonists as vectors V1=(1,2), V2=(4,6).
_V1, _V2 = np.array([1.0, 2.0]), np.array([4.0, 6.0])
_cos = float(_V1 @ _V2 / (np.linalg.norm(_V1) * np.linalg.norm(_V2)))
_theta_deg = math.degrees(math.acos(min(1.0, _cos)))

_MINK_P = 3.0  # representative p for the finale readout

ANSWERS = dict(
    euclidean=_minkowski(_dx, _dy, 2),
    manhattan=abs(_dx) + abs(_dy),
    chebyshev=max(abs(_dx), abs(_dy)),
    minkowski=_minkowski(_dx, _dy, _MINK_P),
    cosine=1 - _cos,
    hamming=_ham,
    jaccard=1 - _inter / _union,
    dice=1 - 2 * _inter / (len(_setA) + len(_setB)),
    haversine=_haversine(PARAMS["city_a"][1], PARAMS["city_a"][2],
                         PARAMS["city_b"][1], PARAMS["city_b"][2], PARAMS["earth_r_km"]),
)


# ======================================================================
# SMALL HELPERS
# ======================================================================
def clamp(x, lo=0.0, hi=1.0):
    return max(lo, min(hi, x))


def smooth(x):
    x = clamp(x)
    return x * x * (3 - 2 * x)


def ease_out(x):
    x = clamp(x)
    return 1 - (1 - x) ** 3


def lerp(a, b, t):
    return a + (b - a) * t


def fade_envelope(p, fin=0.16, fout=0.10):
    """Whole-scene alpha: fade content in at start, out at end (smooth cuts)."""
    return smooth(min(p / fin, 1.0)) * smooth(min((1 - p) / fout, 1.0))


def partial_polyline(pts, frac):
    """Return points of polyline drawn up to `frac` of its total arc length."""
    pts = np.asarray(pts, float)
    if frac <= 0:
        return pts[:1]
    seg = np.linalg.norm(np.diff(pts, axis=0), axis=1)
    total = seg.sum()
    if total == 0:
        return pts
    target = frac * total
    out = [pts[0]]
    acc = 0.0
    for i in range(len(seg)):
        if acc + seg[i] >= target:
            r = (target - acc) / seg[i]
            out.append(pts[i] + r * (pts[i + 1] - pts[i]))
            return np.asarray(out)
        acc += seg[i]
        out.append(pts[i + 1])
    return np.asarray(out)


def txt(ax, x, y, s, size, color, weight="normal", ha="center", va="center",
        alpha=1.0, style="normal", family="sans-serif"):
    return ax.text(x, y, s, fontsize=size, color=color, weight=weight, ha=ha, va=va,
                   alpha=alpha, style=style, family=family, zorder=20)


def card(ax, x, y, w, h, face, edge, lw=1.6, alpha=1.0, rad=0.28, z=2):
    box = FancyBboxPatch((x, y), w, h, boxstyle=f"round,pad=0,rounding_size={rad}",
                         linewidth=lw, facecolor=face, edgecolor=edge, alpha=alpha, zorder=z)
    ax.add_patch(box)
    return box


# Spatial grid mapping: data (i,j) -> stage coordinates (equal aspect)
_GU = 1.24                      # grid unit in coordinate space
_ORIGIN = (2.60, 7.25)          # where data point A sits


def gx(i):
    return _ORIGIN[0] + i * _GU


def gy(j):
    return _ORIGIN[1] + j * _GU


# ======================================================================
# CHROME (background, progress dots, watermark)
# ======================================================================
def draw_footer(ax, active, P, wm_alpha=0.6):
    # progress dots for the nine metrics
    n = 9
    x0, x1 = 2.0, 7.0
    for k in range(n):
        x = lerp(x0, x1, k / (n - 1))
        if active == "all":
            c, r = P["accent"], 0.085
        elif active == k:
            c, r = P["accent"], 0.13
        else:
            c, r = P["panel_edge"], 0.075
        ax.add_patch(Circle((x, LAYOUT["dots_y"]), r, color=c, zorder=15))
    # watermark (bottom-left, tiny; dimmed further on busy summary frames)
    txt(ax, 0.35, LAYOUT["wm_y"], P["watermark"], 11.5, P["muted"],
        ha="left", alpha=wm_alpha)


def new_axes(P):
    fig = plt.figure(figsize=(P["fig_w_in"], P["fig_h_in"]), dpi=P["dpi"])
    fig.patch.set_facecolor(P["bg"])
    ax = fig.add_axes([0, 0, 1, 1])
    return fig, ax


def reset_axes(ax, P):
    ax.clear()
    ax.set_xlim(0, 9)
    ax.set_ylim(0, 16)
    ax.set_aspect("equal")
    ax.axis("off")
    ax.set_facecolor(P["bg"])


# ======================================================================
# CAPTION CARD  (name -> formula -> Distance = value)
# ======================================================================
def caption(ax, P, name, name_color, formula, value_str, reveal,
            formula_size=25, note=None):
    x, y, w, h = LAYOUT["caption"]
    card(ax, x, y, w, h, P["panel"], P["panel_edge"], alpha=0.96)
    cx = x + w / 2
    txt(ax, cx, y + h - 0.62, name, 31, name_color, weight="bold")
    txt(ax, cx, y + h - 1.55, formula, formula_size, P["ink"])
    if note:
        txt(ax, cx, y + 1.02, note, 17, P["muted"], style="italic")
        txt(ax, cx, y + 0.50, value_str, 32, P["accent"], weight="bold",
            alpha=ease_out(reveal))
    else:
        txt(ax, cx, y + 0.66, value_str, 34, P["accent"], weight="bold",
            alpha=ease_out(reveal))


def spatial_stage(ax, P, alpha=1.0, board=False):
    """Faint grid + the two protagonist points A and B."""
    sx, sy, sw, sh = LAYOUT["stage"]
    card(ax, sx, sy, sw, sh, P["panel"], P["panel_edge"], alpha=0.55 * alpha)
    # grid lines
    for i in range(-1, 5):
        ax.plot([gx(i), gx(i)], [gy(-1), gy(5)], color=P["panel_edge"],
                lw=1.0, alpha=0.45 * alpha, zorder=3)
    for j in range(-1, 6):
        ax.plot([gx(-1), gx(4)], [gy(j), gy(j)], color=P["panel_edge"],
                lw=1.0, alpha=0.45 * alpha, zorder=3)
    if board:
        for i in range(-1, 4):
            for j in range(-1, 5):
                if (i + j) % 2 == 0:
                    ax.add_patch(Rectangle((gx(i), gy(j)), _GU, _GU,
                                 color=P["panel_edge"], alpha=0.30 * alpha, zorder=2))
    A, B = P["A"], P["B"]
    ax.add_patch(Circle((gx(A[0]), gy(A[1])), 0.20, color=P["point_a"],
                 zorder=12, alpha=alpha))
    ax.add_patch(Circle((gx(B[0]), gy(B[1])), 0.20, color=P["point_b"],
                 zorder=12, alpha=alpha))
    txt(ax, gx(A[0]) - 0.45, gy(A[1]) - 0.05, "A", 22, P["point_a"],
        weight="bold", alpha=alpha)
    txt(ax, gx(B[0]) + 0.45, gy(B[1]) + 0.05, "B", 22, P["point_b"],
        weight="bold", alpha=alpha)


# ======================================================================
# SCENES
# ======================================================================
def scene_cue(ax, p, P, color, idx):
    """A short accent underline that wipes in beneath the header — a distinct
    per-metric scene-change beat. The wipe direction alternates by metric so
    each transition reads differently."""
    a = fade_envelope(p) * 0.9
    w = ease_out(clamp(p / 0.22))
    x_lo, x_hi, y = 1.7, 7.3, 14.28
    if idx % 2 == 0:
        x0, x1 = x_lo, lerp(x_lo, x_hi, w)
    else:
        x1, x0 = x_hi, lerp(x_hi, x_lo, w)
    ax.plot([x0, x1], [y, y], color=color, lw=3.2, alpha=a, zorder=18,
            solid_capstyle="round")


def scene_intro(ax, p, P):
    # Quick onset: title and both points land almost immediately.
    a = fade_envelope(p, fin=0.08, fout=0.08)
    txt(ax, 4.5, LAYOUT["header_y"], "How far apart", 40, P["ink"], weight="bold", alpha=a)
    txt(ax, 4.5, LAYOUT["header_y"] - 0.95, "are these points?", 40, P["ink"],
        weight="bold", alpha=a)

    sx, sy, sw, sh = LAYOUT["stage"]
    card(ax, sx, sy, sw, sh, P["panel"], P["panel_edge"], alpha=0.55 * a)

    # points pop in right away with a tiny overshoot for life
    pop = ease_out(clamp(p / 0.16))
    over = 1.0 + 0.18 * math.sin(math.pi * clamp(p / 0.16))
    A, B = P["A"], P["B"]
    ax.add_patch(Circle((gx(A[0]), gy(A[1])), 0.26 * pop * over, color=P["point_a"], zorder=12, alpha=a))
    ax.add_patch(Circle((gx(B[0]), gy(B[1])), 0.26 * pop * over, color=P["point_b"], zorder=12, alpha=a))
    txt(ax, gx(A[0]) - 0.5, gy(A[1]) - 0.1, "A", 26, P["point_a"], weight="bold", alpha=a * pop)
    txt(ax, gx(B[0]) + 0.5, gy(B[1]) + 0.1, "B", 26, P["point_b"], weight="bold", alpha=a * pop)

    qa = ease_out(clamp((p - 0.22) / 0.28))
    mx = (gx(A[0]) + gx(B[0])) / 2 + 0.6
    my = (gy(A[1]) + gy(B[1])) / 2
    txt(ax, mx, my, "?", 70, P["accent"], weight="bold", alpha=a * qa)

    sub = ease_out(clamp((p - 0.38) / 0.30))
    txt(ax, 4.5, 3.0, "The answer depends on", 26, P["muted"], alpha=a * sub)
    txt(ax, 4.5, 2.3, "how you define distance.", 26, P["ink"], weight="bold", alpha=a * sub)
    draw_footer(ax, "none", P)


def _spatial_header(ax, P, a):
    txt(ax, 4.5, LAYOUT["header_y"], "Same A and B...", 30, P["muted"], alpha=a)


def scene_euclidean(ax, p, P):
    a = fade_envelope(p)
    _spatial_header(ax, P, a)
    spatial_stage(ax, P, alpha=a)
    A, B = P["A"], P["B"]
    grow = ease_out(clamp((p - 0.12) / 0.45))
    # legs
    ax.plot([gx(A[0]), gx(B[0])], [gy(A[1]), gy(A[1])], color=P["muted"], lw=2.2,
            ls=(0, (4, 3)), alpha=a * grow, zorder=8)
    ax.plot([gx(B[0]), gx(B[0])], [gy(A[1]), gy(B[1])], color=P["muted"], lw=2.2,
            ls=(0, (4, 3)), alpha=a * grow, zorder=8)
    txt(ax, (gx(A[0]) + gx(B[0])) / 2, gy(A[1]) - 0.45, "Δx = 3", 20, P["muted"], alpha=a * grow)
    txt(ax, gx(B[0]) + 0.7, (gy(A[1]) + gy(B[1])) / 2, "Δy = 4", 20, P["muted"], alpha=a * grow)
    # hypotenuse
    seg = partial_polyline([(gx(A[0]), gy(A[1])), (gx(B[0]), gy(B[1]))], grow)
    ax.plot(seg[:, 0], seg[:, 1], color=P["accent"], lw=4.0, alpha=a, zorder=10)
    rev = clamp((p - 0.45) / 0.4)
    val = ANSWERS["euclidean"] * ease_out(rev)
    caption(ax, P, "Euclidean", P["accent"],
            r"$d=\sqrt{\Delta x^{2}+\Delta y^{2}}$",
            f"Distance = {val:.2f}", rev)
    scene_cue(ax, p, P, P["accent"], 0)
    draw_footer(ax, 0, P)


def scene_manhattan(ax, p, P):
    a = fade_envelope(p)
    _spatial_header(ax, P, a)
    spatial_stage(ax, P, alpha=a)
    A, B = P["A"], P["B"]
    # city-block staircase: up/right interleaved
    pts = [(A[0], A[1])]
    seq = ["U", "R", "U", "R", "U", "R", "U"]
    ci, cj = A
    for s in seq:
        if s == "R":
            ci += 1
        else:
            cj += 1
        pts.append((ci, cj))
    spts = [(gx(i), gy(j)) for i, j in pts]
    grow = ease_out(clamp((p - 0.12) / 0.5))
    seg = partial_polyline(spts, grow)
    ax.plot(seg[:, 0], seg[:, 1], color=P["accent"], lw=4.0, alpha=a, zorder=10)
    rev = clamp((p - 0.45) / 0.4)
    val = ANSWERS["manhattan"] * ease_out(rev)
    caption(ax, P, "Manhattan", P["accent"],
            r"$d=|\Delta x|+|\Delta y|$",
            f"Distance = {val:.2f}", rev, note="Example: city-block path")
    scene_cue(ax, p, P, P["accent"], 1)
    draw_footer(ax, 1, P)


def scene_chebyshev(ax, p, P):
    a = fade_envelope(p)
    _spatial_header(ax, P, a)
    spatial_stage(ax, P, alpha=a, board=True)
    A, B = P["A"], P["B"]
    # king moves: 3 diagonal then 1 vertical
    pts = [(A[0], A[1])]
    ci, cj = A
    for _ in range(3):
        ci += 1; cj += 1
        pts.append((ci, cj))
    cj += 1
    pts.append((ci, cj))
    spts = [(gx(i), gy(j)) for i, j in pts]
    grow = ease_out(clamp((p - 0.12) / 0.5))
    seg = partial_polyline(spts, grow)
    ax.plot(seg[:, 0], seg[:, 1], color=P["accent"], lw=4.0, alpha=a, zorder=10)
    for i, j in pts:
        ax.add_patch(Circle((gx(i), gy(j)), 0.10, color=P["accent"], alpha=a * grow, zorder=11))
    rev = clamp((p - 0.45) / 0.4)
    val = ANSWERS["chebyshev"] * ease_out(rev)
    caption(ax, P, "Chebyshev", P["accent"],
            r"$d=\max(|\Delta x|,\,|\Delta y|)$",
            f"Distance = {val:.2f}", rev, note="Example: king's moves on a chessboard")
    scene_cue(ax, p, P, P["accent"], 2)
    draw_footer(ax, 2, P)


def _lp_ball(cx, cy, r, p_exp, n=240):
    t = np.linspace(0, 2 * np.pi, n)
    ct, st = np.cos(t), np.sin(t)
    x = np.sign(ct) * np.abs(ct) ** (2.0 / p_exp)
    y = np.sign(st) * np.abs(st) ** (2.0 / p_exp)
    return cx + r * x, cy + r * y


def scene_minkowski(ax, p, P):
    a = fade_envelope(p)
    _spatial_header(ax, P, a)
    spatial_stage(ax, P, alpha=a)
    A, B = P["A"], P["B"]
    ax.plot([gx(A[0]), gx(B[0])], [gy(A[1]), gy(B[1])], color=P["muted"],
            lw=2.0, alpha=a * 0.7, zorder=8)
    # p morphs 1 -> 2 -> 3
    pp = lerp(1.0, _MINK_P, smooth(clamp((p - 0.10) / 0.7)))
    bx, by = _lp_ball(gx(A[0]), gy(A[1]), 1.5, pp)
    ax.plot(bx, by, color=P["primary"], lw=3.0, alpha=a, zorder=9)
    val = _minkowski(_dx, _dy, pp)
    txt(ax, gx(A[0]), gy(A[1]) + 2.05, f"p = {pp:.1f}", 20, P["primary"], weight="bold", alpha=a)
    rev = clamp((p - 0.2) / 0.4)
    caption(ax, P, "Minkowski", P["primary"],
            r"$d=\left(|\Delta x|^{p}+|\Delta y|^{p}\right)^{1/p}$",
            f"Distance = {val:.2f}", rev,
            note="p=1 Manhattan  -  p=2 Euclidean  -  p→∞ Chebyshev")
    scene_cue(ax, p, P, P["primary"], 3)
    draw_footer(ax, 3, P)


def scene_cosine(ax, p, P):
    a = fade_envelope(p)
    txt(ax, 4.5, LAYOUT["header_y"], "Now compare direction", 30, P["muted"], alpha=a)
    sx, sy, sw, sh = LAYOUT["stage"]
    card(ax, sx, sy, sw, sh, P["panel"], P["panel_edge"], alpha=0.55 * a)
    O = (2.2, 6.3)
    scale = 0.95
    grow = ease_out(clamp((p - 0.12) / 0.45))
    for v, col, lab in ((_V1, P["point_a"], "A"), (_V2, P["point_b"], "B")):
        tip = (O[0] + v[0] * scale * grow, O[1] + v[1] * scale * grow)
        ax.add_patch(FancyArrowPatch(O, tip, arrowstyle="-|>", mutation_scale=22,
                     color=col, lw=3.5, alpha=a, zorder=10))
        txt(ax, tip[0] + 0.25, tip[1] + 0.15, lab, 20, col, weight="bold", alpha=a * grow)
    ang = ease_out(clamp((p - 0.4) / 0.35))
    a1 = math.degrees(math.atan2(_V1[1], _V1[0]))
    a2 = math.degrees(math.atan2(_V2[1], _V2[0]))
    ax.add_patch(Arc(O, 2.4, 2.4, angle=0, theta1=min(a1, a2),
                 theta2=min(a1, a2) + (max(a1, a2) - min(a1, a2)) * ang,
                 color=P["accent"], lw=3.0, alpha=a, zorder=9))
    txt(ax, O[0] + 1.7, O[1] + 1.0, f"θ ≈ {_theta_deg:.1f}°", 20,
        P["accent"], alpha=a * ang)
    rev = clamp((p - 0.5) / 0.4)
    val = ANSWERS["cosine"] * ease_out(rev)
    caption(ax, P, "Cosine", P["accent"],
            r"$d=1-\dfrac{A\cdot B}{\|A\|\,\|B\|}$",
            f"Distance = {val:.3f}", rev, note="It's more about direction, not length")
    scene_cue(ax, p, P, P["accent"], 4)
    draw_footer(ax, 4, P)


def scene_hamming(ax, p, P):
    a = fade_envelope(p)
    txt(ax, 4.5, LAYOUT["header_y"], "Re-encode as bits", 30, P["muted"], alpha=a)
    sx, sy, sw, sh = LAYOUT["stage"]
    card(ax, sx, sy, sw, sh, P["panel"], P["panel_edge"], alpha=0.55 * a)
    ca, cb = P["code_a"], P["code_b"]
    n = len(ca)
    cw = 0.78
    x0 = 4.5 - (n * cw) / 2 + cw / 2
    yA, yB = 11.4, 8.2
    txt(ax, x0 - 1.05, yA, "A", 22, P["point_a"], weight="bold", alpha=a)
    txt(ax, x0 - 1.05, yB, "B", 22, P["point_b"], weight="bold", alpha=a)
    sweep = clamp((p - 0.15) / 0.55) * n
    for k in range(n):
        cx = x0 + k * cw
        shown = k < sweep
        diff = ca[k] != cb[k]
        for (yy, ch, base) in ((yA, ca[k], P["point_a"]), (yB, cb[k], P["point_b"])):
            ec = P["warn"] if (diff and shown) else P["panel_edge"]
            ax.add_patch(FancyBboxPatch((cx - cw / 2 + 0.06, yy - 0.42), cw - 0.12, 0.84,
                         boxstyle="round,pad=0,rounding_size=0.10",
                         facecolor=P["bg"], edgecolor=ec, lw=2.0, alpha=a, zorder=6))
            txt(ax, cx, yy, ch, 24, base if shown else P["panel_edge"], weight="bold", alpha=a)
        if diff and shown:
            ax.plot([cx, cx], [yA - 0.5, yB + 0.5], color=P["warn"], lw=2.0,
                    ls=(0, (3, 3)), alpha=a, zorder=5)
            txt(ax, cx, (yA + yB) / 2, "≠", 22, P["warn"], weight="bold", alpha=a)
    rev = clamp((p - 0.55) / 0.35)
    val = int(round(ANSWERS["hamming"] * ease_out(rev)))
    caption(ax, P, "Hamming", P["accent"],
            r"$d=\sum_i [\,a_i \neq b_i\,]$",
            f"Distance = {val}", rev, note="Counts positions that disagree")
    scene_cue(ax, p, P, P["accent"], 5)
    draw_footer(ax, 5, P)


def _venn(ax, P, a, emphasize_inter=1.0):
    sx, sy, sw, sh = LAYOUT["stage"]
    card(ax, sx, sy, sw, sh, P["panel"], P["panel_edge"], alpha=0.55 * a)
    cy = sy + sh / 2 + 0.3
    r = 2.05
    cA, cB = (3.55, cy), (5.45, cy)
    ax.add_patch(Circle(cA, r, facecolor=P["point_a"], edgecolor=P["point_a"],
                 alpha=0.22 * a, lw=3, zorder=5))
    ax.add_patch(Circle(cB, r, facecolor=P["point_b"], edgecolor=P["point_b"],
                 alpha=0.22 * a, lw=3, zorder=5))
    ax.add_patch(Circle(cA, r, facecolor="none", edgecolor=P["point_a"], lw=3, alpha=a, zorder=6))
    ax.add_patch(Circle(cB, r, facecolor="none", edgecolor=P["point_b"], lw=3, alpha=a, zorder=6))
    txt(ax, cA[0] - 1.15, cy + 2.55, "Ana", 22, P["point_a"], weight="bold", alpha=a)
    txt(ax, cB[0] + 1.15, cy + 2.55, "Ben", 22, P["point_b"], weight="bold", alpha=a)
    # counts
    txt(ax, cA[0] - 0.95, cy, "2", 30, P["ink"], weight="bold", alpha=a)
    txt(ax, cB[0] + 0.95, cy, "2", 30, P["ink"], weight="bold", alpha=a)
    ax.add_patch(Circle(((cA[0] + cB[0]) / 2, cy), 0.62 * emphasize_inter,
                 color=P["accent"], alpha=0.30 * a, zorder=6))
    txt(ax, (cA[0] + cB[0]) / 2, cy, "2", 30, P["accent"], weight="bold", alpha=a)
    txt(ax, 4.5, cy - 2.75, "shared: pop, folk", 18, P["muted"], style="italic", alpha=a)


def scene_jaccard(ax, p, P):
    a = fade_envelope(p)
    txt(ax, 4.5, LAYOUT["header_y"], "Compare as sets", 30, P["muted"], alpha=a)
    _venn(ax, P, a)
    rev = clamp((p - 0.45) / 0.4)
    val = ANSWERS["jaccard"] * ease_out(rev)
    caption(ax, P, "Jaccard", P["accent"],
            r"$d=1-\dfrac{|A\cap B|}{|A\cup B|}$",
            f"Distance = {val:.2f}", rev, note="overlap divided by total")
    scene_cue(ax, p, P, P["accent"], 6)
    draw_footer(ax, 6, P)


def scene_dice(ax, p, P):
    a = fade_envelope(p)
    txt(ax, 4.5, LAYOUT["header_y"], "Same sets, more weight", 30, P["muted"], alpha=a)
    pulse = 1.0 + 0.25 * math.sin(p * 12) * smooth(clamp((p - 0.2) / 0.4))
    _venn(ax, P, a, emphasize_inter=pulse)
    rev = clamp((p - 0.45) / 0.4)
    val = ANSWERS["dice"] * ease_out(rev)
    caption(ax, P, "Sørensen–Dice", P["accent"],
            r"$d=1-\dfrac{2\,|A\cap B|}{|A|+|B|}$",
            f"Distance = {val:.2f}", rev, note="overlap is counted twice")
    scene_cue(ax, p, P, P["accent"], 7)
    draw_footer(ax, 7, P)


def _globe(ax, P, a):
    sx, sy, sw, sh = LAYOUT["stage"]
    card(ax, sx, sy, sw, sh, P["panel"], P["panel_edge"], alpha=0.55 * a)
    cx, cy, R = 4.5, sy + sh / 2 + 0.2, 3.0
    ax.add_patch(Circle((cx, cy), R, facecolor="#0c2a4a", edgecolor=P["point_b"],
                 lw=3, alpha=a, zorder=5))
    for k in range(-2, 3):
        w = R * math.cos(k * 0.5)
        ax.add_patch(Arc((cx, cy + R * math.sin(k * 0.5) * 0.0), 2 * w, 2 * R,
                     theta1=0, theta2=360, color=P["panel_edge"], lw=1.0, alpha=0.5 * a, zorder=6))
    for k in range(-2, 3):
        yy = cy + (k / 2.5) * R
        hw = math.sqrt(max(R * R - (yy - cy) ** 2, 0))
        ax.plot([cx - hw, cx + hw], [yy, yy], color=P["panel_edge"], lw=1.0, alpha=0.5 * a, zorder=6)
    return cx, cy, R


def scene_haversine(ax, p, P):
    a = fade_envelope(p)
    txt(ax, 4.5, LAYOUT["header_y"], "What if space is curved?", 30, P["muted"], alpha=a)
    cx, cy, R = _globe(ax, P, a)
    # two cities placed on the disc (stylized)
    pA = (cx - 0.6, cy - 1.2)
    pB = (cx + 1.1, cy + 1.6)
    ax.add_patch(Circle(pA, 0.16, color=P["point_a"], zorder=11, alpha=a))
    ax.add_patch(Circle(pB, 0.16, color=P["point_b"], zorder=11, alpha=a))
    txt(ax, pA[0] - 0.1, pA[1] - 0.5, P["city_a"][0], 17, P["point_a"], weight="bold", alpha=a)
    txt(ax, pB[0], pB[1] + 0.45, P["city_b"][0], 17, P["point_b"], weight="bold", alpha=a)
    # great-circle arc (curved)
    grow = ease_out(clamp((p - 0.15) / 0.5))
    t = np.linspace(0, 1, 60)
    mid = ((pA[0] + pB[0]) / 2 - 0.7, (pA[1] + pB[1]) / 2 + 0.7)
    arc = [( (1 - tt) ** 2 * pA[0] + 2 * (1 - tt) * tt * mid[0] + tt ** 2 * pB[0],
             (1 - tt) ** 2 * pA[1] + 2 * (1 - tt) * tt * mid[1] + tt ** 2 * pB[1]) for tt in t]
    seg = partial_polyline(arc, grow)
    ax.plot(seg[:, 0], seg[:, 1], color=P["accent"], lw=4.0, alpha=a, zorder=10)
    rev = clamp((p - 0.5) / 0.4)
    val = ANSWERS["haversine"] * ease_out(rev)
    caption(ax, P, "Haversine", P["accent"],
            r"$d=2r\arcsin\!\sqrt{\sin^{2}\frac{\Delta\phi}{2}+\cos\phi_1\cos\phi_2\sin^{2}\frac{\Delta\lambda}{2}}$",
            f"Distance = {val:,.0f} km", rev, formula_size=17,
            note="Example: great-circle on a sphere")
    scene_cue(ax, p, P, P["accent"], 8)
    draw_footer(ax, 8, P)


FINALE_ITEMS = [
    ("Euclidean", f"{ANSWERS['euclidean']:.2f}"),
    ("Manhattan", f"{ANSWERS['manhattan']:.2f}"),
    ("Chebyshev", f"{ANSWERS['chebyshev']:.2f}"),
    ("Minkowski", f"{ANSWERS['minkowski']:.2f}"),
    ("Cosine", f"{ANSWERS['cosine']:.3f}"),
    ("Hamming", f"{int(ANSWERS['hamming'])}"),
    ("Jaccard", f"{ANSWERS['jaccard']:.2f}"),
    ("Sørensen–Dice", f"{ANSWERS['dice']:.2f}"),
    ("Haversine", f"{ANSWERS['haversine']:,.0f} km"),
]


def scene_finale(ax, p, P):
    a = fade_envelope(p, fin=0.10, fout=0.08)
    txt(ax, 4.5, LAYOUT["header_y"], "Same question.", 36, P["ink"], weight="bold", alpha=a)
    txt(ax, 4.5, LAYOUT["header_y"] - 0.95, "Nine answers.", 36, P["accent"], weight="bold", alpha=a)
    cols = 3
    gw, gh = 2.45, 2.30
    x0 = 4.5 - (cols * gw) / 2 + gw / 2
    y_top = 12.6

    # The grid rises in as one block (gentle top-down row stagger keeps it lively).
    grid_in = ease_out(clamp((p - 0.06) / 0.24))
    # Every answer then lands together as a group, with a single shared punch.
    vt = clamp((p - 0.34) / 0.16)
    val_in = ease_out(vt)
    punch = 1.0 + 0.14 * math.sin(math.pi * clamp(vt, 0.0, 1.0))
    for idx, (name, val) in enumerate(FINALE_ITEMS):
        r, c = divmod(idx, cols)
        cx = x0 + c * gw
        cy = y_top - r * gh
        rise = (1.0 - ease_out(clamp((p - 0.06 - r * 0.04) / 0.22))) * 0.55
        cyy = cy - rise
        ap = grid_in * a
        card(ax, cx - gw / 2 + 0.12, cyy - gh / 2 + 0.12, gw - 0.24, gh - 0.24,
             P["panel"], P["panel_edge"], alpha=0.96 * ap)
        # accent halo that arrives with the grouped value reveal -> emphasis
        if val_in > 0.01:
            card(ax, cx - gw / 2 + 0.12, cyy - gh / 2 + 0.12, gw - 0.24, gh - 0.24,
                 "none", P["accent"], lw=2.4, alpha=0.85 * val_in * a, z=3)
        txt(ax, cx, cyy + 0.45, name, 18, P["muted"], weight="bold", alpha=ap)
        txt(ax, cx, cyy - 0.42, val, 24 * punch, P["accent"], weight="bold",
            alpha=val_in * a)
    end = ease_out(clamp((p - 0.62) / 0.22)) * a
    txt(ax, 4.5, 2.55, "Distance is not universal.", 31, P["ink"], weight="bold", alpha=end)
    txt(ax, 4.5, 1.85, "It is a choice.", 24, P["accent"], style="italic", alpha=end)
    draw_footer(ax, "all", P, wm_alpha=0.32)


def scene_closing(ax, p, P):
    # Cross-fade in from the (dark) finale: the primary panel washes up over the
    # background, then the credit lines settle a beat later.
    bg_in = smooth(clamp(p / 0.32))
    ax.add_patch(Rectangle((0, 0), 9, 16, color=P["primary"], alpha=bg_in, zorder=1))
    a = ease_out(clamp((p - 0.22) / 0.30)) * smooth(min((1 - p) / 0.20, 1.0))
    txt(ax, 4.5, 9.3, "Made by Mugambi Ndwiga", 34, "#FFFFFF", weight="bold", alpha=a)
    txt(ax, 4.5, 8.2, "@craftsandengineering", 26, "#FFFFFF", alpha=a * 0.92)


SCENE_FUNCS = dict(
    intro=scene_intro, euclidean=scene_euclidean, manhattan=scene_manhattan,
    chebyshev=scene_chebyshev, minkowski=scene_minkowski, cosine=scene_cosine,
    hamming=scene_hamming, jaccard=scene_jaccard, dice=scene_dice,
    haversine=scene_haversine, finale=scene_finale, closing=scene_closing,
)
SCENE_ORDER = ["intro", "euclidean", "manhattan", "chebyshev", "minkowski",
               "cosine", "hamming", "jaccard", "dice", "haversine",
               "finale", "closing"]


# ======================================================================
# RENDER
# ======================================================================
def build_timeline(P):
    starts, t = [], 0.0
    for key in SCENE_ORDER:
        starts.append(t)
        t += P["dur"][key]
    return starts, t


def render(output="nine_distances.mp4", dpi=None, fps=None, crf=24):
    P = dict(PARAMS)
    if dpi is not None:
        P["dpi"] = dpi
    if fps is not None:
        P["fps"] = fps
    starts, total_t = build_timeline(P)
    total_frames = int(round(total_t * P["fps"]))
    fig, ax = new_axes(P)

    def update(frame):
        t = frame / P["fps"]
        i = bisect.bisect_right(starts, t) - 1
        i = max(0, min(i, len(SCENE_ORDER) - 1))
        key = SCENE_ORDER[i]
        local = (t - starts[i]) / P["dur"][key]
        reset_axes(ax, P)
        SCENE_FUNCS[key](ax, clamp(local, 0, 1), P)
        return []

    writer = FFMpegWriter(
        fps=P["fps"], codec="libx264",
        extra_args=["-vf", "scale=trunc(iw/2)*2:trunc(ih/2)*2",
                    "-pix_fmt", "yuv420p", "-crf", str(crf),
                    "-preset", "medium", "-movflags", "+faststart"],
    )
    anim = FuncAnimation(fig, update, frames=total_frames, interval=1000 / P["fps"], blit=False)
    anim.save(output, writer=writer, dpi=P["dpi"])
    plt.close(fig)
    return output


# ======================================================================
# DISPLAY + DOWNLOAD  (notebook-friendly, least-problematic path)
# ======================================================================
def display_video(path="nine_distances.mp4", width=360):
    from IPython.display import Video, display
    display(Video(path, embed=True, width=width))


def offer_download(path="nine_distances.mp4"):
    try:
        from google.colab import files  # type: ignore
        files.download(path)
    except Exception:
        from IPython.display import FileLink, display
        display(FileLink(path))


# --- run: render, show, then offer the file for download ---
out = render()
import os
print("Wrote", out, f"({os.path.getsize(out) / 1e6:.2f} MB)")
display_video(out)
offer_download(out)
````
