# The Factorial Extension: Visualizing the Gamma Function

This notebook provides an educational animation explaining how the discrete factorial operation ($n!$) is extended to all real numbers using the **Gamma Function** ($\Gamma(x)$).

### Key Concepts:
- **Discrete Factorials**: Points like $3! = 6$ and $4! = 24$.
- **The Gamma Function**: $\Gamma(n) = (n-1)!$, providing a smooth interpolation.
- **Half-Factorials**: Calculating non-integer values like $3.5! \approx 11.63$.

### Requirements:
- `numpy`
- `matplotlib`
- `scipy`

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Data preparation
x_vals = np.linspace(0, 5, 200)
y_gamma = gamma(x_vals + 1)
factorial_points = np.arange(0, 6)
y_fact = [gamma(p + 1) for p in factorial_points]

# Plot setup
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 5.5)
ax.set_ylim(0, 130)
ax.set_xlabel("x", fontsize=12)
ax.set_ylabel("x! (Γ(x+1))", fontsize=12)
ax.set_title("The Factorial Extension: What is 3.5!?", fontsize=14)

# Static elements
ax.scatter(factorial_points, y_fact, color='blue', s=80, label="Integer factorials", zorder=5)
line, = ax.plot([], [], 'r-', lw=2.5, label="Γ(x+1)", alpha=0.8)
v_line = ax.axvline(0, color='gray', linestyle='--', alpha=0.5)
anotation = ax.annotate('', xy=(0, 0), xytext=(20, 20),
                        textcoords='offset points',
                        bbox=dict(boxstyle='round', fc='white', alpha=0.7),
                        arrowprops=dict(arrowstyle='->'))

plt.legend(loc='upper left')
plt.grid(True, which='both', linestyle=':', alpha=0.6)

def animate(i):
    current_x = x_vals[i]
    current_y = y_gamma[i]

    # Update curve
    line.set_data(x_vals[:i], y_gamma[:i])

    # Update vertical scanner and annotation
    v_line.set_xdata([current_x])
    anotation.xy = (current_x, current_y)
    anotation.set_text(f"x = {current_x:.2f}\nx! ≈ {current_y:.2f}")

    # Highlight 3.5 specifically when reached
    if abs(current_x - 3.5) < 0.03:
        anotation.get_bbox_patch().set_facecolor('yellow')
    else:
        anotation.get_bbox_patch().set_facecolor('white')

    return line, v_line, anotation

# Create animation
ani = FuncAnimation(fig, animate, frames=len(x_vals), interval=40, blit=True)

# Close plot to prevent static display
plt.close()

# Display in Colab
HTML(ani.to_html5_video())
```
