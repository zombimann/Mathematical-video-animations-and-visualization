# Cinematic General Relativity Visualization

This notebook generates a high-fidelity 3D animation illustrating the fundamental geometric principles of Einstein's General Theory of Relativity. Using Matplotlib's 3D engine, it visualizes how mass-energy content determines the curvature of spacetime.

### Concepts Represented:

1. **Minkowski Space**: The baseline of Special Relativity. A flat, four-dimensional manifold where spacetime remains unwarped in the absence of mass.
2. **Schwarzschild Singularity**: A visualization of the curvature around a non-rotating, spherically symmetric mass. It illustrates the 'gravity well' effect where spacetime geometry becomes increasingly distorted near the event horizon.
3. **Binary System Inspiral**: Represents two massive objects (like black holes or neutron stars) orbiting each other. As they lose energy to gravitational radiation, their orbit decays, creating a dynamic, rotating deformation in the local geometry.
4. **Gravitational Waves**: As predicted by the linearized Einstein field equations, these are ripples in the curvature of spacetime that propagate outward from the source at the speed of light.

### Visual Details:
- **Grid**: Neon-cyan wireframe representing the spatial dimensions of the metric.
- **Masses**: Navy blue spheres indicating the location of energy density.
- **Annotations**: Cyan for conceptual descriptions and High-contrast Yellow for the corresponding mathematical formulations (Einstein Field Equations, Wave Equations, etc.).

```python
r"""
General Relativity Visualization
-------------------------------
Author: Mugambi Ndwiga (@craftsandengineering)
Description: A cinematic 3D animation illustrating Minkowski space, Schwarzschild
singularity, binary inspirals, and gravitational waves using Matplotlib.
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib

matplotlib.rcParams['animation.embed_limit'] = 100.0

# Animation parameters
FPS = 20
DURATION = 30
TOTAL_FRAMES = FPS * DURATION

# Spacetime grid setup
x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
X, Y = np.meshgrid(x, y)

# Figure initialization
fig = plt.figure(figsize=(12, 10), facecolor='black')
ax = fig.add_subplot(111, projection='3d')
ax.set_facecolor('black')

def get_scene_data(t):
    """Calculates grid deformation and metadata based on progress (0.0 to 1.0)."""
    if t < 0.2:
        z = np.zeros_like(X)
        desc = "Minkowski Space: Flat spacetime geometry."
        math = r"$\mathbf{ds^2 = -c^2dt^2 + dx^2 + dy^2 + dz^2}$"
        return z, [], "I. Minkowski Space", desc, math

    elif t < 0.45:
        prog = (t - 0.2) / 0.25
        r = np.sqrt(X**2 + Y**2)
        depth = -2.5 * prog
        z = depth * np.exp(-r**2)
        desc = "Schwarzschild Geometry: Vacuum solution for spherical mass."
        math = r"$\mathbf{G_{\mu\nu} = \frac{8\pi G}{c^4} T_{\mu\nu}}$"
        return z, [(0, 0, 0.4)], "II. Schwarzschild Singularity", desc, math

    elif t < 0.75:
        prog = (t - 0.45) / 0.30
        angle = prog * 4 * np.pi
        x1, y1 = 1.5 * np.cos(angle), 1.5 * np.sin(angle)
        x2, y2 = -1.5 * np.cos(angle), -1.5 * np.sin(angle)
        r1, r2 = np.sqrt((X-x1)**2 + (Y-y1)**2), np.sqrt((X-x2)**2 + (Y-y2)**2)
        z = -1.8 * (np.exp(-r1**2) + np.exp(-r2**2))
        desc = "Binary Inspiral: Quadrupole radiation from orbiting masses."
        math = r"$\mathbf{P = \frac{32}{5} \frac{G^4}{c^5} \frac{M_1^2 M_2^2 (M_1+M_2)}{r^5}}$"
        return z, [(x1, y1, 0.4), (x2, y2, 0.4)], "III. Binary System Inspiral", desc, math

    else:
        prog = (t - 0.75) / 0.25
        r = np.sqrt(X**2 + Y**2)
        z = 0.6 * np.sin(2 * r - prog * 12 * np.pi) * np.exp(-r*0.2)
        desc = r"Gravitational Waves: Perturbations in the metric tensor $h_{\mu\nu}$."
        math = r"$\mathbf{\nabla^2 h_{\mu\nu} = 0}$"
        return z, [], "IV. Gravitational Wave Propagation", desc, math

def update(frame):
    """Animation frame update function."""
    ax.clear()
    ax.set_facecolor('black')
    t = frame / TOTAL_FRAMES
    Z, sources, subtitle, desc, math = get_scene_data(t)

    surf = ax.plot_wireframe(X, Y, Z, color='#00ffcc', linewidth=0.6, alpha=0.4)

    for sx, sy, sz in sources:
        ax.scatter([sx], [sy], [sz], color='#000080', s=200, edgecolors='#0000CD', lw=1.5, zorder=20)

    ax.view_init(elev=28 + 4 * np.sin(t * np.pi), azim=frame * 0.4)

    ax.text2D(0.5, 0.96, "GENERAL RELATIVITY: SPACETIME GEOMETRY", transform=ax.transAxes, color='white',\
              fontsize=16, ha='center', fontweight='bold', alpha=0.9)

    ax.text2D(0.5, 0.92, subtitle, transform=ax.transAxes, color='#00ffcc',\
              fontsize=12, ha='center', style='italic', alpha=0.8)

    # Watermark
    watermark = "Visualized by Mugambi Ndwiga\n@craftsandengineering"
    ax.text2D(0.99, 0.01, watermark, transform=ax.transAxes, color='white', \
              fontsize=9, alpha=0.7, ha='right', va='bottom', family='sans-serif')

    # Annotations positioning
    ax.text2D(0.5, 0.08, desc, transform=ax.transAxes, color='#00ffcc',\
              fontsize=15, ha='center', va='bottom', weight='normal')
    ax.text2D(0.5, 0.03, math, transform=ax.transAxes, color='#ffffaa',\
              fontsize=15, ha='center', va='bottom', weight='normal')

    ax.set_axis_off()
    ax.set_zlim(-4, 2)
    return surf,

print("Generating cinematic relativity animation...")
ani = FuncAnimation(fig, update, frames=TOTAL_FRAMES, interval=1000/FPS)
plt.close()
HTML(ani.to_jshtml())
```
