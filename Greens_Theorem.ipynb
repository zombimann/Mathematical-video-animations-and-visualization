# Green's Theorem Animation in Manim

This notebook contains a high-quality visualization of Green's Theorem, implemented using the Manim community engine. The goal is to provide a clear geometric interpretation of a fundamental result in vector calculus.

## Theoretical Background

Green's Theorem establishes a fundamental relationship between a line integral around a simple closed curve $C$ and a double integral over the plane region $D$ bounded by $C$. It is a two-dimensional specialization of the generalized Stokes' Theorem.

### The Theorem

Let $C$ be a positively oriented, piecewise smooth, simple closed curve in a plane, and let $D$ be the region bounded by $C$. If $L$ and $M$ are functions of $(x, y)$ defined on an open region containing $D$ and have continuous partial derivatives there, then:

$$\oint_C (L\,dx + M\,dy) = \iint_D \left(\frac{\partial M}{\partial x} - \frac{\partial L}{\partial y}\right) dA$$

In vector notation, where $\mathbf{F} = \langle L, M \rangle$, this is expressed as:

$$\oint_C \mathbf{F} \cdot d\mathbf{r} = \iint_D (\text{curl } \mathbf{F}) \cdot \mathbf{k} \, dA$$

### Geometric Interpretation

*   **Line Integral (Left Side):** Represents the macroscopic circulation of the vector field $\mathbf{F}$ along the boundary $C$.
*   **Double Integral (Right Side):** Represents the sum of all microscopic counter-clockwise rotations (curl) within the region $D$.

Green's Theorem proves that the total amount of 'swirl' inside the region is exactly equal to the flow along the boundary.

## Animation Overview

The following Python code generates a video that:
1.  Displays the vector field and the defined region $D$.
2.  Highlights the boundary $C$ to illustrate the line integral component.
3.  Visualizes the curl within the interior to illustrate the double integral component.
4.  Synchronizes these visual elements with the mathematical formula.

```python
from manim import *
import numpy as np
from google.colab import files

class GreensTheoremAnimation(Scene):
    """
    An educational animation demonstrating Green's Theorem.

    This scene visualizes the relationship between a line integral around a
    closed curve C and a double integral over the plane region D it encloses.
    Visuals include a vector field, highlighted boundary, and the interior curl.
    """
    def construct(self):
        # 1. Subtle Integrated Watermark
        watermark = VGroup(
            Text("@craftsandengineering", font_size=16),
            Text("Mugambi Ndwiga", font_size=14, weight=LIGHT)
        ).arrange(DOWN, aligned_edge=RIGHT, buff=0.1).to_corner(DR, buff=0.4).set_opacity(0.4)
        self.add(watermark)

        # 2. Setup Header & Formula (Top Left to avoid center)
        header = Text("Green's Theorem", font_size=32, color=BLUE_B).to_edge(UP, buff=0.3)
        formula = MathTex(
            r"\oint_C \mathbf{F} \cdot d\mathbf{r} = \iint_D (\text{curl } \mathbf{F}) \, dA",
            color=YELLOW_B
        ).scale(0.7).to_corner(UL, buff=0.5)

        self.add(header)
        self.play(FadeIn(formula, shift=RIGHT), run_time=1)

        # 3. Vector Field and Region Setup
        func = lambda pos: np.array([-pos[1], pos[0], 0])
        vector_field = ArrowVectorField(
            func,
            x_range=[-3.5, 3.5],
            y_range=[-2.5, 2.5],
            colors=[BLUE_A, GREEN_A]
        ).set_opacity(0.4)

        circle = Circle(radius=1.8, color=WHITE, stroke_width=4)
        region = Circle(radius=1.8, fill_opacity=0.2, fill_color=BLUE_E, stroke_width=0)

        label_c = MathTex("C", color=WHITE).next_to(circle, UR, buff=-0.2)
        label_d = MathTex("D", color=BLUE_B).move_to(region.get_center())

        self.play(FadeIn(vector_field), Create(circle), run_time=2)
        self.play(FadeIn(region), Write(label_c), Write(label_d), run_time=1)

        # 4. Educational Annotations Linked to Visuals
        ann_1 = Text("1. Measuring the flow ALONG the boundary (C)", font_size=22).to_edge(DOWN, buff=1.0)
        ann_2 = Text("2. Is the same as summing the 'swirl' INSIDE (D)", font_size=22).to_edge(DOWN, buff=1.0)
        ann_3 = Text("Green's Theorem connects the edge to the surface.", font_size=24, color=YELLOW_B).to_edge(DOWN, buff=1.0)

        # Step 1: Highlight Boundary
        self.play(Write(ann_1), run_time=1.5)
        self.play(circle.animate.set_stroke(color=YELLOW, width=8), run_time=1)
        self.wait(2)

        # Step 2: Highlight Interior/Swirl
        self.play(ReplacementTransform(ann_1, ann_2), run_time=1)
        self.play(
            circle.animate.set_stroke(color=WHITE, width=4),
            region.animate.set_fill(color=YELLOW, opacity=0.4),
            run_time=1
        )

        # Show the 'swirl' (curl) visually
        rotating_field = vector_field.copy().set_color(YELLOW).set_opacity(0.8)
        self.play(rotating_field.animate.scale(1.05).scale(1/1.05), run_time=2, rate_func=there_and_back)
        self.wait(1.5)

        # Step 3: Final Implication
        self.play(ReplacementTransform(ann_2, ann_3), run_time=1)
        self.play(Indicate(formula), run_time=2)
        self.wait(4)

        # 5. Clean Exit
        self.play(FadeOut(Group(*self.mobjects)), run_time=1.5)

# Render
%manim -qm -v WARNING GreensTheoremAnimation

# Download
print("Processing video...")
!find ./media/videos -name "GreensTheoremAnimation.mp4" -exec cp {} ./GreensTheorem_Educational.mp4 \;
files.download('GreensTheorem_Educational.mp4')
```
