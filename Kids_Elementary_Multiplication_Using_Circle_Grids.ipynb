# Project Overview: The Splitting Trick

## Purpose
This notebook contains the source code for an animated educational short designed to teach the distributive property of multiplication to elementary students. By utilizing the Manim animation engine, the lesson bridges the gap between concrete counting and abstract algebraic reasoning using the 'Area Model' or 'Grid Method'.

## Educational Model
The animation follows the Concrete-Representational-Abstract (CRA) sequence, a research-based instructional framework:
1. **Concrete Stage**: Students see individual units (circles) organized into rows and columns.
2. **Representational Stage**: The grid is visually split into two distinct color-coded blocks, representing the decomposition of a large factor.
3. **Abstract Stage**: The visual manipulation is tied to numerical equations (a × b = a × c + a × d), allowing students to internalize the mathematical property.

## Multimedia Learning Principles
The design adheres to Mayer's Cognitive Theory of Multimedia Learning:
- **Segmenting**: The lesson is broken into small, digestible 'beats' to prevent cognitive overload.
- **Signaling**: Colors and arrows highlight critical information, such as specific rows or columns.
- **Spatial Contiguity**: Captions and equations are placed near the visual elements they describe.

## Technical Implementation
- **Framework**: Manim (Community Edition) 0.20.1
- **Resolution**: Optimized for mobile viewing (9:16 portrait aspect ratio, 810x1440).
- **Assets**: The notebook automatically downloads 'Caveat' and 'Patrick Hand' fonts from Google Fonts to ensure a friendly, hand-drawn aesthetic.
- **Procedural Textures**: The paper-like background and watercolor effects are generated programmatically via NumPy to minimize external file dependencies.

## How to Use
1. Run the setup cells to install the necessary system dependencies (LaTeX, FFmpeg, Cairo) and Python libraries.
2. Adjust the `LESSON` dictionary in the code to change the multiplication factors or the 'split' point.
3. Execute the final cell to render the video. The output will be displayed directly in the notebook.

```python
%%manim -v WARNING TheSplittingTrick

"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Level: Elementary
Concept: The Distributive Property as "Splitting" an Array (Grid / Area Model)
Source:  github.com/zombimann/Mathematical-video-animations-and-visualization

Pedagogical model: Concrete -> Representational -> Abstract (CRA), the standard
elementary-math sequence for introducing the distributive property via arrays
(Bruner; Milton et al. 2019, "Using the CRA Sequence to Teach Conceptual
Understanding of Basic Multiplication and Division"). Multimedia pacing and
signal/caption placement follow Mayer's Cognitive Theory of Multimedia Learning
(coherence, signaling, spatial contiguity, segmenting).
"""

from manim import *
import numpy as np

# ============================================================
# RENDER SETUP - portrait 9:16, fixed to avoid any stretching
# ============================================================
config.pixel_width = 810
config.pixel_height = 1440
config.frame_rate = 24
config.frame_height = 8.0
config.frame_width = config.frame_height * config.pixel_width / config.pixel_height
config.background_color = "#F7EFDD"

# NOTE: fonts are registered inside construct() using a `with` block -
# manim.register_font() is a context manager, so calling it as a bare
# statement (as earlier versions of this file did) is a silent no-op.

# ============================================================
# QUICK CUSTOMIZATION GUIDE
# ------------------------------------------------------------
#   Change the numbers taught      -> edit LESSON below
#   Change the pacing (all beats)  -> edit PACE["speed"]
#   Change the color language      -> edit PALETTE
#   Change the fonts               -> edit FONTS
#   Change panel/grid positions    -> edit LAYOUT
#   Change wording of captions     -> search for self.set_caption(...)
# Every other part of the file reacts automatically to these values.
# ============================================================

LESSON = {
    "rows": 4,                     # first factor
    "cols": 6,                     # second factor
    "split_at": 5,                 # split cols into split_at + (cols - split_at)
    "title": "The Splitting Trick",
    "subtitle": "Multiply Big Numbers the Easy Way",
    "level_badge": "For Kids  \u00b7  Elementary",
    "unit_singular": "circle",
    "unit_plural": "circles",
    "practice_rows": 4,            # a second, unsolved example for active recall
    "practice_cols": 7,
    "practice_split_at": 5,        # the practice answer is checked using the same trick
    "author": "Mugambi Ndwiga",
    "instagram": "@craftsandengineering",
}

PACE = {"speed": 1.05}              # >1.0 = slower overall video, <1.0 = faster


def T(seconds):
    """Single knob used everywhere for timing so the whole video can be
    sped up or slowed down from one place (PACE['speed'])."""
    return seconds * PACE["speed"]


PALETTE = {
    "paper": "#F7EFDD",            # premium cream paper background
    "ink": "#33291F",              # warm near-black ink (text & outlines)
    "blue": "#4D96FF",             # rows, and the larger split-part
    "red": "#FF6B6B",              # the smaller "leftover" split-part
    "green": "#6BCB77",            # the whole, unified array / area
    "yellow": "#FFD93D",           # columns / general accents
    "panel_fill": "#FFFDF6",
}

FONTS = {
    "display": "Caveat",           # titles & equations - expressive handwriting
    "body": "Patrick Hand",        # captions & small labels - reads clearly small
}

LAYOUT = {
    "cell": 0.52,                  # size of one circle, in scene units
    "hero_y": 0.30,                # vertical center of the main grid graphic
    "caption_y": 2.80,             # vertical center of the top caption dock
    "equation_y": -2.70,           # vertical center of the bottom equation dock
    "panel_w": 3.75,
    "panel_h": 0.85,
}

SEED = 7                            # fixed seed -> identical "hand-drawn" look every render


# ============================================================
# LOW-LEVEL VISUAL HELPERS (paper texture, watercolor washes, cues)
# ============================================================

def make_paper_texture(width_px=540, height_px=960, base_rgb=(247, 239, 221),
                        grain_sigma=4.0, seed=SEED):
    """Procedurally builds a soft cream paper-grain texture as a numpy array,
    so the deliverable never depends on an external image file."""
    rng = np.random.default_rng(seed)
    base = np.array(base_rgb, dtype=np.float64)
    grain = rng.normal(0, grain_sigma, size=(height_px, width_px, 1))
    yy, xx = np.mgrid[0:height_px, 0:width_px]
    cy, cx = height_px / 2, width_px / 2
    dist = np.sqrt(((yy - cy) / cy) ** 2 + ((xx - cx) / cx) ** 2)
    vignette = -14.0 * np.clip(dist - 0.55, 0, None)[..., None]
    img = np.clip(base[None, None, :] + grain + vignette, 0, 255)
    alpha = np.full((height_px, width_px, 1), 255)
    return np.concatenate([img, alpha], axis=-1).astype(np.uint8)


def hand_wobble(mobject, amplitude=0.02, seed=0):
    """Perturbs a shape's anchor points with a fixed random offset so its
    outline reads as hand-inked rather than machine-perfect."""
    rng = np.random.default_rng(seed)
    points = mobject.points
    noise = rng.normal(0, amplitude, points.shape)
    noise[:, 2] = 0
    mobject.set_points(points + noise)
    return mobject


def watercolor_unit(kind, color, size, seed):
    """One 'circle' unit (or optionally a rounded square, via kind='square'):
    two soft translucent
    color washes plus a single wobbly ink outline on top - the vector
    approximation of a watercolor-and-pen sketch used throughout this scene."""
    rng = np.random.default_rng(seed)
    group = VGroup()
    for scale_j, opacity in ((1.05, 0.30), (0.95, 0.40)):
        if kind == "circle":
            wash = Circle(radius=size / 2 * scale_j)
        else:
            wash = RoundedRectangle(width=size * scale_j, height=size * scale_j,
                                     corner_radius=size * 0.18)
        wash.set_fill(color, opacity=opacity)
        wash.set_stroke(width=0)
        dx, dy = rng.uniform(-0.015, 0.015, 2)
        wash.shift([dx, dy, 0])
        group.add(wash)

    if kind == "circle":
        outline = Circle(radius=size / 2)
    else:
        outline = RoundedRectangle(width=size, height=size, corner_radius=size * 0.18)
    outline.set_fill(opacity=0)
    outline.set_stroke(color=PALETTE["ink"], width=3.0, opacity=0.85)
    hand_wobble(outline, amplitude=size * 0.035, seed=seed)
    group.add(outline)
    return group


def recolor_unit_animation(unit, color):
    """Returns an .animate call that recolors only the wash layers of a
    watercolor unit, leaving its ink outline untouched."""
    return unit.animate.set_fill(color)


def build_grid(kind, rows, cols, size, color, seed0=0):
    """Builds a rows x cols grid of watercolor units, centered on the origin.
    Returns (grid[r][c] lookup, flat VGroup of every unit)."""
    grid = [[None for _ in range(cols)] for _ in range(rows)]
    flat = VGroup()
    for r in range(rows):
        for c in range(cols):
            unit = watercolor_unit(kind, color, size, seed0 + r * cols + c)
            unit.move_to(RIGHT * c * size + DOWN * r * size)
            grid[r][c] = unit
            flat.add(unit)
    flat.center()
    return grid, flat


def keep_in_frame_x(mobject, margin=0.12):
    """Defensive safety net: nudges a mobject back inside the visible frame
    horizontally if it would otherwise clip off the edge. This keeps the
    scene robust if LESSON values are edited to less balanced numbers."""
    half_w = config.frame_width / 2 - margin
    right_edge = mobject.get_right()[0]
    left_edge = mobject.get_left()[0]
    if right_edge > half_w:
        mobject.shift(LEFT * (right_edge - half_w))
    elif left_edge < -half_w:
        mobject.shift(RIGHT * (-half_w - left_edge))
    return mobject


def translucent_panel(width, height, corner_radius=0.22, opacity=0.62):
    """A soft, mostly-see-through card. Text placed on it stays fully solid -
    only the panel background is translucent, per the paper/pen aesthetic."""
    return RoundedRectangle(width=width, height=height, corner_radius=corner_radius,
                             fill_color=PALETTE["panel_fill"], fill_opacity=opacity,
                             stroke_color=PALETTE["ink"], stroke_width=2, stroke_opacity=0.30)


def cue_arrow(start, end, color, curve_angle=-0.4, stroke_width=6):
    """A short curved arrow with a dot at its tail, as specified for all
    pointing/annotation cues in this lesson."""
    arrow = CurvedArrow(start_point=np.array(start), end_point=np.array(end),
                         angle=curve_angle, color=color, stroke_width=stroke_width)
    tail = Dot(np.array(start), radius=0.05, color=color)
    return VGroup(arrow, tail)


def make_background_flecks(n=10, seed=SEED):
    """A handful of near-invisible drifting paper flecks that keep the frame
    gently alive at all times, so no moment of the video is a static slide."""
    rng = np.random.default_rng(seed)
    flecks = VGroup()
    for _ in range(n):
        x = rng.uniform(-config.frame_width / 2 * 0.85, config.frame_width / 2 * 0.85)
        y = rng.uniform(-config.frame_height / 2 * 0.85, config.frame_height / 2 * 0.85)
        radius = rng.uniform(0.02, 0.05)
        fleck = Circle(radius=radius, fill_color=PALETTE["ink"],
                        fill_opacity=rng.uniform(0.035, 0.08), stroke_width=0)
        origin = np.array([x, y, 0])
        fleck.move_to(origin)
        phase = rng.uniform(0, TAU)
        speed = rng.uniform(0.12, 0.24)
        amp = rng.uniform(0.05, 0.11)

        def updater(mob, dt, phase=phase, speed=speed, amp=amp, origin=origin):
            mob._t = getattr(mob, "_t", 0.0) + dt
            offset = np.array([np.sin(mob._t * speed + phase) * amp,
                                np.cos(mob._t * speed * 0.7 + phase) * amp * 0.6, 0])
            mob.move_to(origin + offset)

        fleck.add_updater(updater)
        flecks.add(fleck)
    return flecks


# ============================================================
# SCENE
# ============================================================

class TheSplittingTrick(Scene):
    """A short, mostly-silent-friendly, caption-driven lesson teaching the
    distributive property (a x b = a x c + a x (b - c)) through a concrete
    array that is built, counted, and then physically split and rejoined."""

    def construct(self):
        with register_font("assets/fonts/Caveat-Variable.ttf"), \
             register_font("assets/fonts/PatrickHand-Regular.ttf"):
            self.rows = LESSON["rows"]
            self.cols = LESSON["cols"]
            self.split_at = LESSON["split_at"]
            self.split_b = self.cols - self.split_at
            self.cell = LAYOUT["cell"]

            self.setup_stage()
            self.beat_title()
            self.beat_hook()
            self.beat_build_rows()
            self.beat_show_columns()
            self.beat_count_units()
            self.beat_reveal_equation()
            self.beat_challenge()
            self.beat_bracket_split()
            self.beat_physical_split()
            self.beat_sub_equations()
            self.beat_recombine()
            self.beat_final_sum()
            self.beat_takeaway()
            self.beat_practice()
            self.beat_closing()

    # ------------------------------------------------------
    # Persistent stage furniture: background, watermark, badge,
    # and the two text docks that stay on screen the whole lesson.
    # ------------------------------------------------------
    def setup_stage(self):
        background = ImageMobject(make_paper_texture())
        background.stretch_to_fit_width(config.frame_width)
        background.stretch_to_fit_height(config.frame_height)
        self.add(background)
        self.add(make_background_flecks())

        self.watermark = Text(
            f"\u00a9 {LESSON['author']} / {LESSON['instagram']}",
            font=FONTS["body"], font_size=15, color=PALETTE["ink"],
        )
        self.watermark.set_opacity(0.6)
        self.watermark.to_corner(DL, buff=0.18)
        self.add(self.watermark)

        badge_pill = RoundedRectangle(width=2.55, height=0.5, corner_radius=0.25,
                                       fill_color=PALETTE["blue"], fill_opacity=0.85,
                                       stroke_width=0)
        badge_text = Text(LESSON["level_badge"], font=FONTS["body"],
                           font_size=20, color=WHITE, weight=BOLD)
        badge_text.move_to(badge_pill)
        self.level_badge = VGroup(badge_pill, badge_text)
        self.level_badge.to_corner(UR, buff=0.16)
        self.add(self.level_badge)

        self.caption_dock = translucent_panel(LAYOUT["panel_w"], LAYOUT["panel_h"])
        self.caption_dock.move_to(UP * LAYOUT["caption_y"])
        self.caption_text = Text("", font=FONTS["body"], font_size=1, color=PALETTE["ink"])
        self.caption_text.move_to(self.caption_dock)

        self.equation_dock = translucent_panel(LAYOUT["panel_w"], LAYOUT["panel_h"])
        self.equation_dock.move_to(UP * LAYOUT["equation_y"])
        self.equation_text = Text("", font=FONTS["display"], font_size=1, color=PALETTE["ink"])
        self.equation_text.move_to(self.equation_dock)

    def set_caption(self, message, run_time=0.6, font_size=30):
        new_text = Text(message, font=FONTS["body"], font_size=font_size,
                         color=PALETTE["ink"], weight=BOLD)
        max_w = LAYOUT["panel_w"] - 0.5
        if new_text.width > max_w:
            new_text.scale(max_w / new_text.width)
        new_text.move_to(self.caption_dock)
        self.play(FadeOut(self.caption_text, shift=UP * 0.12),
                   FadeIn(new_text, shift=UP * 0.12), run_time=run_time)
        self.caption_text = new_text

    def set_equation(self, message, color=None, run_time=0.6, font_size=48):
        new_text = Text(message, font=FONTS["display"], weight=BOLD, font_size=font_size,
                         color=color or PALETTE["ink"])
        max_w = LAYOUT["panel_w"] - 0.4
        if new_text.width > max_w:
            new_text.scale(max_w / new_text.width)
        new_text.move_to(self.equation_dock)
        self.play(FadeOut(self.equation_text, shift=UP * 0.12),
                   FadeIn(new_text, shift=UP * 0.12), run_time=run_time)
        self.equation_text = new_text

    def idle_pulse(self, run_time=1.0, amount=1.015):
        """A gentle breathing pulse on the hero graphic, used instead of a
        bare self.wait() whenever the lesson needs a beat to let an idea
        land - the frame keeps moving even while nothing new is taught."""
        self.play(self.hero_flat.animate.scale(amount), rate_func=there_and_back,
                   run_time=run_time)

    # ------------------------------------------------------
    # BEAT 1 - Title
    # ------------------------------------------------------
    def beat_title(self):
        title = Text(LESSON["title"], font=FONTS["display"], weight=BOLD,
                      font_size=52, color=PALETTE["ink"])
        subtitle = Text(LESSON["subtitle"], font=FONTS["body"], font_size=24,
                         color=PALETTE["blue"])
        title_group = VGroup(title, subtitle).arrange(DOWN, buff=0.22)
        max_title_w = config.frame_width - 0.7
        if title_group.width > max_title_w:
            title_group.scale(max_title_w / title_group.width)
        title_group.move_to(UP * 0.9)

        preview_grid, preview_flat = build_grid("circle", 1, 3, self.cell * 1.3, PALETTE["green"], seed0=100)
        preview_flat.next_to(title_group, DOWN, buff=0.9)

        self.play(FadeIn(preview_flat, lag_ratio=0.15), run_time=T(0.6))
        self.play(Write(title_group), run_time=T(2.0))
        self.play(preview_flat.animate.scale(1.1), rate_func=there_and_back, run_time=T(0.8))
        self.wait(T(1.1))

        self.play(FadeOut(preview_flat), run_time=T(0.5))
        self.play(title_group.animate.scale(0.001).move_to(UP * LAYOUT["caption_y"]),
                   run_time=T(0.9))
        self.remove(title_group)

        self.caption_text.move_to(self.caption_dock)
        self.add(self.caption_dock, self.caption_text)

    # ------------------------------------------------------
    # BEAT 2 - Hook: introduce the concrete context
    # ------------------------------------------------------
    def beat_hook(self):
        self.add(self.equation_dock, self.equation_text)
        self.set_caption(f"Let's grow some {LESSON['unit_plural']}!", run_time=T(0.7))

        first_unit = watercolor_unit("circle", PALETTE["green"], self.cell, seed=999)
        first_unit.move_to(UP * LAYOUT["hero_y"] +
                            LEFT * (self.cols - 1) * self.cell / 2 +
                            UP * (self.rows - 1) * self.cell / 2)
        self.play(GrowFromCenter(first_unit), run_time=T(0.6))
        self._hook_unit = first_unit
        self.wait(T(0.4))

    # ------------------------------------------------------
    # BEAT 3 - Build the array row by row (concrete stage)
    # ------------------------------------------------------
    def beat_build_rows(self):
        self.counters, self.hero_flat = build_grid(
            "circle", self.rows, self.cols, self.cell, PALETTE["green"], seed0=200)
        self.hero_flat.move_to(UP * LAYOUT["hero_y"])
        self.remove(self._hook_unit)

        row_groups = [VGroup(*self.counters[r]) for r in range(self.rows)]
        for row in row_groups:
            row.set_opacity(0)

        self.add(self.hero_flat)

        for i, row in enumerate(row_groups):
            row.set_opacity(1)
            self.play(LaggedStart(*[GrowFromCenter(u) for u in row], lag_ratio=0.12),
                       run_time=T(1.1))
            self.set_caption(f"{i + 1} row{'s' if i else ''} of {LESSON['unit_plural']}",
                              run_time=T(0.5))
            self.wait(T(0.4))

        row_brace = Brace(self.hero_flat, LEFT, buff=0.14)
        row_brace.set_color(PALETTE["blue"])
        row_label = Text(f"{self.rows} rows", font=FONTS["body"], font_size=18,
                          color=PALETTE["blue"], weight=BOLD)
        row_label.rotate(PI / 2)
        row_label.next_to(self.hero_flat, LEFT, buff=0.05)
        self.row_brace_group = VGroup(row_brace, row_label)

        self.play(GrowFromCenter(row_brace), FadeIn(row_label, shift=LEFT * 0.08), run_time=T(0.8))
        self.wait(T(0.5))

    # ------------------------------------------------------
    # BEAT 4 - Show the columns
    # ------------------------------------------------------
    def beat_show_columns(self):
        self.set_caption(f"{self.cols} {LESSON['unit_plural']} in every row", run_time=T(0.6))

        top_row = VGroup(*self.counters[0])
        sweep = cue_arrow(
            top_row.get_left() + UP * 0.55,
            top_row.get_right() + UP * 0.55,
            color=PALETTE["yellow"], curve_angle=-0.9, stroke_width=7,
        )
        self.play(Create(sweep), run_time=T(1.0))

        col_brace = Brace(self.hero_flat, DOWN, buff=0.14)
        col_brace.set_color(PALETTE["yellow"])
        col_label = Text(f"{self.cols} columns", font=FONTS["body"], font_size=24,
                          color=PALETTE["yellow"], weight=BOLD)
        col_label.next_to(col_brace, DOWN, buff=0.12)
        self.col_brace_group = VGroup(col_brace, col_label)

        self.play(GrowFromCenter(col_brace), FadeIn(col_label, shift=DOWN * 0.1), run_time=T(0.8))
        self.wait(T(0.4))
        self.play(FadeOut(sweep), run_time=T(0.4))

    # ------------------------------------------------------
    # BEAT 5 - Count every single counter, one at a time. Each unit
    #          gets its own highlight beat and the equation dock ticks
    #          up by exactly 1 each time, so kids can follow the count
    #          instead of being told a row total all at once.
    # ------------------------------------------------------
    def beat_count_units(self):
        # the row/column braces already made their point; clear them so the
        # per-unit highlight ring never has to fight them for space,
        # especially along the left edge where the row brace sits.
        self.play(FadeOut(self.row_brace_group), FadeOut(self.col_brace_group),
                   run_time=T(0.5))
        self.set_caption(f"Let's count every {LESSON['unit_singular']}, one by one",
                          run_time=T(0.7), font_size=27)
        self.wait(T(0.5))

        counted_color = interpolate_color(
            ManimColor(PALETTE["green"]), ManimColor(PALETTE["yellow"]), 0.5)
        running_total = 0
        for r in range(self.rows):
            for c in range(self.cols):
                unit = self.counters[r][c]
                running_total += 1
                ring = SurroundingRectangle(unit, color=PALETTE["yellow"], stroke_width=5,
                                             corner_radius=0.14, buff=0.05)
                self.play(Create(ring), unit.animate.scale(1.2), run_time=T(0.22))
                self.set_equation(str(running_total), color=PALETTE["yellow"],
                                   run_time=T(0.15), font_size=54)
                self.play(
                    unit.animate.scale(1 / 1.2),
                    recolor_unit_animation(unit, counted_color),
                    FadeOut(ring),
                    run_time=T(0.2),
                )
            self.wait(T(0.3))   # a small breath at the end of every row

        self.wait(T(0.6))
        # tidy the grid back to a single clean color before the next idea
        self.play(*[recolor_unit_animation(self.counters[r][c], PALETTE["green"])
                     for r in range(self.rows) for c in range(self.cols)],
                   run_time=T(0.6))

    # ------------------------------------------------------
    # BEAT 6 - Reveal the total and the multiplication sentence
    # ------------------------------------------------------
    def beat_reveal_equation(self):
        total = self.rows * self.cols
        self.set_caption(f"{total} {LESSON['unit_plural']} in all!", run_time=T(0.7))
        self.idle_pulse(run_time=T(1.0))
        self.wait(T(0.4))

        equation = Text(f"{self.rows} \u00d7 {self.cols} = {total}",
                         font=FONTS["display"], weight=BOLD, font_size=52, color=PALETTE["ink"])
        rows_part = equation[0:1]
        cols_part = equation[4:5]
        rows_part.set_color(PALETTE["blue"])
        cols_part.set_color(PALETTE["yellow"])
        equation.move_to(self.equation_dock)

        self.play(FadeOut(self.equation_text, shift=UP * 0.1),
                   FadeIn(equation, shift=UP * 0.1), run_time=T(0.7))
        self.equation_text = equation
        self.wait(T(1.0))

    # ------------------------------------------------------
    # BEAT 8 - The challenge that motivates splitting
    # ------------------------------------------------------
    def beat_challenge(self):
        self.tiles = self.counters   # split beats work on this grid directly - no shape change needed
        self.set_caption(f"{self.cols} {LESSON['unit_plural']} across is a lot to picture!", run_time=T(0.7))
        self.play(self.hero_flat.animate.rotate(0.02), rate_func=there_and_back, run_time=T(0.5))
        self.play(self.hero_flat.animate.rotate(-0.02), rate_func=there_and_back, run_time=T(0.5))

        question = Text("?", font=FONTS["display"], weight=BOLD, font_size=64,
                         color=PALETTE["red"])
        question.next_to(self.hero_flat, UR, buff=0.05).shift(LEFT * 0.3 + DOWN * 0.2)
        self.play(FadeIn(question, scale=1.4), run_time=T(0.5))
        self.play(FadeOut(question), run_time=T(0.4))

    # ------------------------------------------------------
    # BEAT 9 - Bracket the split conceptually before moving anything
    #          (CRA practice: "circle the factor you break apart")
    # ------------------------------------------------------
    def beat_bracket_split(self):
        self.set_caption(f"Let's split {self.cols} into {self.split_at} and {self.split_b}",
                          run_time=T(0.8))

        left_units = [self.tiles[r][c] for r in range(self.rows) for c in range(self.split_at)]
        right_units = [self.tiles[r][c] for r in range(self.rows) for c in range(self.split_at, self.cols)]
        self.left_units, self.right_units = left_units, right_units

        left_group = VGroup(*left_units)
        right_group = VGroup(*right_units)

        left_brace = Brace(left_group, DOWN, buff=0.14)
        left_brace.set_color(PALETTE["blue"])
        left_label = Text(str(self.split_at), font=FONTS["display"], weight=BOLD,
                           font_size=32, color=PALETTE["blue"])
        left_brace.put_at_tip(left_label)

        right_brace = Brace(right_group, DOWN, buff=0.14)
        right_brace.set_color(PALETTE["red"])
        right_label = Text(str(self.split_b), font=FONTS["display"], weight=BOLD,
                            font_size=32, color=PALETTE["red"])
        right_brace.put_at_tip(right_label)

        self.split_braces = VGroup(left_brace, left_label, right_brace, right_label)
        self.play(GrowFromCenter(left_brace), GrowFromCenter(right_brace),
                   FadeIn(left_label, shift=DOWN * 0.1), FadeIn(right_label, shift=DOWN * 0.1),
                   run_time=T(0.9))

        self.play(*[recolor_unit_animation(u, PALETTE["blue"]) for u in left_units],
                   *[recolor_unit_animation(u, PALETTE["red"]) for u in right_units],
                   run_time=T(1.0))
        self.wait(T(0.4))

    # ------------------------------------------------------
    # BEAT 10 - Physically slide the two parts apart
    # ------------------------------------------------------
    def beat_physical_split(self):
        self.set_caption("Now let's pull them apart", run_time=T(0.6))
        right_group = VGroup(*self.right_units)
        self.split_gap = self.cell * 0.65   # shared with beat_recombine

        self.play(
            right_group.animate.shift(RIGHT * self.split_gap),
            self.split_braces[2].animate.shift(RIGHT * self.split_gap),   # right brace
            self.split_braces[3].animate.shift(RIGHT * self.split_gap),  # right label
            run_time=T(1.4),
        )
        self.wait(T(0.3))

    # ------------------------------------------------------
    # BEAT 11 - Sub-equations tied directly to each block
    # ------------------------------------------------------
    def beat_sub_equations(self):
        left_group = VGroup(*self.left_units)
        right_group = VGroup(*self.right_units)
        part_a = self.rows * self.split_at
        part_b = self.rows * self.split_b

        left_eq = Text(f"{self.rows}\u00d7{self.split_at}={part_a}", font=FONTS["display"],
                        weight=BOLD, font_size=28, color=PALETTE["blue"])
        left_eq.next_to(left_group, UP, buff=0.35)
        keep_in_frame_x(left_eq)
        left_cue = cue_arrow(left_eq.get_bottom() + DOWN * 0.05, left_group.get_top() + UP * 0.05,
                              color=PALETTE["blue"], curve_angle=0.6, stroke_width=5)

        right_eq = Text(f"{self.rows}\u00d7{self.split_b}={part_b}", font=FONTS["display"],
                         weight=BOLD, font_size=28, color=PALETTE["red"])
        right_eq.next_to(right_group, UP, buff=0.35)
        keep_in_frame_x(right_eq)
        right_cue = cue_arrow(right_eq.get_bottom() + DOWN * 0.05, right_group.get_top() + UP * 0.05,
                               color=PALETTE["red"], curve_angle=-0.6, stroke_width=5)

        self.set_caption("Each part is an easy multiplication", run_time=T(0.6))
        self.play(FadeIn(left_eq, shift=DOWN * 0.15), Create(left_cue), run_time=T(0.8))
        self.set_equation(f"{self.rows}\u00d7{self.split_at} = {part_a}",
                           color=PALETTE["blue"], run_time=T(0.5), font_size=40)
        self.wait(T(0.3))
        self.play(FadeIn(right_eq, shift=DOWN * 0.15), Create(right_cue), run_time=T(0.8))
        self.set_equation(f"{self.rows}\u00d7{self.split_b} = {part_b}",
                           color=PALETTE["red"], run_time=T(0.5), font_size=40)
        self.wait(T(0.4))

        self.sub_equation_group = VGroup(left_eq, left_cue, right_eq, right_cue)
        self.part_a, self.part_b = part_a, part_b

    # ------------------------------------------------------
    # BEAT 12 - Recombine: proves the split didn't change anything
    # ------------------------------------------------------
    def beat_recombine(self):
        self.set_caption("Same tray, just split for a moment!", run_time=T(0.6))
        self.play(FadeOut(self.sub_equation_group), FadeOut(self.split_braces), run_time=T(0.5))

        right_group = VGroup(*self.right_units)
        self.play(right_group.animate.shift(LEFT * self.split_gap), run_time=T(1.2))
        self.play(*[recolor_unit_animation(u, PALETTE["green"])
                     for u in self.left_units + self.right_units],
                   run_time=T(0.9))
        self.idle_pulse(run_time=T(0.9))

    # ------------------------------------------------------
    # BEAT 13 - Final sum, tied back to the original total
    # ------------------------------------------------------
    def beat_final_sum(self):
        total = self.rows * self.cols
        self.set_equation(f"{self.part_a} + {self.part_b} = {total}",
                           color=PALETTE["green"], run_time=T(0.7), font_size=44)
        self.set_caption(f"Still {total} {LESSON['unit_plural']} - nothing changed!", run_time=T(0.7))
        self.idle_pulse(run_time=T(1.0))

    # ------------------------------------------------------
    # BEAT 14 - The takeaway (big idea), graphic still present
    # ------------------------------------------------------
    def beat_takeaway(self):
        self.set_caption("Splitting a big number makes multiplying easy!",
                          run_time=T(0.8), font_size=27)
        self.play(self.hero_flat.animate.scale(1.06), rate_func=there_and_back, run_time=T(1.0))
        self.wait(T(0.6))

    # ------------------------------------------------------
    # BEAT 15 - Practice prompt (active recall), keeps a worked
    #           example visible while a fresh, unsolved grid appears.
    # ------------------------------------------------------
    def beat_practice(self):
        self.set_caption("Now you try it yourself!", run_time=T(0.6))
        self.play(self.hero_flat.animate.set_opacity(0.25), run_time=T(0.5))
        self.play(FadeOut(self.hero_flat), run_time=T(0.5))

        practice_rows = LESSON["practice_rows"]
        practice_cols = LESSON["practice_cols"]
        # keep the practice grid the same overall width as the taught example
        practice_cell = self.cell * self.cols / practice_cols
        self.set_equation(f"{practice_rows} \u00d7 {practice_cols} = ?",
                           color=PALETTE["ink"], run_time=T(0.5), font_size=42)

        _, practice_flat = build_grid("circle", practice_rows, practice_cols,
                                       practice_cell, PALETTE["green"], seed0=900)
        practice_flat.move_to(UP * LAYOUT["hero_y"])
        self.play(LaggedStart(*[GrowFromCenter(u) for u in practice_flat], lag_ratio=0.03),
                   run_time=T(1.3))
        self.set_caption(f"Pause the video and try {practice_rows} \u00d7 {practice_cols}!",
                          run_time=T(0.6), font_size=27)
        self.wait(T(1.3))

        # check the answer using the SAME splitting trick just taught,
        # instead of simply stating the total
        self.set_caption("Good job! Let's check with our trick",
                          run_time=T(0.6), font_size=24)
        self.wait(T(0.3))

        practice_split_at = LESSON["practice_split_at"]
        practice_split_b = practice_cols - practice_split_at
        practice_grid = [[practice_flat[r * practice_cols + c] for c in range(practice_cols)]
                          for r in range(practice_rows)]
        left_units = [practice_grid[r][c] for r in range(practice_rows) for c in range(practice_split_at)]
        right_units = [practice_grid[r][c] for r in range(practice_rows)
                        for c in range(practice_split_at, practice_cols)]
        practice_gap = practice_cell * 0.65

        self.play(
            *[recolor_unit_animation(u, PALETTE["blue"]) for u in left_units],
            *[recolor_unit_animation(u, PALETTE["red"]) for u in right_units],
            run_time=T(0.7),
        )
        self.play(VGroup(*right_units).animate.shift(RIGHT * practice_gap), run_time=T(0.7))

        part_a = practice_rows * practice_split_at
        part_b = practice_rows * practice_split_b
        self.set_equation(f"{practice_rows}\u00d7{practice_split_at} + {practice_rows}\u00d7{practice_split_b}",
                           color=PALETTE["ink"], run_time=T(0.6), font_size=32)
        self.wait(T(0.5))
        practice_total = practice_rows * practice_cols
        self.set_equation(f"{part_a} + {part_b} = {practice_total}",
                           color=PALETTE["green"], run_time=T(0.6), font_size=38)
        self.play(VGroup(*left_units, *right_units).animate.scale(1.05),
                   rate_func=there_and_back, run_time=T(0.5))
        self.wait(T(0.5))
        self.practice_flat = practice_flat

    # ------------------------------------------------------
    # BEAT 16 - Closing card (the one graphic-free moment)
    # ------------------------------------------------------
    def beat_closing(self):
        self.play(
            FadeOut(self.practice_flat),
            FadeOut(self.caption_dock), FadeOut(self.caption_text),
            FadeOut(self.equation_dock), FadeOut(self.equation_text),
            FadeOut(self.level_badge),
            run_time=T(0.4),
        )
        closing_bg = Rectangle(width=config.frame_width + 1, height=config.frame_height + 1,
                                fill_color=PALETTE["blue"], fill_opacity=1, stroke_width=0)

        credits = VGroup(
            Text(f"Made by {LESSON['author']}", font=FONTS["display"], weight=BOLD,
                 font_size=36, color=WHITE),
            Text(LESSON["instagram"], font=FONTS["body"], font_size=24, color=WHITE),
        ).arrange(DOWN, buff=0.15)

        cta = VGroup(
            Text("Contribute to this OpenSource Project:", font=FONTS["body"], font_size=16, color=WHITE),
            Text("See GitHub Sponsors or DM for details", font=FONTS["body"], font_size=16, color=PALETTE["yellow"]),
        ).arrange(DOWN, buff=0.1)

        closing_text = VGroup(credits, cta).arrange(DOWN, buff=0.6)

        max_closing_w = config.frame_width - 0.7
        if closing_text.width > max_closing_w:
            closing_text.scale(max_closing_w / closing_text.width)

        # Closing-card timing is intentionally NOT scaled by T(): the brief
        # caps this card at 2 seconds regardless of the overall lesson pace.
        self.play(FadeIn(closing_bg), FadeIn(closing_text), run_time=0.4)
        self.wait(1.3)
```
