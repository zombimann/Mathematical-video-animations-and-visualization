## Koch Snowflake Animation

This notebook demonstrates the generation of a Koch Snowflake using the Manim animation library. The Koch Snowflake is a classic example of a fractal curve, notable for its intriguing mathematical properties.

### Background Concepts

#### Fractals
A fractal is a never-ending pattern that is self-similar across different scales. This means that a fractal pattern looks roughly the same no matter how zoomed in or out it is. Fractals can be found in nature (e.g., coastlines, snowflakes, trees) and are often generated by repeating a simple process in an ongoing feedback loop.

#### Koch Snowflake
The Koch Snowflake is one of the earliest fractal curves described. It is constructed by starting with an equilateral triangle and iteratively adding smaller equilateral triangles to the middle third of each segment of the previous iteration. This process leads to:

*   **Infinite Perimeter:** As the iterations continue, the length of the curve increases by a factor of 4/3 at each step, resulting in an infinitely long perimeter.
*   **Finite Area:** Despite having an infinite perimeter, the Koch Snowflake encloses a finite area. This paradox highlights the counter-intuitive nature of fractals.

### Work Done in this Notebook

This notebook presents a Manim animation that visually constructs the Koch Snowflake. It highlights the iterative process and the mathematical formula for its perimeter. The animation is designed with a sketchbook aesthetic, aiming to make the mathematical concepts more engaging and visually appealing.

```python
%%manim -v WARNING -r 1080,1920 --fps 30 --progress_bar None KochCurveAnimation

"""
Author: Mugambi Ndwiga
Aesthetic: Paper & Bleeding Fountain Pen Sketchbook Layout
Optimization: Prominent/Maximized Graphic, Highly Detailed Iterations (Level 5)
Mod: Square-based Koch Snowflake Transformation
"""

from manim import *
import numpy as np

class KochCurveAnimation(Scene):
    def construct(self):
        # --- Force 9:16 Aspect Ratio Coordinate Space ---
        config.frame_width = 9
        config.frame_height = 16

        # --- Parchment & Fountain Pen Palette ---
        PAPER_BG = "#f4f1ea"        # Warm, textured sketchbook paper tone
        INK_MAIN = "#1c2d42"        # Deep Prussian Blue ink
        INK_ACCENT = "#8b1e2f"      # Bleeding Burgundy ink for active changes
        INK_MUTED = "#5c6b73"       # Soft graphite grey for secondary text

        self.camera.background_color = PAPER_BG

        # --- Hand-drawn Typography Configuration ---
        HANDWRITING_FONT = "Ink Free"
        HANDWRITTEN_MATH = TexFontTemplates.french_cursive

        # --- Watermark (Elegant margin signature) ---
        watermark = Text(
            "© Mugambi Ndwiga / @craftsandengineering",
            font=HANDWRITING_FONT,
            font_size=18,
            color=INK_MUTED,
            fill_opacity=0.6,
            warn_missing_font=False
        ).move_to([0, -7.6, 0])
        self.add(watermark)

        # --- Zone 1: Title Block ---
        title_main = Text("INFINITE PERIMETER", font=HANDWRITING_FONT, weight=BOLD, font_size=46, color=INK_MAIN, warn_missing_font=False)
        title_main.scale_to_fit_width(config.frame_width * 0.85)
        title_sub = Text("The Koch Snowflake", font=HANDWRITING_FONT, font_size=34, color=INK_MUTED, warn_missing_font=False)
        title = VGroup(title_main, title_sub).arrange(DOWN, buff=0.25).move_to([0, 6.3, 0])

        # --- Zone 2: Hand-written Static Formula Panel (Perimeter Equation) ---
        info_panel = MathTex(
            r"P_n = P_0 \cdot \left(\frac{4}{3}\right)^n",
            font_size=46,
            color=INK_MAIN,
            tex_template=HANDWRITTEN_MATH
        ).move_to([0, 4.3, 0])

        # --- Zone 3: Maximized Closed Square Geometry Base ---
        # Scaled and centered to allow outward growth to safely fit the 9:16 frame aspect ratios
        side_length = 3.5
        h_side = side_length / 2
        center_y = -1.1

        # Clockwise vertex ordering directs fractal growth outwards
        p1 = np.array([-h_side, center_y + h_side, 0])  # Top-Left
        p2 = np.array([ h_side, center_y + h_side, 0])  # Top-Right
        p3 = np.array([ h_side, center_y - h_side, 0])  # Bottom-Right
        p4 = np.array([-h_side, center_y - h_side, 0])  # Bottom-Left
        base_points = [p1, p2, p3, p4, p1]

        curve = VMobject(color=INK_MAIN, stroke_width=6.0, stroke_opacity=0.95)
        curve.set_points_as_corners(base_points)

        # --- Geometry Logic ---
        def get_koch_points(points, order):
            for _ in range(order):
                new_points = []
                for i in range(len(points) - 1):
                    p1 = points[i]
                    p2 = points[i+1]
                    v = p2 - p1
                    a = p1 + v / 3
                    b = p1 + 2 * v / 3
                    rotation_matrix = np.array([
                        [np.cos(PI/3), -np.sin(PI/3), 0],
                        [np.sin(PI/3), np.cos(PI/3), 0],
                        [0, 0, 1]
                    ])
                    peak = a + np.dot(rotation_matrix, v / 3)
                    new_points.extend([p1, a, peak, b])
                new_points.append(points[-1])
                points = new_points
            return points

        # --- Narrative Flow ---
        self.play(Write(title), run_time=1.4)
        self.play(FadeIn(info_panel, shift=DOWN * 0.2))
        self.play(Create(curve))
        self.wait(0.6)

        # --- Zone 4: High-Iteration Dynamic Ink Transformations ---
        displayed_length = None

        # 5 deep iterations for highly complex structural fractal detail
        for n in range(1, 6):
            new_pts = get_koch_points(base_points, n)

            # Stroke width dynamically scales down as complexity builds to prevent bleeding blots
            current_stroke = max(1.5, 6.5 - (n * 1.0))
            new_curve = VMobject(color=INK_MAIN, stroke_width=current_stroke, stroke_opacity=0.95)
            new_curve.set_points_as_corners(new_pts)

            # Repositioned tracker zone safely below the expanding lower base of the square fractal
            next_length = MathTex(
                f"P_{n} = {round((4/3)**n, 2)} \\cdot P_0",
                font_size=42,
                color=INK_ACCENT,
                tex_template=HANDWRITTEN_MATH
            ).move_to([0, -4.6, 0])

            if displayed_length is None:
                self.play(
                    Transform(curve, new_curve),
                    FadeIn(next_length, shift=UP * 0.2),
                    run_time=1.6
                )
                displayed_length = next_length
            else:
                self.play(
                    Transform(curve, new_curve),
                    Transform(displayed_length, next_length),
                    run_time=1.5
                )
            self.wait(0.8)

        # --- Zone 5: Lower Final Statement ---
        conclusion = Text(
            "As n → ∞, Perimeter → ∞",
            font=HANDWRITING_FONT,
            font_size=42,
            color=INK_ACCENT,
            weight=BOLD,
            warn_missing_font=False
        ).move_to([0, -6.3, 0])
        conclusion.scale_to_fit_width(config.frame_width * 0.85)

        self.play(FadeIn(conclusion, shift=UP * 0.3))
        self.wait(2.5)

        # --- Outro Title Card (Consistent Sketchbook Branding) ---
        self.play(FadeOut(Group(*[m for m in self.mobjects if m != watermark])))

        closing_text = VGroup(
            Text("Made by Mugambi Ndwiga", font=HANDWRITING_FONT, font_size=44, weight=BOLD, color=INK_MAIN, warn_missing_font=False),
            Text("@craftsandengineering", font=HANDWRITING_FONT, font_size=32, color=INK_ACCENT, warn_missing_font=False)
        ).arrange(DOWN, buff=0.4).move_to(ORIGIN)

        self.play(Write(closing_text), run_time=1.8)
        self.wait(2.5)
```
