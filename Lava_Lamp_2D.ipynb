```python
%%manim -v WARNING -r 720,1280 LavaLampAnimation
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Concept: 2D Lava Lamp Animation with Real Fluid Interactions
Repository reference: github.com/zombimann/Mathematical-video-animations-and-visualization
"""

from manim import *
import numpy as np
from dataclasses import dataclass

config.pixel_width = 720
config.pixel_height = 1280
config.frame_width = 7.2
config.frame_height = 12.8
config.frame_rate = 24
config.background_color = "#08111d"


@dataclass
class LampPalette:
    background: str = "#08111d"
    panel_fill: str = "#0f1a2a"
    panel_stroke: str = "#2f425c"
    glass_fill: str = "#93b8d1"
    glass_stroke: str = "#dcecff"
    wax_hot: str = "#ff8a3d"
    wax_core: str = "#ffb15e"
    wax_cool: str = "#d96b34"
    heat_glow: str = "#ff6a2a"
    cool_glow: str = "#79a9ff"
    text_main: str = "#eef5ff"
    text_muted: str = "#b7c6d8"


@dataclass
class LampConfig:
    scene_seconds: float = 14.0
    closing_seconds: float = 1.8
    blob_count: int = 4
    blob_speed: float = 0.13
    drift_speed: float = 0.18
    lamp_x: float = -1.35
    lamp_y: float = -1.20
    lamp_width: float = 3.8
    lamp_height: float = 7.1
    panel_width: float = 3.1
    panel_x: float = 1.75
    watermark_scale: float = 0.18


PALETTE = LampPalette()
CFG = LampConfig()


def make_lamp_body(cfg: LampConfig) -> VMobject:
    """Custom lamp silhouette with a smooth, glass-like outline."""
    w = cfg.lamp_width
    h = cfg.lamp_height
    x0 = cfg.lamp_x
    y0 = cfg.lamp_y

    pts = [
        np.array([x0 - 0.34 * w, y0 - 0.47 * h, 0]),
        np.array([x0 - 0.42 * w, y0 - 0.31 * h, 0]),
        np.array([x0 - 0.42 * w, y0 + 0.18 * h, 0]),
        np.array([x0 - 0.34 * w, y0 + 0.38 * h, 0]),
        np.array([x0 - 0.18 * w, y0 + 0.47 * h, 0]),
        np.array([x0 - 0.11 * w, y0 + 0.58 * h, 0]),
        np.array([x0 - 0.10 * w, y0 + 0.73 * h, 0]),
        np.array([x0 - 0.15 * w, y0 + 0.85 * h, 0]),
        np.array([x0 + 0.15 * w, y0 + 0.85 * h, 0]),
        np.array([x0 + 0.10 * w, y0 + 0.73 * h, 0]),
        np.array([x0 + 0.11 * w, y0 + 0.58 * h, 0]),
        np.array([x0 + 0.18 * w, y0 + 0.47 * h, 0]),
        np.array([x0 + 0.34 * w, y0 + 0.38 * h, 0]),
        np.array([x0 + 0.42 * w, y0 + 0.18 * h, 0]),
        np.array([x0 + 0.42 * w, y0 - 0.31 * h, 0]),
        np.array([x0 + 0.34 * w, y0 - 0.47 * h, 0]),
    ]
    body = VMobject()
    body.set_points_smoothly(pts)
    body.set_fill(color=PALETTE.glass_fill, opacity=0.17)
    body.set_stroke(color=PALETTE.glass_stroke, width=3.0, opacity=0.58)
    return body


def make_lamp_base(cfg: LampConfig) -> VGroup:
    """Dynamically anchors base elements to the computed bottom of the glass."""
    glass_bottom = cfg.lamp_y - 0.47 * cfg.lamp_height

    base = RoundedRectangle(
        width=4.3,
        height=0.78,
        corner_radius=0.18,
        stroke_width=2.2,
        stroke_color=PALETTE.glass_stroke,
        fill_color="#182638",
        fill_opacity=0.95,
    ).move_to([cfg.lamp_x, glass_bottom - 0.65, 0])

    cap = RoundedRectangle(
        width=3.0,
        height=0.54,
        corner_radius=0.15,
        stroke_width=1.8,
        stroke_color=PALETTE.glass_stroke,
        fill_color="#24364f",
        fill_opacity=0.96,
    ).move_to([cfg.lamp_x, glass_bottom - 0.1, 0])

    heat_band = Rectangle(
        width=2.8,
        height=0.15,
        stroke_width=0,
        fill_color=PALETTE.heat_glow,
        fill_opacity=0.65,
    ).move_to([cfg.lamp_x, glass_bottom + 0.18, 0])

    return VGroup(base, cap, heat_band)


def clamp_to_lamp_walls(pos, r, cfg: LampConfig):
    """Dynamic wall collision avoidance system based on height profile."""
    x, y, _ = pos
    w = cfg.lamp_width
    h = cfg.lamp_height
    x0 = cfg.lamp_x
    y0 = cfg.lamp_y

    dy = (y - y0) / h

    if dy < 0.18:
        max_w = 0.36 * w
    elif dy < 0.47:
        t = (dy - 0.18) / (0.47 - 0.18)
        max_w = interpolate(0.36 * w, 0.14 * w, t)
    else:
        max_w = 0.12 * w

    allowed_half_width = max(0.05, max_w - r)
    cx = np.clip(x, x0 - allowed_half_width, x0 + allowed_half_width)
    return np.array([cx, y, 0])


def get_fluid_centers(t: float, cfg: LampConfig):
    """Calculates positions and radii of core fluid nodes with wall collision buffers."""
    centers = []
    bot_y = cfg.lamp_y - cfg.lamp_height * 0.44
    top_y = cfg.lamp_y + cfg.lamp_height * 0.42

    # [0] Bottom anchor pool
    centers.append((np.array([cfg.lamp_x, bot_y, 0]), 1.1))

    # [1] Top reservoir pool
    centers.append((np.array([cfg.lamp_x, top_y + 0.1, 0]), 0.8))

    # [2, 3] Undulating base masses
    centers.append((clamp_to_lamp_walls(np.array([cfg.lamp_x + 0.3 * np.sin(t * 0.25), bot_y + 0.1, 0]), 0.6, cfg), 0.6))
    centers.append((clamp_to_lamp_walls(np.array([cfg.lamp_x - 0.2 * np.sin(t * 0.31), bot_y - 0.05, 0]), 0.5, cfg), 0.5))

    # [4+] Traveling active blobs
    for i in range(cfg.blob_count):
        phase = i * 14.2
        speed_mult = 0.7 + 0.5 * np.sin(i * 4.1)

        raw_swing = np.sin(cfg.blob_speed * speed_mult * t + phase)
        swing = np.sign(raw_swing) * (abs(raw_swing) ** 0.8)
        normalized_swing = 0.5 * (1 + swing)

        y_val = interpolate(bot_y + 0.4, top_y - 0.4, normalized_swing)
        x_val = cfg.lamp_x + 0.42 * np.sin(cfg.drift_speed * t * 1.1 + phase * 1.6)

        size = 0.36 + 0.12 * np.cos(phase * 0.7)
        raw_pos = np.array([x_val, y_val, 0])

        centers.append((clamp_to_lamp_walls(raw_pos, size, cfg), size))

    return centers


def make_interactable_fluid_surface(centers, color: str, opacity: float, radius_mult: float = 1.0) -> VGroup:
    """Renders highly stable, gorgeous fluid profiles using contextual vector geometry.

    Completely eliminates vector tearing, inversions, and layout-destroying artifacts.
    """
    surface_group = VGroup()

    for idx, (center_pos, radius) in enumerate(centers):
        adj_radius = radius * radius_mult

        if idx == 0:
            # Bottom pool - flattened resting mass
            blob = Ellipse(width=adj_radius * 2.4, height=adj_radius * 1.3).move_to(center_pos + DOWN * 0.05)
        elif idx == 1:
            # Top reservoir - flattened upper ceiling
            blob = Ellipse(width=adj_radius * 2.2, height=adj_radius * 1.1).move_to(center_pos + UP * 0.05)
        elif idx in [2, 3]:
            # Base undulations - organic resting blobs
            blob = Ellipse(width=adj_radius * 2.1, height=adj_radius * 1.5).move_to(center_pos)
        else:
            # Traveling droplets - vertically elongated to simulate surface tension stretch
            blob = Ellipse(width=adj_radius * 1.7, height=adj_radius * 2.1).move_to(center_pos)

        blob.set_fill(color=color, opacity=opacity)
        blob.set_stroke(width=0)
        surface_group.add(blob)

    return surface_group


def make_panel(cfg: LampConfig) -> VGroup:
    """Constructs the sidebar information panel with fully readable text bounds."""
    panel = RoundedRectangle(
        width=cfg.panel_width,
        height=6.05,
        corner_radius=0.18,
        stroke_width=2.0,
        stroke_color=PALETTE.panel_stroke,
        fill_color=PALETTE.panel_fill,
        fill_opacity=0.80,
    ).move_to([cfg.panel_x, 0.38, 0])

    header = Text("What keeps it moving?", font_size=15, color=PALETTE.text_main, weight=BOLD)
    header.next_to(panel.get_top(), DOWN, buff=0.3).align_to(panel, LEFT).shift(RIGHT * 0.25)

    lines = VGroup(
        Text("Warm wax expands", font_size=13, color=PALETTE.text_main),
        Text("Less dense, so it rises", font_size=13, color=PALETTE.text_main),
        Text("Cooling makes it sink", font_size=13, color=PALETTE.text_main),
    ).arrange(DOWN, aligned_edge=LEFT, buff=0.22)
    lines.next_to(header, DOWN, buff=0.35).align_to(header, LEFT)

    expert = VGroup(
        Text("Expert note", font_size=12, color=PALETTE.text_muted, weight=BOLD),
        MathTex(r"T \uparrow \;\Rightarrow\; \rho \downarrow", font_size=22, color=PALETTE.text_main),
        Text("T = temperature, ρ = density", font_size=11, color=PALETTE.text_muted),
    ).arrange(DOWN, aligned_edge=LEFT, buff=0.18)
    expert.next_to(lines, DOWN, buff=0.45).align_to(lines, LEFT)

    divider = Line(
        panel.get_left() + RIGHT * 0.2,
        panel.get_right() + LEFT * 0.2,
        color=PALETTE.panel_stroke,
        stroke_width=1.4,
    )
    divider.move_to(panel.get_center() + DOWN * 0.8)

    return VGroup(panel, header, lines, expert, divider)


def make_watermark(cfg: LampConfig) -> Text:
    wm = Text("© Mugambi Ndwiga / @craftsandengineering", font_size=15, color=PALETTE.text_main)
    wm.set_opacity(0.6)
    wm.scale(cfg.watermark_scale)
    wm.to_corner(DR, buff=0.18)
    return wm


def create_background_glows(cfg: LampConfig) -> VGroup:
    warm_glow = Circle(radius=3.0).set_fill(color=PALETTE.heat_glow, opacity=0.08).set_stroke(width=0)
    warm_glow.move_to([cfg.lamp_x, cfg.lamp_y - 3.2, 0])

    cool_glow = Circle(radius=2.6).set_fill(color=PALETTE.cool_glow, opacity=0.06).set_stroke(width=0)
    cool_glow.move_to([cfg.lamp_x, cfg.lamp_y + 3.2, 0])

    lamp_shadow = Ellipse(width=5.0, height=7.8).set_fill(color="#000000", opacity=0.2).set_stroke(width=0)
    lamp_shadow.move_to([cfg.lamp_x, cfg.lamp_y, 0])

    return VGroup(warm_glow, cool_glow, lamp_shadow)


class LavaLampAnimation(Scene):
    """2D lava lamp explainer for vertical social distribution."""

    def construct(self):
        cfg = CFG

        background = Rectangle(width=config.frame_width * 2, height=config.frame_height * 2)
        background.set_fill(color=PALETTE.background, opacity=1)
        background.set_stroke(width=0)
        self.add(background)

        self.add(create_background_glows(cfg))

        lamp_body = make_lamp_body(cfg)
        lamp_base = make_lamp_base(cfg)
        panel = make_panel(cfg)
        watermark = make_watermark(cfg)

        title_band = RoundedRectangle(
            width=6.6,
            height=1.3,
            corner_radius=0.18,
            stroke_width=0,
            fill_color="#0d1724",
            fill_opacity=0.50,
        ).move_to([0, 5.6, 0])

        title = Text(
            "Why a lava lamp\nnever sits still",
            font_size=36,
            color=PALETTE.text_main,
            weight=BOLD,
            line_spacing=0.85
        ).move_to(title_band.get_center() + UP * 0.1)

        subtitle = Text("Heat, density, rise, cooling", font_size=16, color=PALETTE.text_muted)
        subtitle.next_to(title, DOWN, buff=0.15)

        self.add(title_band, lamp_base, lamp_body, panel, watermark)
        self.play(FadeIn(title, shift=DOWN * 0.08), FadeIn(subtitle, shift=DOWN * 0.05), run_time=0.9)

        time = ValueTracker(0)
        fluid_composite = VGroup()

        def fluid_updater(mob):
            t_val = time.get_value()
            centers = get_fluid_centers(t_val, cfg)

            # Concentric multi-layered depths with zero structural tearing risks
            layer_cool = make_interactable_fluid_surface(centers, PALETTE.wax_cool, 0.25, radius_mult=1.08)
            layer_hot = make_interactable_fluid_surface(centers, PALETTE.wax_hot, 0.55, radius_mult=0.98)
            layer_core = make_interactable_fluid_surface(centers, PALETTE.wax_core, 0.70, radius_mult=0.76)

            fresh_composite = VGroup(layer_cool, layer_hot, layer_core)
            mob.become(fresh_composite)

        fluid_updater(fluid_composite)
        fluid_composite.add_updater(fluid_updater)
        self.add(fluid_composite)

        # Clean dynamic labels placed flawlessly inside the lower container bounds
        hook_note = Text("Heat starts the climb", font_size=14, color=PALETTE.text_main, weight=BOLD)
        hook_note.move_to(panel[0].get_center() + DOWN * 1.8).align_to(panel[1], LEFT)
        self.play(FadeIn(hook_note, shift=UP * 0.1), run_time=0.5)

        self.play(time.animate.set_value(cfg.scene_seconds * 0.35), run_time=4.0, rate_func=linear)
        self.play(time.animate.set_value(cfg.scene_seconds * 0.68), run_time=4.4, rate_func=linear)

        mid_note = Text("Cooling sends it back", font_size=14, color=PALETTE.text_main, weight=BOLD)
        mid_note.move_to(panel[0].get_center() + DOWN * 1.8).align_to(panel[1], LEFT)
        self.play(Transform(hook_note, mid_note), run_time=0.5)

        self.play(time.animate.set_value(cfg.scene_seconds), run_time=3.9, rate_func=linear)

        self.play(FadeOut(title), FadeOut(subtitle), FadeOut(hook_note), run_time=0.6)

        closing_bg = Rectangle(width=config.frame_width * 2, height=config.frame_height * 2)
        closing_bg.set_fill(color="#10192a", opacity=1)
        closing_bg.set_stroke(width=0)

        closing_text_1 = Text("Made by Mugambi Ndwiga", font_size=30, color=PALETTE.text_main, weight=BOLD)
        closing_text_2 = Text("@craftsandengineering", font_size=24, color=PALETTE.text_muted)
        closing_group = VGroup(closing_text_1, closing_text_2).arrange(DOWN, buff=0.2)
        closing_group.move_to(ORIGIN)

        self.play(FadeIn(closing_bg), FadeIn(closing_group, scale=0.96), run_time=0.55)
        self.wait(cfg.closing_seconds - 0.55)
```
