# Mountains of Infinity: Visualizing the Complex Gamma Function

This project provides a visually compelling 3D animation of the magnitude of the Gamma function $|\Gamma(z)|$ over the complex plane. It is designed to demonstrate the concept of **poles**—points where a mathematical function approaches infinity—in a way that is accessible to the general public.

## 🌌 Overview
The Gamma function is an extension of the factorial function to complex numbers. One of its most striking features is that it 'explodes' at zero and negative integers ($0, -1, -2, -3, \dots$). In this visualization, these points appear as towering 'Mountains of Infinity.'

## 🛠️ Features
- **3D Surface Mapping**: Uses `scipy.special.gamma` to calculate values across a complex grid.
- **Dynamic Animation**: A 360-degree camera rotation built with Matplotlib's `FuncAnimation`.
- **Cinematic Styling**: Features the 'inferno' colormap on a dark background for maximum visual impact.
- **Interactive Playback**: Rendered as an HTML5 video for seamless viewing in Jupyter/Colab environments.

## 📚 Mathematical Context
The 'peaks' you see are called **Poles**. In complex analysis, a pole is a type of singularity where a function behaves like $1/z^n$. For the Gamma function, these occur at every non-positive integer, creating the infinite towers seen in the plot.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# 1. Setup the coordinate grid
x = np.linspace(-3.8, 3.5, 300)
y = np.linspace(-2, 2, 300)
X, Y = np.meshgrid(x, y)
Z_complex = X + 1j * Y

# 2. Calculate |Gamma(z)| and clip for visual stability
Z = np.abs(gamma(Z_complex))
Z_limit = 10
Z_clipped = np.clip(Z, 0, Z_limit)

# 3. Create the plot
fig = plt.figure(figsize=(12, 8), facecolor='black')
ax = fig.add_subplot(111, projection='3d', facecolor='black')

# Initial plot
surf = ax.plot_surface(X, Y, Z_clipped, cmap='inferno',
                       linewidth=0, antialiased=True,
                       shade=True, alpha=0.9)

# Styling the axes to be dark and minimalist
ax.set_axis_off()
ax.set_zlim(0, Z_limit)

# Adding a title that glows
ax.text2D(0.5, 0.95, "The Mountains of Infinity: |Γ(z)|",
          transform=ax.transAxes, color='white',
          ha='center', fontsize=18, fontweight='bold')

# 4. Animation function: Rotate the camera
def update(frame):
    ax.view_init(elev=30, azim=frame)
    return surf,

# 5. Generate animation
# 360 frames for a full rotation, 20ms interval for smooth motion
anim = FuncAnimation(fig, update, frames=np.arange(0, 360, 2), interval=40)

# Close the initial static plot so it doesn't double-render
plt.close()

# Display as HTML5 video
HTML(anim.to_html5_video())
```

### What are you seeing?

*   **The Peaks:** Each of those glowing towers represents a **pole**. Notice they are located exactly at $x = 0, -1, -2, -3$. At these precise points, the Gamma function has no finite value; it literally 'explodes' to infinity.
*   **The Colors:** The 'inferno' colormap uses dark purples for low values and bright yellows/whites for high values, emphasizing the intensity of the function as it approaches a pole.
*   **The Complex Plane:** The flat 'floor' is the complex plane (Real and Imaginary numbers). By looking at the magnitude of the function in 3D, we can see the mathematical structure in a way a 2D graph can't capture.
