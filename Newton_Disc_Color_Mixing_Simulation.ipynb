# Newton Disc Color Mixing Simulation

This notebook demonstrates **additive color theory** through a high-fidelity animation of a spinning Newton Disc.

### Features
- **Temporal Super-Sampling:** Renders multiple sub-steps per frame to simulate realistic motion blur.
- **Exponential Acceleration:** Uses a non-linear speed ramp to simulate a motor starting up.
- **Optical Blending:** Shows how primary colors (RGB) merge into white when rotated at high speeds.
- **MP4 Export:** Includes code to render and download the final simulation.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib

"""
Newton Disc Animation with Temporal Super-Sampling (Motion Blur).

This script creates a 45-second animation of a spinning disk with Red, Green,
and Blue segments. To achieve realistic color blending on digital displays,
it implements motion blur by averaging 8 sub-step positions per frame.

Physics Logic:
- total_rot: Calculated using a 2.5 power scaling for a smooth, natural acceleration.
- intensity: Dynamically adjusted based on speed to simulate persistence of vision.
"""

# Increase limit for high-quality blur rendering
matplotlib.rcParams['animation.embed_limit'] = 150.0

fig, ax = plt.subplots(figsize=(10, 10), facecolor='#050505')
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 1.2)
ax.axis('off')

# Setup colors and segments
colors = ['#FF0000', '#00FF00', '#0000FF']
num_segments = 3
theta_span = 360 / num_segments

# Physics/Camera settings
fps = 30
duration = 45
frames = fps * duration
max_speed = 180
sub_steps = 8

# Container for all patch artists (main and blur samples)
all_patches = []
for _ in range(sub_steps):
    step_patches = []
    for i in range(num_segments):
        w = patches.Wedge((0, 0), 1.0, 0, 0, color=colors[i], alpha=1.0/sub_steps)
        ax.add_artist(w)
        step_patches.append(w)
    all_patches.append(step_patches)

def update(frame):
    t = frame / frames
    speed = max_speed * (t**1.5)
    total_rot = (max_speed * frames / 2.5) * (t**2.5)

    for s in range(sub_steps):
        sub_offset = (s / sub_steps) * speed
        current_rot = total_rot + sub_offset

        for i, wedge in enumerate(all_patches[s]):
            start = (i * theta_span + current_rot) % 360
            end = start + theta_span
            wedge.set_theta1(start)
            wedge.set_theta2(end)

            # Stay in 2D space
            wedge.set_transform(ax.transData)

            intensity = 0.5 + 0.5 * (1 - min(speed/max_speed, 0.9))
            wedge.set_alpha((1.0 / sub_steps) * intensity)

    return [p for sub in all_patches for p in sub]

ani = FuncAnimation(fig, update, frames=frames, interval=1000/fps, blit=True)
plt.close()
HTML(ani.to_jshtml())
```

```python
from google.colab import files

# Define the filename
filename = 'newton_disc_animation.mp4'

print(f"Saving animation to {filename}... This may take a few minutes.")
# Save the animation using ffmpeg writer
ani.save(filename, writer='ffmpeg', fps=fps)

# Trigger the download to your local machine
print("Download starting...")
files.download(filename)
```
