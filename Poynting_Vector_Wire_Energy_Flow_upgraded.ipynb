# Energy Flow Visualization: The Poynting Vector

## Overview
This project provides an accurate 3D visualization of electromagnetic energy flow around a current-carrying conductor. By calculating and animating the **Poynting Vector ($S = E \times H$)**, it demonstrates the fundamental principle that energy travels through the electromagnetic fields surrounding a wire rather than inside the metal itself.

## Key Features
- **3D Field Mapping**: Simultaneous visualization of Electric ($E$) and Magnetic ($H$) fields.
- **Poynting Streamlines**: Dynamic cyan streamlines showing the path of energy transport.
- **Physical Realism**: Features radial magnetic field decay and longitudinal energy flow.
- **Educational Annotations**: Scene-based commentary explaining the source-to-load energy transfer.
- **Cinematic Rendering**: Smooth orbital camera motion with high-resolution field lines.

## Physics Principles
- **Amperian Loops**: Magnetic field rings circling the wire according to the right-hand rule.
- **Electric Gradient**: Field lines pointing along the conductor towards the resistive load.
- **Energy Influx**: Visual proof of energy entering the conductor surface to compensate for resistive losses ($I^2R$).

````python
"""
ENERGY FLOW INTO A WIRE — PHYSICALLY UPGRADED 3D VISUALIZATION
==============================================================

Author:
Mugambi Ndwiga

Description:
Scientifically upgraded visualization of electromagnetic energy flow
around a current-carrying conductor using the Poynting vector.

This version improves:
- Correct longitudinal Poynting flow
- Explicit electric field visualization
- Radial magnetic field decay
- Streamline-like energy transport
- Scene-based educational annotations
- Source/load interpretation
- Resistive conductor clarification
- Enhanced cinematic camera motion
- Better physical realism

Physics:
S = E % H

The animation demonstrates that electromagnetic energy primarily
flows THROUGH THE SURROUNDING FIELDS and is guided into the conductor.

Platform:
Google Colab / Jupyter

Libraries:
numpy
matplotlib
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.lines import Line2D
from IPython.display import HTML
import matplotlib

# ============================================================
# CONFIGURATION
# ============================================================

# Increased limit to handle longer duration
matplotlib.rcParams['animation.embed_limit'] = 150.0

FPS = 24                 # Slightly reduced FPS to manage file size
DURATION = 26            # 10 seconds longer than original 16
TOTAL_FRAMES = FPS * DURATION

LIMIT = 2.5

# ============================================================
# FIGURE SETUP
# ============================================================

fig = plt.figure(figsize=(12, 12), facecolor='black')
ax = fig.add_subplot(111, projection='3d')
ax.set_facecolor('black')

ax.set_xlim(-LIMIT, LIMIT)
ax.set_ylim(-LIMIT, LIMIT)
ax.set_zlim(-LIMIT, LIMIT)

ax.axis('off')

# ============================================================
# TITLE
# ============================================================

title = ax.text2D(
    0.5,
    0.95,
    "ENERGY FLOW INTO A CURRENT-CARRYING WIRE",
    transform=ax.transAxes,
    ha='center',
    color='white',
    fontsize=18,
    fontweight='bold'
)

subtitle = ax.text2D(
    0.5,
    0.915,
    "Electromagnetic energy flows THROUGH surrounding fields",
    transform=ax.transAxes,
    ha='center',
    color='cyan',
    fontsize=11
)

# ============================================================
# SCENE TEXT (Repositioned to bottom-left to avoid title)
# ============================================================

scene_text = ax.text2D(
    0.03,
    0.15,
    "",
    transform=ax.transAxes,
    color='white',
    fontsize=12,
    verticalalignment='bottom',
    bbox=dict(facecolor='black', alpha=0.6, edgecolor='white')
)

formula_text = ax.text2D(
    0.03,
    0.06,
    r"$\bf{S = E \times H}$"
    "\nPoynting Vector = Electromagnetic Energy Flow",
    transform=ax.transAxes,
    color='cyan',
    fontsize=13
)

# ============================================================
# CENTRAL WIRE
# ============================================================

z_wire = np.linspace(-LIMIT, LIMIT, 80)
theta_wire = np.linspace(0, 2*np.pi, 60)

Tw, Zw = np.meshgrid(theta_wire, z_wire)

wire_radius = 0.18

Xw = wire_radius * np.cos(Tw)
Yw = wire_radius * np.sin(Tw)

wire = ax.plot_surface(
    Xw,
    Yw,
    Zw,
    color='gold',
    alpha=0.95,
    linewidth=0,
    shade=True
)

# ============================================================
# CURRENT DIRECTION ARROW
# ============================================================

current_arrow = ax.quiver(
    0,
    0,
    -1.7,
    0,
    0,
    2.6,
    color='white',
    linewidth=3,
    arrow_length_ratio=0.12
)

current_label = ax.text(
    0.25,
    0,
    1.3,
    "Current I",
    color='white',
    fontsize=11
)

# ============================================================
# ELECTRIC FIELD LINES
# ============================================================

electric_lines = []

radii_e = [0.45, 0.75, 1.05]

for r in radii_e:

    for angle in np.linspace(0, 2*np.pi, 12, endpoint=False):

        x0 = r * np.cos(angle)
        y0 = r * np.sin(angle)

        zline = np.linspace(-2.2, 2.2, 100)

        xline = np.full_like(zline, x0)
        yline = np.full_like(zline, y0)

        line, = ax.plot(
            xline,
            yline,
            zline,
            color='yellow',
            alpha=0.12,
            lw=1
        )

        electric_lines.append(line)

# ============================================================
# MAGNETIC FIELD RINGS
# ============================================================

def get_ring(radius, height, n=200):

    t = np.linspace(0, 2*np.pi, n)

    return (
        radius * np.cos(t),
        radius * np.sin(t),
        np.full_like(t, height)
    )

magnetic_rings = []

ring_heights = np.linspace(-2.0, 2.0, 10)

for h in ring_heights:

    line, = ax.plot([], [], [],
                    color='magenta',
                    lw=2,
                    alpha=0.5)

    magnetic_rings.append((line, h))

# ============================================================
# POYNTING VECTOR FIELD
# ============================================================

grid_n = 8

x = np.linspace(-2.0, 2.0, grid_n)
y = np.linspace(-2.0, 2.0, grid_n)
z = np.linspace(-2.0, 2.0, grid_n)

X, Y, Z = np.meshgrid(x, y, z)

R = np.sqrt(X**2 + Y**2)

mask = (R > 0.35) & (R < 1.8)

Xm = X[mask]
Ym = Y[mask]
Zm = Z[mask]

radial_strength = 0.45 / (R[mask] + 0.25)

Sx = -Xm * radial_strength
Sy = -Ym * radial_strength

Sz = 1.4 / (R[mask] + 0.2)

norm = np.sqrt(Sx**2 + Sy**2 + Sz**2)

Sx /= norm
Sy /= norm
Sz /= norm

alpha_vals = np.clip(1.6 - R[mask], 0.15, 1.0)

quiver = ax.quiver(
    Xm,
    Ym,
    Zm,
    Sx,
    Sy,
    Sz,
    length=0.32,
    normalize=True,
    color='cyan',
    linewidth=1.2,
    alpha=0.7
)

# ============================================================
# ENERGY STREAMLINES
# ============================================================

stream_lines = []

for angle in np.linspace(0, 2*np.pi, 10, endpoint=False):

    r0 = 1.6

    zstream = np.linspace(-2.0, 2.0, 200)

    spiral_r = np.linspace(1.6, 0.35, 200)

    theta = angle + 0.9 * zstream

    xs = spiral_r * np.cos(theta)
    ys = spiral_r * np.sin(theta)

    line, = ax.plot(
        xs,
        ys,
        zstream,
        color='cyan',
        alpha=0.12,
        lw=2
    )

    stream_lines.append(line)

# ============================================================
# BATTERY / LOAD INDICATORS
# ============================================================

battery_text = ax.text(
    0,
    0,
    -2.35,
    "Voltage Source",
    color='lime',
    fontsize=11,
    ha='center'
)

load_text = ax.text(
    0,
    0,
    2.25,
    "Resistive Load",
    color='orange',
    fontsize=11,
    ha='center'
)

# ============================================================
# LEGEND
# ============================================================

legend_elements = [

    Line2D(
        [0],
        [0],
        color='gold',
        lw=6,
        label='Copper Conductor'
    ),

    Line2D(
        [0],
        [0],
        color='yellow',
        lw=2,
        label='Electric Field (E)'
    ),

    Line2D(
        [0],
        [0],
        color='magenta',
        lw=2,
        label='Magnetic Field (H)'
    ),

    Line2D(
        [0],
        [0],
        color='cyan',
        marker='>',
        linestyle='None',
        markersize=10,
        label='Poynting Flow (S)'
    )
]

ax.legend(
    handles=legend_elements,
    loc='lower right',
    facecolor='black',
    edgecolor='white',
    labelcolor='white',
    fontsize=11,
    framealpha=0.7
)

# ============================================================
# SCENE DEFINITIONS (Updated: Removed 'SCENE X' labels)
# ============================================================

# Adjusted frame windows for 24 FPS and 26s duration
SCENES = [
    (
        0,
        int(TOTAL_FRAMES * 0.2),
        "A current flows upward through the conductor.\n"
        "The wire guides electromagnetic energy."
    ),
    (
        int(TOTAL_FRAMES * 0.2),
        int(TOTAL_FRAMES * 0.4),
        "Electric fields (yellow) exist around the wire.\n"
        "They point mainly along the conductor."
    ),
    (
        int(TOTAL_FRAMES * 0.4),
        int(TOTAL_FRAMES * 0.6),
        "Magnetic fields (magenta) circle the wire\n"
        "according to the right-hand rule."
    ),
    (
        int(TOTAL_FRAMES * 0.6),
        int(TOTAL_FRAMES * 0.8),
        "The Poynting vector S = E % H\n"
        "describes electromagnetic energy flow."
    ),
    (
        int(TOTAL_FRAMES * 0.8),
        TOTAL_FRAMES,
        "Energy flows mostly ALONG the wire,\n"
        "while gradually entering the conductor."
    ),
]

# ============================================================
# ANIMATION UPDATE
# ============================================================

def update(frame):

    azim = frame * 0.45
    elev = 22 + 10*np.sin(frame * 0.02)

    ax.view_init(elev=elev, azim=azim)

    for line, h in magnetic_rings:

        radius = 0.55 + 0.55/(abs(h)+0.6)
        pulse = 0.05*np.sin(frame*0.08 + h)
        radius += pulse

        xr, yr, zr = get_ring(radius, h)
        decay = 1/(radius + 0.3)

        line.set_data(xr, yr)
        line.set_3d_properties(zr)
        line.set_alpha(0.25 + 0.45*decay)
        line.set_linewidth(1 + 2*decay)

    for idx, line in enumerate(stream_lines):

        zstream = np.linspace(-2.0, 2.0, 200)
        spiral_r = np.linspace(1.6, 0.35, 200)

        theta = (idx*0.6 + 1.2*zstream + frame*0.04)

        xs = spiral_r * np.cos(theta)
        ys = spiral_r * np.sin(theta)

        line.set_data(xs, ys)
        line.set_3d_properties(zstream)

    for start, end, text in SCENES:
        if start <= frame < end:
            scene_text.set_text(text)
            break

    return []

# ============================================================
# CREATE ANIMATION
# ============================================================

ani = FuncAnimation(
    fig,
    update,
    frames=TOTAL_FRAMES,
    interval=1000/FPS,
    blit=False
)

plt.close()

# ============================================================
# EXPORT
# ============================================================

HTML(ani.to_jshtml())
````
