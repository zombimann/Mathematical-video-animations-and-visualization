# Visualizing Quantum Interference: The Wigner Function Morph

## Project Overview
This notebook contains a narrative visualization of a quantum state transition. It focuses on the morphing of a coherent state into a Schrodinger cat state, visualized through the lens of the Wigner quasi-probability distribution. The output is a high-fidelity animation designed for technical communication and educational purposes.

## Background and Concepts

### Coherent States
A coherent state is a specific quantum state of the harmonic oscillator that most closely resembles classical behavior. In phase space, it is represented as a localized 'blob' of probability density. It maintains a constant uncertainty and a minimum uncertainty product, making it the standard representation for classical-like light or particles.

### Schrodinger Cat States
In quantum optics, a cat state is a quantum superposition of two coherent states with opposite phases. This state is a cornerstone of quantum mechanics, illustrating the principle of superposition on a macroscopic scale. It represents a system that exists in two distinct classical-like states simultaneously.

### The Wigner Quasi-probability Distribution
The Wigner function provides a way to represent quantum states in phase space (position and momentum). Unlike classical probability distributions, the Wigner function can take on negative values. These negative regions are a hallmark of non-classicality and serve as a direct signature of quantum interference between the superposed components of the state.

### Quantum Interference
When two coherent states are superposed, they interfere. In the Wigner representation, this interference manifests as oscillatory patterns (fringes) between the two main lobes. These fringes contain the characteristic negative values that distinguish a true quantum superposition from a classical statistical mixture.

```python
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Concept: Wigner-function morph from a coherent blob into a Schrödinger cat state
Repo: github.com/zombimann/Mathematical-video-animations-and-visualization
"""

from __future__ import annotations

import shutil
import subprocess
import logging
from dataclasses import dataclass, field
from pathlib import Path

import matplotlib
matplotlib.use("Agg")

import matplotlib.pyplot as plt
from matplotlib import patheffects as pe
from matplotlib.animation import FFMpegWriter, FuncAnimation
from matplotlib.colors import LinearSegmentedColormap, TwoSlopeNorm
from matplotlib.font_manager import FontProperties
from matplotlib.patches import ConnectionPatch, Ellipse, FancyArrowPatch, FancyBboxPatch
from matplotlib.path import Path as MplPath

import numpy as np
from scipy.ndimage import gaussian_filter
from qutip import coherent, wigner

# Suppress repetitive font manager warnings to keep console clean
logging.getLogger('matplotlib.font_manager').setLevel(logging.ERROR)

# ======================================================================================
# CONFIG  -- every dial for the film lives here
# ======================================================================================

@dataclass(frozen=True)
class Config:
    # --- quantum parameters ---
    dimension: int = 50
    alpha: float = 2.15
    grid_extent: float = 4.6
    grid_points: int = 121

    # --- canvas (YouTube Shorts) ---
    width_in: float = 9.0
    height_in: float = 16.0
    dpi: int = 120
    fps: int = 24

    # --- pacing, in seconds, of each narrative beat ---
    t_hook: float = 2.6
    t_scene: float = 3.6
    t_morph: float = 15.6
    t_payoff: float = 4.2
    t_closing: float = 1.8
    fade_s: float = 0.45  # crossfade duration between beats, seconds

    # --- output / compression ---
    output_name: str = "wigner_cat_transition.mp4"
    target_mb: float = 9.4

    # --- copy ---
    title: str = "From One Blob to a Quantum Cat"
    watermark: str = "\u00a9 Mugambi Ndwiga / @craftsandengineering"


cfg = Config()


# ======================================================================================
# FONTS -- calligraphic, loaded directly by path with safe, fast system fallbacks
# ======================================================================================

FONT_DIR = Path("/home/claude/fonts")
path_reg = FONT_DIR / "Caveat-Regular.ttf"
path_bold = FONT_DIR / "Caveat-Bold.ttf"
path_title = FONT_DIR / "PermanentMarker.ttf"

# Check if the custom font files exist; if not, fall back to avoid the costly filesystem loop
if path_reg.exists() and path_bold.exists() and path_title.exists():
    hand_reg = FontProperties(fname=str(path_reg))
    hand_bold = FontProperties(fname=str(path_bold))
    hand_title = FontProperties(fname=str(path_title))
else:
    print("\n[!] Custom font files were not found at /home/claude/fonts/")
    print("[!] Falling back to safe standard system fonts to prevent rendering slowdowns.\n")
    hand_reg = FontProperties(family="sans-serif", weight="normal")
    hand_bold = FontProperties(family="sans-serif", weight="bold")
    hand_title = FontProperties(family="sans-serif", weight="bold")

ffmpeg_path = shutil.which("ffmpeg")
if ffmpeg_path:
    matplotlib.rcParams["animation.ffmpeg_path"] = ffmpeg_path

# --- palette: bled ink on aged parchment ---
ink = "#2c1810"
ink_soft = "#4a2f20"
paper = "#e9dcc0"
card_face = (1.0, 0.97, 0.90, 0.50)
card_edge = (0.26, 0.16, 0.10, 0.30)
text_fx = [pe.withStroke(linewidth=2.0, foreground=(1, 1, 1, 0.20))]

ink_paper_cmap = LinearSegmentedColormap.from_list(
    "ink_paper",
    ["#1c0f1a", "#5a3152", "#e9dcc0", "#b3753f", "#3c2114"],
    N=256,
)


# ======================================================================================
# HELPERS
# ======================================================================================

def smoothstep(x: float) -> float:
    x = float(np.clip(x, 0.0, 1.0))
    return x * x * (3.0 - 2.0 * x)


def ease_window(frame: int, start: int, end: int, fade_frames: int) -> float:
    """0->1->0 opacity envelope for a [start, end) frame window with cosine fades."""
    if frame < start - fade_frames or frame >= end + fade_frames:
        return 0.0
    if frame < start:
        t = (frame - (start - fade_frames)) / max(fade_frames, 1)
        return 0.5 - 0.5 * np.cos(np.pi * np.clip(t, 0, 1))
    if frame >= end:
        t = (frame - end) / max(fade_frames, 1)
        return 1.0 - (0.5 - 0.5 * np.cos(np.pi * np.clip(t, 0, 1)))
    return 1.0


def make_paper_texture(height: int, width: int, seed: int = 7) -> np.ndarray:
    rng = np.random.default_rng(seed)
    base = np.empty((height, width, 3), dtype=np.float32)
    base[..., 0], base[..., 1], base[..., 2] = 0.93, 0.87, 0.74

    fine = gaussian_filter(rng.normal(size=(height, width)), sigma=10)
    fine = (fine - fine.min()) / (fine.max() - fine.min() + 1e-9)
    blotch = gaussian_filter(rng.normal(size=(height, width)), sigma=42)
    blotch = (blotch - blotch.min()) / (blotch.max() - blotch.min() + 1e-9)

    yy, xx = np.mgrid[0:height, 0:width]
    vignette = np.sqrt(((xx - width / 2) / (0.60 * width)) ** 2 + ((yy - height / 2) / (0.74 * height)) ** 2)
    vignette = np.clip(vignette, 0.0, 1.0)

    stains = np.zeros((height, width), dtype=np.float32)
    for _ in range(22):
        cx, cy = rng.uniform(0, width), rng.uniform(0, height)
        sx, sy = rng.uniform(width * 0.03, width * 0.12), rng.uniform(height * 0.02, height * 0.07)
        amp = rng.uniform(0.02, 0.08)
        stains += amp * np.exp(-0.5 * (((xx - cx) / sx) ** 2 + ((yy - cy) / sy) ** 2))

    tex = base - 0.045 * fine[..., None] - 0.030 * blotch[..., None]
    tex -= 0.095 * vignette[..., None]
    tex -= stains[..., None] * np.array([0.22, 0.13, 0.08], dtype=np.float32)
    return np.clip(tex, 0.0, 1.0)


def card(ax, face=card_face, edge=card_edge, lw=1.3, pad=0.035, rounding=0.05):
    patch = FancyBboxPatch(
        (0, 0), 1, 1, transform=ax.transAxes,
        boxstyle=f"round,pad={pad},rounding_size={rounding}",
        facecolor=face, edgecolor=edge, linewidth=lw, zorder=0, clip_on=False,
    )
    ax.add_patch(patch)
    ax.set_axis_off()
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_autoscale_on(False)
    return patch


def set_card_alpha(patch, text_objs, opacity: float, base_face=card_face, base_edge=card_edge):
    patch.set_facecolor((base_face[0], base_face[1], base_face[2], base_face[3] * opacity))
    patch.set_edgecolor((base_edge[0], base_edge[1], base_edge[2], base_edge[3] * opacity))
    for t, base_alpha in text_objs:
        t.set_alpha(base_alpha * opacity)


# ======================================================================================
# FIGURE CONSTRUCTION -- explicit rects (inches -> fractions), guarantees no overlap
# ======================================================================================

def build_figure():
    W, H = cfg.width_in, cfg.height_in
    fig = plt.figure(figsize=(W, H), dpi=cfg.dpi)
    fig.patch.set_facecolor(paper)

    bg_ax = fig.add_axes([0, 0, 1, 1], zorder=0)
    bg_ax.imshow(make_paper_texture(int(cfg.dpi * H), int(cfg.dpi * W), seed=9), aspect="auto")
    bg_ax.set_axis_off()

    def rect(x0, y1, w, h):
        return [x0, y1 - h, w, h]

    margin = 0.045
    content_w = 1.0 - 2 * margin
    y = 0.985

    # ---- title bar ----
    h_title = 0.052
    y -= h_title
    title_ax = fig.add_axes(rect(margin, y + h_title, content_w, h_title), zorder=20)
    title_patch = card(title_ax)
    title_txt = title_ax.text(0.5, 0.5, cfg.title, ha="center", va="center",
                               fontsize=34, color=ink, fontproperties=hand_title,
                               path_effects=text_fx)
    y -= 0.008

    # ---- square phase-space plot, centered ----
    plot_w = content_w * 0.94
    plot_h = plot_w * (W / H)
    y -= plot_h
    plot_ax = fig.add_axes(rect(margin + (content_w - plot_w) / 2, y + plot_h, plot_w, plot_h), zorder=5)
    plot_ax.set_facecolor((1, 1, 1, 0.0))
    plot_frame = FancyBboxPatch(
        (0, 0), 1, 1, transform=plot_ax.transAxes,
        boxstyle="round,pad=0.012,rounding_size=0.025",
        facecolor=(1, 1, 1, 0.14), edgecolor=(0.26, 0.16, 0.10, 0.35), linewidth=1.6, zorder=0,
    )
    plot_ax.add_patch(plot_frame)
    y -= 0.038

    # ---- story card ----
    h_story = 0.135
    y -= h_story
    story_ax = fig.add_axes(rect(margin, y + h_story, content_w, h_story), zorder=20)
    story_patch = card(story_ax)
    story_kicker = story_ax.text(0.07, 0.83, "", ha="left", va="center", fontsize=18,
                                  color=ink_soft, fontproperties=hand_bold, alpha=0.85, path_effects=text_fx)
    story_title = story_ax.text(0.07, 0.57, "", ha="left", va="center", fontsize=30,
                                 color=ink, fontproperties=hand_bold, path_effects=text_fx)
    story_body = story_ax.text(0.07, 0.22, "", ha="left", va="center", fontsize=21,
                                color=ink, fontproperties=hand_reg, wrap=True, path_effects=text_fx)
    y -= 0.010

    # ---- math card ----
    h_math = 0.118
    y -= h_math
    math_ax = fig.add_axes(rect(margin, y + h_math, content_w, h_math), zorder=20)
    math_patch = card(math_ax)
    math_kicker = math_ax.text(0.07, 0.82, "THE MATH", ha="left", va="center", fontsize=16,
                                color=ink_soft, fontproperties=hand_bold, alpha=0.80, path_effects=text_fx)
    math_expr = math_ax.text(0.07, 0.50, "", ha="left", va="center", fontsize=25,
                              color=ink, fontproperties=hand_reg, path_effects=text_fx)
    math_note = math_ax.text(0.07, 0.19, "", ha="left", va="center", fontsize=17,
                              color=ink_soft, fontproperties=hand_reg, alpha=0.9, path_effects=text_fx)
    track_x0, track_x1, track_y = 0.07, 0.93, 0.10
    math_ax.plot([track_x0, track_x1], [track_y, track_y], color=(0.26, 0.16, 0.10, 0.30),
                 linewidth=3.0, solid_capstyle="round", zorder=1)
    math_dot, = math_ax.plot([track_x0], [track_y], marker="o", markersize=11,
                              markerfacecolor=ink, markeredgecolor=paper, markeredgewidth=1.2,
                              alpha=0.0, zorder=2)
    y -= 0.008

    # ---- legend strip ----
    h_legend = 0.056
    y -= h_legend
    legend_ax = fig.add_axes(rect(margin, y + h_legend, content_w, h_legend), zorder=20)
    legend_patch = card(legend_ax, pad=0.05, rounding=0.10)
    grad_ax = fig.add_axes(rect(margin + content_w * 0.32, y + h_legend - 0.038, content_w * 0.36, 0.020), zorder=21)
    grad_ax.imshow(np.linspace(-1, 1, 400)[None, :], aspect="auto", cmap=ink_paper_cmap,
                    norm=TwoSlopeNorm(0.0, -1.0, 1.0))
    grad_ax.set_axis_off()
    legend_left = legend_ax.text(0.07, 0.5, "− negative", ha="left", va="center", fontsize=16,
                                  color=ink, fontproperties=hand_bold, path_effects=text_fx)
    legend_right = legend_ax.text(0.93, 0.5, "positive +", ha="right", va="center", fontsize=16,
                                   color=ink, fontproperties=hand_bold, path_effects=text_fx)
    y -= 0.010

    # ---- footer / disclaimer + progress bar ----
    h_footer = 0.048
    y -= h_footer
    footer_ax = fig.add_axes(rect(margin, y + h_footer, content_w, h_footer), zorder=20)
    footer_ax.set_axis_off()
    footer_txt = footer_ax.text(0.0, 0.78, "Narrative visualization, not a literal time evolution.",
                                 ha="left", va="center", fontsize=14, color=ink_soft,
                                 fontproperties=hand_reg, alpha=0.85, path_effects=text_fx)
    bar_bg = FancyBboxPatch((0.0, 0.10), 1.0, 0.16, transform=footer_ax.transAxes,
                             boxstyle="round,pad=0,rounding_size=0.08",
                             facecolor=(0.26, 0.16, 0.10, 0.18), edgecolor="none", zorder=1)
    footer_ax.add_patch(bar_bg)
    bar_fg = plt.Rectangle((0.0, 0.10), 0.001, 0.16, transform=footer_ax.transAxes,
                            facecolor=(0.45, 0.20, 0.16, 0.55), edgecolor="none", zorder=2)
    footer_ax.add_patch(bar_fg)

    # ---- closing title card ----
    closing_ax = fig.add_axes([0, 0, 1, 1], zorder=40)
    closing_ax.set_axis_off()
    closing_ax.set_xlim(0, 1)
    closing_ax.set_ylim(0, 1)
    closing_ax.set_visible(False)
    closing_ax.add_patch(plt.Rectangle((0, 0), 1, 1, transform=closing_ax.transAxes,
                                        facecolor=paper, edgecolor="none", zorder=0))
    closing_ax.imshow(make_paper_texture(int(cfg.dpi * H), int(cfg.dpi * W), seed=21),
                       extent=[0, 1, 0, 1], aspect="auto", zorder=1)
    closing_name = closing_ax.text(0.5, 0.54, "Made by Mugambi Ndwiga", ha="center", va="center",
                                    fontsize=34, color=ink, fontproperties=hand_title,
                                    path_effects=text_fx, zorder=2)
    closing_handle = closing_ax.text(0.5, 0.45, "@craftsandengineering", ha="center", va="center",
                                      fontsize=24, color=ink_soft, fontproperties=hand_bold,
                                      path_effects=text_fx, zorder=2)

    # ---- watermark ----
    wm_ax = fig.add_axes([0, 0, 1, 1], zorder=30)
    wm_ax.set_axis_off()
    wm_ax.set_xlim(0, 1)
    wm_ax.set_ylim(0, 1)
    watermark_txt = wm_ax.text(0.965, 0.012, cfg.watermark, ha="right", va="bottom",
                                fontsize=13, color=ink, alpha=0.60, fontproperties=hand_reg,
                                path_effects=text_fx, zorder=31)

    elems = dict(
        fig=fig, plot_ax=plot_ax, plot_frame=plot_frame,
        title_ax=title_ax, title_patch=title_patch, title_txt=title_txt,
        story_ax=story_ax, story_patch=story_patch,
        story_kicker=story_kicker, story_title=story_title, story_body=story_body,
        math_ax=math_ax, math_patch=math_patch,
        math_kicker=math_kicker, math_expr=math_expr, math_note=math_note, math_dot=math_dot,
        legend_ax=legend_ax, legend_patch=legend_patch, legend_left=legend_left, legend_right=legend_right,
        footer_ax=footer_ax, footer_txt=footer_txt, bar_bg=bar_bg, bar_fg=bar_fg,
        watermark_txt=watermark_txt,
        closing_ax=closing_ax, closing_name=closing_name, closing_handle=closing_handle,
    )
    return elems


def style_plot_axes(plot_ax):
    g = cfg.grid_extent
    plot_ax.set_xlim(-g, g)
    plot_ax.set_ylim(-g, g)
    plot_ax.set_aspect("equal", adjustable="box")
    ticks = [-4, -2, 0, 2, 4]
    plot_ax.set_xticks(ticks)
    plot_ax.set_yticks(ticks)
    plot_ax.tick_params(length=4, width=1.0, pad=5, labelsize=17)
    for lbl in plot_ax.get_xticklabels() + plot_ax.get_yticklabels():
        lbl.set_fontproperties(hand_reg)
    plot_ax.grid(True, alpha=0.10, linewidth=0.9)
    plot_ax.axhline(0, color=ink, alpha=0.16, linewidth=1.1)
    plot_ax.axvline(0, color=ink, alpha=0.16, linewidth=1.1)
    for spine in plot_ax.spines.values():
        spine.set_color((0.26, 0.16, 0.10, 0.4))
        spine.set_linewidth(1.1)
    plot_ax.set_xlabel("x   (position)", fontsize=21, fontproperties=hand_bold, color=ink, labelpad=6)
    plot_ax.set_ylabel("p   (momentum)", fontsize=21, fontproperties=hand_bold, color=ink, labelpad=4)


def build_annotations(e):
    plot_ax = e["plot_ax"]
    g = cfg.grid_extent
    a = cfg.alpha

    sep_arrow = FancyArrowPatch(
        (-a, -g * 0.74), (a, -g * 0.74), transform=plot_ax.transData,
        arrowstyle="<|-|>", mutation_scale=14, linewidth=1.6,
        color=ink, alpha=0.0, zorder=8,
    )
    plot_ax.add_patch(sep_arrow)
    sep_label = plot_ax.text(0, -g * 0.74, "2 alpha", ha="center", va="bottom", fontsize=15,
                              color=ink, fontproperties=hand_bold, alpha=0.0, zorder=9,
                              path_effects=text_fx)

    pos_callout = ConnectionPatch(
        xyA=(0.82, 0.95), coordsA="axes fraction", axesA=e["legend_ax"],
        xyB=(g * 0.62, g * 0.08), coordsB="data", axesB=plot_ax,
        connectionstyle="arc3,rad=-0.18", arrowstyle="-|>", mutation_scale=12,
        linewidth=1.3, color=ink_soft, alpha=0.0, zorder=25,
    )
    e["fig"].add_artist(pos_callout)

    fringe_ellipse = Ellipse((0, 0), width=a * 1.55, height=g * 0.62, transform=plot_ax.transData,
                              facecolor="none", edgecolor=ink, linewidth=2.0, linestyle=(0, (5, 3)),
                              alpha=0.0, zorder=10)
    plot_ax.add_patch(fringe_ellipse)

    leader = ConnectionPatch(
        xyA=(0.30, 0.99), coordsA="axes fraction", axesA=e["story_ax"],
        xyB=(-a * 0.05, -g * 0.30), coordsB="data", axesB=plot_ax,
        connectionstyle="arc3,rad=0.12", arrowstyle="-|>", mutation_scale=14,
        linewidth=1.5, color=ink_soft, alpha=0.0, zorder=25,
    )
    e["fig"].add_artist(leader)

    e["sep_arrow"], e["sep_label"] = sep_arrow, sep_label
    e["pos_callout"] = pos_callout
    e["fringe_ellipse"] = fringe_ellipse
    e["leader"] = leader
    return e


# ======================================================================================
# NARRATIVE TIMELINE -- the "screenplay"
# ======================================================================================

def seconds_to_frames(seconds: float) -> int:
    return max(1, int(round(seconds * cfg.fps)))


F_HOOK = seconds_to_frames(cfg.t_hook)
F_SCENE = seconds_to_frames(cfg.t_scene)
F_MORPH = seconds_to_frames(cfg.t_morph)
F_PAYOFF = seconds_to_frames(cfg.t_payoff)
F_CLOSING = seconds_to_frames(cfg.t_closing)
FADE = seconds_to_frames(cfg.fade_s)

B_HOOK = (0, F_HOOK)
B_SCENE = (B_HOOK[1], B_HOOK[1] + F_SCENE)
B_MORPH = (B_SCENE[1], B_SCENE[1] + F_MORPH)
B_PAYOFF = (B_MORPH[1], B_MORPH[1] + F_PAYOFF)
B_CLOSING = (B_PAYOFF[1], B_PAYOFF[1] + F_CLOSING)
TOTAL_FRAMES = B_CLOSING[1]
TOTAL_SECONDS = TOTAL_FRAMES / cfg.fps

MORPH_SPLIT_1, MORPH_SPLIT_2 = 0.08, 0.62


def morph_substage(s: float) -> tuple[str, str, str]:
    if s < MORPH_SPLIT_1:
        return ("THE SHIFT", "Here we go", "Slowly, the blob is nudged toward existing in two places at once.")
    if s < MORPH_SPLIT_2:
        return ("THE STAGE", "Splitting", "A second lobe grows from the first, and the two halves begin to interfere.")
    return ("THE REVEAL", "A Schr\u00f6dinger cat state", "Two lobes remain \u2014 and the space between them turns to stripes.")


# ======================================================================================
# PRECOMPUTE WIGNER FRAMES
# ======================================================================================

print("Precomputing Wigner functions ...")
g = cfg.grid_extent
xvec = np.linspace(-g, g, cfg.grid_points)

blob_state = coherent(cfg.dimension, cfg.alpha)
W_blob = np.asarray(wigner(blob_state, xvec, xvec), dtype=np.float32)

n_morph = B_MORPH[1] - B_MORPH[0]
morph_W: list[np.ndarray] = []
morph_s: list[float] = []
for i in range(n_morph):
    progress = i / max(n_morph - 1, 1)
    s = smoothstep(progress)
    ket_pos = coherent(cfg.dimension, cfg.alpha)
    ket_neg = coherent(cfg.dimension, -cfg.alpha)
    psi = (ket_pos + s * ket_neg).unit()
    W = np.asarray(wigner(psi, xvec, xvec), dtype=np.float32)
    morph_W.append(W)
    morph_s.append(s)

W_cat = morph_W[-1]
wmax = float(max(np.max(np.abs(W_blob)), np.max(np.abs(W_cat))))
plot_norm = TwoSlopeNorm(vcenter=0.0, vmin=-wmax, vmax=wmax)
print(f"Done. {n_morph} unique morph frames computed. Total runtime: {TOTAL_SECONDS:.2f}s "
      f"({TOTAL_FRAMES} frames @ {cfg.fps} fps).")


# ======================================================================================
# BUILD THE FIGURE + STATEFUL ARTISTS
# ======================================================================================

E = build_figure()
style_plot_axes(E["plot_ax"])
build_annotations(E)

plot_image = E["plot_ax"].imshow(
    W_blob, extent=[-g, g, -g, g], origin="lower", cmap=ink_paper_cmap,
    norm=plot_norm, interpolation="bicubic", alpha=0.90, zorder=2,
)

TITLE_GROUP = (E["title_patch"], [(E["title_txt"], 1.0)])
STORY_GROUP = (E["story_patch"], [(E["story_kicker"], 0.85), (E["story_title"], 1.0), (E["story_body"], 1.0)])
MATH_GROUP = (E["math_patch"], [(E["math_kicker"], 0.80), (E["math_expr"], 1.0), (E["math_note"], 0.9)])
LEGEND_GROUP = (E["legend_patch"], [(E["legend_left"], 1.0), (E["legend_right"], 1.0)])


def update(frame: int):
    if frame < B_HOOK[1]:
        beat = "hook"
        s = 0.0
    elif frame < B_SCENE[1]:
        beat = "scene"
        s = 0.0
    elif frame < B_MORPH[1]:
        beat = "morph"
        idx = frame - B_MORPH[0]
        s = morph_s[idx]
    elif frame < B_PAYOFF[1]:
        beat = "payoff"
        s = 1.0
    else:
        beat = "closing"
        s = 1.0

    closing_alpha = ease_window(frame, B_CLOSING[0], B_CLOSING[1], FADE)
    if closing_alpha > 0.0:
        E["closing_ax"].set_visible(True)
        for ax_key in ("title_ax", "plot_ax", "story_ax", "math_ax", "legend_ax", "footer_ax"):
            E[ax_key].set_visible(closing_alpha < 0.5)
        E["closing_name"].set_alpha(min(1.0, closing_alpha * 1.4))
        E["closing_handle"].set_alpha(min(1.0, closing_alpha * 1.4))
        if closing_alpha >= 0.999:
            return (plot_image,)
    else:
        E["closing_ax"].set_visible(False)
        for ax_key in ("title_ax", "plot_ax", "story_ax", "math_ax", "legend_ax", "footer_ax"):
            E[ax_key].set_visible(True)

    if beat in ("hook", "scene"):
        plot_image.set_data(W_blob)
        lim = g
    elif beat == "morph":
        idx = frame - B_MORPH[0]
        plot_image.set_data(morph_W[idx])
        lim = g * (1.0 - 0.10 * smoothstep(s))
    else:
        plot_image.set_data(W_cat)
        if beat == "payoff":
            zp = smoothstep((frame - B_PAYOFF[0]) / max(F_PAYOFF - 1, 1))
            lim = (g * 0.90) - zp * (g * 0.90 - g * 0.46)
        else:
            lim = g * 0.46
    E["plot_ax"].set_xlim(-lim, lim)
    E["plot_ax"].set_ylim(-lim, lim)

    if beat == "hook":
        kicker, title, body = "THE SETUP", "Meet the calmest quantum state", (
            "This is a coherent state \u2014 the quantum state that behaves\n"
            "most like a single classical point of light."
        )
        math_expr = "state:   |alpha>          alpha = 2.15"
        math_note = "alpha sets the blob's size \u2014 and soon, how far it splits."
        s_for_dot = None
    elif beat == "scene":
        kicker, title, body = "THE STAGE", "Reading a phase-space portrait", (
            "Brightness shows how likely the state is to be found at\n"
            "each position x and momentum p. Warm = likely."
        )
        math_expr = "state:   |alpha>          alpha = 2.15"
        math_note = "Watch what happens when this single blob splits in two."
        s_for_dot = None
    elif beat == "morph":
        kicker, title, body = morph_substage(s)
        math_expr = "|psi(s)> = N [ |alpha> + s|\u2212alpha> ]"
        math_note = f"s = {s:0.2f}     alpha = 2.15"
        s_for_dot = s
    elif beat == "payoff":
        kicker, title, body = "THE PUNCHLINE", "Negative probability: pure quantum", (
            "Those dark stripes between the lobes dip below zero \u2014\n"
            "no classical probability can ever do that."
        )
        math_expr = "W(x, p) < 0   only in quantum phase space"
        math_note = "Wigner negativity = the fingerprint of quantum interference."
        s_for_dot = 1.0
    else:
        kicker, title, body, math_expr, math_note, s_for_dot = "", "", "", "", "", None

    if beat != "closing":
        E["story_kicker"].set_text(kicker)
        E["story_title"].set_text(title)
        E["story_body"].set_text(body)
        E["math_expr"].set_text(math_expr)
        E["math_note"].set_text(math_note)
        if s_for_dot is None:
            E["math_dot"].set_alpha(0.0)
        else:
            E["math_dot"].set_alpha(0.9)
            tx = 0.07 + s_for_dot * (0.93 - 0.07)
            E["math_dot"].set_data([tx], [0.10])

    boundaries = [B_HOOK[0], B_HOOK[1], B_SCENE[1], B_MORPH[0] + int(n_morph * 0.0),
                  B_MORPH[0] + next((i for i, sv in enumerate(morph_s) if sv >= MORPH_SPLIT_1), n_morph),
                  B_MORPH[0] + next((i for i, sv in enumerate(morph_s) if sv >= MORPH_SPLIT_2), n_morph),
                  B_MORPH[1], B_PAYOFF[1]]
    nearest_gap = min(abs(frame - b) for b in boundaries)
    fade_in = 1.0 if nearest_gap > FADE else (0.15 + 0.85 * (nearest_gap / FADE))

    set_card_alpha(STORY_GROUP[0], STORY_GROUP[1], fade_in)
    set_card_alpha(MATH_GROUP[0], MATH_GROUP[1], fade_in)
    set_card_alpha(TITLE_GROUP[0], TITLE_GROUP[1], 1.0)
    set_card_alpha(LEGEND_GROUP[0], LEGEND_GROUP[1], 1.0)

    pos_callout_alpha = ease_window(frame, B_HOOK[0] + FADE, B_SCENE[1], FADE)
    E["pos_callout"].set_alpha(0.55 * pos_callout_alpha)

    sep_alpha = ease_window(frame, B_MORPH[0] + int(n_morph * 0.30), B_MORPH[1], FADE) if beat == "morph" else 0.0
    if beat == "morph" and s < 0.32:
        sep_alpha = 0.0
    E["sep_arrow"].set_alpha(0.85 * sep_alpha)
    E["sep_label"].set_alpha(0.95 * sep_alpha)

    fringe_alpha = ease_window(frame, B_PAYOFF[0] + FADE, B_PAYOFF[1], FADE) if beat == "payoff" else 0.0
    E["fringe_ellipse"].set_alpha(0.85 * fringe_alpha)
    E["leader"].set_alpha(0.80 * fringe_alpha)
    E["legend_left"].set_color(ink if fringe_alpha > 0.5 else ink_soft)

    progress = min(1.0, frame / max(TOTAL_FRAMES - F_CLOSING - 1, 1))
    E["bar_fg"].set_width(max(0.001, progress))

    return (plot_image,)


# ======================================================================================
# RENDER
# ======================================================================================

if __name__ == "__main__":
    anim = FuncAnimation(
        E["fig"], update, frames=TOTAL_FRAMES, interval=1000 / cfg.fps,
        blit=False, repeat=False, cache_frame_data=False,
    )

    writer = FFMpegWriter(
        fps=cfg.fps, codec="libx264", bitrate=3000,
        extra_args=["-preset", "veryfast", "-pix_fmt", "yuv420p", "-movflags", "+faststart"],
    )

    output_path = Path("/home/claude/preview") / cfg.output_name
    anim.save(str(output_path), writer=writer, dpi=cfg.dpi)
    plt.close(E["fig"])

    size_mb = output_path.stat().st_size / (1024 * 1024)
    print(f"Saved: {output_path.name}  ({size_mb:.2f} MB)")

    target_bytes = cfg.target_mb * 1024 * 1024
    if output_path.stat().st_size > target_bytes and shutil.which("ffmpeg"):
        squeezed = output_path.with_name(output_path.stem + "_squeezed.mp4")
        target_kbps = max(1800, int((cfg.target_mb * 8192 / TOTAL_SECONDS) * 0.80))
        cmd = [
            "ffmpeg", "-y", "-i", str(output_path),
            "-c:v", "libx264", "-preset", "veryfast",
            "-b:v", f"{target_kbps}k", "-maxrate", f"{target_kbps}k", "-bufsize", f"{target_kbps * 2}k",
            "-pix_fmt", "yuv420p", "-movflags", "+faststart", str(squeezed),
        ]
        subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        if squeezed.exists() and squeezed.stat().st_size < output_path.stat().st_size:
            squeezed.replace(output_path)
        size_mb = output_path.stat().st_size / (1024 * 1024)
        print(f"Compressed to: {size_mb:.2f} MB")
```

```python
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Concept: Wigner-function morph from a coherent blob into a Schrödinger cat state
Repo: github.com/zombimann/Mathematical-video-animations-and-visualization
"""

from __future__ import annotations

import shutil
import subprocess
from dataclasses import dataclass
from pathlib import Path

import matplotlib
matplotlib.use("Agg")

import matplotlib.pyplot as plt
from matplotlib import patheffects as pe
from matplotlib.animation import FFMpegWriter, FuncAnimation
from matplotlib.colors import LinearSegmentedColormap, TwoSlopeNorm
from matplotlib.font_manager import FontProperties
from matplotlib.patches import ConnectionPatch, Ellipse, FancyArrowPatch, FancyBboxPatch

import numpy as np
from scipy.ndimage import gaussian_filter
from qutip import coherent, wigner


# ======================================================================================
# CONFIG  -- every dial for the film lives here
# ======================================================================================

@dataclass(frozen=True)
class Config:
    # --- quantum parameters ---
    dimension: int = 50
    alpha: float = 2.15
    grid_extent: float = 4.6
    grid_points: int = 121

    # --- canvas (YouTube Shorts) ---
    width_in: float = 9.0
    height_in: float = 16.0
    dpi: int = 120
    fps: int = 24

    # --- pacing, in seconds, of each narrative beat ---
    t_hook: float = 2.6
    t_scene: float = 3.6
    t_morph: float = 15.6
    t_payoff: float = 4.2
    t_closing: float = 1.8
    fade_s: float = 0.45  # crossfade duration between beats, seconds

    # --- output / compression ---
    output_name: str = "wigner_cat_transition.mp4"
    target_mb: float = 9.4

    # --- copy ---
    title: str = "From One Blob to a Quantum Cat"
    watermark: str = "\u00a9 Mugambi Ndwiga / @craftsandengineering"


cfg = Config()


# ======================================================================================
# FONTS -- a cursive system font, selected by family name only (no font files are
# downloaded or bundled by this script). "Lobster Two" ships with the standard
# Colab / Jupyter font set; if it isn't present we fall back to a legible serif
# rather than fail.
# ======================================================================================

from matplotlib import font_manager as _fm

_CURSIVE_FAMILY = "Lobster Two"
_available = {f.name for f in _fm.fontManager.ttflist}
if _CURSIVE_FAMILY not in _available:
    print(f"[wigner_cat] '{_CURSIVE_FAMILY}' not found on this system; "
          f"falling back to 'Gentium Book Basic'.")
    _CURSIVE_FAMILY = "Gentium Book Basic" if "Gentium Book Basic" in _available else "DejaVu Serif"

hand_reg = FontProperties(family=_CURSIVE_FAMILY, weight="normal")
hand_bold = FontProperties(family=_CURSIVE_FAMILY, weight="bold")
hand_italic = FontProperties(family=_CURSIVE_FAMILY, style="italic")
hand_title = FontProperties(family=_CURSIVE_FAMILY, weight="bold")

matplotlib.rcParams["font.family"] = _CURSIVE_FAMILY
# a gentle hand-drawn wobble on every plotted line/patch edge -- subtle, not cartoonish
matplotlib.rcParams["path.sketch"] = (0.9, 110, 1.6)

ffmpeg_path = shutil.which("ffmpeg")
if ffmpeg_path:
    matplotlib.rcParams["animation.ffmpeg_path"] = ffmpeg_path

# --- palette: bled ink on aged parchment ---
ink = "#2c1810"
ink_soft = "#4a2f20"
paper = "#e9dcc0"
card_face = (1.0, 0.97, 0.90, 0.50)
card_edge = (0.26, 0.16, 0.10, 0.30)
text_fx = [pe.withStroke(linewidth=2.0, foreground=(1, 1, 1, 0.20))]

ink_paper_cmap = LinearSegmentedColormap.from_list(
    "ink_paper",
    ["#1c0f1a", "#5a3152", "#e9dcc0", "#b3753f", "#3c2114"],
    N=256,
)


# ======================================================================================
# HELPERS
# ======================================================================================

def smoothstep(x: float) -> float:
    x = float(np.clip(x, 0.0, 1.0))
    return x * x * (3.0 - 2.0 * x)


def ease_window(frame: int, start: int, end: int, fade_frames: int) -> float:
    """0->1->0 opacity envelope for a [start, end) frame window with cosine fades."""
    if frame < start - fade_frames or frame >= end + fade_frames:
        return 0.0
    if frame < start:
        t = (frame - (start - fade_frames)) / max(fade_frames, 1)
        return 0.5 - 0.5 * np.cos(np.pi * np.clip(t, 0, 1))
    if frame >= end:
        t = (frame - end) / max(fade_frames, 1)
        return 1.0 - (0.5 - 0.5 * np.cos(np.pi * np.clip(t, 0, 1)))
    return 1.0


def make_paper_texture(height: int, width: int, seed: int = 7) -> np.ndarray:
    rng = np.random.default_rng(seed)
    base = np.empty((height, width, 3), dtype=np.float32)
    base[..., 0], base[..., 1], base[..., 2] = 0.93, 0.87, 0.74

    fine = gaussian_filter(rng.normal(size=(height, width)), sigma=10)
    fine = (fine - fine.min()) / (fine.max() - fine.min() + 1e-9)
    blotch = gaussian_filter(rng.normal(size=(height, width)), sigma=42)
    blotch = (blotch - blotch.min()) / (blotch.max() - blotch.min() + 1e-9)

    yy, xx = np.mgrid[0:height, 0:width]
    vignette = np.sqrt(((xx - width / 2) / (0.60 * width)) ** 2 + ((yy - height / 2) / (0.74 * height)) ** 2)
    vignette = np.clip(vignette, 0.0, 1.0)

    stains = np.zeros((height, width), dtype=np.float32)
    for _ in range(22):
        cx, cy = rng.uniform(0, width), rng.uniform(0, height)
        sx, sy = rng.uniform(width * 0.03, width * 0.12), rng.uniform(height * 0.02, height * 0.07)
        amp = rng.uniform(0.02, 0.08)
        stains += amp * np.exp(-0.5 * (((xx - cx) / sx) ** 2 + ((yy - cy) / sy) ** 2))

    tex = base - 0.045 * fine[..., None] - 0.030 * blotch[..., None]
    tex -= 0.095 * vignette[..., None]
    tex -= stains[..., None] * np.array([0.22, 0.13, 0.08], dtype=np.float32)
    return np.clip(tex, 0.0, 1.0)


def card(ax, face=card_face, edge=card_edge, lw=1.3, pad=0.035, rounding=0.05):
    patch = FancyBboxPatch(
        (0, 0), 1, 1, transform=ax.transAxes,
        boxstyle=f"round,pad={pad},rounding_size={rounding}",
        facecolor=face, edgecolor=edge, linewidth=lw, zorder=0, clip_on=False,
    )
    ax.add_patch(patch)
    ax.set_axis_off()
    # fix the data frame to (0,1)x(0,1) so any later ax.plot() call cannot
    # silently autoscale this axes and shift/hide text anchored in that space
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_autoscale_on(False)
    return patch


def set_card_alpha(patch, text_objs, opacity: float, base_face=card_face, base_edge=card_edge):
    patch.set_facecolor((base_face[0], base_face[1], base_face[2], base_face[3] * opacity))
    patch.set_edgecolor((base_edge[0], base_edge[1], base_edge[2], base_edge[3] * opacity))
    for t, base_alpha in text_objs:
        t.set_alpha(base_alpha * opacity)


# ======================================================================================
# FIGURE CONSTRUCTION -- explicit rects (inches -> fractions), guarantees no overlap
# ======================================================================================

def build_figure():
    W, H = cfg.width_in, cfg.height_in
    fig = plt.figure(figsize=(W, H), dpi=cfg.dpi)
    fig.patch.set_facecolor(paper)

    bg_ax = fig.add_axes([0, 0, 1, 1], zorder=0)
    bg_ax.imshow(make_paper_texture(int(cfg.dpi * H), int(cfg.dpi * W), seed=9), aspect="auto")
    bg_ax.set_axis_off()

    def rect(x0, y1, w, h):
        """x0, y1 given as fractions; y1 is the TOP edge, h grows downward."""
        return [x0, y1 - h, w, h]

    margin = 0.045
    content_w = 1.0 - 2 * margin
    y = 0.985

    # ---- title bar ----
    h_title = 0.052
    y -= h_title
    title_ax = fig.add_axes(rect(margin, y + h_title, content_w, h_title), zorder=20)
    title_patch = card(title_ax)
    title_txt = title_ax.text(0.5, 0.5, cfg.title, ha="center", va="center",
                               fontsize=34, color=ink, fontproperties=hand_title,
                               rotation=-1.1, path_effects=text_fx)
    y -= 0.008

    # ---- square phase-space plot, centered ----
    plot_w = content_w * 0.94
    plot_h = plot_w * (W / H)
    y -= plot_h
    plot_ax = fig.add_axes(rect(margin + (content_w - plot_w) / 2, y + plot_h, plot_w, plot_h), zorder=5)
    plot_ax.set_facecolor((1, 1, 1, 0.0))
    plot_frame = FancyBboxPatch(
        (0, 0), 1, 1, transform=plot_ax.transAxes,
        boxstyle="round,pad=0.012,rounding_size=0.025",
        facecolor=(1, 1, 1, 0.14), edgecolor=(0.26, 0.16, 0.10, 0.35), linewidth=1.6, zorder=0,
    )
    plot_ax.add_patch(plot_frame)
    y -= 0.038

    # ---- story card ----
    h_story = 0.135
    y -= h_story
    story_ax = fig.add_axes(rect(margin, y + h_story, content_w, h_story), zorder=20)
    story_patch = card(story_ax)
    story_kicker = story_ax.text(0.07, 0.83, "", ha="left", va="center", fontsize=18,
                                  color=ink_soft, fontproperties=hand_italic, alpha=0.85,
                                  rotation=-1.4, path_effects=text_fx)
    story_title = story_ax.text(0.07, 0.57, "", ha="left", va="center", fontsize=30,
                                 color=ink, fontproperties=hand_bold, rotation=0.8, path_effects=text_fx)
    story_body = story_ax.text(0.07, 0.22, "", ha="left", va="center", fontsize=21,
                                color=ink, fontproperties=hand_reg, wrap=True, path_effects=text_fx)
    y -= 0.010

    # ---- math card ----
    h_math = 0.118
    y -= h_math
    math_ax = fig.add_axes(rect(margin, y + h_math, content_w, h_math), zorder=20)
    math_patch = card(math_ax)
    math_kicker = math_ax.text(0.07, 0.82, "THE MATH", ha="left", va="center", fontsize=16,
                                color=ink_soft, fontproperties=hand_italic, alpha=0.80,
                                rotation=1.2, path_effects=text_fx)
    math_expr = math_ax.text(0.07, 0.50, "", ha="left", va="center", fontsize=25,
                              color=ink, fontproperties=hand_reg, path_effects=text_fx)
    math_note = math_ax.text(0.07, 0.19, "", ha="left", va="center", fontsize=17,
                              color=ink_soft, fontproperties=hand_reg, alpha=0.9, path_effects=text_fx)
    # live s-progress track: a thin rail with a sliding dot, directly under the equation
    track_x0, track_x1, track_y = 0.07, 0.93, 0.10
    math_ax.plot([track_x0, track_x1], [track_y, track_y], color=(0.26, 0.16, 0.10, 0.30),
                 linewidth=3.0, solid_capstyle="round", zorder=1)
    math_dot, = math_ax.plot([track_x0], [track_y], marker="o", markersize=11,
                              markerfacecolor=ink, markeredgecolor=paper, markeredgewidth=1.2,
                              alpha=0.0, zorder=2)
    y -= 0.008

    # ---- legend strip ----
    h_legend = 0.056
    y -= h_legend
    legend_ax = fig.add_axes(rect(margin, y + h_legend, content_w, h_legend), zorder=20)
    legend_patch = card(legend_ax, pad=0.05, rounding=0.10)
    grad_ax = fig.add_axes(rect(margin + content_w * 0.32, y + h_legend - 0.038, content_w * 0.36, 0.020), zorder=21)
    grad_ax.imshow(np.linspace(-1, 1, 400)[None, :], aspect="auto", cmap=ink_paper_cmap,
                    norm=TwoSlopeNorm(0.0, -1.0, 1.0))
    grad_ax.set_axis_off()
    legend_left = legend_ax.text(0.07, 0.5, "− negative", ha="left", va="center", fontsize=16,
                                  color=ink, fontproperties=hand_bold, path_effects=text_fx)
    legend_right = legend_ax.text(0.93, 0.5, "positive +", ha="right", va="center", fontsize=16,
                                   color=ink, fontproperties=hand_bold, path_effects=text_fx)
    y -= 0.010

    # ---- footer / disclaimer + progress bar ----
    h_footer = 0.048
    y -= h_footer
    footer_ax = fig.add_axes(rect(margin, y + h_footer, content_w, h_footer), zorder=20)
    footer_ax.set_axis_off()
    footer_txt = footer_ax.text(0.0, 0.78, "Narrative visualization, not a literal time evolution.",
                                 ha="left", va="center", fontsize=14, color=ink_soft,
                                 fontproperties=hand_reg, alpha=0.85, path_effects=text_fx)
    bar_bg = FancyBboxPatch((0.0, 0.10), 1.0, 0.16, transform=footer_ax.transAxes,
                             boxstyle="round,pad=0,rounding_size=0.08",
                             facecolor=(0.26, 0.16, 0.10, 0.18), edgecolor="none", zorder=1)
    footer_ax.add_patch(bar_bg)
    bar_fg = plt.Rectangle((0.0, 0.10), 0.001, 0.16, transform=footer_ax.transAxes,
                            facecolor=(0.45, 0.20, 0.16, 0.55), edgecolor="none", zorder=2)
    footer_ax.add_patch(bar_fg)

    # ---- closing title card (full-bleed, shown only for the final beat) ----
    closing_ax = fig.add_axes([0, 0, 1, 1], zorder=40)
    closing_ax.set_axis_off()
    closing_ax.set_xlim(0, 1)
    closing_ax.set_ylim(0, 1)
    closing_ax.set_visible(False)
    closing_ax.add_patch(plt.Rectangle((0, 0), 1, 1, transform=closing_ax.transAxes,
                                        facecolor=paper, edgecolor="none", zorder=0))
    closing_ax.imshow(make_paper_texture(int(cfg.dpi * H), int(cfg.dpi * W), seed=21),
                       extent=[0, 1, 0, 1], aspect="auto", zorder=1)
    closing_name = closing_ax.text(0.5, 0.54, "Made by Mugambi Ndwiga", ha="center", va="center",
                                    fontsize=34, color=ink, fontproperties=hand_title,
                                    rotation=-1.0, path_effects=text_fx, zorder=2)
    closing_handle = closing_ax.text(0.5, 0.45, "@craftsandengineering", ha="center", va="center",
                                      fontsize=24, color=ink_soft, fontproperties=hand_bold,
                                      rotation=0.8, path_effects=text_fx, zorder=2)

    # ---- cinematic fade overlay: solid curtain for the opening/closing fade ----
    fade_ax = fig.add_axes([0, 0, 1, 1], zorder=50)
    fade_ax.set_axis_off()
    fade_curtain = plt.Rectangle((0, 0), 1, 1, transform=fade_ax.transAxes,
                                  facecolor="#15100a", edgecolor="none", alpha=1.0, zorder=0)
    fade_ax.add_patch(fade_curtain)

    # ---- watermark (bottom-right corner, modest, never collides with footer) ----
    wm_ax = fig.add_axes([0, 0, 1, 1], zorder=30)
    wm_ax.set_axis_off()
    wm_ax.set_xlim(0, 1)
    wm_ax.set_ylim(0, 1)
    watermark_txt = wm_ax.text(0.965, 0.012, cfg.watermark, ha="right", va="bottom",
                                fontsize=13, color=ink, alpha=0.60, fontproperties=hand_reg,
                                path_effects=text_fx, zorder=31)

    elems = dict(
        fig=fig, plot_ax=plot_ax, plot_frame=plot_frame,
        title_ax=title_ax, title_patch=title_patch, title_txt=title_txt,
        story_ax=story_ax, story_patch=story_patch,
        story_kicker=story_kicker, story_title=story_title, story_body=story_body,
        math_ax=math_ax, math_patch=math_patch,
        math_kicker=math_kicker, math_expr=math_expr, math_note=math_note, math_dot=math_dot,
        legend_ax=legend_ax, legend_patch=legend_patch, legend_left=legend_left, legend_right=legend_right,
        footer_ax=footer_ax, footer_txt=footer_txt, bar_bg=bar_bg, bar_fg=bar_fg,
        watermark_txt=watermark_txt,
        closing_ax=closing_ax, closing_name=closing_name, closing_handle=closing_handle,
        fade_curtain=fade_curtain,
    )
    return elems


def style_plot_axes(plot_ax):
    g = cfg.grid_extent
    plot_ax.set_xlim(-g, g)
    plot_ax.set_ylim(-g, g)
    plot_ax.set_aspect("equal", adjustable="box")
    ticks = [-4, -2, 0, 2, 4]
    plot_ax.set_xticks(ticks)
    plot_ax.set_yticks(ticks)
    plot_ax.tick_params(length=4, width=1.0, pad=5, labelsize=17)
    for lbl in plot_ax.get_xticklabels() + plot_ax.get_yticklabels():
        lbl.set_fontproperties(hand_reg)
    plot_ax.grid(True, alpha=0.10, linewidth=0.9)
    plot_ax.axhline(0, color=ink, alpha=0.16, linewidth=1.1)
    plot_ax.axvline(0, color=ink, alpha=0.16, linewidth=1.1)
    for spine in plot_ax.spines.values():
        spine.set_color((0.26, 0.16, 0.10, 0.4))
        spine.set_linewidth(1.1)
    plot_ax.set_xlabel("x   (position)", fontsize=21, fontproperties=hand_bold, color=ink, labelpad=6)
    plot_ax.set_ylabel("p   (momentum)", fontsize=21, fontproperties=hand_bold, color=ink, labelpad=4)


def build_annotations(e):
    """Create (once) every connector / highlight artist, all starting invisible.
    Each is fully visible-state-controlled later via .set_alpha in update()."""
    plot_ax = e["plot_ax"]
    g = cfg.grid_extent
    a = cfg.alpha

    # (1) measurement arrow tying the math card's "alpha" to the physical gap between lobes
    sep_arrow = FancyArrowPatch(
        (-a, -g * 0.74), (a, -g * 0.74), transform=plot_ax.transData,
        arrowstyle="<|-|>", mutation_scale=14, linewidth=1.6,
        color=ink, alpha=0.0, zorder=8,
    )
    plot_ax.add_patch(sep_arrow)
    sep_label = plot_ax.text(0, -g * 0.74, "2 alpha", ha="center", va="bottom", fontsize=15,
                              color=ink, fontproperties=hand_bold, alpha=0.0, zorder=9,
                              path_effects=text_fx)

    # (2) legend -> blob callout, used early (only "positive" exists while s is small)
    pos_callout = ConnectionPatch(
        xyA=(0.82, 0.95), coordsA="axes fraction", axesA=e["legend_ax"],
        xyB=(g * 0.62, g * 0.08), coordsB="data", axesB=plot_ax,
        connectionstyle="arc3,rad=-0.18", arrowstyle="-|>", mutation_scale=12,
        linewidth=1.3, color=ink_soft, alpha=0.0, zorder=25,
    )
    e["fig"].add_artist(pos_callout)

    # (3) climax fringe highlight: dashed ellipse + leader line from story title to it
    fringe_ellipse = Ellipse((0, 0), width=a * 1.55, height=g * 0.62, transform=plot_ax.transData,
                              facecolor="none", edgecolor=ink, linewidth=2.0, linestyle=(0, (5, 3)),
                              alpha=0.0, zorder=10)
    plot_ax.add_patch(fringe_ellipse)

    leader = ConnectionPatch(
        xyA=(0.30, 0.99), coordsA="axes fraction", axesA=e["story_ax"],
        xyB=(-a * 0.05, -g * 0.30), coordsB="data", axesB=plot_ax,
        connectionstyle="arc3,rad=0.12", arrowstyle="-|>", mutation_scale=14,
        linewidth=1.5, color=ink_soft, alpha=0.0, zorder=25,
    )
    e["fig"].add_artist(leader)

    e["sep_arrow"], e["sep_label"] = sep_arrow, sep_label
    e["pos_callout"] = pos_callout
    e["fringe_ellipse"] = fringe_ellipse
    e["leader"] = leader
    return e


# ======================================================================================
# NARRATIVE TIMELINE -- the "screenplay": every beat's frame range + copy
# ======================================================================================

def seconds_to_frames(seconds: float) -> int:
    return max(1, int(round(seconds * cfg.fps)))


F_HOOK = seconds_to_frames(cfg.t_hook)
F_SCENE = seconds_to_frames(cfg.t_scene)
F_MORPH = seconds_to_frames(cfg.t_morph)
F_PAYOFF = seconds_to_frames(cfg.t_payoff)
F_CLOSING = seconds_to_frames(cfg.t_closing)
FADE = seconds_to_frames(cfg.fade_s)

B_HOOK = (0, F_HOOK)
B_SCENE = (B_HOOK[1], B_HOOK[1] + F_SCENE)
B_MORPH = (B_SCENE[1], B_SCENE[1] + F_MORPH)
B_PAYOFF = (B_MORPH[1], B_MORPH[1] + F_PAYOFF)
B_CLOSING = (B_PAYOFF[1], B_PAYOFF[1] + F_CLOSING)
TOTAL_FRAMES = B_CLOSING[1]
TOTAL_SECONDS = TOTAL_FRAMES / cfg.fps

MORPH_SPLIT_1, MORPH_SPLIT_2 = 0.08, 0.62  # smoothstep(s) thresholds for sub-beat copy


def morph_substage(s: float) -> tuple[str, str, str]:
    if s < MORPH_SPLIT_1:
        return ("THE SHIFT", "Here we go", "Slowly, the blob is nudged toward existing in two places at once.")
    if s < MORPH_SPLIT_2:
        return ("THE STAGE", "Splitting", "A second lobe grows from the first, and the two halves begin to interfere.")
    return ("THE REVEAL", "A Schr\u00f6dinger cat state", "Two lobes remain \u2014 and the space between them turns to stripes.")


# ======================================================================================
# PRECOMPUTE WIGNER FRAMES (only the unique ones we actually need)
# ======================================================================================

print("Precomputing Wigner functions ...")
g = cfg.grid_extent
xvec = np.linspace(-g, g, cfg.grid_points)

blob_state = coherent(cfg.dimension, cfg.alpha)
W_blob = np.asarray(wigner(blob_state, xvec, xvec), dtype=np.float32)

n_morph = B_MORPH[1] - B_MORPH[0]
morph_W: list[np.ndarray] = []
morph_s: list[float] = []
for i in range(n_morph):
    progress = i / max(n_morph - 1, 1)
    s = smoothstep(progress)
    ket_pos = coherent(cfg.dimension, cfg.alpha)
    ket_neg = coherent(cfg.dimension, -cfg.alpha)
    # s=0 -> pure |alpha> (one blob); s=1 -> equal-weight |alpha>+|-alpha> (symmetric cat).
    # NOTE: this is NOT linear interpolation of coefficients -- (1-s, s) weighting would
    # drift the state back to a single (shifted) blob at s=1 and lose the two-lobe payoff.
    psi = (ket_pos + s * ket_neg).unit()
    W = np.asarray(wigner(psi, xvec, xvec), dtype=np.float32)
    morph_W.append(W)
    morph_s.append(s)

W_cat = morph_W[-1]
wmax = float(max(np.max(np.abs(W_blob)), np.max(np.abs(W_cat))))
plot_norm = TwoSlopeNorm(vcenter=0.0, vmin=-wmax, vmax=wmax)
print(f"Done. {n_morph} unique morph frames computed. Total runtime: {TOTAL_SECONDS:.2f}s "
      f"({TOTAL_FRAMES} frames @ {cfg.fps} fps).")


# ======================================================================================
# BUILD THE FIGURE + STATEFUL ARTISTS
# ======================================================================================

E = build_figure()
style_plot_axes(E["plot_ax"])
build_annotations(E)

plot_image = E["plot_ax"].imshow(
    W_blob, extent=[-g, g, -g, g], origin="lower", cmap=ink_paper_cmap,
    norm=plot_norm, interpolation="bicubic", alpha=0.90, zorder=2,
)

# group bookkeeping for crossfades: (patch, [(text_obj, base_alpha), ...])
TITLE_GROUP = (E["title_patch"], [(E["title_txt"], 1.0)])
STORY_GROUP = (E["story_patch"], [(E["story_kicker"], 0.85), (E["story_title"], 1.0), (E["story_body"], 1.0)])
MATH_GROUP = (E["math_patch"], [(E["math_kicker"], 0.80), (E["math_expr"], 1.0), (E["math_note"], 0.9)])
LEGEND_GROUP = (E["legend_patch"], [(E["legend_left"], 1.0), (E["legend_right"], 1.0)])


def update(frame: int):
    # ---------------- figure out which beat we're in, and s ----------------
    if frame < B_HOOK[1]:
        beat = "hook"
        s = 0.0
    elif frame < B_SCENE[1]:
        beat = "scene"
        s = 0.0
    elif frame < B_MORPH[1]:
        beat = "morph"
        idx = frame - B_MORPH[0]
        s = morph_s[idx]
    elif frame < B_PAYOFF[1]:
        beat = "payoff"
        s = 1.0
    else:
        beat = "closing"
        s = 1.0

    closing_alpha = ease_window(frame, B_CLOSING[0], B_CLOSING[1], FADE)
    if closing_alpha > 0.0:
        E["closing_ax"].set_visible(True)
        for ax_key in ("title_ax", "plot_ax", "story_ax", "math_ax", "legend_ax", "footer_ax"):
            E[ax_key].set_visible(closing_alpha < 0.5)
        E["closing_name"].set_alpha(min(1.0, closing_alpha * 1.4))
        E["closing_handle"].set_alpha(min(1.0, closing_alpha * 1.4))
        if closing_alpha >= 0.999:
            CLOSE_FADE = seconds_to_frames(0.45)
            curtain_alpha = 0.0
            if frame >= TOTAL_FRAMES - CLOSE_FADE:
                curtain_alpha = smoothstep((frame - (TOTAL_FRAMES - CLOSE_FADE)) / CLOSE_FADE)
            E["fade_curtain"].set_alpha(curtain_alpha)
            return (plot_image,)
    else:
        E["closing_ax"].set_visible(False)
        for ax_key in ("title_ax", "plot_ax", "story_ax", "math_ax", "legend_ax", "footer_ax"):
            E[ax_key].set_visible(True)

    # ---------------- the phase-space image + zoom ----------------
    if beat in ("hook", "scene"):
        plot_image.set_data(W_blob)
        lim = g
    elif beat == "morph":
        idx = frame - B_MORPH[0]
        plot_image.set_data(morph_W[idx])
        lim = g * (1.0 - 0.10 * smoothstep(s))
    else:  # payoff / closing hold the cat state, payoff zooms in on the fringes
        plot_image.set_data(W_cat)
        if beat == "payoff":
            zp = smoothstep((frame - B_PAYOFF[0]) / max(F_PAYOFF - 1, 1))
            lim = (g * 0.90) - zp * (g * 0.90 - g * 0.46)
        else:
            lim = g * 0.46
    E["plot_ax"].set_xlim(-lim, lim)
    E["plot_ax"].set_ylim(-lim, lim)

    # ---------------- story + math copy per beat ----------------
    if beat == "hook":
        kicker, title, body = "THE SETUP", "Meet the calmest quantum state", (
            "This is a coherent state \u2014 the quantum state that behaves\n"
            "most like a single classical point of light."
        )
        math_expr = "state:   |alpha>          alpha = 2.15"
        math_note = "alpha sets the blob's size \u2014 and soon, how far it splits."
        s_for_dot = None
    elif beat == "scene":
        kicker, title, body = "THE STAGE", "Reading a phase-space portrait", (
            "Brightness shows how likely the state is to be found at\n"
            "each position x and momentum p. Warm = likely."
        )
        math_expr = "state:   |alpha>          alpha = 2.15"
        math_note = "Watch what happens when this single blob splits in two."
        s_for_dot = None
    elif beat == "morph":
        kicker, title, body = morph_substage(s)
        math_expr = "|psi(s)> = N [ |alpha> + s|\u2212alpha> ]"
        math_note = f"s = {s:0.2f}     alpha = 2.15"
        s_for_dot = s
    elif beat == "payoff":
        kicker, title, body = "THE PUNCHLINE", "Negative probability: pure quantum", (
            "Those dark stripes between the lobes dip below zero \u2014\n"
            "no classical probability can ever do that."
        )
        math_expr = "W(x, p) < 0   only in quantum phase space"
        math_note = "Wigner negativity = the fingerprint of quantum interference."
        s_for_dot = 1.0
    else:
        kicker, title, body, math_expr, math_note, s_for_dot = "", "", "", "", "", None

    if beat != "closing":
        E["story_kicker"].set_text(kicker)
        E["story_title"].set_text(title)
        E["story_body"].set_text(body)
        E["math_expr"].set_text(math_expr)
        E["math_note"].set_text(math_note)
        if s_for_dot is None:
            E["math_dot"].set_alpha(0.0)
        else:
            E["math_dot"].set_alpha(0.9)
            tx = 0.07 + s_for_dot * (0.93 - 0.07)
            E["math_dot"].set_data([tx], [0.10])

    # ---------------- crossfade story + math card whenever copy changes ----------------
    # re-trigger a short fade at every beat boundary inside the main narrative
    boundaries = [B_HOOK[0], B_HOOK[1], B_SCENE[1], B_MORPH[0] + int(n_morph * 0.0),
                  B_MORPH[0] + next((i for i, sv in enumerate(morph_s) if sv >= MORPH_SPLIT_1), n_morph),
                  B_MORPH[0] + next((i for i, sv in enumerate(morph_s) if sv >= MORPH_SPLIT_2), n_morph),
                  B_MORPH[1], B_PAYOFF[1]]
    nearest_gap = min(abs(frame - b) for b in boundaries)
    fade_in = 1.0 if nearest_gap > FADE else (0.15 + 0.85 * (nearest_gap / FADE))

    set_card_alpha(STORY_GROUP[0], STORY_GROUP[1], fade_in)
    set_card_alpha(MATH_GROUP[0], MATH_GROUP[1], fade_in)
    set_card_alpha(TITLE_GROUP[0], TITLE_GROUP[1], 1.0)
    set_card_alpha(LEGEND_GROUP[0], LEGEND_GROUP[1], 1.0)

    # ---------------- connector annotations ----------------
    pos_callout_alpha = ease_window(frame, B_HOOK[0] + FADE, B_SCENE[1], FADE)
    E["pos_callout"].set_alpha(0.55 * pos_callout_alpha)

    sep_alpha = ease_window(frame, B_MORPH[0] + int(n_morph * 0.30), B_MORPH[1], FADE) if beat == "morph" else 0.0
    if beat == "morph" and s < 0.32:
        sep_alpha = 0.0
    E["sep_arrow"].set_alpha(0.85 * sep_alpha)
    E["sep_label"].set_alpha(0.95 * sep_alpha)

    fringe_alpha = ease_window(frame, B_PAYOFF[0] + FADE, B_PAYOFF[1], FADE) if beat == "payoff" else 0.0
    E["fringe_ellipse"].set_alpha(0.85 * fringe_alpha)
    E["leader"].set_alpha(0.80 * fringe_alpha)
    neg_emphasis = 0.4 + 0.6 * fringe_alpha
    E["legend_left"].set_color(ink if fringe_alpha > 0.5 else ink_soft)

    # ---------------- footer progress bar ----------------
    progress = min(1.0, frame / max(TOTAL_FRAMES - F_CLOSING - 1, 1))
    E["bar_fg"].set_width(max(0.001, progress))

    # ---------------- cinematic curtain: brief fade-in at open, fade-out at the very end ----------------
    OPEN_FADE = seconds_to_frames(0.45)
    CLOSE_FADE = seconds_to_frames(0.45)
    if frame < OPEN_FADE:
        curtain_alpha = 1.0 - smoothstep(frame / OPEN_FADE)
    elif frame >= TOTAL_FRAMES - CLOSE_FADE:
        curtain_alpha = smoothstep((frame - (TOTAL_FRAMES - CLOSE_FADE)) / CLOSE_FADE)
    else:
        curtain_alpha = 0.0
    E["fade_curtain"].set_alpha(curtain_alpha)

    return (plot_image,)


# ======================================================================================
# RENDER
# ======================================================================================

if __name__ == "__main__":
    import os
    _preview_frame = os.environ.get("PREVIEW_FRAME")
    if _preview_frame is not None:
        update(int(_preview_frame))
        out = Path.cwd() / "quick_preview.png"
        E["fig"].savefig(str(out), facecolor=paper)
        print(f"Saved single-frame preview: {out}")
        raise SystemExit(0)

    anim = FuncAnimation(
        E["fig"], update, frames=TOTAL_FRAMES, interval=1000 / cfg.fps,
        blit=False, repeat=False, cache_frame_data=False,
    )

    writer = FFMpegWriter(
        fps=cfg.fps, codec="libx264", bitrate=3000,
        extra_args=["-preset", "veryfast", "-pix_fmt", "yuv420p", "-movflags", "+faststart"],
    )

    output_path = Path.cwd() / cfg.output_name
    anim.save(str(output_path), writer=writer, dpi=cfg.dpi)
    plt.close(E["fig"])

    size_mb = output_path.stat().st_size / (1024 * 1024)
    print(f"Saved: {output_path.name}  ({size_mb:.2f} MB)")

    target_bytes = cfg.target_mb * 1024 * 1024
    if output_path.stat().st_size > target_bytes and shutil.which("ffmpeg"):
        squeezed = output_path.with_name(output_path.stem + "_squeezed.mp4")
        target_kbps = max(1800, int((cfg.target_mb * 8192 / TOTAL_SECONDS) * 0.80))
        cmd = [
            "ffmpeg", "-y", "-i", str(output_path),
            "-c:v", "libx264", "-preset", "veryfast",
            "-b:v", f"{target_kbps}k", "-maxrate", f"{target_kbps}k", "-bufsize", f"{target_kbps * 2}k",
            "-pix_fmt", "yuv420p", "-movflags", "+faststart", str(squeezed),
        ]
        subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        if squeezed.exists() and squeezed.stat().st_size < output_path.stat().st_size:
            squeezed.replace(output_path)
        size_mb = output_path.stat().st_size / (1024 * 1024)
        print(f"Compressed to: {size_mb:.2f} MB")

    # ---- priority: play the video inline, with a simple download link underneath ----
    try:
        from IPython.display import FileLink, Video, display
        display(Video(str(output_path), embed=True, width=360))
        display(FileLink(str(output_path)))
    except ImportError:
        print(f"Video ready at: {output_path}")
```
