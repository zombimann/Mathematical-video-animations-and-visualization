# Mathematical Functions Visualization

This notebook provides an educational visualization of 12 distinct mathematical functions. It features a 4x3 grid of subplots where static curves representing each function are traced by a dynamic point, illustrating their behavior across a shared domain.

### Functions Included:
* Trigonometric (Sine, Cosine, Sine-Cosine product)
* Exponential & Logarithmic
* Polynomial (Cubic, Quadratic)
* Special Functions (Sinc, Tanh, Arctan)

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib

# Configure plot style for better visibility on GitHub
plt.rcParams.update({'font.size': 10, 'figure.facecolor': 'white'})

# Define functions, their LaTeX expressions, and custom y-limits
functions = [
    (lambda x, t: np.sin(x + t), r'$\sin(x + t)$', (-1.5, 1.5)),
    (lambda x, t: np.exp(-0.1 * (x - t)**2), r'$e^{-0.1(x-t)^2}$', (-0.2, 1.2)),
    (lambda x, t: 0.1 * (x**3 - 3*t*x), r'$0.1(x^3 - 3tx)$', (-10, 10)),
    (lambda x, t: np.sin(x) * np.cos(t), r'$\sin(x)\cos(t)$', (-1.5, 1.5)),
    (lambda x, t: np.abs(x)**(np.sin(t) + 1.1), r'$|x|^{\sin(t)+1.1}$', (-1, 20)),
    (lambda x, t: np.exp(-0.1 * x) * np.cos(x - 5*t), r'$e^{-0.1x}\cos(x-5t)$', (-1, 3)),
    (lambda x, t: np.sinc(x - t), r'$\text{sinc}(x - t)$', (-0.5, 1.2)),
    (lambda x, t: np.tanh(x * (np.sin(t) + 0.5)), r'$\tanh(x(\sin(t)+0.5))$', (-1.2, 1.2)),
    (lambda x, t: np.cos(x * (t + 0.5)), r'$\cos(x(t+0.5))$', (-1.5, 1.5)),
    (lambda x, t: np.log(np.abs(x) + 1) * (np.sin(t) + 0.5), r'$\ln(|x|+1)(\sin(t)+0.5)$', (-3, 6)),
    (lambda x, t: np.arctan(x + np.tan(t)), r'$\arctan(x + \tan(t))$', (-2, 2)),
    (lambda x, t: (x**2 / 10) * np.cos(t), r'$0.1x^2\cos(t)$', (-2, 11))
]

x_vals = np.linspace(-10, 10, 400)
frames = 100
t_values = np.linspace(0, 2*np.pi, frames)

fig, axes = plt.subplots(4, 3, figsize=(15, 14), constrained_layout=True)
fig.suptitle("Mathematical Functions: Moving Point Visualization", fontsize=20, fontweight='bold')

# Using a non-zero t for background skeleton
background_t = 1.0
dots = []

for i, (func, expr, ylim) in enumerate(functions):
    ax = axes.flat[i]
    ax.plot(x_vals, func(x_vals, background_t), color='steelblue', alpha=0.3, linestyle='--')
    ax.plot(x_vals, func(x_vals, background_t), color='steelblue', lw=2)
    dot, = ax.plot([], [], 'ro', markersize=8)
    ax.set_title(expr, fontsize=14)
    ax.set_ylim(ylim)
    ax.grid(True, linestyle='--', alpha=0.5)
    dots.append(dot)

def update(frame):
    current_x = -10 + (20 * (frame / frames))
    for i, (func, expr, ylim) in enumerate(functions):
        y = func(current_x, background_t)
        dots[i].set_data([current_x], [y])
    return dots

ani = FuncAnimation(fig, update, frames=frames, interval=50, blit=True)
plt.close(fig)
HTML(ani.to_html5_video())
```
