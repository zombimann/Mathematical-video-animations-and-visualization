# Sphere Construction via Triple Integrals

### Project Overview
This project provides a visual, 3D exploration of how a sphere's volume is constructed using calculus—specifically **triple integrals in spherical coordinates**. By breaking down the integration process into sequential geometric transformations, the animation makes abstract mathematical concepts accessible and engaging for the general public.

### Mathematical Premise
The volume $V$ of a sphere with radius $R$ is given by the triple integral:

$$V = \color{blue}{\int_{0}^{\pi}} \color{green}{\int_{0}^{2\pi}} \color{red}{\int_{0}^{R}} r^2 \sin(\phi) \color{red}{dr} \color{green}{d\theta} \color{blue}{d\phi}$$

**Visual Logic:**
1.  **Radial Growth ($\color{red}{dr}$):** Construction starts with a point extending into a line of length $R$.
2.  **Angular Sweep ($\color{green}{d\theta}$):** The line rotates $360^\circ$ ($2\pi$ radians) in the $xy$-plane to form a circular area.
3.  **Volume Revolution ($\color{blue}{d\phi}$):** That circular area revolves from the North Pole to the South Pole ($\pi$ radians) to sweep out the full 3D volume.

```python
from manim import *
import numpy as np

"""
Author: Mugambi Ndwiga
Socials: @craftsandengineering (Instagram)
Description: A mesmerizing animation showing the construction of a sphere
using triple integrals in spherical coordinates with explicit color mapping.
"""

class SphereConstruction(ThreeDScene):
    def construct(self):
        # --- Configuration ---
        self.set_camera_orientation(phi=75 * DEGREES, theta=-45 * DEGREES)
        radius = 2.5
        color_r = RED        # r, dr
        color_theta = GREEN  # theta, dtheta
        color_phi = BLUE     # phi, dphi

        # --- Watermark & Credits ---
        credits = Text("Author: Mugambi Ndwiga\nIG: @craftsandengineering",
                      font_size=16, color=GRAY).to_corner(UR)
        self.add_fixed_in_frame_mobjects(credits)

        # --- Formula Setup ---
        # Mapping integral signs and differentials to colors using raw strings
        formula = MathTex(
            r"V = ",
            r"\int_{0}^{\pi}", r"\int_{0}^{2\pi}", r"\int_{0}^{R}",
            r"r^2 \sin(\phi) ", r"dr", r"d\theta", r"d\phi",
            font_size=36
        ).to_corner(DL)

        # Color-code formula: Integral signs match their differentials and the scene elements
        formula[3].set_color(color_r)      # Radial Integral (Inner)
        formula[5].set_color(color_r)      # dr

        formula[2].set_color(color_theta)  # Theta Integral (Middle)
        formula[6].set_color(color_theta)  # dtheta

        formula[1].set_color(color_phi)    # Phi Integral (Outer)
        formula[7].set_color(color_phi)    # dphi

        self.add_fixed_in_frame_mobjects(formula)

        # --- 1. The Radial Line (dr) ---
        line = Line(ORIGIN, [radius, 0, 0], color=color_r, stroke_width=6)
        label_r = Tex(r"Radial Growth ($dr$)", color=color_r).scale(0.8)
        self.add_fixed_in_frame_mobjects(label_r.to_edge(UP))

        self.play(
            Create(line),
            formula[3].animate.scale(1.3),
            formula[5].animate.scale(1.3)
        )
        self.wait(1)
        self.play(
            formula[3].animate.scale(1/1.3),
            formula[5].animate.scale(1/1.3),
            FadeOut(label_r)
        )

        # --- 2. The Area Revolution (d theta) ---
        arc = Arc(radius=radius, angle=TAU, color=color_theta, stroke_width=2)
        label_theta = Tex(r"Angular Sweep ($d\theta$)", color=color_theta).scale(0.8)
        self.add_fixed_in_frame_mobjects(label_theta.to_edge(UP))

        self.play(
            formula[2].animate.scale(1.3),
            formula[6].animate.scale(1.3)
        )
        self.play(
            Rotate(line, angle=TAU, about_point=ORIGIN, run_time=3, rate_func=linear),
            Create(arc),
            run_time=3
        )

        circle = Circle(radius=radius, color=color_theta, fill_opacity=0.2).rotate(90*DEGREES, axis=RIGHT)
        self.add(circle)
        self.wait(1)
        self.play(
            formula[2].animate.scale(1/1.3),
            formula[6].animate.scale(1/1.3),
            FadeOut(label_theta)
        )

        # --- 3. The Volume Revolution (d phi) ---
        label_phi = Tex(r"Volume Revolution ($d\phi$)", color=color_phi).scale(0.8)
        self.add_fixed_in_frame_mobjects(label_phi.to_edge(UP))

        self.play(
            formula[1].animate.scale(1.3),
            formula[7].animate.scale(1.3)
        )

        sphere = Sphere(
            center=ORIGIN,
            radius=radius,
            resolution=(20, 20),
            fill_opacity=0.6,
            checkerboard_colors=[BLUE_D, BLUE_E]
        )

        self.move_camera(phi=75 * DEGREES, theta=30 * DEGREES, run_time=2)
        self.play(
            Create(sphere),
            run_time=4,
            rate_func=slow_into
        )

        # --- Final Polish ---
        self.play(FadeOut(label_phi))
        self.play(formula.animate.set_color(YELLOW).scale(1.1))

        self.begin_ambient_camera_rotation(rate=0.2)
        self.wait(5)
        self.stop_ambient_camera_rotation()

# Run the scene
%manim -v WARNING -qm SphereConstruction
```
