# Visual Guide to Calculus Integrals

## Project Overview
This project presents a parametric, high-fidelity animation of six fundamental types of integrals used in mathematics, physics, and engineering. Designed for educational outreach by Mugambi Ndwiga (@craftsandengineering), the visualization bridges the gap between abstract mathematical expressions and their physical interpretations.

## Types of Integrals

### 1. Single Variable Integral
*   **Mathematical Form**: $\int_{a}^{b} f(x) dx$
*   **Description**: Represents the signed area between a function and the x-axis over a specific interval.
*   **Application**: Calculating the total distance traveled by an object given its velocity over time.

### 2. Line Integral
*   **Mathematical Form**: $\int_{C} \mathbf{F} \cdot d\mathbf{r}$
*   **Description**: Accumulates a quantity along a specific path or curve within a scalar or vector field.
*   **Application**: Determining the work performed by a force field as an object moves along a trajectory.

### 3. Lebesgue Integral
*   **Mathematical Form**: $\int_{X} f \, d\mu$
*   **Description**: Generalizes the Riemann integral by partitioning the range (output) rather than the domain, allowing for the integration of highly discontinuous functions.
*   **Application**: Foundational in modern probability theory and digital signal processing.

### 4. Double Integral
*   **Mathematical Form**: $\iint_{R} f(x,y) dA$
*   **Description**: Extends integration to two dimensions, summing values over a surface area R.
*   **Application**: Calculating the volume under a 3D surface or finding the area of a non-standard 2D region.

### 5. Flux Integral
*   **Mathematical Form**: $\iint_{S} \mathbf{F} \cdot \mathbf{n} \, dS$
*   **Description**: Calculates the amount of a vector field passing through a given surface S.
*   **Application**: Measuring the rate of fluid flow through a pipe or the intensity of an electric field through a boundary.

### 6. Triple Integral
*   **Mathematical Form**: $\iiint_{V} \rho dV$
*   **Description**: Sums values throughout a three-dimensional volume V.
*   **Application**: Calculating the total mass or center of gravity of an object with varying density.

```python
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Description: Cinematic educational animation with high-contrast UI and color-coded math.
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from mpl_toolkits.mplot3d import Axes3D
import os
from IPython.display import HTML
from base64 import b64encode

# --- Parametric Setup ---
AUTHOR = "Mugambi Ndwiga"
INSTA = "@craftsandengineering"
FILENAME = "integral_cinematic.mp4"

def create_animation():
    # High DPI for crisp text and graphics
    fig = plt.figure(figsize=(16, 9), facecolor='#050505', dpi=100)

    data = [
        {"t": "SINGLE VARIABLE", "f": r"$\int_{a}^{b} f(x) dx$", "app": "Measure: Area under a trajectory", "c": "#00FBFF"},
        {"t": "LINE INTEGRAL", "f": r"$\int_{C} \mathbf{F} \cdot d\mathbf{r}$", "app": "Measure: Energy/Work along a path", "c": "#FFD700"},
        {"t": "LEBESGUE INTEGRAL", "f": r"$\int_{X} f \, d\mu$", "app": "Measure: Advanced probability sets", "c": "#FF00FF"},
        {"t": "DOUBLE INTEGRAL", "f": r"$\iint_{R} f(x,y) dA$", "app": "Measure: Volume of a landscape", "c": "#39FF14"},
        {"t": "FLUX INTEGRAL", "f": r"$\iint_{S} \mathbf{F} \cdot \mathbf{n} \, dS$", "app": "Measure: Fluid flow through a mesh", "c": "#FF3131"},
        {"t": "TRIPLE INTEGRAL", "f": r"$\iiint_{V} \rho dV$", "app": "Measure: Total mass of an object", "c": "#FF8C00"}
    ]

    def update(frame):
        plt.clf()
        stage = min(frame // 100, len(data) - 1)
        sub_frame = frame % 100
        info = data[stage]

        # 1. TOP HEADER
        fig.text(0.5, 0.92, "THE CALCULUS UNIVERSE: INTEGRALS", color='white', ha='center', fontsize=26, weight='bold', alpha=0.9)

        # 2. BRANDING (BOTTOM RIGHT)
        fig.text(0.98, 0.02, f"{AUTHOR} | {INSTA}", color='white', ha='right', fontsize=12, alpha=0.4)

        # 3. GLOSSARY/MATH CARD (LEFT)
        # Semi-transparent dark background for text
        fig.add_artist(plt.Rectangle((0.02, 0.1), 0.28, 0.75, transform=fig.transFigure, color='#111111', alpha=0.7, zorder=0))

        fig.text(0.05, 0.8, "TYPE:", color='gray', fontsize=12, weight='bold')
        fig.text(0.05, 0.76, info['t'], color=info['c'], fontsize=18, weight='black')

        fig.text(0.05, 0.65, "EXPRESSION:", color='gray', fontsize=12, weight='bold')
        fig.text(0.05, 0.58, info['f'], color='white', fontsize=28)

        fig.text(0.05, 0.45, "APPLICATION:", color='gray', fontsize=12, weight='bold')
        fig.text(0.05, 0.35, info['app'], color='white', fontsize=14, wrap=True)

        # 4. VISUALIZATION WINDOW (RIGHT)
        if stage == 0:
            ax = fig.add_axes([0.35, 0.15, 0.6, 0.7], facecolor='none')
            x = np.linspace(0, 5, 100); y = np.sin(x) + 2
            ax.plot(x, y, color=info['c'], lw=5)
            ax.fill_between(x[:sub_frame], 0, y[:sub_frame], color=info['c'], alpha=0.3)
            ax.set_ylim(0, 4); ax.axis('off')

        elif stage == 1:
            ax = fig.add_axes([0.35, 0.15, 0.6, 0.7], projection='3d', facecolor='none')
            t = np.linspace(0, (sub_frame/100)*6*np.pi, 300)
            ax.plot(np.cos(t), np.sin(t), t/6, color=info['c'], lw=4)
            ax.view_init(30, 45); ax.set_axis_off()

        elif stage == 2:
            ax = fig.add_axes([0.35, 0.15, 0.6, 0.7], facecolor='none')
            x = np.linspace(0, 4, 100); y = np.arctan(x*2)
            ax.plot(x, y, color=info['c'], lw=3)
            for i in range(1, int(sub_frame/8)+1):
                lvl = i * 0.15
                ax.axhline(lvl, color='white', alpha=0.2, ls='--')
            ax.axis('off')

        elif stage == 3:
            ax = fig.add_axes([0.35, 0.1, 0.6, 0.8], projection='3d', facecolor='none')
            X, Y = np.meshgrid(np.linspace(-2, 2, 25), np.linspace(-2, 2, 25))
            Z = np.sin(X) * np.cos(Y)
            ax.plot_surface(X, Y, Z, cmap='magma', alpha=0.8, antialiased=True)
            ax.view_init(30, sub_frame*3.6); ax.set_axis_off()

        elif stage == 4:
            ax = fig.add_axes([0.35, 0.1, 0.6, 0.8], projection='3d', facecolor='none')
            u, v = np.mgrid[0:2*np.pi:12j, 0:np.pi:10j]
            x, y, z = np.cos(u)*np.sin(v), np.sin(u)*np.sin(v), np.cos(v)
            ax.quiver(x, y, z, x*0.5, y*0.5, z*0.5, color=info['c'], alpha=0.9)
            ax.plot_wireframe(x, y, z, color='white', alpha=0.1)
            ax.view_init(20, sub_frame*3.6); ax.set_axis_off()

        else:
            ax = fig.add_axes([0.35, 0.1, 0.6, 0.8], projection='3d', facecolor='none')
            r = 0.1 + (sub_frame/100)*0.8
            u, v = np.mgrid[0:2*np.pi:18j, 0:np.pi:12j]
            ax.plot_surface(r*np.cos(u)*np.sin(v), r*np.sin(u)*np.sin(v), r*np.cos(v), color=info['c'], alpha=0.6)
            ax.set_xlim(-1, 1); ax.set_ylim(-1, 1); ax.set_zlim(-1, 1); ax.set_axis_off()

    ani = FuncAnimation(fig, update, frames=len(data)*100, interval=40)
    writer = FFMpegWriter(fps=25, bitrate=5000)
    print("Rendering Cinematic Version...")
    ani.save(FILENAME, writer=writer)
    plt.close()
    print("Render Complete.")

create_animation()

# Display
mp4 = open(FILENAME,'rb').read()
data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
HTML(f"<video width=900 controls autoplay loop><source src='{data_url}' type='video/mp4'></video>")
```
