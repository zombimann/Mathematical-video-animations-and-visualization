```python
%%manim -r 720,1280 -v WARNING ResonanceShort
"""
Author: Mugambi Ndwiga
Elevated to Gold-Standard Scientific Visual Art
Concept: Wave System Resonator (Phase Coherence & Boundary Harmony)

Optimized for mobile viewports with enhanced font sizes and a distinct
physical separation between Phase Velocity (Top) and Group Wave Packets (Bottom).
"""

from manim import *
import numpy as np

# -----------------------------------------------------------------
# Production Canvas Configuration (9:16 Vertical Video)
# -----------------------------------------------------------------
config.pixel_width = 720
config.pixel_height = 1280
config.frame_width = 9.0
config.frame_height = 16.0
config.frame_rate = 30
config.background_color = "#F5EFE4"

# Force standard Sans-Serif globally for Tex objects to eliminate Pango bugs
config.tex_template = TexTemplate()
config.tex_template.add_to_preamble(r"\usepackage{helvet}\renewcommand{\familydefault}{\sfdefault}")

PAPER = "#F5EFE4"
PAPER_CARD = "#FFF8EE"
INK = "#2E2925"
INK_SOFT = "#6C6259"
ACCENT_RED = "#B83A2C"  # Dispersive / Dissonant State
ACCENT_BLUE = "#2A6496" # Pure Phase / Consonant State

# -----------------------------------------------------------------
# Clean Layout Component Factories (Using Tex for pristine kerning)
# -----------------------------------------------------------------
def create_card(w, h=1.4, fill=PAPER_CARD, stroke=INK_SOFT, stroke_w=1.2, opacity=0.98):
    return RoundedRectangle(
        width=w, height=h, corner_radius=0.15,
        stroke_color=stroke, stroke_width=stroke_w,
        fill_color=fill, fill_opacity=opacity,
    )

def create_formula_chip(formula, caption, w, h=1.6):
    c = create_card(w, h, fill=PAPER_CARD, stroke="#A09182", stroke_w=1.2)
    m = MathTex(formula, font_size=24, color=INK)
    m.move_to(c.get_center() + UP * 0.18)
    cap = Tex(caption, font_size=15, color=INK_SOFT)
    cap.next_to(m, DOWN, buff=0.12)
    return VGroup(c, m, cap)

def create_story_paragraph(header_text, body_lines):
    """Constructs an enlarged text block for small screen scannability."""
    header = Tex(r"\textbf{" + header_text + "}", font_size=24, color=INK)
    lines_group = VGroup(header)
    for line in body_lines:
        l_tex = Tex(line, font_size=19, color=INK_SOFT)
        lines_group.add(l_tex)
    lines_group.arrange(DOWN, buff=0.14)
    return lines_group

class ResonanceShort(Scene):
    def construct(self):
        # -----------------------------------------------------------------
        # 1. Modular Vertical Layout Grid Setup
        # -----------------------------------------------------------------
        title = Tex(r"\textbf{Wave System Resonator}", font_size=32, color=INK)
        subtitle = Tex("How phase alignment controls system boundaries.", font_size=20, color=INK_SOFT)
        title.to_edge(UP, buff=0.8)
        subtitle.next_to(title, DOWN, buff=0.15)
        self.add(title, subtitle)

        # Content Region Anchors (Optimized Y-Axis Partitioning)
        Y_STAGE = 2.6
        Y_MATH  = -1.6
        Y_TEXT  = -5.2

        # Main Visualization Arena
        stage_box = create_card(8.4, 4.8, fill="#FCF6EA", stroke="#8A7C6F", stroke_w=1.5, opacity=0.95).move_to(UP * Y_STAGE)
        self.add(stage_box)

        # -----------------------------------------------------------------
        # 2. Permanent Formula Panel (Enlarged for Mobile)
        # -----------------------------------------------------------------
        math_strip = create_card(8.4, 2.2, fill=PAPER_CARD, stroke="#A09182", stroke_w=1.5).move_to(UP * Y_MATH)
        chip1 = create_formula_chip(r"\phi = kx - \omega t", "Wave Phase", 2.6, h=1.7)
        chip2 = create_formula_chip(r"v_p = \frac{\omega}{k}", "Phase Speed", 2.5, h=1.7)
        chip3 = create_formula_chip(r"v_g = \frac{d\omega}{dk}", "Group Speed", 2.5, h=1.7)
        chips = VGroup(chip1, chip2, chip3).arrange(RIGHT, buff=0.15).move_to(math_strip.get_center())

        self.add(math_strip, chips)

        # -----------------------------------------------------------------
        # 3. Physics Simulation Engine Parameters
        # -----------------------------------------------------------------
        time_track = ValueTracker(0.0)
        phase_diff = ValueTracker(-np.pi / 2) # Highly out of phase initially

        x_left, x_right = -3.2, 3.2
        y_top_row, y_bot_row = Y_STAGE + 1.1, Y_STAGE - 1.1

        k1, w1 = 1.5 * np.pi, 3.0
        k2, w2 = 1.0 * np.pi, 1.5
        base_amp = 0.22
        circle_radius = 0.35

        # Structural Isolation Boundaries
        left_wall = Rectangle(width=0.14, height=3.6, stroke_width=0, fill_color="#7A7065", fill_opacity=1.0).move_to(np.array([x_left, Y_STAGE, 0]))
        right_wall = Rectangle(width=0.14, height=3.6, stroke_width=0, fill_color="#7A7065", fill_opacity=1.0).move_to(np.array([x_right, Y_STAGE, 0]))
        self.add(left_wall, right_wall)

        # -----------------------------------------------------------------
        # 4. Dynamic Resonator System Definition (Permanent Differentiation)
        # -----------------------------------------------------------------
        num_circles = 7
        x_centers = np.linspace(x_left + 0.45, x_right - 0.45, num_circles)

        def generate_resonator_frame():
            t = time_track.get_value()
            d_phi = phase_diff.get_value()
            frame_group = VGroup()

            for row_idx, y_center in enumerate([y_top_row, y_bot_row]):
                if row_idx == 0:
                    # Top Row: Pure Monochromatic Carrier Wave (Phase Velocity Reference)
                    color_theme = ACCENT_BLUE
                    xs = np.linspace(x_left, x_right, 250)
                    ys_wave = y_center + base_amp * np.cos(k1 * xs - w1 * t)

                    for xc in x_centers:
                        ring = Circle(radius=circle_radius, stroke_color=INK_SOFT, stroke_width=0.8, stroke_opacity=0.15)
                        ring.move_to(np.array([xc, y_center, 0]))
                        frame_group.add(ring)

                        phi_phase = k1 * xc - w1 * t
                        dot_x = xc + circle_radius * np.cos(phi_phase)
                        dot_y = y_center + circle_radius * np.sin(phi_phase)
                        frame_group.add(Dot(point=np.array([dot_x, dot_y, 0]), radius=0.07, color=color_theme))
                else:
                    # Bottom Row: Dispersive Superposition Wave Packet (Group Velocity Engine)
                    color_theme = ACCENT_RED
                    amp_mod = base_amp * (0.50 + 0.50 * np.cos(d_phi))
                    xs = np.linspace(x_left, x_right, 250)
                    ys_wave = y_center + amp_mod * (np.cos(k1 * xs - w1 * t + d_phi) + np.cos(k2 * xs - w2 * t))

                    for xc in x_centers:
                        ring = Circle(radius=circle_radius, stroke_color=INK_SOFT, stroke_width=0.8, stroke_opacity=0.15)
                        ring.move_to(np.array([xc, y_center, 0]))
                        frame_group.add(ring)

                        phi_group = 0.5 * ((k1 + k2) * xc - (w1 + w2) * t + d_phi)
                        dot_x = xc + circle_radius * np.cos(phi_group)
                        dot_y = y_center + circle_radius * np.sin(phi_group)
                        frame_group.add(Dot(point=np.array([dot_x, dot_y, 0]), radius=0.07, color=color_theme))

                    # Wave Packet Envelope Line
                    ys_env = y_center + 2 * amp_mod * np.cos(0.5 * ((k1 - k2) * xs - (w1 - w2) * t + d_phi))
                    envelope_line = VMobject()
                    envelope_line.set_points_smoothly([np.array([x, y, 0]) for x, y in zip(xs, ys_env)])
                    envelope_line.set_stroke(INK_SOFT, width=1.2, opacity=0.40)
                    frame_group.add(DashedVMobject(envelope_line, num_dashes=35, equal_lengths=True))

                wave_line = VMobject()
                wave_line.set_points_smoothly([np.array([x, y, 0]) for x, y in zip(xs, ys_wave)])
                wave_line.set_stroke(color_theme, width=3.5, opacity=0.95)
                frame_group.add(wave_line)

            return frame_group

        active_simulation = always_redraw(generate_resonator_frame)
        self.add(active_simulation)

        # -----------------------------------------------------------------
        # 5. Velocity Vectors & Tracking Labels (Enlarged)
        # -----------------------------------------------------------------
        vp_arrow = always_redraw(lambda: Vector(RIGHT * 0.5, color=INK_SOFT, stroke_width=3.5).move_to(np.array([np.sin(time_track.get_value() * 0.4) * 0.6 - 0.4, y_top_row + 0.7, 0])))
        vp_lbl = always_redraw(lambda: MathTex("v_{p}", font_size=20, color=INK_SOFT).next_to(vp_arrow, UP, buff=0.06))
        lbl_sys1 = MathTex(r"\phi_1 = 0", font_size=20, color=ACCENT_BLUE).move_to(np.array([x_left + 0.8, y_top_row + 0.7, 0]))
        consonance_lbl = Tex(r"\textbf{PHASE MODE}", font_size=18, color=ACCENT_BLUE).move_to(np.array([x_right - 1.4, y_top_row + 0.7, 0]))

        vg_arrow = always_redraw(lambda: Vector(RIGHT * 0.6, color=ACCENT_RED, stroke_width=3.5).move_to(np.array([np.cos(time_track.get_value() * 0.5) * 0.6 + 0.4, y_bot_row - 0.7, 0])))
        vg_lbl = always_redraw(lambda: MathTex("v_{g}", font_size=20, color=ACCENT_RED).next_to(vg_arrow, DOWN, buff=0.06))

        lbl_sys2 = always_redraw(
            lambda: MathTex(
                r"\phi_2 = " + f"{phase_diff.get_value()/np.pi:.2f}" + r"\pi",
                font_size=18, color=ACCENT_RED
            ).move_to(np.array([x_left + 0.9, y_bot_row - 0.7, 0]))
        )

        dissonance_lbl = always_redraw(
            lambda: Tex(
                r"\textbf{DISPERSIVE PACKET}" if abs(phase_diff.get_value()) > 0.05 else r"\textbf{RESONANT PACKET}",
                font_size=16,
                color=ACCENT_RED if abs(phase_diff.get_value()) > 0.05 else ACCENT_BLUE
            ).move_to(np.array([x_right - 1.6, y_bot_row - 0.7, 0]))
        )

        self.add(vp_arrow, vp_lbl, lbl_sys1, consonance_lbl, vg_arrow, vg_lbl, lbl_sys2, dissonance_lbl)

        # -----------------------------------------------------------------
        # 6. Structured Storyboard Sequences & Fluid Pacing
        # -----------------------------------------------------------------
        story_1 = create_story_paragraph(
            "1. PHASE VS. GROUP VELOCITY",
            ["The top row tracks uniform monochromatic phase,", "while the bottom forms a complex dispersive packet."]
        ).move_to(np.array([0, Y_TEXT, 0]))

        story_2 = create_story_paragraph(
            "2. BOUNDARY PHASE MISMATCH",
            ["Phase misalignment causes destructive interference,", "suppressing energy flow across the packet boundary."]
        ).move_to(np.array([0, Y_TEXT, 0]))

        story_3 = create_story_paragraph(
            "3. RESONANT ENVELOPE LOCK",
            ["Modulating phase alignment updates boundary conditions,", "locking the packet into high-amplitude resonance."]
        ).move_to(np.array([0, Y_TEXT, 0]))

        # Phase 1: Dynamic Demonstration of Separated Systems
        self.play(FadeIn(story_1, shift=UP * 0.15), run_time=1.0)
        self.play(time_track.animate.set_value(5.0), run_time=5.0, rate_func=linear)

        # Phase 2: Highlight Dissonance / Boundary Constraint
        self.play(
            FadeOut(story_1, shift=DOWN * 0.15),
            FadeIn(story_2, shift=UP * 0.15),
            run_time=1.0
        )
        self.play(
            phase_diff.animate.set_value(-np.pi * 0.20),
            time_track.animate.set_value(11.0),
            run_time=5.5,
            rate_func=linear
        )

        # Phase 3: Transition into Coherence (Packet expands without copying top row)
        self.play(
            FadeOut(story_2, shift=DOWN * 0.15),
            FadeIn(story_3, shift=UP * 0.15),
            run_time=1.0
        )
        self.play(
            phase_diff.animate.set_value(0.0),
            time_track.animate.set_value(17.0),
            run_time=5.5,
            rate_func=smooth
        )

        # Perfect stable hold
        self.play(time_track.animate.set_value(20.0), run_time=3.5, rate_func=linear)
        self.play(FadeOut(story_3), run_time=0.6)

        # -----------------------------------------------------------------
        # 7. Clean Production Outro Card
        # -----------------------------------------------------------------
        self.play(
            *[FadeOut(obj) for obj in self.mobjects if obj != title],
            run_time=0.8
        )

        closing_block = VGroup(
            Tex(r"\textbf{Mugambi Ndwiga}", font_size=36, color=INK),
            Tex(r"@craftsandengineering", font_size=24, color=ACCENT_RED)
        ).arrange(DOWN, buff=0.25).move_to(ORIGIN)

        self.play(
            Transform(title, closing_block[0]),
            FadeIn(closing_block[1], shift=UP * 0.1),
            run_time=0.7
        )
        self.wait(2.0)
```

```python
%%manim -r 720,1280 -v WARNING ResonanceRigorous
"""
Author: Mugambi Ndwiga
Theme: Deeply Integrated Physics & Math Harmonizer
Concept: Direct Wave Superposition, Dispersion Mechanics & Phase Resonance

Optimized for 9:16 mobile viewports. Connects math formulas directly
to live-calculated physical wave mechanics using color-coded pointer arrows.
"""

from manim import *
import numpy as np

# -----------------------------------------------------------------
# Production Canvas Configuration (9:16 Vertical Video)
# -----------------------------------------------------------------
config.pixel_width = 720
config.pixel_height = 1280
config.frame_width = 9.0
config.frame_height = 16.0
config.frame_rate = 30
config.background_color = "#F5EFE4"

# Set global clean sans-serif layout for Tex objects
config.tex_template = TexTemplate()
config.tex_template.add_to_preamble(r"\usepackage{helvet}\renewcommand{\familydefault}{\sfdefault}")

# Premium Minimalist Color Palette
PAPER = "#F5EFE4"
PAPER_CARD = "#FFF8EE"
INK = "#2E2925"
INK_SOFT = "#6C6259"

# Color Codes for Explicit Mapping
COLOR_W1 = "#1A659E"     # Wave 1: Reference
COLOR_W2 = "#F26419"     # Wave 2: Phase-Shifted Component
COLOR_SUM = "#6A4C93"    # Resultant Superposition Wave
COLOR_VP = "#2A9D8F"     # Phase Velocity Tracking
COLOR_VG = "#E63946"     # Group Velocity Tracking

# -----------------------------------------------------------------
# Layout & UI Component Factories
# -----------------------------------------------------------------
def create_panel(w, h, fill=PAPER_CARD, stroke=INK_SOFT, stroke_w=1.2):
    return RoundedRectangle(
        width=w, height=h, corner_radius=0.12,
        stroke_color=stroke, stroke_width=stroke_w,
        fill_color=fill, fill_opacity=0.98,
    )

def create_formula_chip(formula, caption, color, w, h=1.5):
    c = create_panel(w, h, fill=PAPER_CARD, stroke=color, stroke_w=2.0)
    m = MathTex(formula, font_size=24, color=INK)
    m.move_to(c.get_center() + UP * 0.16)
    cap = Tex(caption, font_size=15, color=INK_SOFT)
    cap.next_to(m, DOWN, buff=0.10)
    return VGroup(c, m, cap)

def create_story_block(header_text, body_lines):
    header = Tex(r"\textbf{" + header_text + "}", font_size=22, color=INK)
    lines_group = VGroup(header)
    for line in body_lines:
        l_tex = Tex(line, font_size=17, color=INK_SOFT)
        lines_group.add(l_tex)
    lines_group.arrange(DOWN, buff=0.12)
    return lines_group

class ResonanceRigorous(Scene):
    def construct(self):
        # -----------------------------------------------------------------
        # 1. Structural Screen Partitioning (9:16 Optimized Layout)
        # -----------------------------------------------------------------
        title = Tex(r"\textbf{Wave Interference \& Resonance}", font_size=28, color=INK)
        subtitle = Tex("Linking mathematics directly to wave mechanics.", font_size=18, color=INK_SOFT)
        title.to_edge(UP, buff=0.6)
        subtitle.next_to(title, DOWN, buff=0.12)
        self.add(title, subtitle)

        # Dynamic Content Anchors
        Y_TRACK1 = 3.2    # Interaction Arena (Component Waves)
        Y_TRACK2 = -0.2   # Resultant & Envelope Packet Arena
        Y_MATH   = -3.4   # Formula Mapping Panel
        Y_TEXT   = -6.2   # Explanatory Storyboard

        # Draw Arena Enclosures
        panel1 = create_panel(8.4, 2.5, stroke="#A69B8F").move_to(UP * Y_TRACK1)
        panel2 = create_panel(8.4, 2.5, stroke="#A69B8F").move_to(UP * Y_TRACK2)
        lbl_p1 = Tex(r"\textbf{TRACK 1: Constituent Sources Interacting}", font_size=15, color=INK_SOFT).next_to(panel1, UP, buff=0.06)
        lbl_p2 = Tex(r"\textbf{TRACK 2: Superposition \& Envelope Packet}", font_size=15, color=INK_SOFT).next_to(panel2, UP, buff=0.06)
        self.add(panel1, panel2, lbl_p1, lbl_p2)

        # -----------------------------------------------------------------
        # 2. Physics Engine Setup (Rigorous Normal Dispersion)
        # -----------------------------------------------------------------
        time_track = ValueTracker(0.0)
        phase_diff = ValueTracker(-np.pi)  # Starts in perfect destructive dissonance (-pi)

        x_min, x_max = -3.6, 3.6
        amp = 0.22  # Individual amplitude inside rows

        # Dispersion parameters: v_p = 2/pi (~0.64), v_g = 1/pi (~0.32) -> Phase moves exactly 2x faster than group
        k1, w1 = 3.0 * np.pi, 5.0
        k2, w2 = 1.0 * np.pi, 3.0

        # -----------------------------------------------------------------
        # 3. Dynamic Wave Generators (Track 1 & Track 2)
        # -----------------------------------------------------------------
        def get_track1_components():
            t = time_track.get_value()
            phi = phase_diff.get_value()
            xs = np.linspace(x_min, x_max, 250)

            # Actual Wave 1 (Reference)
            y1 = Y_TRACK1 + amp * np.cos(k1 * xs - w1 * t)
            line1 = VMobject().set_points_smoothly([np.array([x, y, 0]) for x, y in zip(xs, y1)]).set_stroke(COLOR_W1, width=2.5)

            # Actual Wave 2 (Phase-Shifted)
            y2 = Y_TRACK1 + amp * np.cos(k2 * xs - w2 * t + phi)
            line2 = VMobject().set_points_smoothly([np.array([x, y, 0]) for x, y in zip(xs, y2)]).set_stroke(COLOR_W2, width=2.5)

            return VGroup(line1, line2)

        def get_track2_interference():
            t = time_track.get_value()
            phi = phase_diff.get_value()
            xs = np.linspace(x_min, x_max, 300)

            # Superposition Resultant Wave
            y_sum = Y_TRACK2 + amp * np.cos(k1 * xs - w1 * t) + amp * np.cos(k2 * xs - w2 * t + phi)
            sum_line = VMobject().set_points_smoothly([np.array([x, y, 0]) for x, y in zip(xs, y_sum)]).set_stroke(COLOR_SUM, width=3.5)

            # Analytical Envelope Guide
            y_env_top = Y_TRACK2 + 2 * amp * np.cos(0.5 * ((k1 - k2) * xs - (w1 - w2) * t - phi))
            y_env_bot = Y_TRACK2 - 2 * amp * np.cos(0.5 * ((k1 - k2) * xs - (w1 - w2) * t - phi))

            env_top = VMobject().set_points_smoothly([np.array([x, y, 0]) for x, y in zip(xs, y_env_top)]).set_stroke(COLOR_VG, width=1.2, opacity=0.4)
            env_bot = VMobject().set_points_smoothly([np.array([x, y, 0]) for x, y in zip(xs, y_env_bot)]).set_stroke(COLOR_VG, width=1.2, opacity=0.4)

            dashed_env = VGroup(
                DashedVMobject(env_top, num_dashes=35, equal_lengths=True),
                DashedVMobject(env_bot, num_dashes=35, equal_lengths=True)
            )
            return VGroup(sum_line, dashed_env)

        track1_live = always_redraw(get_track1_components)
        track2_live = always_redraw(get_track2_interference)
        self.add(track1_live, track2_live)

        # -----------------------------------------------------------------
        # 4. Rigorous Velocity Coordinate Functions & Trackers
        # -----------------------------------------------------------------
        def get_vp_pos():
            t = time_track.get_value()
            x_p = (2 * t / np.pi)
            x_p_wrapped = ((x_p - x_min) % (x_max - x_min)) + x_min
            y_val = Y_TRACK2 + amp * np.cos(k1 * x_p_wrapped - w1 * t) + amp * np.cos(k2 * x_p_wrapped - w2 * t + phase_diff.get_value())
            return np.array([x_p_wrapped, y_val, 0])

        def get_vg_pos():
            t = time_track.get_value()
            phi = phase_diff.get_value()
            x_g = ((t + phi) / np.pi)
            x_g_wrapped = ((x_g - x_min) % (x_max - x_min)) + x_min
            return np.array([x_g_wrapped, Y_TRACK2 + 0.6, 0])

        # Render moving indicators on top of the waves
        def get_vg_indicator():
            pos = get_vg_pos()
            bracket = VGroup(
                Line(pos + LEFT * 0.4, pos + RIGHT * 0.4),
                Line(pos + LEFT * 0.4, pos + LEFT * 0.4 + DOWN * 0.2),
                Line(pos + RIGHT * 0.4, pos + RIGHT * 0.4 + DOWN * 0.2)
            ).set_stroke(COLOR_VG, width=2.5)
            lbl = MathTex("v_g", font_size=18, color=COLOR_VG).next_to(bracket, UP, buff=0.04)
            return VGroup(bracket, lbl)

        def get_vp_indicator():
            pos = get_vp_pos()
            dot = Dot(point=pos, radius=0.08, color=COLOR_VP)
            lbl = MathTex("v_p", font_size=18, color=COLOR_VP).next_to(dot, DOWN, buff=0.08)
            return VGroup(dot, lbl)

        vg_track = always_redraw(get_vg_indicator)
        vp_track = always_redraw(get_vp_indicator)
        self.add(vg_track, vp_track)

        # -----------------------------------------------------------------
        # 5. Mathematical Panel Setup
        # -----------------------------------------------------------------
        math_box = create_panel(8.4, 2.0, stroke="#A69B8F").move_to(UP * Y_MATH)
        chip_phi = create_formula_chip(r"\Delta \phi = \pm \pi", "Phase Sync", COLOR_W2, 2.5, h=1.4)
        chip_vp  = create_formula_chip(r"v_p = \frac{\omega_p}{k_p}", "Crest Speed", COLOR_VP, 2.6, h=1.4)
        chip_vg  = create_formula_chip(r"v_g = \frac{\Delta\omega}{\Delta k}", "Group Speed", COLOR_VG, 2.5, h=1.4)

        chips_group = VGroup(chip_phi, chip_vp, chip_vg).arrange(RIGHT, buff=0.15).move_to(math_box.get_center())
        self.add(math_box, chips_group)

        # Labels for Source Waves on Track 1
        lbl_w1 = Tex("Source 1", font_size=16, color=COLOR_W1).move_to(np.array([x_min + 0.6, Y_TRACK1 + 0.8, 0]))
        lbl_w2 = Tex("Source 2", font_size=16, color=COLOR_W2).move_to(np.array([x_max - 0.6, Y_TRACK1 + 0.8, 0]))
        self.add(lbl_w1, lbl_w2)

        # Status Overlay Indicators
        status_dissonance = Tex(r"\textbf{STATE: DESTRUCTIVE DISSONANCE}", font_size=16, color=COLOR_W2).move_to(np.array([0, Y_TRACK2 - 0.9, 0]))
        status_resonance = Tex(r"\textbf{STATE: PERFECT HARMONIC LOCK}", font_size=16, color=COLOR_VP).move_to(np.array([0, Y_TRACK2 - 0.9, 0]))
        self.add(status_dissonance)

        # -----------------------------------------------------------------
        # 6. Dynamic Redraw Pointers (Solving the "Pointing to Nothing" bug)
        # -----------------------------------------------------------------
        pointer_phi = Arrow(start=chip_phi.get_top(), end=lbl_w2.get_bottom(), buff=0.12, color=COLOR_W2, stroke_width=3)

        pointer_vp = always_redraw(lambda: Arrow(
            start=chip_vp.get_top(),
            end=get_vp_pos(),
            buff=0.15,
            color=COLOR_VP,
            stroke_width=3,
            max_tip_length_to_length_ratio=0.15
        ))

        pointer_vg = always_redraw(lambda: Arrow(
            start=chip_vg.get_top(),
            end=get_vg_pos() + DOWN * 0.1,
            buff=0.15,
            color=COLOR_VG,
            stroke_width=3,
            max_tip_length_to_length_ratio=0.15
        ))

        # -----------------------------------------------------------------
        # 7. Instructional Storyboard Execution
        # -----------------------------------------------------------------
        story_1 = create_story_block(r"1. INDEPENDENT WAVE SOURCES", [r"Track 1 shows Source 1 (Blue) \& Source 2 (Orange) arriving.", r"Because they are $\Delta \phi = \pi$ out of phase, they actively oppose each other."]).move_to(UP * Y_TEXT)
        story_2 = create_story_block(r"2. DISPERSION \& PACKET SPEEDS", [r"Track 2 shows their combined physical field interference.", r"Notice individual ripples ($v_p$) travelling 2x faster than the packet envelope ($v_g$)."]).move_to(UP * Y_TEXT)
        story_3 = create_story_block(r"3. DYNAMIC RESONANCE LOCK", [r"Watch Phase Sync smoothly shift to zero ($\Delta \phi \to 0$).", r"The physical waves align perfectly, locking into maximum resonant energy."]).move_to(UP * Y_TEXT)

        # Phase 1: Setup and Wave Identity
        self.play(FadeIn(story_1, shift=UP * 0.15), run_time=1.0)

        # Point directly from Phase Chip to Source 2
        self.play(Create(pointer_phi), run_time=0.8)
        self.play(time_track.animate.set_value(2.5), run_time=4.5, rate_func=linear)
        self.play(FadeOut(pointer_phi), run_time=0.5)

        # Phase 2: Demonstrate Velocity Concept with Dynamic Pointers
        self.play(
            FadeOut(story_1, shift=DOWN * 0.15),
            FadeIn(story_2, shift=UP * 0.15),
            run_time=1.0
        )

        # Pointers continuously lock onto the moving targets
        self.play(FadeIn(pointer_vp), FadeIn(pointer_vg), run_time=0.6)
        self.play(time_track.animate.set_value(6.5), run_time=5.5, rate_func=linear)
        self.play(FadeOut(pointer_vp), FadeOut(pointer_vg), run_time=0.5)

        # Phase 3: Dynamic Tuning into Resonance Lock
        self.play(
            FadeOut(story_2, shift=DOWN * 0.15),
            FadeIn(story_3, shift=UP * 0.15),
            run_time=1.0
        )

        # Modulate phase discrepancy to zero while running time forward
        self.play(
            phase_diff.animate.set_value(0.0),
            time_track.animate.set_value(11.5),
            FadeOut(status_dissonance),
            FadeIn(status_resonance),
            run_time=6.0,
            rate_func=smooth
        )

        # Hold simulation in full resonance state
        self.play(time_track.animate.set_value(15.0), run_time=3.5, rate_func=linear)
        self.play(FadeOut(story_3), FadeOut(status_resonance), run_time=0.6)

        # -----------------------------------------------------------------
        # 8. Production Outro
        # -----------------------------------------------------------------
        self.play(
            *[FadeOut(obj) for obj in self.mobjects if obj != title],
            run_time=0.8
        )

        closing_block = VGroup(
            Tex(r"\textbf{Mugambi Ndwiga}", font_size=34, color=INK),
            Tex(r"@craftsandengineering", font_size=22, color=COLOR_W2)
        ).arrange(DOWN, buff=0.25).move_to(ORIGIN)

        self.play(
            Transform(title, closing_block[0]),
            FadeIn(closing_block[1], shift=UP * 0.15),
            run_time=0.8
        )
        self.wait(2.0)
```
