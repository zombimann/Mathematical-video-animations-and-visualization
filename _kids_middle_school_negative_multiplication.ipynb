# Educator and Parent Guide: Negative Multiplication Visualized

This notebook generates a high-quality educational animation explaining why negative multiplied by negative equals positive.

### Video Overview
The animation uses a visual "number line reversal" method, which is often more intuitive for middle school students than memorizing rules.

### Key Concepts Covered
1.  **Directionality**: Positive numbers move right; multiplying by -1 is a "180-degree flip" to the left.
2.  **Step-by-Step Logic**:
    *   **3 multiplied by 1**: Shows a standard positive vector.
    *   **3 multiplied by -1**: Demonstrates the first flip (results in -3).
    *   **-3 multiplied by -2**: The "Secret" step—starting at -3 (facing left) and multiplying by another negative flips the direction again, pointing back to the positive side.
3.  **The Rule Summary**: Explicitly states the sign patterns for classroom reinforcement.

```python
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Level: Middle School
Concept: Negative × negative via number line reversal
Github: github.com/zombimann/Mathematical-video-animations-and-visualization
"""

from manim import *
import os

class MathAnimation(Scene):
    def construct(self):
        C = {
            "bg": "#FFFFFF",
            "grid": "#E0E0E0",
            "pos": "#86A788", # Sage
            "neg": "#D62828", # Red
            "highlight": "#E9C46A", # Mustard
            "navy": "#2A3D66",
            "watermark": "© Mugambi Ndwiga / @craftsandengineering"
        }
        self.camera.background_color = C["bg"]

        # --- Setup UI Elements ---
        wm = Text(C["watermark"], font="sans-serif", font_size=16, color=BLACK).set_opacity(0.6).to_corner(DR, buff=0.4)
        # Increased width from 4 to 5 to fit text properly
        label_bg = RoundedRectangle(corner_radius=0.1, height=0.6, width=5, fill_color=C["navy"], fill_opacity=0.8, stroke_width=0)
        label_t = Text("For Kids: Middle School", font="sans-serif", font_size=24, color=WHITE)
        header = VGroup(label_bg, label_t).to_corner(UR, buff=0.5)
        self.add(wm, header)

        # --- Number Line ---
        nl = NumberLine(x_range=[-8, 8, 1], length=12, color=BLACK, include_numbers=True, font_size=32, label_constructor=Text).shift(UP*0.5)
        self.add(nl)

        # --- 1. Example: 3 x 1 ---
        title1 = Text("Step 1: 3 × 1 = 3", font_size=42, color=C["navy"]).to_edge(UP, buff=1.2)
        self.play(Write(title1))
        v1 = Arrow(nl.n2p(0), nl.n2p(3), buff=0, color=C["pos"], stroke_width=8)
        self.play(GrowArrow(v1))
        self.wait(1.5)
        self.play(FadeOut(title1), FadeOut(v1))

        # --- 2. Example: 3 x -1 ---
        title2 = Text("Step 2: 3 × (-1) = -3", font_size=42, color=C["navy"]).to_edge(UP, buff=1.2)
        v2 = Arrow(nl.n2p(0), nl.n2p(3), buff=0, color=C["pos"], stroke_width=8)
        self.play(Write(title2), GrowArrow(v2))
        self.wait(0.5)

        flip_note = Text("Negative flips direction!", font_size=32, color=C["neg"]).shift(DOWN*2)
        self.play(Write(flip_note))
        self.play(Rotate(v2, angle=PI, about_point=nl.n2p(0)), v2.animate.set_color(C["neg"]))
        self.wait(1.5)
        self.play(FadeOut(title2), FadeOut(v2), FadeOut(flip_note))

        # --- 3. The 'Why': -3 x -2 ---
        title3 = Text("The Secret: (-3) × (-2)", font_size=42, color=C["navy"]).to_edge(UP, buff=1.2)
        v3 = Arrow(nl.n2p(0), nl.n2p(-3), buff=0, color=C["neg"], stroke_width=8)
        self.play(Write(title3), GrowArrow(v3))
        self.wait(0.5)

        logic1 = Text("• Start at -3", font_size=32, color=BLACK).to_edge(LEFT, buff=1).shift(DOWN*1.5)
        logic2 = Text("• Multiply by -2: Flip and Double", font_size=32, color=BLACK).next_to(logic1, DOWN, aligned_edge=LEFT)
        self.play(FadeIn(logic1))
        self.wait(0.5)
        self.play(FadeIn(logic2))

        self.play(
            Rotate(v3, angle=PI, about_point=nl.n2p(0)),
            v3.animate.put_start_and_end_on(nl.n2p(0), nl.n2p(6)).set_color(C["pos"]),
            run_time=2
        )

        summary_logic = Text("Two flips = Forward!", font_size=38, color=C["navy"], weight=BOLD).shift(DOWN*0.5)
        self.play(Write(summary_logic))
        self.wait(2)
        self.play(FadeOut(VGroup(title3, v3, logic1, logic2, summary_logic, nl)))

        # --- 4. Final Rule Card ---
        rule_bg = Rectangle(width=10, height=5, fill_color=C["navy"], fill_opacity=0.9)
        line1 = Text("THE RULE", font_size=48, color=C["highlight"])
        line2 = Text("Positive × Negative = Negative", font_size=36, color=WHITE).next_to(line1, DOWN, buff=0.5)
        line3 = Text("Negative × Negative = Positive", font_size=36, color=WHITE).next_to(line2, DOWN, buff=0.3)
        self.play(FadeIn(rule_bg), Write(line1))
        self.play(Write(line2))
        self.play(Write(line3))
        self.wait(3)
        self.play(FadeOut(VGroup(rule_bg, line1, line2, line3)))

        # --- Closing Card ---
        final_bg = Rectangle(width=15, height=10, fill_color=C["pos"], fill_opacity=1)
        final_text = Text("Made by Mugambi Ndwiga\n@craftsandengineering", font_size=36, color=WHITE)
        self.play(FadeIn(final_bg))
        self.play(Write(final_text))
        self.wait(2)

# Render Command
%manim -v WARNING -r 854,480 --progress_bar None MathAnimation
```
