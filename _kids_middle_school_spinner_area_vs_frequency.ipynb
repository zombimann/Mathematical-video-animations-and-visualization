# Educational Simulation: Probability as Area vs. Frequency

### Overview
This notebook contains a complete Python script to generate a high-definition, portrait-oriented (9:16) educational animation. The simulation is designed for Middle School and High School students to visualize the Law of Large Numbers through a spinner experiment.

### Learning Objectives
1. **Theoretical Probability**: Understanding that the area of a sector (1/3 for blue, 2/3 for red) represents the 'true chance' of an outcome.
2. **Experimental Frequency**: Observing how the results of a small number of trials can be 'wobbly' or inconsistent.
3. **The Law of Large Numbers**: Demonstrating that as the number of trials increases (from 1 to 1000), the experimental frequency converges toward the theoretical probability.

### Simulation Structure
- **Scene 1: Concept Introduction**: Explains the relationship between circle area and probability.
- **Scene 2: Small Sample Size**: Shows 10 manual spins and the resulting variance in frequency.
- **Scene 3: Medium Sample Size**: Rapidly executes 100 spins to show the initial settling of the data.
- **Scene 4: Large Sample Size**: Uses a grid of 40 simultaneous spinners to reach 1,000 trials, showing high precision relative to the 1/3 target.
- **Scene 5: Conclusion**: Summarizes the 'Big Idea' that frequency tracks area over time.

### Technical Requirements
- **Libraries**: NumPy (data), Matplotlib (visualization), and ImageIO (FFMpeg handling).
- **Output**: A 30-second MP4 video (1080x1920) optimized for mobile devices and social media platforms.

```python
"""
Author: Mugambi Ndwiga
Instagram: @craftsandengineering
Level: Elementary | Middle School | High School
Concept: Probability as spinner area vs frequency
"""

import os
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from matplotlib.patches import Wedge, Circle, Rectangle, Polygon, FancyBboxPatch, FancyArrow
from matplotlib.ticker import MaxNLocator

try:
    import imageio_ffmpeg
    matplotlib.rcParams["animation.ffmpeg_path"] = imageio_ffmpeg.get_ffmpeg_exe()
except Exception:
    pass

# -----------------------------------------------------------------------------
# 1) PARAMETERS  (vertical 9:16 for phones / YouTube Shorts)
# -----------------------------------------------------------------------------
W_PX, H_PX   = 1080, 1920
DPI          = 120                 # 9 x 16 inches -> 1080 x 1920 px
FPS          = 24
BITRATE      = 2200
OUT_FILE     = "spinner_area_vs_frequency.mp4"

P_BLUE       = 1.0 / 3.0
BLUE_DEG     = 360.0 * P_BLUE      # blue sector = 120 degrees = 1 of 3 equal parts
MAX_SPINS    = 1000
SEED         = 9                   # 10->0.20, 100->0.31, 1000->0.333 (exact)

DUR          = [6, 6, 6, 8, 4]     # five scenes, 30 s total

WHITE   = "#FFFFFF"
GRID    = "#ECEFF1"
NAVY    = "#2A3D66"
SAGE    = "#86A788"
CORAL   = "#E5989B"
MUSTARD = "#E9C46A"
RED     = "#D62828"
INK     = "#333333"
BLUE    = NAVY
PANEL_BG= "#F5F7F5"

# Request a font only if it is actually installed (avoids findfont warnings).
from matplotlib import font_manager as _fm
_AVAIL = {f.name for f in _fm.fontManager.ttflist}
plt.rcParams["font.family"] = [f for f in ("Inter", "Roboto", "DejaVu Sans")
                               if f in _AVAIL] or ["DejaVu Sans"]

# Type scale (portrait => big, legible on a phone)
TITLE_FS  = 40
CAPTION_FS= 34
SUB_FS    = 22
BIG_FS    = 40
MID_FS    = 26
SMALL_FS  = 16
TINY_FS   = 14

CXC = 45.0           # horizontal centre of the whole canvas (0..90)

# -----------------------------------------------------------------------------
# 2) SIMULATION  (the OUTCOME is the sector the wheel lands on)
# -----------------------------------------------------------------------------
rng       = np.random.default_rng(SEED)
land_phi  = rng.uniform(0.0, 360.0, MAX_SPINS)   # wheel-local angle under pointer
is_blue   = land_phi < BLUE_DEG
idx       = np.arange(1, MAX_SPINS + 1)
cum_blue  = np.cumsum(is_blue)
cum_freq  = cum_blue / idx

# Pointer is fixed at the top (screen angle 90 deg). To put wheel-local angle phi
# under the pointer the wheel must rotate so that (rot mod 360) == 90 - phi.
START_ROT = 90.0

def ease_out(p):                                 # decelerating flick (stops dead)
    p = min(max(p, 0.0), 1.0)
    return 1.0 - (1.0 - p) ** 3

def build_schedule(phi_slice, spin_start, last_land, min_turns, start_rot=START_ROT):
    """Land times (evenly spaced) plus the wheel rotation that lands each spin."""
    n = len(phi_slice)
    land_times = spin_start + np.arange(1, n + 1) * ((last_land - spin_start) / n)
    rot, prev = np.empty(n), start_rot
    for i in range(n):
        target = (90.0 - phi_slice[i]) % 360.0
        base = prev + min_turns * 360.0          # guarantee several full turns
        rot[i] = base + ((target - base) % 360.0)
        prev = rot[i]
    return land_times, rot

def spin_state(land_times, rot, spin_start, t, start_rot=START_ROT):
    """(rotation_deg, completed_spins) at local time t. Count advances only AFTER
    a spin has fully landed -- exactly like a real wheel."""
    n = len(land_times)
    k = int(np.searchsorted(land_times, t, side="right"))
    if k >= n:
        return float(rot[-1]), n
    r0 = float(rot[k - 1]) if k > 0 else start_rot
    t0 = float(land_times[k - 1]) if k > 0 else spin_start
    span = float(land_times[k]) - t0
    p = 0.0 if span <= 0 else (t - t0) / span
    return r0 + (rot[k] - r0) * ease_out(p), k

def build_rotations_var(phi_slice, turns_arr, start_rot=START_ROT):
    """Landing rotations when each spin gets its own number of full turns
    (so spins can accelerate while every one still lands on its true sector)."""
    n = len(phi_slice)
    rot, prev = np.empty(n), start_rot
    for i in range(n):
        target = (90.0 - phi_slice[i]) % 360.0
        base = prev + turns_arr[i] * 360.0
        rot[i] = base + ((target - base) % 360.0)
        prev = rot[i]
    return rot

# -- Scene 1 & 2: every spin lands one-by-one (real stop, then +1) -------------
S1_PHI          = np.array([232.0])                       # lands in the red sector
S1_LAND, S1_ROT = build_schedule(S1_PHI, 2.0, 3.6, min_turns=4)
S2_LAND, S2_ROT = build_schedule(land_phi[:10], 0.5, 5.0, min_turns=2)

# -- Scene 3: 100 REAL spins, sped up. The gap between stops shrinks, so the
# wheel spins clearly at first then in a fast flurry -- but each spin still
# decelerates to a dead stop and is counted (+1) ONLY at that stop. ------------
S3_N     = 100
_gap     = 0.9 ** np.arange(S3_N)                         # shrinking gaps
S3_LAND  = 0.30 + np.cumsum(_gap / _gap.sum() * 4.9)      # last stop ~5.2 s
S3_TURNS = np.maximum(1, np.round(4.0 * _gap)).astype(int)  # 4 turns -> 1 turn
S3_ROT   = build_rotations_var(land_phi[:S3_N], S3_TURNS)

# -- Scene 4: 1000 is too many to stop one-at-a-time, so spin MANY wheels at
# once. 40 wheels x 25 synchronised rounds = 1000. Each round the whole grid
# decelerates and stops together, and the count jumps (+40) ONLY at that stop --
# still 'count only after it has stopped', just 40 stops at the same instant. --
S4_COLS, S4_ROWS = 5, 8
S4_G     = S4_COLS * S4_ROWS                              # 40 wheels per round
S4_R     = MAX_SPINS // S4_G                              # 25 rounds -> 1000
S4_REST  = ((90.0 - land_phi[:S4_G * S4_R]) % 360.0).reshape(S4_R, S4_G)
S4_BLUEG = is_blue[:S4_G * S4_R].reshape(S4_R, S4_G)
S4_START = 0.5                                            # first round begins
S4_RDUR  = 0.235                                          # seconds per round (~5.9 s)
S4_TURNS = 6                                              # turns each wheel unwinds

def count_at(land_times, t):
    return int(min(np.searchsorted(land_times, t, side="right"), len(land_times)))

# -----------------------------------------------------------------------------
# 3) FIGURE & PERSISTENT ARTISTS
# -----------------------------------------------------------------------------
fig = plt.figure(figsize=(W_PX / DPI, H_PX / DPI), dpi=DPI)
fig.patch.set_facecolor(WHITE)

ax_main  = fig.add_axes([0, 0, 1, 1])
CHART_BOX = [0.12, 0.10, 0.78, 0.17]      # bottom strip (figure fraction)
ax_chart  = fig.add_axes(CHART_BOX)

for_kids = fig.text(0.5, 0.972, "For Kids: Middle School",
                    fontsize=18, color=WHITE, ha="center", va="center",
                    weight="bold",
                    bbox=dict(boxstyle="round,pad=0.5", fc=NAVY, ec="none"))
watermark = fig.text(0.5, 0.012, "(c) Mugambi Ndwiga / @craftsandengineering",
                     fontsize=13, color="#888888", alpha=0.6,
                     ha="center", va="bottom")

# -----------------------------------------------------------------------------
# 4) HELPERS
# -----------------------------------------------------------------------------
frames_per = [int(d * FPS) for d in DUR]
starts     = np.cumsum([0] + frames_per)
TOTAL      = int(starts[-1])

def locate(f):
    for i in range(5):
        if f < starts[i + 1]:
            return i + 1, (f - starts[i]) / FPS
    return 5, (f - starts[4]) / FPS

def setup_main(scene):
    ax_main.clear()
    ax_main.set_xlim(0, 90)
    ax_main.set_ylim(0, 160)
    ax_main.axis("off")
    if scene == 5:
        ax_main.add_patch(Rectangle((0, 0), 90, 160, color=NAVY, zorder=0))
    else:
        ax_main.set_facecolor(WHITE)
        for gx in range(0, 91, 10):
            ax_main.axvline(gx, color=GRID, lw=0.8, zorder=0)
        for gy in range(0, 161, 10):
            ax_main.axhline(gy, color=GRID, lw=0.8, zorder=0)

def draw_spinner(cx, cy, r, rot=0.0, highlight=None, thirds=False):
    """Wheel. highlight in {None,'blue','red'} rings the winning sector.
    thirds=True draws the equal-part dividers so kids SEE three equal slices."""
    ax_main.add_patch(Circle((cx + 0.7, cy - 0.9), r, fc="#D9D9D9",
                             ec="none", zorder=3))                 # soft shadow
    if highlight:
        ax_main.add_patch(Circle((cx, cy), r * 1.06, fill=False,
                                 ec=MUSTARD, lw=4.0, alpha=0.95, zorder=3.5))
    b_lw = 6.0 if highlight == "blue" else 3.0
    r_lw = 6.0 if highlight == "red" else 3.0
    b_ec = MUSTARD if highlight == "blue" else WHITE
    r_ec = MUSTARD if highlight == "red" else WHITE
    ax_main.add_patch(Wedge((cx, cy), r, rot, rot + 120, fc=BLUE,
                            ec=b_ec, lw=b_lw, zorder=4))
    ax_main.add_patch(Wedge((cx, cy), r, rot + 120, rot + 360, fc=RED,
                            ec=r_ec, lw=r_lw, zorder=4))
    if thirds:                                                    # 3 equal slices
        for a in (rot, rot + 120, rot + 240):
            rad = np.radians(a)
            ax_main.plot([cx, cx + r * np.cos(rad)], [cy, cy + r * np.sin(rad)],
                         color=WHITE, lw=2.0, zorder=4.5)
    ax_main.add_patch(Circle((cx, cy), r, fill=False, ec=NAVY, lw=3.0, zorder=5))
    ax_main.add_patch(Circle((cx, cy), r * 0.07, fc=NAVY, ec=WHITE,
                             lw=2.0, zorder=6))
    tip = r * 0.16
    ax_main.add_patch(Polygon([(cx - tip, cy + r + tip * 1.5),
                               (cx + tip, cy + r + tip * 1.5),
                               (cx,       cy + r - tip * 0.6)],
                              fc=MUSTARD, ec=NAVY, lw=2.0, zorder=7))

def draw_counters(n):
    """Big count + frequency, centred."""
    k = int(cum_blue[n - 1]) if n > 0 else 0
    f = cum_freq[n - 1] if n > 0 else 0.0
    ax_main.text(CXC, 83, f"Blue landed: {k}", fontsize=BIG_FS - 4, color=NAVY,
                 weight="bold", ha="center", va="center", zorder=6)
    ax_main.text(CXC, 76, f"Frequency = {f:0.3f}", fontsize=MID_FS, color=RED,
                 weight="bold", ha="center", va="center", zorder=6)

def draw_meter(n, scale):
    """Number line from 0..scale showing the TRUE chance (1/3) and where the
    spins currently sit -- the clearest 'are we close yet?' cue for kids."""
    f = cum_freq[n - 1] if n > 0 else 0.0
    x0, x1, y = 14.0, 76.0, 67.0
    ax_main.plot([x0, x1], [y, y], color=INK, lw=2.5, zorder=4)
    for v in (0.0, scale):                                        # end ticks
        xx = x0 + (x1 - x0) * (v / scale)
        ax_main.plot([xx, xx], [y - 1, y + 1], color=INK, lw=2.0, zorder=4)
    xt = x0 + (x1 - x0) * (P_BLUE / scale)                        # true-chance tick
    ax_main.plot([xt, xt], [y - 2.4, y + 2.4], color=SAGE, lw=3.5, zorder=5)
    ax_main.text(xt, y + 4.2, "true chance 1/3", fontsize=SMALL_FS, color=SAGE,
                 weight="bold", ha="center", va="bottom", zorder=6)
    xf = x0 + (x1 - x0) * (min(f, scale) / scale)                 # observed marker
    ax_main.add_patch(Polygon([(xf, y), (xf - 1.6, y - 4.2), (xf + 1.6, y - 4.2)],
                              fc=RED, ec=WHITE, lw=1.2, zorder=6))
    ax_main.text(xf, y - 5.2, f"{f:0.3f}", fontsize=SMALL_FS, color=RED,
                 weight="bold", ha="center", va="top", zorder=6)

def draw_chart(n, xmax, ymax):
    ax_chart.set_visible(True)
    ax_chart.clear()
    ax_chart.set_facecolor("none")
    if n >= 2:
        ax_chart.plot(idx[:n], cum_freq[:n], color=NAVY, lw=3.0, zorder=3)
        ax_chart.plot(n, cum_freq[n - 1], "o", color=RED, ms=8, zorder=4)
    elif n == 1:
        ax_chart.plot(1, cum_freq[0], "o", color=RED, ms=8, zorder=4)
    ax_chart.axhline(P_BLUE, color=SAGE, lw=2.6, ls="--", zorder=2)
    ax_chart.text(xmax * 0.98, P_BLUE + ymax * 0.04, "target 1/3 = 0.333",
                  color=SAGE, fontsize=SMALL_FS, ha="right", weight="bold")
    ax_chart.set_xlim(1, xmax)
    ax_chart.set_ylim(0, ymax)
    ax_chart.set_yticks([0, P_BLUE, ymax])
    ax_chart.set_yticklabels(["0", "1/3", f"{ymax:g}"], fontsize=SMALL_FS)
    ax_chart.xaxis.set_major_locator(MaxNLocator(integer=True, nbins=5))
    ax_chart.tick_params(labelsize=SMALL_FS, colors=INK)
    for s in ("top", "right"):
        ax_chart.spines[s].set_visible(False)
    for s in ("left", "bottom"):
        ax_chart.spines[s].set_color(INK)
    ax_chart.set_xlabel("Number of spins", fontsize=SMALL_FS + 1, color=INK)
    ax_chart.set_ylabel("Blue frequency", fontsize=SMALL_FS + 1, color=INK)
    arrow = dict(arrowstyle="->", lw=1.6, color=INK)
    ax_chart.annotate("", xy=(1.03, 0), xytext=(0, 0),
                      xycoords="axes fraction", arrowprops=arrow)
    ax_chart.annotate("", xy=(0, 1.08), xytext=(0, 0),
                      xycoords="axes fraction", arrowprops=arrow)

def header(caption, sub):
    ax_main.text(CXC, 148, caption, fontsize=CAPTION_FS, color=NAVY,
                 weight="bold", ha="center", va="center")
    if sub:
        ax_main.text(CXC, 141, sub, fontsize=SUB_FS, color=SAGE,
                     weight="bold", ha="center", va="center")

def transition(scene, lt):
    T = 0.35
    if scene > 1 and lt < T:
        col = NAVY if scene == 5 else WHITE
        ax_main.add_patch(Rectangle((0, 0), 90, 160, color=col,
                                    alpha=1 - lt / T, zorder=50))

# -----------------------------------------------------------------------------
# 5) SCENES
# -----------------------------------------------------------------------------
def scene1(lt):
    """Teach the idea: area = chance. Static divided wheel, then one real spin."""
    rot, done = spin_state(S1_LAND, S1_ROT, 2.0, lt)
    landed = done >= 1
    teaching = lt < 2.0                                   # static teaching phase
    hl = "red" if landed else None
    draw_spinner(CXC, 99, 26, rot, highlight=hl, thirds=True)
    ax_main.text(CXC, 149, "Spinner: area = chance", fontsize=TITLE_FS - 4,
                 color=NAVY, weight="bold", ha="center", va="center")
    ax_main.text(CXC, 140, "How much space a colour fills\n= how likely it is",
                 fontsize=SUB_FS, color=INK, ha="center", va="center",
                 linespacing=1.25)
    if teaching:
        a = float(np.clip(lt / 0.6, 0, 1))
        ax_main.text(CXC, 65, "The circle has 3 equal parts",
                     fontsize=MID_FS, color=INK, ha="center", weight="bold",
                     alpha=a)
        ax_main.text(CXC, 58, "1 part blue,  2 parts red",
                     fontsize=SUB_FS, color=INK, ha="center", alpha=a)
    if landed:
        rb = float(np.clip((lt - S1_LAND[-1]) / 0.4, 0, 1))
        ax_main.text(CXC, 65, "This spin landed on RED", fontsize=MID_FS,
                     color=RED, weight="bold", ha="center", alpha=rb)
    a = float(np.clip((lt - 4.0) / 0.6, 0, 1))           # fractions + hook
    if a > 0:
        ax_main.text(CXC - 16, 48, "Blue = 1/3", fontsize=BIG_FS - 6, color=BLUE,
                     weight="bold", ha="center", alpha=a)
        ax_main.text(CXC + 16, 48, "Red = 2/3", fontsize=BIG_FS - 6, color=RED,
                     weight="bold", ha="center", alpha=a)
        ax_main.add_patch(FancyBboxPatch((9, 20), 72, 18,
                          boxstyle="round,pad=0.6,rounding_size=2.5",
                          fc=MUSTARD, ec=NAVY, lw=2.0, zorder=8, alpha=a))
        ax_main.text(CXC, 29, "Spin many times - will Blue\ncome up about 1/3?",
                     fontsize=MID_FS, color=NAVY, weight="bold",
                     ha="center", va="center", linespacing=1.3,
                     zorder=9, alpha=a)

def scene2(lt):
    """10 real spins, each lands then counts. Small sample = wobble."""
    rot, n = spin_state(S2_LAND, S2_ROT, 0.5, lt)
    flash = None
    if n > 0 and (lt - float(S2_LAND[n - 1])) < 0.30:
        flash = "blue" if is_blue[n - 1] else "red"
    draw_spinner(CXC, 113, 18, rot, highlight=flash)
    header("Try 10 spins", "Few spins = wobbly results")
    status = "Spinning..." if n == 0 else f"Spin {n} of 10"
    ax_main.text(CXC, 91, status, fontsize=SUB_FS, color=INK, ha="center",
                 style=("italic" if n == 0 else "normal"))
    draw_counters(n)
    draw_meter(n, 1.0)
    # outcome tiles, one per completed spin
    ntiles, w, gap = 10, 5.5, 1.4
    total = ntiles * w + (ntiles - 1) * gap
    x0, yt = CXC - total / 2, 48
    ax_main.text(CXC, 54, "Each spin:  blue or red", fontsize=SMALL_FS + 1,
                 color=INK, ha="center", va="bottom")
    for i in range(n):
        ax_main.add_patch(Rectangle((x0 + i * (w + gap), yt), w, 5,
                          fc=(BLUE if is_blue[i] else RED), ec=WHITE,
                          lw=1.5, zorder=4))
    draw_chart(n, 10, 1.0)

def scene3(lt):
    """100 REAL spins. Each one decelerates to a dead stop, then +1. The gaps
    between stops shrink, so it speeds into a flurry -- but the count still only
    moves when a spin has actually stopped."""
    rot, n = spin_state(S3_LAND, S3_ROT, 0.30, lt)
    flash = None
    if 0 < n <= S3_N and (lt - float(S3_LAND[n - 1])) < 0.10:
        flash = "blue" if is_blue[n - 1] else "red"
    draw_spinner(CXC, 113, 18, rot, highlight=flash)
    header("Now 100 spins", "Faster now - each still stops")
    status = "Spinning..." if n == 0 else f"Spin {n} of 100"
    ax_main.text(CXC, 91, status, fontsize=SUB_FS, color=INK, ha="center",
                 style=("italic" if n == 0 else "normal"))
    draw_counters(n)
    draw_meter(n, 0.6)
    ax_main.text(CXC, 51, "Wobble shrinks - it settles toward 1/3",
                 fontsize=SUB_FS - 2, color=INK, ha="center", style="italic")
    draw_chart(n, 100, 0.6)

_GX = np.linspace(16.0, 74.0, S4_COLS)
_GY = np.linspace(133.0, 88.0, S4_ROWS)
_GR = 2.9

def draw_wheel_grid(cr, spin_amt, settling):
    """Draw the 40 wheels of round cr. They share one rotation offset spin_amt
    (so the whole grid spins and stops together). When settling, blue-result
    wheels get a mustard ring so the blue outcomes pop out."""
    j = 0
    for yy in _GY:
        for xx in _GX:
            rot = S4_REST[cr, j] + spin_amt
            if settling and S4_BLUEG[cr, j]:
                ax_main.add_patch(Circle((xx, yy), _GR * 1.18, fill=False,
                                         ec=MUSTARD, lw=1.6, zorder=4.5))
            ax_main.add_patch(Wedge((xx, yy), _GR, rot, rot + 120, fc=BLUE,
                                    ec=WHITE, lw=0.6, zorder=5))
            ax_main.add_patch(Wedge((xx, yy), _GR, rot + 120, rot + 360, fc=RED,
                                    ec=WHITE, lw=0.6, zorder=5))
            ax_main.add_patch(Circle((xx, yy), _GR, fill=False, ec=NAVY,
                                     lw=1.0, zorder=6))
            ax_main.add_patch(Polygon([(xx - 0.7, yy + _GR + 1.0),
                                       (xx + 0.7, yy + _GR + 1.0),
                                       (xx,       yy + _GR - 0.3)],
                                      fc=MUSTARD, ec=NAVY, lw=0.5, zorder=7))
            j += 1

def scene4(lt):
    """1000 spins, the elegant way: 40 wheels spin at once, 25 synchronised
    rounds (40x25 = 1000). The count jumps +40 ONLY when a whole round has
    stopped -- still 'count only after it stops', just 40 stops at once."""
    rt = lt - S4_START
    rd = int(min(max(rt, 0.0) // S4_RDUR, S4_R))          # rounds completed
    n  = rd * S4_G                                        # +40 per stopped round
    cr = min(rd, S4_R - 1)                                # round on screen now
    if rd >= S4_R:
        p = 1.0
    else:
        p = float(np.clip((rt - rd * S4_RDUR) / S4_RDUR, 0.0, 1.0))
    spin_amt = S4_TURNS * 360.0 * (1.0 - ease_out(p))
    settling = (p > 0.78)
    draw_wheel_grid(cr, spin_amt, settling)
    header("1000 spins!", "Spin 40 wheels at once")
    k = int(cum_blue[n - 1]) if n > 0 else 0
    ax_main.text(CXC, 80, f"Spins {n}   Blue {k}", fontsize=SUB_FS, color=NAVY,
                 weight="bold", ha="center", va="center")
    draw_meter(n, 0.6)
    a = float(np.clip((rt - S4_R * S4_RDUR) / 0.5, 0, 1))  # reveal after last stop
    if a > 0 and n > 0:
        ax_main.add_patch(FancyBboxPatch((8, 47), 74, 9,
                          boxstyle="round,pad=0.5,rounding_size=2",
                          fc=MUSTARD, ec=NAVY, lw=2.0, zorder=8, alpha=a))
        ax_main.text(CXC, 51.5, f"Frequency {cum_freq[n-1]:0.3f}  ~  1/3 = 0.333",
                     fontsize=SUB_FS + 1, color=NAVY, weight="bold",
                     ha="center", va="center", zorder=9, alpha=a)
    draw_chart(n, 1000, 0.6)

def scene5(lt):
    """Recap the big idea, then credits."""
    a = float(np.clip(lt / 0.5, 0, 1))
    ax_main.text(CXC, 138, "THE BIG IDEA", fontsize=TITLE_FS - 4, color=MUSTARD,
                 weight="bold", ha="center", va="center", alpha=a)
    draw_idea_card(a)
    ax_main.text(CXC, 78,
                 "More spins ->\nfrequency gets closer to the\ntrue chance (the area).",
                 fontsize=MID_FS, color=WHITE, ha="center", va="center",
                 linespacing=1.35, alpha=a)
    ax_main.text(CXC, 40, "Made by Mugambi Ndwiga", fontsize=MID_FS + 2,
                 color=WHITE, weight="bold", ha="center", va="center", alpha=a)
    ax_main.text(CXC, 32, "@craftsandengineering", fontsize=SUB_FS, color=SAGE,
                 ha="center", va="center", alpha=a)

def draw_idea_card(a):
    cx, cy, r = 26, 108, 13                               # mini spinner: 1/3 area
    ax_main.add_patch(Wedge((cx, cy), r, 90, 210, fc=BLUE, ec=WHITE, lw=2,
                            zorder=4, alpha=a))
    ax_main.add_patch(Wedge((cx, cy), r, 210, 450, fc=RED, ec=WHITE, lw=2,
                            zorder=4, alpha=a))
    ax_main.text(cx, cy - r - 5, "area = 1/3", fontsize=SUB_FS, color=WHITE,
                 ha="center", weight="bold", alpha=a)
    ax_main.add_patch(FancyArrow(40, 108, 12, 0, width=1.2, head_width=4,
                                 head_length=3, fc=MUSTARD, ec="none",
                                 zorder=5, alpha=a))
    ax_main.text(66, 113, "0.333", fontsize=BIG_FS - 6, color=WHITE,
                 weight="bold", ha="center", alpha=a)
    ax_main.text(66, 103, "frequency", fontsize=SUB_FS, color=SAGE,
                 ha="center", weight="bold", alpha=a)

# -----------------------------------------------------------------------------
# 6) FRAME DISPATCH
# -----------------------------------------------------------------------------
def update(f):
    scene, lt = locate(f)
    for_kids.set_visible(scene != 5)
    watermark.set_visible(True)
    setup_main(scene)
    ax_chart.set_visible(False)

    if   scene == 1: scene1(lt)
    elif scene == 2: scene2(lt)
    elif scene == 3: scene3(lt)
    elif scene == 4: scene4(lt)
    else:            scene5(lt)

    transition(scene, lt)
    return []

# -----------------------------------------------------------------------------
# 7) RENDER
# -----------------------------------------------------------------------------
ani = FuncAnimation(fig, update, frames=TOTAL, interval=1000 / FPS, blit=False)
writer = FFMpegWriter(fps=FPS, bitrate=BITRATE, codec="libx264",
                      extra_args=["-pix_fmt", "yuv420p"])
ani.save(OUT_FILE, writer=writer, dpi=DPI)
plt.close(fig)
print(f"Saved {OUT_FILE} ({os.path.getsize(OUT_FILE)/1e6:0.2f} MB, "
      f"{TOTAL} frames, {TOTAL/FPS:0.1f} s, {W_PX}x{H_PX})")

# -----------------------------------------------------------------------------
# 8) SHOW & DOWNLOAD  (Colab helpers)
# -----------------------------------------------------------------------------
def show_video(path=OUT_FILE, width=360):
    """Play the video inline in the Colab notebook."""
    import base64
    from IPython.display import HTML, display
    with open(path, "rb") as fh:
        b64 = base64.b64encode(fh.read()).decode()
    display(HTML(f'''
        <video width="{width}" controls loop autoplay muted playsinline
               style="border-radius:12px;box-shadow:0 4px 18px rgba(0,0,0,.25)">
          <source src="data:video/mp4;base64,{b64}" type="video/mp4">
        </video>'''))

def download_video(path=OUT_FILE):
    """Trigger a browser download of the mp4 (Colab only)."""
    try:
        from google.colab import files
        files.download(path)
    except Exception as e:
        print("Download helper is for Google Colab. File is at:", path, "|", e)

show_video()        # preview inline
# download_video()  # <- uncomment to save the mp4 to your computer
```
