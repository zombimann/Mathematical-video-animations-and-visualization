# README: The River of Radiance: Visualizing How Light Carries Energy


**Overview:**
This notebook creates a mesmerizing 3D animation of a light wave. It demonstrates a fundamental concept: light is not just a vibration, but a 'river' of energy moving through space.

**Visual Key:**
*   **Electric Field:** Oscillates side-to-side. Color shifts from dark blue to bright cyan based on its strength.
*   **Magnetic Field:** Oscillates up-and-down. Color shifts from deep red to bright yellow based on its strength.
*   **Energy Flow (Poynting Vector):** Flows forward along the path of the wave. Represented by thick golden arrows indicating the direction and magnitude of energy transfer.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from matplotlib.colors import Normalize

def create_public_em_animation():
    """
    Generates a 3D animation of a light wave with clear, non-technical labels.
    - Electric Field: Blue/Cyan gradient
    - Magnetic Field: Orange/Red gradient
    - Energy Flow: Glowing Gold arrows
    """
    plt.style.use('dark_background')
    fig = plt.figure(figsize=(14, 9))
    ax = fig.add_subplot(111, projection='3d')

    # Parameters
    z = np.linspace(0, 4 * np.pi, 25)
    Electric_Max, Magnetic_Max = 1.0, 1.0
    norm = Normalize(vmin=0, vmax=1)

    # Distinct colormaps
    cmap_Electric = plt.cm.cool
    cmap_Magnetic = plt.cm.YlOrRd

    def update(frame):
        ax.clear()
        ax.set_facecolor('black')
        time = frame * 0.1

        # Field calculations
        Electric_X = Electric_Max * np.sin(z - time)
        Magnetic_Y = Magnetic_Max * np.sin(z - time)
        Energy_Z = Electric_X * Magnetic_Y
        zeros = np.zeros_like(z)

        # 1. Electric Field - Side-to-side movement
        ax.quiver(zeros, zeros, z, Electric_X, zeros, zeros,
                  color=cmap_Electric(norm(np.abs(Electric_X))), label='Electric Field', alpha=0.8, arrow_length_ratio=0.2)

        # 2. Magnetic Field - Up-and-down movement
        ax.quiver(zeros, zeros, z, zeros, Magnetic_Y, zeros,
                  color=cmap_Magnetic(norm(np.abs(Magnetic_Y))), label='Magnetic Field', alpha=0.8, arrow_length_ratio=0.2)

        # 3. Energy Flow - Moving forward
        Forward_Flow = np.maximum(Energy_Z, 0)
        ax.quiver(zeros, zeros, z, zeros, zeros, Forward_Flow * 2,
                  color='gold', label='Energy Flow', lw=3, alpha=0.9)

        # Aesthetics
        ax.set_xlim(-1.5, 1.5)
        ax.set_ylim(-1.5, 1.5)
        ax.set_zlim(0, 4 * np.pi)
        ax.set_title("The River of Radiance: Energy Flowing Through Space", fontsize=18, color='gold', pad=30)
        ax.set_axis_off()

        # Rotation for 3D effect
        ax.view_init(elev=20, azim=45 + frame * 0.7)

        if frame == 0:
            ax.legend(loc='upper left', bbox_to_anchor=(0, 1.1), frameon=False)
            sm_electric = plt.cm.ScalarMappable(cmap=cmap_Electric, norm=norm)
            sm_magnetic = plt.cm.ScalarMappable(cmap=cmap_Magnetic, norm=norm)
            cax1 = fig.add_axes([0.85, 0.2, 0.02, 0.2])
            cax2 = fig.add_axes([0.92, 0.2, 0.02, 0.2])
            fig.colorbar(sm_electric, cax=cax1, label='Electric Strength')
            fig.colorbar(sm_magnetic, cax=cax2, label='Magnetic Strength')

    anim = FuncAnimation(fig, update, frames=100, interval=50)
    plt.close()
    return anim

wave_anim = create_public_em_animation()
HTML(wave_anim.to_html5_video())
```
