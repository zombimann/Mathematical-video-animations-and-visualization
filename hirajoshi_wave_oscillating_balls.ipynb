# Hirajoshi Wave Animation

This notebook generates a visual representation of a polyrhythmic pendulum wave.

### The Mechanics
* **Scale**: The frequencies are mapped to the Japanese **Hirajoshi scale**.
* **Synchronization**: 15 distinct 'balls' oscillate at mathematically related frequencies such that they all realign perfectly at a set interval.
* **Visuals**: Uses `matplotlib.animation` to render the motion in an ultra-slow, meditative pace.

### Navigation
1. Run the **Configuration & Animation** cell to view the interactive JSHTML player.
2. Run the **Export** cell at the bottom to generate and download an MP4 video file.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import matplotlib

"""
Hirajoshi Wave: A Polyrhythmic Pendulum Animation

This script creates a 2D animation of 15 points oscillating on a Hirajoshi-scale
frequency distribution. The parameters are tuned for 'Hyper Slow' motion,
ensuring realignment every 4 seconds.
"""

# Increase the animation embed limit for high-quality notebook rendering
matplotlib.rcParams['animation.embed_limit'] = 100.0

# --- Configuration ---
NUM_BALLS = 15
DURATION_PER_CYCLE = 4  # The interval (seconds) where all balls sync up
NUM_CYCLES = 3          # Total number of realignment events to show
DURATION = DURATION_PER_CYCLE * NUM_CYCLES
FPS = 30
TOTAL_FRAMES = DURATION * FPS

# Calculate frequencies based on the realignment interval
# BASE_CYCLES_PER_REALIGNMENT controls the 'drift' speed
BASE_CYCLES_PER_REALIGNMENT = 0.171875
frequencies = [(BASE_CYCLES_PER_REALIGNMENT + i) / DURATION_PER_CYCLE for i in range(NUM_BALLS)]

# --- Plot Setup ---
fig, ax = plt.subplots(figsize=(12, 7), facecolor='black')
ax.set_facecolor('black')
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1, NUM_BALLS + 1)
ax.axis('off')

lines = []
points = []
colors = plt.cm.viridis(np.linspace(0.2, 1.0, NUM_BALLS))

# Initialize lines (strings) and points (pendulum bobs)
for i in range(NUM_BALLS):
    line, = ax.plot([], [], color=colors[i], alpha=0.2, lw=0.5)
    point, = ax.plot([], [], 'o', color=colors[i], markersize=10,
                     markeredgecolor='white', markeredgewidth=0.5)
    lines.append(line)
    points.append(point)

ax.text(0.5, 1.05, f"Hirajoshi Wave: {DURATION_PER_CYCLE}s Cycle (Hyper Slow Speed)",
        transform=ax.transAxes, color='white', fontsize=16,
        ha='center', fontweight='light')

def init():
    """Reset line and point data for animation start."""
    for line, point in zip(lines, points):
        line.set_data([], [])
        point.set_data([], [])
    return lines + points

def update(frame):
    """Update the position of each ball based on its unique frequency."""
    t = (frame / TOTAL_FRAMES) * DURATION
    for i in range(NUM_BALLS):
        # x position follows a sine wave: x = sin(2 * pi * f * t)
        x = np.sin(2 * np.pi * frequencies[i] * t)
        y = i
        lines[i].set_data([0, x], [NUM_BALLS, y])
        points[i].set_data([x], [y])
    return lines + points

# Create animation object
ani = FuncAnimation(fig, update, frames=TOTAL_FRAMES, init_func=init,
                    blit=True, interval=1000/FPS)

plt.close()
HTML(ani.to_jshtml())
```

```python
from google.colab import files

# Note: This requires 'ffmpeg' to be installed (standard in Colab environments)
print("Saving animation as MP4...")
filename = "hirajoshi_wave_slow.mp4"

# Use the FFMpegWriter to save the animation
ani.save(filename, writer='ffmpeg', fps=FPS, dpi=150)

print(f"Done! Downloading {filename}...")
files.download(filename)
```
