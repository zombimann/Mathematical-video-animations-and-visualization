# 🌌 Mesmerizing Polar Curve Symphony

### Overview
This notebook is a generative art project that transforms mathematical equations into a high-definition (1080p) visual experience. By plotting diverse polar curves simultaneously in a synchronized grid, it creates a rhythmic 'dance' of geometry and light.

### Visual Design
- **Neon Aesthetic**: Each curve is rendered using a 'Neon Stack'—a high-intensity core line layered over a soft, translucent glow to simulate light emission.
- **Cinematic Palette**: A curated set of vibrant colors (Cyan, Magenta, Lime, Gold, etc.) is assigned to different curve families to provide high contrast against the deep black background.
- **Atmospheric Shimmer**: A background particle field pulses with randomized alpha transparency, creating a sense of depth and cosmic atmosphere.

### Mathematical Gallery
The symphony features 12 distinct curves, including:
- **Rose Curves ($r = \sin(n\theta)$)**: Elegant symmetric patterns.
- **Cardioids & Limaçons**: Heart-shaped and looped trajectories.
- **Spirals**: Archimedean and Fermat variants that expand infinitely.
- **Specialty Curves**: The complex Butterfly curve and the infinite Lituus.

### Technical Specifications
- **Engine**: Matplotlib Polar Projection.
- **Resolution**: 1920x1080 (Full HD).
- **Frame Rate**: 30 FPS.
- **Encoding**: H.264 MP4 via FFMpegWriter with a 3000kbps bitrate.


```python
"""
Core animation engine for Polar Curve Symphony.
Utilizes Matplotlib's Polar projection and FFMpegWriter for high-bitrate video encoding.
"""
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FFMpegWriter
import matplotlib.patches as patches
from IPython.display import HTML
import base64

# Configuration
fps = 30
duration = 18
total_frames = fps * duration
n_rows, n_cols = 3, 4
n_curves = n_rows * n_cols
plt.rcParams['axes.facecolor'] = 'black'
plt.rcParams['savefig.facecolor'] = 'black'

def get_curve(idx, theta):
    """Return (radius, name, color) for a specific index in the grid."""
    if idx == 0: return np.sin(3*theta), "3-Petal Rose", "cyan"
    if idx == 1: return 1 - np.cos(theta), "Cardioid", "magenta"
    if idx == 2: return 0.5 + np.cos(theta), "Limacon", "lime"
    if idx == 3: return 0.2 * theta, "Archimedean Spiral", "orange"
    if idx == 4: return np.sqrt(np.abs(np.cos(2*theta))), "Lemniscate", "#FF00FF"
    if idx == 5: return np.exp(np.cos(theta)) - 2*np.cos(4*theta) + np.sin(theta/12)**5, "Butterfly", "#FF1493"
    if idx == 6: return np.sin(2*theta), "4-Petal Rose", "#00BFFF"
    if idx == 7: return 1.5 + np.sin(theta), "Convex Limacon", "#ADFF2F"
    if idx == 8: return np.cos(5*theta), "5-Petal Rose", "#FFD700"
    if idx == 9: return theta**0.5, "Fermat's Spiral", "#DA70D6"
    if idx == 10: return 1 / (theta + 0.1), "Lituus", "#FF4500"
    if idx == 11: return np.sin(1.2*theta), "Rose-like Curve", "#7FFFD4"

# Plot Initialization
fig, axes = plt.subplots(n_rows, n_cols, figsize=(19.2, 10.8), subplot_kw={'projection': 'polar'})
fig.patch.set_facecolor('black')
axes = axes.flatten()

lines, glows, points = [], [], []
particles_list = []

for i in range(n_curves):
    ax = axes[i]
    ax.set_theta_zero_location("N")
    ax.set_yticklabels([]); ax.set_xticklabels([])
    ax.grid(False); ax.spines['polar'].set_visible(False)

    _, name, color = get_curve(i, 0)

    # Neon Stack: Glow layer + Sharp core layer
    glow, = ax.plot([], [], color=color, lw=8, alpha=0.15, zorder=3)
    line, = ax.plot([], [], color=color, lw=2.5, alpha=0.8, zorder=5)
    # Tracer point
    point, = ax.plot([], [], color='white', marker='o', markersize=6,
                     markeredgecolor=color, markeredgewidth=2, zorder=10)

    lines.append(line); glows.append(glow); points.append(point)
    ax.text(0.5, -0.1, name, color='white', transform=ax.transAxes,
            ha='center', fontsize=11, family='monospace', alpha=0.7)

    # Initialize random starfield shimmer
    p_scat = ax.scatter(np.random.rand(15)*2*np.pi, np.random.rand(15)*2, s=np.random.rand(15)*5, color='white', alpha=0.1)
    particles_list.append(p_scat)

plt.tight_layout(pad=4.0)

def animate(frame):
    """Frame update function for drawing curves and pulsing particles."""
    progress = frame / total_frames
    theta_max = 6 * np.pi * progress

    for i in range(n_curves):
        theta = np.linspace(0, theta_max, 1000)
        r, _, _ = get_curve(i, theta)

        lines[i].set_data(theta, r)
        glows[i].set_data(theta, r)

        if len(r) > 0:
            points[i].set_data([theta[-1]], [r[-1]])

        # Background shimmer intensity pulse
        new_alpha = 0.05 + 0.15 * np.abs(np.sin(frame/15 + i))
        particles_list[i].set_alpha(new_alpha)

    return lines + glows + points

# Rendering Pipeline
writer = FFMpegWriter(fps=fps, metadata=dict(artist='Colab Composer'), bitrate=3000)
output_file = "mesmerizing_polar_curves.mp4"
with writer.saving(fig, output_file, 100):
    for f in range(total_frames):
        animate(f)
        writer.grab_frame()
plt.close()

# Embedded Display
video_file = open(output_file, "rb").read()
video_url = f"data:video/mp4;base64,{base64.b64encode(video_file).decode()}"
display(HTML(f'''<video width="960" height="540" controls autoplay loop><source src="{video_url}" type="video/mp4"></video>'''))
```
