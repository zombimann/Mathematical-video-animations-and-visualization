```python
"""
Trigonometric Synchrony Visualization
------------------------------------
This script generates a 20-second animation comparing four unique polar coordinate
functions (Rose, Spiral, Cardioid, and Lemniscate) with a singular Cartesian
driver. The parameter 'k' modulates over time to create a 'blooming' effect,
while the phase shift ensures continuous motion. All systems are synchronized
to a shared harmonic pulse shown on the right axis.
"""
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Timing and Resolution
FPS, SECONDS = 30, 20
TOTAL_FRAMES = FPS * SECONDS
theta = np.linspace(0, 2 * np.pi, 1000)

# Design the Layout: 4 Polar (Left 2x2), 1 Singular Cartesian (Right)
fig = plt.figure(figsize=(16, 9), facecolor='black')
fig.suptitle("Harmonic Resonance: Polar-Cartesian Synchrony", color='white', fontsize=18, y=0.98)
gs = fig.add_gridspec(2, 3, wspace=0.3, hspace=0.3)

axes_polar = [
    fig.add_subplot(gs[0, 0], projection='polar', facecolor='#050505'),
    fig.add_subplot(gs[0, 1], projection='polar', facecolor='#050505'),
    fig.add_subplot(gs[1, 0], projection='polar', facecolor='#050505'),
    fig.add_subplot(gs[1, 1], projection='polar', facecolor='#050505')
]

# Singular Cartesian plot representing the driver
ax_cart = fig.add_subplot(gs[:, 2], facecolor='#050505')

lines_polar = [ax.plot([], [], lw=2)[0] for ax in axes_polar]
line_cart, = ax_cart.plot([], [], lw=2.5, color='#FFA500', alpha=0.9)

titles = ["Rose Curve", "Archimedean Spiral", "Cardioid", "Lemniscate"]
telemetry = fig.text(0.5, 0.02, '', color='#FF8800', fontfamily='monospace', fontsize=9, ha='center')

for i, ax in enumerate(axes_polar):
    ax.set_ylim(0, 2.5)
    ax.set_title(titles[i], color='#FFA500', fontsize=10, pad=10)
    ax.grid(True, color='#222222', alpha=0.3)
    ax.set_xticklabels([])
    ax.set_yticklabels([])

ax_cart.set_xlim(0, 2 * np.pi)
ax_cart.set_ylim(-1.5, 1.5)
ax_cart.set_title("Core Harmonic Driver", color='white', pad=15)
ax_cart.tick_params(colors='#555555', labelsize=8)
ax_cart.grid(True, color='#151515')

def get_data(frame):
    # k and phase drive all systems together
    k = 2 + 1.5 * np.sin(frame / 40)
    phase = frame / 20

    r1 = np.cos(k * (theta + phase))
    r2 = 0.2 * (1.2 + 0.4 * np.sin(frame/30)) * theta
    r3 = (1.1 + 0.2 * np.cos(frame/40)) * (1 + np.cos(theta + phase))
    r4 = np.sqrt(np.abs(2.2 * np.cos(2 * theta + phase)))

    # The single representative sine wave
    y_cart = np.sin(k * (theta + phase))

    return [r1, r2, r3, r4], y_cart, k

def update(frame):
    datasets, y_cart, k = get_data(frame)
    base_hue = 0.4 + 0.3 * np.sin(frame/100)

    for i in range(4):
        color = plt.cm.copper(base_hue + (i * 0.1))
        lines_polar[i].set_data(theta, datasets[i])
        lines_polar[i].set_color(color)

    line_cart.set_data(theta, y_cart)
    line_cart.set_color(plt.cm.copper(base_hue + 0.2))

    telemetry.set_text(f"SYSTEM SYNC: OK | FREQ-K: {k:.3f} | T: {frame/FPS:.1f}s")
    return lines_polar + [line_cart, telemetry]

ani = FuncAnimation(fig, update, frames=TOTAL_FRAMES, blit=True, interval=1000/FPS)
plt.tight_layout(rect=[0, 0.05, 1, 0.93])
plt.close()
HTML(ani.to_html5_video())
```
